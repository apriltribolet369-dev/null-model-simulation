```python
import h5py
import pandas as pd

data_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

def decode_categories(group):
  categories = group["categories"][:]
  return [
      x.decode("utf-8") if isinstance(x, bytes) else str(x)
      for x in categories
  ]

with h5py.File(data_path, "r") as f:
  obs = f["obs"]

  print("BROAD CELL TYPE COUNTS")

  group = obs["cell_type.broad"]
  labels = decode_categories(group)
  codes = group["codes"][:]
  counts = pd.Series(codes).value_counts().sort_index()

  for code, count in counts.items():
    if code == -1:
      print(f"Missing/unassigned: {count:,}")
    elif 0 <= code < len(labels):
      print(f"{labels[code]}: {count:,}")

  print("\nDETAILED CELL TYPES OF INTEREST")

  group = obs["cell_type.detailed"]
  labels = decode_categories(group)
  codes = group["codes"][:]
  counts = pd.Series(codes).value_counts().sort_index()

  for code, count in counts.items():
    if 0 <= code < len(labels):
      label = labels[code]
      if any(term in label.lower() for term in
             ["mast", "schwann", "fib", "peri"]):
        print(f"{label}: {count:,}")

  print("\nANATOMIC SITE COUNTS")

  group = obs["anatomic_site"]
  labels = decode_categories(group)
  codes = group["codes"][:]
  counts = pd.Series(codes).value_counts().sort_index()

  for code, count in counts.items():
    if code == -1:
      print(f"Missing/unassigned: {count:,}")
    elif 0 <= code < len(labels):
      print(f"{labels[code]}: {count:,}")
```

```python
import h5py
import numpy as np

data_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

def decode_value(value):
  if isinstance(value, bytes):
    return value.decode("utf-8")
  return str(value)

with h5py.File(data_path, "r") as f:
  obs = f["obs"]

  for name in ["donor_id", "sample_id", "sample_barcode", "EntityOD"]:
    print(f"\n--- {name} ---")
    group = obs[name]

    if isinstance(group, h5py.Group) and "categories" in group and "codes" in group:
      categories = [decode_value(x) for x in group ["categories"][:]]
      codes = group["codes"][:]

      valid_codes = codes[codes >= 0]
      used_codes = np.unique(valid_codes)

      print("Unique categories defined:", len(categories))
      print("Categories actually used:", len(used_codes))
      print("Missing/unassigned cells:", int(np.sum(codes < 0)))
      print("Example categories:")
      for code in used_codes[:12]:
          print(" ", categories[int(code)])
    else:
      print("Storage format:", type(group).__name__)
```

```python
import h5py
import pandas as pd

data_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

def decode_categories(group):
  return [
      x.decode("utf-8") if isinstance(x, bytes) else str(x)
      for x in group["categories"][:]
  ]

with h5py.File(data_path, "r") as f:
  obs = f["obs"]

  donor_group = obs["donor_id"]
  donor_labels = decode_categories(donor_group)
  donor_codes = donor_group["codes"][:]

  sample_group = obs["sample_id"]
  sample_labels = decode_categories(sample_group)
  sample_codes = sample_group["codes"][:]

  pairs = pd.DataFrame ({
      "donor_code": donor_codes,
      "sample_code": sample_codes
  }).drop_duplicates()

  pairs = pairs[
      (pairs["donor_code"] >= 0) &
      (pairs["sample_code"] >= 0)
  ]

  pairs["donor"] = pairs["donor_code"].map(
      lambda x: donor_labels[int(x)]
  )
  pairs["sample"] = pairs["sample_code"].map(
      lambda x: sample_labels[int(x)]
  )

  print("Samples per donor:")
  print(
      pairs.groupby("donor")["sample"]
      .nunique()
      .sort_values(ascending=False)
      .to_string()
  )

  print("\nTotal unique donor-sample pairs:", len(pairs))
```

```python
import h5py
import pandas as pd

data_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

def decode_categories(group):
  return [
      x.decode("utf-8") if isinstance(x, bytes) else str(x)
      for x in group["categories"][:]
  ]

with h5py.File(data_path, "r") as f:
  obs = f["obs"]

  sample_group = obs["sample_id"]
  sample_labels = decode_categories(sample_group)
  sample_codes = sample_group["codes"][:]

  type_group = obs["cell_type.broad"]
  type_labels = decode_categories(type_group)
  type_codes = type_group["codes"][:]

  sample_counts = pd.DataFrame({
      "sample_code": sample_codes,
      "type_code": type_codes
  })

  sample_counts = sample_counts[
      (sample_counts["sample_code"] >= 0) &
      (sample_counts["type_code"] >= 0)
  ]

  counts = sample_counts.groupby (
      ["sample_code", "type_code"]
  ).size().unstack(fill_value=0)

  mast_code = type_labels.index("Mast")
  schwann_code = type_labels.index("Schwann")

  for code in [mast_code, schwann_code]:
    if code not in counts.columns:
      counts[code] = 0

  counts = counts.rename(columns={
      mast_code: "Mast",
      schwann_code: "Schwann"
  })

  counts["sample"] = [
      sample_labels[int(code)] for code in counts.index
  ]

  counts = counts.sort_values(
      ["Mast", "Schwann"], ascending=False
  )

  print("Samples with the most mast cells and Schwann cells:")
  print(counts[["sample", "Mast", "Schwann"]].head(15).to_string(index=False))

  print("\nSamples containing both cell types:",
        int(((counts["Mast"] > 0) & (counts["Schwann"] > 0)).sum()))
```

```python
import h5py
import numpy as np
import matplotlib.pyplot as plt

data_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

target_sample = "20230421534_MSSM-Ji-Skin3-VA227_VMSCO2701_region_0"

def decode_categories(group):
  return [
      x.decode("utf-8") if isinstance(x, bytes) else str(x)
      for x in group["categories"][:]
  ]

  with h5py.File(data_path, "r") as f:
    obs = f["obs"]

    sample_group = obs["sample_id"]
    sample_labels = decode_categories(sample_group)
    sample_codes = sample_group["codes"][:]
    target_code = sample_labels.index(target_sample)

    type_group = obs["cell_type.broad"]
    type_labels = decode_categories(type_group)
    type_codes = type_group["codes"][:]

    mast_code = type_labels.index("Mast")
    schwann_code = type_labels.index("Schwann")

    sample_mask = sample_codes == target_code
    cell_mask = sample_mask & np.isin(
        type_codes, [mast_code, schwann_code]
    )

    coords = f["obsm"]["spatial"]

    # Read coordinates only for the selected cells.
    selected_indices = np.flatnonzero(cell_mask)
    selected_coords = coords[selected_indices, :]

    seleted_types = type_codes[selected_indices]

    mast_mask = selected_types == mast_code
    schwann_mask = selected_types == schwann_code

  print("Sample:", target_sample)
  print("Mast cells plotted:", int(mast_mask.sum()))
  print("Schwann cells plotted:", int(schwann_mask.sum()))

  plt.figure(figsize=(9, 7))
  plt.scatter(
      selected_coords[schwann_mask, 0],
      selected_coords[schwann_mask, 1],
      s=5, alpha=0.55, label="Schwann cells"
  )
  plt.scatter(
      selected_coords[mast_mask, 0],
      selected_coords[mast_mask, 1],
      s=7, alpha=0.7, label="Mast cells"
  )
  plt.xlabel("Spatial coordinate 1")
  plt.ylabel("Spatial coordinate 2")
  plt.title("Mast and Schwann Cells in One Skin Sample")
  plt.legend(markerscale=2)
  plt.gca().set_aspect("equal", adjustable="datalim")
  plt.tight_layout()
  plt.show()
```

```python
import h5py
import numpy as np
import matplotlib.pyplot as plt

data_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

def decode_categories(group):
  return [
      x.decode("utf-8") if isinstance(x, bytes) else str(x)
      for x in group["categories"][:]
  ]

with h5py.File(data_path, "r") as f:
  obs = f["obs"]

  sample_group = obs["sample_id"]
  sample_labels = decode_categories(sample_group)
  sample_codes = sample_group["codes"][:]

  type_group = obs["cell_type.broad"]
  type_labels = decode_categories(type_group)
  type_codes = type_group["codes"][:]

  #Select the first sample from the list of samples.
  target_sample = "202304121534_MSSM-Ji-Skin3-VA227_VMSC02701_region_0"
  target_code = sample_labels.index(target_sample)

  mast_code = type_labels.index("Mast")
  schwann_code = type_labels.index("Schwann")

  sample_mask = sample_codes == target_code
  cell_mask = sample_mask & np.isin(
      type_codes, [mast_code, schwann_code]
  )

  selected_indices = np.flatnonzero(cell_mask)
  selected_coords = f["obsm"]["spatial"][selected_indices, :]
  selected_types = type_codes[selected_indices]

mast_mask = selected_types == mast_code
schwann_mask = selected_types == schwann_code

print("Sample:", target_sample)
print("Coordinates shape:", selected_coords.shape)
print("Mast cells:", int(mast_mask.sum()))
print("Schwann cells:", int(schwann_mask.sum()))

plt.figure(figsize=(8, 6))
plt.scatter(
    selected_coords[schwann_mask, 0],
    selected_coords[schwann_mask, 1],
    s=5, alpha=0.6, label="Schwann cells"
)
plt.scatter(
    selected_coords[mast_mask, 0],
    selected_coords[mast_mask, 1],
    s=7, alpha=0.7, label="Mast cells"
)
plt.xlabel("Spatial coordinate 1")
plt.ylabel("Spatial coordinate 2")
plt.title("Mast Cells and Schwann Cells in One Skin Sample")
plt.legend()
plt.gca().set_aspect("equal", adjustable="datalim")
plt.tight_layout()
plt.show()
```

```python
import h5py

data_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"
target_sample = "202304121534_MSSM-Ji-Skin3-VA227_VMSCO2701_region_0"

def decode_categories(group):
  return [
      x.decode("utf-8") if isinstance(x, bytes) else str(x)
      for x in group["categories"][:]
  ]

with h5py.File(data_path, "r") as f:
  obs = f["obs"]

  sample_group = obs["sample_id"]
  sample_labels = decode_categories(sample_group)
  sample_codes = sample_group["codes"][:]
  target_code = sample_labels.index(target_sample)

  mask = sample_codes == target_code

  for field in ["donor_id", "anatomic site", "sample_compartment",
                "tissue_compartment"]:
      group = obs[field]
      labels = decode_categories(group)
      codes = group["codes"][:]

      values = codes[mask]
      valid = values[values >= 0]

      print(f"\n{field}:")
      for code in set(valid.tolist()):
        print(f" {labels[int(code)]}: {(valid == code).sum():,} cells")
      print(" Missing:", int((values < 0).sum()))
```

```python
import h5py
import pandas as pd

data_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

def decode_categories(group):
  return [
      x.decode(utf-8) if isinstance(x, bytes) else str(x)
      for x in group["categories"][:]
  ]

with h5py.File(data_path, "r") as f:
  obs = f["obs"]

  sample_labels = decode_categories(obs["sample_id"])
  sample_codes = obs["sample_id"]["codes"][:]
  target_sample = next(
      label for label in sample_labels
      if "MSSM-Ji-Skin3" in label and "region_0" in label
  )
  sample_mask = sample_codes == sample_labels.index(target_sample)

  compartment_labels = decode_categories(obs["tissue_compartment"])
  compartment_codes = obs["tissue_compartment"]["codes"][:]

  df = pd.DataFrame({
      "compartment_code": compartment_codes[sample_mask],
      "cell_type_code": type_codes[sample_mask]
  })

  df = df[(df["compartment_code"] >= 0) & (df["cell_type_code"] >= 0)]

  counts = df.groupby(["compartment_code", "cell_type_code"]).size().unstack(fill_value=0)

  mast_code = type_labels.index("Mast")
  schwann_code = type_labels.index("Schwann")

  for code in [mast_code, schwann_code]:
    if code not in counts.columns:
      counts[code] = 0

  print("Mast and Schwann cells by tissue compartment:")
  for compartment_code, row in counts.sort_index().iterrows():
    print(
        f"{compartment_labels[int(compartment_code)]}: "
        f"Mast = {int(row[mast_code]):,}, "
        f"Schwann = {int(row[schwann_code]):,}"
    )
```

```python
import h5py
import pandas as pd

path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

with h5py.File(path, "r") as f:
  obs = f["obs"]

  def labels_for(field):
    return [
        x.decode() if isinstance(x, bytes) else str(x)
        for x in obs[field]["categories"][:]
    ]

  samples = labels_for("sample_id")
  target = next(s for s in samples if "MSSM-Ji-Skin3" in s and "region_0" in s)
  sample_codes = obs["sample_id"]["codes"][:]
  mask = sample_codes == samples.index(target)

  compartments = labels_for("tissue_compartment")
  comp_codes = obs["tissue_compartment"]["codes"][:][mask]

  types = labels_for("cell_type.broad")
  type_codes = obs["cell_type.broad"]["codes"][:][mask]

  df = pd.DataFrame ({
      "compartment": comp_codes,
      "type": type_codes
  })

  df = df[(df["compartment"] >= 0) & (df["type"] >= 0)]

  table = pd.crosstab(df["compartment"], df["type"])

  print("Mast and Schwann cells by tissue compartment:")
  for i, name in enumerate(compartments):
    mast = int(table.loc[i, types.index("Mast")]) if types.index("Mast") in table.columns else 0
    schwann = int(table.loc[i, types.index("Schwann")]) if types.index ("Schwann") in table.columns else 0
    print(f"{name}: Mast = {mast:,}, Schwann = {schwann:,}")
```

```python
import h5py
import numpy as np
import matplotlib.pyplot as plt

path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

target = "202304121534_MSSM-Ji-Skin3-VA227_VMSC02701_region_0"

with h5py.File(path, "r") as f:
    obs = f["obs"]

    def categories(field):
        return [
            x.decode() if isinstance(x, bytes) else str(x)
            for x in obs[field]["categories"][:]
        ]

    # Select the exact sample
    sample_labels = categories("sample_id")
    sample_codes = obs["sample_id"]["codes"][:]
    sample_mask = sample_codes == sample_labels.index(target)

    # Restrict to the dermis
    compartment_labels = categories("tissue_compartment")
    compartment_codes = obs["tissue_compartment"]["codes"][:]
    dermis_code = compartment_labels.index("dermis")
    dermis_mask = sample_mask & (compartment_codes == dermis_code)

    # Identify broad cell types
    type_labels = categories("cell_type.broad")
    type_codes = obs["cell_type.broad"]["codes"][:]

    # Read coordinates for dermal cells only
    indices = np.flatnonzero(dermis_mask)
    coords = f["obsm"]["spatial"][indices, :]
    dermal_types = type_codes[indices]

    mast_code = type_labels.index("Mast")
    schwann_code = type_labels.index("Schwann")

    mast = coords[dermal_types == mast_code]
    schwann = coords[dermal_types == schwann_code]

print("Dermal cells:", len(coords))
print("Mast cells:", len(mast))
print("Schwann cells:", len(schwann))

# Plot the two cell populations
plt.figure(figsize=(9, 8))
plt.scatter(
    mast[:, 0], mast[:, 1],
    s=5, alpha=0.65, label="Mast cells"
)
plt.scatter(
    schwann[:, 0], schwann[:, 1],
    s=5, alpha=0.65, label="Schwann cells"
)

plt.xlabel("Spatial coordinate 1 (dataset units)")
plt.ylabel("Spatial coordinate 2 (dataset units)")
plt.title("Mast and Schwann Cells in Face-Skin Dermis\nDonor D004")
plt.legend(markerscale=2)
plt.axis("equal")
plt.tight_layout()

# Save a high-resolution copy to Google Drive
output_path = "/content/drive/MyDrive/Neurothesis Resoure Library/Figure1_Exploratory_Dermal_Spatial_Map.png"

plt.savefig(output_path, dpi=300, bbox_inches="tight")
plt.show()

print("Figure saved to:", output_path)
```

```python

import h5py
import numpy as np
from scipy.spatial import cKDTree

path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

target_sample = (
    "202304121534_MSSM-Ji-Skin3-VA227_VMSC02701_region_0"
)

def decode_value(value):
    if isinstance(value, bytes):
        return value.decode("utf-8")
    return value

def read_obs_column(obs, name):
    item = obs[name]

    # AnnData categorical columns store codes and categories.
    if isinstance(item, h5py.Group):
        codes = item["codes"][:]
        categories = [
            decode_value(x) for x in item["categories"][:]
        ]
        result = np.empty(len(codes), dtype=object)
        result[:] = "NA"

        valid = codes >= 0
        result[valid] = np.asarray(categories, dtype=object)[
            codes[valid]
        ]
        return result

    return np.asarray([
        decode_value(x) for x in item[:]
    ], dtype=object)

with h5py.File(path, "r") as f:
    obs = f["obs"]

    sample_ids = read_obs_column(obs, "sample_id")
    compartments = read_obs_column(obs, "tissue_compartment")
    cell_types = read_obs_column(obs, "cell_type.broad")

    # AnnData stores obsm rows in observation order.
    sample_mask = sample_ids == target_sample
    dermis_mask = np.char.lower(
        compartments.astype(str)
    ) == "dermis"

    selected = sample_mask & dermis_mask

    print("Selected sample cells:", int(sample_mask.sum()))
    print("Selected dermal cells:", int(selected.sum()))

    # Check the available labels before calculating distances.
    print("\nCell types in selected dermis:")
    types, counts = np.unique(
        cell_types[selected].astype(str),
        return_counts=True
    )
    for cell_type, count in zip(types, counts):
        print(f"{cell_type}: {count}")

    mast_mask = selected & (cell_types == "Mast")
    schwann_mask = selected & (cell_types == "Schwann")

    if mast_mask.sum() == 0 or schwann_mask.sum() == 0:
        print("\nSTOP: Expected Mast or Schwann labels were not found.")
    else:
        coords = f["obsm"]["spatial"]

        # Read coordinates only for cells in the selected sample.
        all_coords = coords[:]
        mast_coords = all_coords[mast_mask]
        schwann_coords = all_coords[schwann_mask]

        # Find each mast cell's nearest Schwann cell.
        tree = cKDTree(schwann_coords)
        distances, _ = tree.query(mast_coords, k=1)

        print("\nEXPLORATORY PROXIMITY RESULTS")
        print("Mast cells:", len(mast_coords))
        print("Schwann cells:", len(schwann_coords))
        print("Distance units: dataset coordinate units")
        print("Median nearest distance:", np.median(distances))
        print("Mean nearest distance:", np.mean(distances))
        print("25th percentile:", np.percentile(distances, 25))
        print("75th percentile:", np.percentile(distances, 75))
        print("90th percentile:", np.percentile(distances, 90))
```

```python

import h5py
import numpy as np
import pandas as pd
import os
from scipy.spatial import cKDTree

path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"
folder = "/content/drive/MyDrive/Neurothesis Resoure Library"

target_samples = [
    "202304121534_MSSM-Ji-Skin3-VA227_VMSC02701_region_0",
    "202304141439_MSSM-Ji-Skin2-VA227_VMSC02701_region_0",
    "202306070957_AH21-NSFFPE-060723-AH_VMSC02301_region_0"
]

def decode(x):
    return x.decode("utf-8") if isinstance(x, bytes) else str(x)

def read_column(obs, name):
    item = obs[name]

    if isinstance(item, h5py.Group):
        codes = item["codes"][:]
        categories = [decode(x) for x in item["categories"][:]]
        values = np.full(len(codes), "NA", dtype=object)
        valid = codes >= 0
        values[valid] = np.asarray(categories, dtype=object)[codes[valid]]
        return values

    return np.array([decode(x) for x in item[:]], dtype=object)

results = []

with h5py.File(path, "r") as f:
    obs = f["obs"]

    sample_ids = read_column(obs, "sample_id")
    donors = read_column(obs, "donor_id")
    sites = read_column(obs, "anatomic_site")
    compartments = read_column(obs, "tissue_compartment")
    cell_types = read_column(obs, "cell_type.broad")

    coords_ds = f["obsm"]["spatial"]

    for sample in target_samples:
        print("\nProcessing:", sample)

        selected = (
            (sample_ids == sample) &
            (np.char.lower(compartments.astype(str)) == "dermis")
        )

        mast_mask = selected & (cell_types == "Mast")
        schwann_mask = selected & (cell_types == "Schwann")

        mast_count = int(mast_mask.sum())
        schwann_count = int(schwann_mask.sum())

        print("Dermal cells:", int(selected.sum()))
        print("Mast cells:", mast_count)
        print("Schwann cells:", schwann_count)

        if mast_count == 0 or schwann_count == 0:
            print("Skipping: one or both cell types are absent.")
            continue

        # Read coordinates for the selected sample's cells only.
        # The stored coordinate array is indexed in observation order.
        row_indices = np.flatnonzero(selected)
        sample_coords = coords_ds[row_indices, :]

        mast_coords = sample_coords[cell_types[selected] == "Mast"]
        schwann_coords = sample_coords[cell_types[selected] == "Schwann"]

        tree = cKDTree(schwann_coords)
        distances, _ = tree.query(mast_coords, k=1)

        result = {
            "sample_id": sample,
            "donor_id": donors[np.flatnonzero(selected)[0]],
            "anatomic_site": sites[np.flatnonzero(selected)[0]],
            "mast_cells": mast_count,
            "schwann_cells": schwann_count,
            "median_nearest_distance": float(np.median(distances)),
            "mean_nearest_distance": float(np.mean(distances)),
            "p25_nearest_distance": float(np.percentile(distances, 25)),
            "p75_nearest_distance": float(np.percentile(distances, 75)),
            "p90_nearest_distance": float(np.percentile(distances, 90)),
            "distance_units": "unverified dataset coordinate units"
        }

        results.append(result)
        print("Median nearest distance:", result["median_nearest_distance"])

summary = pd.DataFrame(results)

output_path = os.path.join(
    folder, "Exploratory_Proximity_Three_Samples.csv"
)
summary.to_csv(output_path, index=False)

print("\nTHREE-SAMPLE SUMMARY")
print(summary.to_string(index=False))
print("\nSaved:", output_path)
```

```python
import h5py
import numpy as np
import pandas as pd
import os
from scipy.spatial import cKDTree

path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"
folder = "/content/drive/MyDrive/Neurothesis Resoure Library"

target_samples = [
    "202304121534_MSSM-Ji-Skin3-VA227_VMSC02701_region_0",
    "202304141439_MSSM-Ji-Skin2-VA227_VMSC02701_region_0",
    "202306070957_AH21-NSFFPE-060723-AH_VMSC02301_region_0"
]

N_PERMUTATIONS = 199
rng = np.random.default_rng(42)

def decode(x):
    return x.decode("utf-8") if isinstance(x, bytes) else str(x)

def read_column(obs, name):
    item = obs[name]
    if isinstance(item, h5py.Group):
        codes = item["codes"][:]
        categories = np.array(
            [decode(x) for x in item["categories"][:]], dtype=object
        )
        values = np.full(len(codes), "NA", dtype=object)
        valid = codes >= 0
        values[valid] = categories[codes[valid]]
        return values
    return np.array([decode(x) for x in item[:]], dtype=object)

results = []

with h5py.File(path, "r") as f:
    obs = f["obs"]

    sample_ids = read_column(obs, "sample_id")
    compartments = read_column(obs, "tissue_compartment")
    cell_types = read_column(obs, "cell_type.broad")
    donors = read_column(obs, "donor_id")
    sites = read_column(obs, "anatomic_site")

    coords_ds = f["obsm"]["spatial"]

    for sample in target_samples:
        print("\nProcessing:", sample)

        mask = (
            (sample_ids == sample) &
            (np.char.lower(compartments.astype(str)) == "dermis")
        )
        idx = np.flatnonzero(mask)

        coords = coords_ds[idx, :]
        labels = cell_types[idx]

        mast_idx = np.flatnonzero(labels == "Mast")
        schwann_idx = np.flatnonzero(labels == "Schwann")

        if len(mast_idx) == 0 or len(schwann_idx) == 0:
            print("Skipping: a cell type is absent.")
            continue

        observed = float(np.median(
            cKDTree(coords[schwann_idx]).query(coords[mast_idx], k=1)[0]
        ))

        null_medians = []

        for _ in range(N_PERMUTATIONS):
            shuffled = rng.permutation(labels)
            perm_mast = np.flatnonzero(shuffled == "Mast")
            perm_schwann = np.flatnonzero(shuffled == "Schwann")

            distances = cKDTree(coords[perm_schwann]).query(
                coords[perm_mast], k=1
            )[0]

            null_medians.append(float(np.median(distances)))

        null_medians = np.asarray(null_medians)

        # One-sided test: are observed distances unusually small?
        p_lower = (
            1 + np.sum(null_medians <= observed)
        ) / (N_PERMUTATIONS + 1)

        result = {
            "sample_id": sample,
            "donor_id": donors[idx[0]],
            "anatomic_site": sites[idx[0]],
            "dermal_cells": len(idx),
            "mast_cells": len(mast_idx),
            "schwann_cells": len(schwann_idx),
            "observed_median_distance": observed,
            "null_median_distance": float(np.median(null_medians)),
            "null_2.5_percentile": float(np.percentile(null_medians, 2.5)),
            "null_97.5_percentile": float(np.percentile(null_medians, 97.5)),
            "lower_tail_permutation_p": float(p_lower),
            "permutations": N_PERMUTATIONS,
            "units": "coordinate units; micrometers not yet confirmed"
        }

        results.append(result)

        print("Observed median:", round(observed, 3))
        print("Null median:", round(result["null_median_distance"], 3))
        print("Null 95% range:",
              round(result["null_2.5_percentile"], 3),
              "to", round(result["null_97.5_percentile"], 3))
        print("Lower-tail permutation p:", round(p_lower, 4))

summary = pd.DataFrame(results)
output_path = os.path.join(
    folder, "Exploratory_Random_Labeling_Null_Model.csv"
)
summary.to_csv(output_path, index=False)

print("\nSUMMARY")
print(summary.to_string(index=False))
print("\nSaved:", output_path)
```

```python
import h5py
import pandas as pd
import numpy as np

path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

with h5py.File(path, "r") as f:
    obs = f["obs"]

    def read_column(name):
        group = obs[name]
        codes = group["codes"][:]
        cat_ds = group["categories"]
        raw = cat_ds[:]

        if cat_ds.dtype.kind in ("S", "O"):
            categories = [
                x.decode("utf-8") if isinstance(x, bytes) else str(x)
                for x in raw
            ]
        else:
            categories = raw.tolist()

        return pd.Categorical.from_codes(codes, categories)

    columns = [
        "cell_type.broad",
        "anatomic_site",
        "neighborhood",
        "region_name",
        "cellcharter_cluster.k_10",
        "cellcharter_cluster.k_20",
    ]

    data = {col: read_column(col) for col in columns}
    df = pd.DataFrame(data)

    for col in columns[2:]:
        print(f"\n--- {col}: category counts ---")
        print(df[col].value_counts().head(20))

    print("\n--- Cell types within neighborhoods ---")
    print(pd.crosstab(
        df["neighborhood"],
        df["cell_type.broad"],
        normalize="index"
    ).round(3).to_string())

    print("\n--- Cell types within regions ---")
    print(pd.crosstab(
        df["region_name"],
        df["cell_type.broad"],
        normalize="index"
    ).round(3).to_string())
```

```python
import h5py
import pandas as pd

path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

with h5py.File(path, "r") as f:
    obs = f["obs"]

    def read_cat(name):
        g = obs[name]
        codes = g["codes"][:]
        raw = g["categories"][:]

        categories = [
            x.decode("utf-8") if isinstance(x, bytes) else x.item() if hasattr(x, "item") else x
            for x in raw
        ]
        return pd.Categorical.from_codes(codes, categories)

    neighborhood = read_cat("neighborhood")
    cell_type = read_cat("cell_type.broad")

df = pd.DataFrame({
    "neighborhood": neighborhood,
    "cell_type": cell_type
})

counts = pd.crosstab(df["neighborhood"], df["cell_type"])

print("Mast and Schwann counts by neighborhood:")
print(counts[["Mast", "Schwann"]].sort_index().to_string())

print("\nMast and Schwann percentages within each neighborhood:")
percentages = counts.div(counts.sum(axis=1), axis=0) * 100
print(percentages[["Mast", "Schwann"]].round(2).sort_index().to_string())
```

```python
import h5py

path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

def show_text(obj, prefix="", depth=0):
    if depth > 3:
        return

    if isinstance(obj, h5py.Dataset):
        if obj.size <= 30 and obj.dtype.kind in ("S", "O", "U"):
            try:
                print(prefix, obj.asstr()[:])
            except (TypeError, AttributeError):
                print(prefix, obj[:])
        return

    if isinstance(obj, h5py.Group):
        for key in obj.keys():
            if any(term in key.lower() for term in [
                "neighborhood", "region", "annotation", "cellcharter"
            ]):
                print(f"\nMetadata path: {prefix}/{key}")
            show_text(obj[key], f"{prefix}/{key}", depth + 1)

with h5py.File(path, "r") as f:
    show_text(f["uns"], "uns")
```

```python
import h5py

path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

def inspect(group, prefix="uns", depth=0):
    if depth > 4:
        return

    for key, obj in group.items():
        name = f"{prefix}/{key}"

        if any(term in key.lower() for term in [
            "neighborhood", "region", "annotation", "cellcharter"
        ]):
            print("\n", name, "|", type(obj).__name__)

        if isinstance(obj, h5py.Group):
            inspect(obj, name, depth + 1)

        elif isinstance(obj, h5py.Dataset):
            if obj.shape == () or obj.size > 30:
                continue
            if obj.dtype.kind in ("S", "O", "U"):
                try:
                    print(name, "=", obj.asstr()[()])
                except (TypeError, ValueError):
                    pass

with h5py.File(path, "r") as f:
    inspect(f["uns"])
```

```python
import pandas as pd
import os

folder = "/content/drive/MyDrive/Neurothesis Resoure Library"

counts = pd.read_csv(
    os.path.join(folder, "Dermal_Mast_Schwann_Counts_By_Sample.csv")
)

coverage = pd.read_csv(
    os.path.join(folder, "Dermal_Sample_Coverage_Summary.csv")
)

# Recalculate sample and donor counts from the original sample table
recalculated = (
    counts.groupby("anatomic_site")
    .agg(
        recalculated_samples=("sample_id", "nunique"),
        recalculated_donors=("donor_id", "nunique")
    )
    .reset_index()
)

# Compare with the previously saved coverage summary
comparison = coverage.merge(
    recalculated,
    on="anatomic_site",
    how="outer"
)

comparison["sample_count_matches"] = (
    comparison["sample_count"] == comparison["recalculated_samples"]
)

comparison["donor_count_matches"] = (
    comparison["donor_count"] == comparison["recalculated_donors"]
)

print("FULL COVERAGE AUDIT")
print(comparison.to_string(index=False))

print("\nDISCREPANCIES:")
print(
    comparison[
        (~comparison["sample_count_matches"]) |
        (~comparison["donor_count_matches"])
    ].to_string(index=False)
)
```

```python
import h5py
import numpy as np
import pandas as pd

file_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

with h5py.File(file_path, "r") as f:
    detailed = f["obs"]["cell_type.detailed"]

    labels = [
        x.decode() if isinstance(x, bytes) else str(x)
        for x in detailed["categories"][:]
    ]

    codes = detailed["codes"][:]

    counts = np.bincount(
        codes[codes >= 0],
        minlength=len(labels)
    )

    detailed_counts = pd.DataFrame({
        "cell_type": labels,
        "total_cells": counts
    }).sort_values("total_cells", ascending=False)

print("DETAILED CELL-TYPE COUNTS")
display(detailed_counts.to_string(index=False))

print("\nFIBROBLAST SUBTYPE COUNTS")
fibro_labels = [
    "Papil Fib",
    "Retic Fib I",
    "Retic Fib II",
    "Retic Fib III",
    "Perivasc Fib I",
    "Perivasc Fib II"
]

display(
    detailed_counts[
        detailed_counts["cell_type"].isin(fibro_labels)
    ].to_string(index=False)
)
```

```python
import h5py
import numpy as np
import pandas as pd

file_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

with h5py.File(file_path, "r") as f:

    broad = f["obs"]["cell_type.broad"]
    detailed = f["obs"]["cell_type.detailed"]

    broad_labels = [
        x.decode() if isinstance(x, bytes) else str(x)
        for x in broad["categories"][:]
    ]

    detailed_labels = [
        x.decode() if isinstance(x, bytes) else str(x)
        for x in detailed["categories"][:]
    ]

    broad_codes = broad["codes"][:]
    detailed_codes = detailed["codes"][:]

    valid = (broad_codes >= 0) & (detailed_codes >= 0)

    n_broad = len(broad_labels)
    n_detailed = len(detailed_labels)

    cross_counts = np.bincount(
        broad_codes[valid] * n_detailed + detailed_codes[valid],
        minlength=n_broad * n_detailed
    ).reshape(n_broad, n_detailed)

    cross_table = pd.DataFrame(
        cross_counts,
        index=broad_labels,
        columns=detailed_labels
    )

print("BROAD-TO-DETAILED ANNOTATION COMPARISON")

for label in ["Mast", "Schwann", "Fibro", "Peri"]:
    if label in cross_table.index:
        print(f"\n{label}:")
        display(
            cross_table.loc[label][
                cross_table.loc[label] > 0
            ].sort_values(ascending=False).to_frame("cell_count")
        )

print("\nTOTAL CELLS WITH VALID ANNOTATIONS:", int(valid.sum()))
```

```python
import h5py
import numpy as np
import pandas as pd

file_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

with h5py.File(file_path, "r") as f:

    broad = f["obs"]["cell_type.broad"]
    detailed = f["obs"]["cell_type.detailed"]

    broad_labels = [
        x.decode() if isinstance(x, bytes) else str(x)
        for x in broad["categories"][:]
    ]

    detailed_labels = [
        x.decode() if isinstance(x, bytes) else str(x)
        for x in detailed["categories"][:]
    ]

    broad_codes = broad["codes"][:].astype(np.int64)
    detailed_codes = detailed["codes"][:].astype(np.int64)

    valid = (
        (broad_codes >= 0) &
        (broad_codes < len(broad_labels)) &
        (detailed_codes >= 0) &
        (detailed_codes < len(detailed_labels))
    )

    n_broad = len(broad_labels)
    n_detailed = len(detailed_labels)

    combined_codes = (
        broad_codes[valid] * n_detailed +
        detailed_codes[valid]
    )

    cross_counts = np.bincount(
        combined_codes,
        minlength=n_broad * n_detailed
    ).reshape(n_broad, n_detailed)

    cross_table = pd.DataFrame(
        cross_counts,
        index=broad_labels,
        columns=detailed_labels
    )

print("BROAD-TO-DETAILED ANNOTATION COMPARISON")

for label in ["Mast", "Schwann", "Fibro", "Peri"]:
    if label in cross_table.index:
        print(f"\n{label}:")
        display(
            cross_table.loc[label][
                cross_table.loc[label] > 0
            ].sort_values(ascending=False).to_frame("cell_count")
        )

print("\nTOTAL CELLS WITH VALID ANNOTATIONS:", int(valid.sum()))
```

```python
print("BROAD CATEGORY TOTALS AND DOMINANT DETAILED LABEL")

for label in ["Mast", "Schwann", "Fibro", "Peri"]:
    row = cross_table.loc[label]
    total = row.sum()
    dominant = row.idxmax()
    dominant_count = row.max()

    print(
        f"{label}: {total:,} cells | "
        f"Most common detailed label: {dominant} "
        f"({dominant_count:,}; {dominant_count/total:.1%})"
    )

print("\nWHERE THE DETAILED MAST AND SCHWANN LABELS OCCUR")

for label in ["Mast", "Schwann"]:
    col = cross_table[label].sort_values(ascending=False)
    print(f"\nDetailed {label}:")
    display(col[col > 0].to_frame("cell_count"))
```

```python
import h5py
import numpy as np
import pandas as pd

file_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

with h5py.File(file_path, "r") as f:

    def get_labels(group):
        return [
            x.decode() if isinstance(x, bytes) else str(x)
            for x in group["categories"][:]
        ]

    compartment = f["obs"]["tissue_compartment"]
    broad = f["obs"]["cell_type.broad"]
    detailed = f["obs"]["cell_type.detailed"]

    compartment_labels = get_labels(compartment)
    broad_labels = get_labels(broad)
    detailed_labels = get_labels(detailed)

    compartment_codes = compartment["codes"][:].astype(np.int64)
    broad_codes = broad["codes"][:].astype(np.int64)
    detailed_codes = detailed["codes"][:].astype(np.int64)

    valid = (
        (compartment_codes >= 0) &
        (broad_codes >= 0) &
        (detailed_codes >= 0)
    )

    results = []

    for i, compartment_name in enumerate(compartment_labels):

        mask = valid & (compartment_codes == i)

        row = {"compartment": compartment_name,
               "total_cells": int(mask.sum())}

        for cell in ["Mast", "Schwann", "Fibro", "Peri"]:
            j = broad_labels.index(cell)
            row[f"broad_{cell}"] = int(
                np.sum(mask & (broad_codes == j))
            )

        for cell in [
            "Mast",
            "Schwann",
            "Papil Fib",
            "Retic Fib I",
            "Retic Fib II",
            "Retic Fib III",
            "Perivasc Fib I",
            "Perivasc Fib II"
        ]:
            j = detailed_labels.index(cell)
            row[f"detailed_{cell}"] = int(
                np.sum(mask & (detailed_codes == j))
            )

        results.append(row)

    compartment_summary = pd.DataFrame(results)

display(compartment_summary)
```

```python
import h5py
import numpy as np
import pandas as pd

file_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

def read_h5ad_column(group, column_name):
    """Read a categorical H5AD column as text labels."""
    col = group[column_name]

    if isinstance(col, h5py.Group) and "codes" in col and "categories" in col:
        codes = col["codes"][:]
        categories = [
            x.decode("utf-8") if isinstance(x, bytes) else str(x)
            for x in col["categories"][:]
        ]
        values = np.array(categories, dtype=object)[np.maximum(codes, 0)]
        values[codes < 0] = None
        return values

    values = col[:]
    return np.array([
        x.decode("utf-8") if isinstance(x, bytes) else str(x)
        for x in values
    ], dtype=object)

with h5py.File(file_path, "r") as f:
    obs = f["obs"]

    # Read only the metadata columns needed for this audit
    compartment = read_h5ad_column(obs, "tissue_compartment")
    sample = read_h5ad_column(obs, "sample_id")
    donor = read_h5ad_column(obs, "donor_id")
    site = read_h5ad_column(obs, "anatomic_site")
    broad = read_h5ad_column(obs, "cell_type.broad")
    detailed = read_h5ad_column(obs, "cell_type.detailed")

    # Restrict the primary analysis to dermal cells
    dermis = compartment == "dermis"

    audit = pd.DataFrame({
        "sample_id": sample[dermis],
        "donor_id": donor[dermis],
        "anatomic_site": site[dermis],
        "broad_Mast": broad[dermis] == "Mast",
        "broad_Schwann": broad[dermis] == "Schwann",
        "detailed_Mast": detailed[dermis] == "Mast",
        "detailed_Schwann": detailed[dermis] == "Schwann",
        "detailed_Papil_Fib": detailed[dermis] == "Papil Fib",
        "detailed_Retic_Fib_I": detailed[dermis] == "Retic Fib I",
        "detailed_Retic_Fib_II": detailed[dermis] == "Retic Fib II",
        "detailed_Retic_Fib_III": detailed[dermis] == "Retic Fib III",
        "detailed_Perivasc_Fib_I": detailed[dermis] == "Perivasc Fib I",
        "detailed_Perivasc_Fib_II": detailed[dermis] == "Perivasc Fib II"
    })

    count_columns = [
        col for col in audit.columns
        if col not in ["sample_id", "donor_id", "anatomic_site"]
    ]

    sample_audit = (
        audit.groupby(
            ["sample_id", "donor_id", "anatomic_site"],
            dropna=False
        )[count_columns]
        .sum()
        .reset_index()
    )

    print("Number of dermal samples:", len(sample_audit))
    print("Number of donors:", sample_audit["donor_id"].nunique())
    print("\nSample-level annotation audit:")
    display(sample_audit.sort_values(
        ["anatomic_site", "donor_id"]
    ).reset_index(drop=True))

    print("\nSummary of broad versus detailed annotations:")
    display(sample_audit[[
        "broad_Mast", "detailed_Mast",
        "broad_Schwann", "detailed_Schwann"
    ]].describe().round(2))
```

```python
import h5py
import pandas as pd
import numpy as np

file_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

with h5py.File(file_path, "r") as f:

    def get_labels(group):
        return [
            x.decode() if isinstance(x, bytes) else str(x)
            for x in group["categories"][:]
        ]

    category_group = f["obs"]["cell_category"]
    detailed_group = f["obs"]["cell_type.detailed"]
    compartment_group = f["obs"]["tissue_compartment"]

    category_labels = get_labels(category_group)
    detailed_labels = get_labels(detailed_group)
    compartment_labels = get_labels(compartment_group)

    category_codes = category_group["codes"][:]
    detailed_codes = detailed_group["codes"][:]
    compartment_codes = compartment_group["codes"][:]

    dermis_code = compartment_labels.index("dermis")

    valid = (
        (category_codes >= 0)
        & (detailed_codes >= 0)
        & (compartment_codes == dermis_code)
    )

    selected_types = [
        "Schwann",
        "Retic Fib III",
        "Retic Fib I",
        "Papil Fib",
        "Retic Fib II",
        "Perivasc Fib I",
        "Perivasc Fib II",
        "Mast"
    ]

    results = []

    for cell_type in selected_types:
        type_code = detailed_labels.index(cell_type)
        mask = valid & (detailed_codes == type_code)

        row = {"detailed_cell_type": cell_type}

        for i, category in enumerate(category_labels):
            row[category] = int(np.sum(mask & (category_codes == i)))

        row["total"] = int(mask.sum())
        results.append(row)

lineage_audit = pd.DataFrame(results)
display(lineage_audit)
```

```python
import h5py
import numpy as np
import pandas as pd

file_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

with h5py.File(file_path, "r") as f:
    obs = f["obs"]

    def get_labels(group):
        return [
            x.decode() if isinstance(x, bytes) else str(x)
            for x in group["categories"][:]
        ]

    broad_group = obs["cell_type.broad"]
    detailed_group = obs["cell_type.detailed"]
    sample_group = obs["sample_id"]
    compartment_group = obs["tissue_compartment"]

    broad_labels = get_labels(broad_group)
    detailed_labels = get_labels(detailed_group)
    sample_labels = get_labels(sample_group)
    compartment_labels = get_labels(compartment_group)

    broad = broad_group["codes"][:]
    detailed = detailed_group["codes"][:]
    samples = sample_group["codes"][:]
    compartments = compartment_group["codes"][:]

    dermis = compartment_labels.index("dermis")
    broad_schwann = broad_labels.index("Schwann")
    detailed_schwann = detailed_labels.index("Schwann")
    retic_iii = detailed_labels.index("Retic Fib III")

    valid = (compartments == dermis) & (samples >= 0)

    broad_mask = valid & (broad == broad_schwann)
    overlap_mask = broad_mask & (detailed == retic_iii)
    detailed_schwann_mask = valid & (detailed == detailed_schwann)

    n_samples = len(sample_labels)

    broad_counts = np.bincount(samples[broad_mask], minlength=n_samples)
    overlap_counts = np.bincount(samples[overlap_mask], minlength=n_samples)
    detailed_counts = np.bincount(
        samples[detailed_schwann_mask], minlength=n_samples
    )

    result = pd.DataFrame({
        "sample_id": sample_labels,
        "broad_Schwann": broad_counts,
        "broad_Schwann_detailed_Retic_Fib_III": overlap_counts,
        "detailed_Schwann": detailed_counts
    })

    result["overlap_percent_of_broad_Schwann"] = (
        100 * result["broad_Schwann_detailed_Retic_Fib_III"]
        / result["broad_Schwann"].replace(0, np.nan)
    )

    print("SUMMARY OF WITHIN-SAMPLE OVERLAP")
    display(result["overlap_percent_of_broad_Schwann"].describe())

    print("\nSAMPLES WITH HIGHEST OVERLAP PERCENTAGES")
    display(
        result.sort_values(
            "overlap_percent_of_broad_Schwann", ascending=False
        ).head(10)
    )

    print("\nSAMPLES WITH LOWEST OVERLAP PERCENTAGES")
    display(
        result.sort_values(
            "overlap_percent_of_broad_Schwann", ascending=True
        ).head(10)
    )
```

```python
import h5py
import pandas as pd
import numpy as np

file_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

with h5py.File(file_path, "r") as f:
    obs = f["obs"]

    def read_column(name):
        node = obs[name]

        if isinstance(node, h5py.Group) and "codes" in node:
            categories = [
                x.decode() if isinstance(x, bytes) else str(x)
                for x in node["categories"][:]
            ]
            codes = node["codes"][:]
            return [
                categories[c] if c >= 0 else None
                for c in codes
            ]

        elif isinstance(node, h5py.Dataset):
            values = node[:]
            return [
                x.decode() if isinstance(x, bytes) else x
                for x in values
            ]

        else:
            return None

    metadata_columns = [
        "sample_id",
        "batch",
        "run_name.short",
        "preservation_type",
        "collection_type",
        "collection_source",
        "gene_panel",
        "segmentation"
    ]

    metadata = pd.DataFrame({
        col: read_column(col)
        for col in metadata_columns
    })

    metadata_summary = (
        metadata.groupby("sample_id")
        .agg({
            col: "nunique"
            for col in metadata_columns
            if col != "sample_id"
        })
    )

    print("NUMBER OF DISTINCT METADATA VALUES PER SAMPLE")
    display(metadata_summary.describe())

    print("\nMETADATA CATEGORIES ACROSS THE DATASET")
    for col in metadata_columns[1:]:
        print(f"\n{col}:")
        print(metadata[col].value_counts(dropna=False).head(15))
```

```python

import h5py
import numpy as np
import pandas as pd
from scipy.sparse import csr_matrix

marker_genes = [
    "ANGPTL7", "ITGA6", "KLF5",
    "PRG4", "COMP", "EGFR"
]

target_cell_types = [
    "Schwann",
    "Retic Fib I",
    "Retic Fib II",
    "Retic Fib III",
    "Papil Fib",
    "Perivasc Fib I",
    "Perivasc Fib II"
]

chunk_size = 25000

# Accumulators for each cell type and marker
results = {
    cell_type: {
        gene: {"sum": 0.0, "detected": 0}
        for gene in marker_genes
    }
    for cell_type in target_cell_types
}

cell_totals = {cell_type: 0 for cell_type in target_cell_types}

with h5py.File(file_path, "r") as f:

    obs = f["obs"]
    X = f["X"]

    # Read annotation categories and codes
    def read_cat(group):
        categories = [
            x.decode("utf-8") if isinstance(x, bytes) else str(x)
            for x in group["categories"][:]
        ]
        return categories, group["codes"][:]

    compartments, compartment_codes = read_cat(
        obs["tissue_compartment"]
    )

    detailed, detailed_codes = read_cat(
        obs["cell_type.detailed"]
    )

    genes = [
        x.decode("utf-8") if isinstance(x, bytes) else str(x)
        for x in f["var"]["_index"][:]
    ]

    gene_indices = [genes.index(gene) for gene in marker_genes]

    dermis_code = compartments.index("dermis")

    # Process the sparse expression matrix in chunks
    n_cells = X.attrs["shape"][0]
    n_genes = X.attrs["shape"][1]

    for start in range(0, n_cells, chunk_size):

        stop = min(start + chunk_size, n_cells)

        indptr = X["indptr"][start:stop + 1]
        data_start = int(indptr[0])
        data_end = int(indptr[-1])

        local_indptr = indptr - data_start

        chunk_matrix = csr_matrix(
            (
                X["data"][data_start:data_end],
                X["indices"][data_start:data_end],
                local_indptr
            ),
            shape=(stop - start, n_genes)
        )

        compartment_chunk = compartment_codes[start:stop]
        detailed_chunk = detailed_codes[start:stop]

        dermis_rows = compartment_chunk == dermis_code

        for cell_type in target_cell_types:

            cell_code = detailed.index(cell_type)

            selected_rows = np.where(
                dermis_rows & (detailed_chunk == cell_code)
            )[0]

            if len(selected_rows) == 0:
                continue

            cell_totals[cell_type] += len(selected_rows)

            expression = chunk_matrix[
                selected_rows, :
            ][:, gene_indices]

            sums = np.asarray(
                expression.sum(axis=0)
            ).ravel()

            detected = np.asarray(
                (expression > 0).sum(axis=0)
            ).ravel()

            for j, gene in enumerate(marker_genes):
                results[cell_type][gene]["sum"] += float(sums[j])
                results[cell_type][gene]["detected"] += int(detected[j])

        if start % 100000 == 0:
            print(f"Processed {stop:,} of {n_cells:,} cells")

# Create summary table
rows = []

for cell_type in target_cell_types:

    n_cells = cell_totals[cell_type]

    for gene in marker_genes:

        total = results[cell_type][gene]["sum"]
        detected = results[cell_type][gene]["detected"]

        rows.append({
            "cell_type": cell_type,
            "gene": gene,
            "cells": n_cells,
            "cells_detected": detected,
            "percent_detected": round(
                100 * detected / n_cells, 2
            ) if n_cells else np.nan,
            "mean_expression": round(
                total / n_cells, 5
            ) if n_cells else np.nan
        })

marker_expression_summary = pd.DataFrame(rows)

print("\nDERMAL MARKER EXPRESSION SUMMARY")
display(marker_expression_summary)

print("\nMATRIX VALUE INFORMATION")
with h5py.File(file_path, "r") as f:
    print("Expression data dtype:", f["X"]["data"].dtype)
    print("Expression matrix shape:", f["X"].attrs["shape"])
```

```python

import h5py
import numpy as np
import pandas as pd
from scipy.sparse import csr_matrix

marker_genes = [
    "ANGPTL7", "ITGA6", "KLF5",
    "PRG4", "COMP", "EGFR"
]

target_cell_types = [
    "Schwann",
    "Retic Fib I",
    "Retic Fib II",
    "Retic Fib III",
    "Papil Fib",
    "Perivasc Fib I",
    "Perivasc Fib II"
]

chunk_size = 25000

results = {
    cell_type: {
        gene: {
            "count_sum": 0.0,
            "log_sum": 0.0,
            "detected": 0
        }
        for gene in marker_genes
    }
    for cell_type in target_cell_types
}

cell_totals = {cell_type: 0 for cell_type in target_cell_types}

with h5py.File(file_path, "r") as f:

    obs = f["obs"]

    def read_cat(group):
        categories = [
            x.decode("utf-8") if isinstance(x, bytes) else str(x)
            for x in group["categories"][:]
        ]
        return categories, group["codes"][:]

    compartments, compartment_codes = read_cat(
        obs["tissue_compartment"]
    )

    detailed, detailed_codes = read_cat(
        obs["cell_type.detailed"]
    )

    genes = [
        x.decode("utf-8") if isinstance(x, bytes) else str(x)
        for x in f["var"]["_index"][:]
    ]

    gene_indices = [genes.index(gene) for gene in marker_genes]

    dermis_code = compartments.index("dermis")

    counts_layer = f["layers"]["counts"]
    log_layer = f["layers"]["log1p"]

    n_cells = int(counts_layer.attrs["shape"][0])
    n_genes = int(counts_layer.attrs["shape"][1])

    for start in range(0, n_cells, chunk_size):

        stop = min(start + chunk_size, n_cells)

        compartment_chunk = compartment_codes[start:stop]
        detailed_chunk = detailed_codes[start:stop]

        dermis_rows = compartment_chunk == dermis_code

        for cell_type in target_cell_types:

            cell_code = detailed.index(cell_type)

            selected_rows = np.where(
                dermis_rows & (detailed_chunk == cell_code)
            )[0]

            if len(selected_rows) == 0:
                continue

            cell_totals[cell_type] += len(selected_rows)

            # Convert global rows to local chunk rows
            def read_layer(layer):

                indptr = layer["indptr"][start:stop + 1]
                data_start = int(indptr[0])
                data_end = int(indptr[-1])

                matrix = csr_matrix(
                    (
                        layer["data"][data_start:data_end],
                        layer["indices"][data_start:data_end],
                        indptr - data_start
                    ),
                    shape=(stop - start, n_genes)
                )

                return matrix[selected_rows, :][:, gene_indices]

            count_values = read_layer(counts_layer)
            log_values = read_layer(log_layer)

            count_sums = np.asarray(
                count_values.sum(axis=0)
            ).ravel()

            log_sums = np.asarray(
                log_values.sum(axis=0)
            ).ravel()

            detected = np.asarray(
                (count_values > 0).sum(axis=0)
            ).ravel()

            for j, gene in enumerate(marker_genes):

                results[cell_type][gene]["count_sum"] += float(
                    count_sums[j]
                )

                results[cell_type][gene]["log_sum"] += float(
                    log_sums[j]
                )

                results[cell_type][gene]["detected"] += int(
                    detected[j]
                )

        if stop % 100000 < chunk_size or stop == n_cells:
            print(f"Processed {stop:,} of {n_cells:,} cells")

rows = []

for cell_type in target_cell_types:

    n = cell_totals[cell_type]

    for gene in marker_genes:

        item = results[cell_type][gene]

        rows.append({
            "cell_type": cell_type,
            "gene": gene,
            "cells": n,
            "detected_cells": item["detected"],
            "percent_detected": round(
                100 * item["detected"] / n, 2
            ) if n else np.nan,
            "mean_raw_count": round(
                item["count_sum"] / n, 4
            ) if n else np.nan,
            "mean_log1p": round(
                item["log_sum"] / n, 4
            ) if n else np.nan
        })

corrected_marker_summary = pd.DataFrame(rows)

print("\nCORRECTED DERMAL MARKER EXPRESSION")
display(corrected_marker_summary)

print("\nCELL POPULATION TOTALS")
display(
    pd.DataFrame(
        cell_totals.items(),
        columns=["cell_type", "cell_count"]
    )
)
```

```python

import h5py
import numpy as np
import pandas as pd
from scipy.sparse import csr_matrix

core_markers = ["ANGPTL7", "ITGA6", "KLF5"]
additional_markers = ["EGFR", "PRG4", "COMP"]
all_markers = core_markers + additional_markers

target_cell_types = [
    "Schwann",
    "Retic Fib I",
    "Retic Fib II",
    "Retic Fib III",
    "Papil Fib",
    "Perivasc Fib I",
    "Perivasc Fib II"
]

chunk_size = 25000

summary = {
    ct: {
        "cells": 0,
        "core_1plus": 0,
        "core_2plus": 0,
        "core_3plus": 0,
        "core_2plus_EGFR": 0,
        "PRG4_COMP_both": 0
    }
    for ct in target_cell_types
}

with h5py.File(file_path, "r") as f:

    obs = f["obs"]

    def read_categories(group):
        categories = [
            x.decode("utf-8") if isinstance(x, bytes) else str(x)
            for x in group["categories"][:]
        ]
        return categories, group["codes"][:]

    compartments, compartment_codes = read_categories(
        obs["tissue_compartment"]
    )

    detailed, detailed_codes = read_categories(
        obs["cell_type.detailed"]
    )

    genes = [
        x.decode("utf-8") if isinstance(x, bytes) else str(x)
        for x in f["var"]["_index"][:]
    ]

    gene_indices = [genes.index(g) for g in all_markers]

    dermis_code = compartments.index("dermis")
    counts_layer = f["layers"]["counts"]

    n_cells = int(counts_layer.attrs["shape"][0])
    n_genes = int(counts_layer.attrs["shape"][1])

    for start in range(0, n_cells, chunk_size):

        stop = min(start + chunk_size, n_cells)

        indptr = counts_layer["indptr"][start:stop + 1]
        data_start = int(indptr[0])
        data_end = int(indptr[-1])

        matrix = csr_matrix(
            (
                counts_layer["data"][data_start:data_end],
                counts_layer["indices"][data_start:data_end],
                indptr - data_start
            ),
            shape=(stop - start, n_genes)
        )

        expression = matrix[:, gene_indices]

        detected = (expression > 0).toarray()

        dermis_mask = compartment_codes[start:stop] == dermis_code
        type_codes = detailed_codes[start:stop]

        for ct in target_cell_types:

            ct_code = detailed.index(ct)

            selected = dermis_mask & (type_codes == ct_code)

            if not selected.any():
                continue

            values = detected[selected, :]

            core = values[:, :len(core_markers)]
            egfr = values[:, all_markers.index("EGFR")]
            prg4 = values[:, all_markers.index("PRG4")]
            comp = values[:, all_markers.index("COMP")]

            summary[ct]["cells"] += len(values)
            summary[ct]["core_1plus"] += int(
                (core.sum(axis=1) >= 1).sum()
            )
            summary[ct]["core_2plus"] += int(
                (core.sum(axis=1) >= 2).sum()
            )
            summary[ct]["core_3plus"] += int(
                (core.sum(axis=1) == 3).sum()
            )
            summary[ct]["core_2plus_EGFR"] += int(
                ((core.sum(axis=1) >= 2) & egfr).sum()
            )
            summary[ct]["PRG4_COMP_both"] += int(
                (prg4 & comp).sum()
            )

        if stop % 100000 < chunk_size or stop == n_cells:
            print(f"Processed {stop:,} of {n_cells:,} cells")

rows = []

for ct, values in summary.items():

    n = values["cells"]

    rows.append({
        "cell_type": ct,
        "cells": n,
        "at_least_1_core_marker_pct": round(
            100 * values["core_1plus"] / n, 2
        ) if n else np.nan,
        "at_least_2_core_markers_pct": round(
            100 * values["core_2plus"] / n, 2
        ) if n else np.nan,
        "all_3_core_markers_pct": round(
            100 * values["core_3plus"] / n, 2
        ) if n else np.nan,
        "2_core_plus_EGFR_pct": round(
            100 * values["core_2plus_EGFR"] / n, 2
        ) if n else np.nan,
        "PRG4_and_COMP_pct": round(
            100 * values["PRG4_COMP_both"] / n, 2
        ) if n else np.nan
    })

coexpression_summary = pd.DataFrame(rows)

print("\nSINGLE-CELL MARKER CO-EXPRESSION")
display(coexpression_summary)
```

```python

import h5py
import numpy as np
import pandas as pd
from scipy.sparse import csr_matrix

markers = ["ANGPTL7", "ITGA6", "KLF5"]

cell_types_to_check = [
    "Schwann",
    "Retic Fib I",
    "Retic Fib II",
    "Retic Fib III",
    "Papil Fib",
    "Perivasc Fib I",
    "Perivasc Fib II"
]

chunk_size = 25000

pattern_counts = {
    ct: {} for ct in cell_types_to_check
}
population_totals = {
    ct: 0 for ct in cell_types_to_check
}

with h5py.File(file_path, "r") as f:

    obs = f["obs"]

    def get_categories(group):
        categories = [
            x.decode("utf-8") if isinstance(x, bytes) else str(x)
            for x in group["categories"][:]
        ]
        return categories, group["codes"][:]

    compartments, compartment_codes = get_categories(
        obs["tissue_compartment"]
    )

    detailed, detailed_codes = get_categories(
        obs["cell_type.detailed"]
    )

    genes = [
        x.decode("utf-8") if isinstance(x, bytes) else str(x)
        for x in f["var"]["_index"][:]
    ]

    gene_indices = [genes.index(g) for g in markers]

    dermis_code = compartments.index("dermis")
    layer = f["layers"]["counts"]

    n_cells, n_genes = map(int, layer.attrs["shape"])

    for start in range(0, n_cells, chunk_size):

        stop = min(start + chunk_size, n_cells)

        indptr = layer["indptr"][start:stop + 1]
        data_start = int(indptr[0])
        data_end = int(indptr[-1])

        matrix = csr_matrix(
            (
                layer["data"][data_start:data_end],
                layer["indices"][data_start:data_end],
                indptr - data_start
            ),
            shape=(stop - start, n_genes)
        )

        expression = (matrix[:, gene_indices] > 0).toarray()

        dermis_mask = compartment_codes[start:stop] == dermis_code
        type_codes = detailed_codes[start:stop]

        for ct in cell_types_to_check:

            selected = (
                dermis_mask &
                (type_codes == detailed.index(ct))
            )

            if not selected.any():
                continue

            values = expression[selected]
            population_totals[ct] += len(values)

            # Encode each marker combination as a three-bit pattern.
            # Bit 1 = ANGPTL7, bit 2 = ITGA6, bit 4 = KLF5.
            codes = (
                values[:, 0].astype(int) +
                2 * values[:, 1].astype(int) +
                4 * values[:, 2].astype(int)
            )

            unique, counts = np.unique(codes, return_counts=True)

            for code, count in zip(unique, counts):
                pattern_counts[ct][int(code)] = (
                    pattern_counts[ct].get(int(code), 0) + int(count)
                )

        if stop % 100000 < chunk_size or stop == n_cells:
            print(f"Processed {stop:,} of {n_cells:,} cells")

pattern_names = {
    0: "None",
    1: "ANGPTL7 only",
    2: "ITGA6 only",
    3: "ANGPTL7 + ITGA6",
    4: "KLF5 only",
    5: "ANGPTL7 + KLF5",
    6: "ITGA6 + KLF5",
    7: "All three"
}

rows = []

for ct in cell_types_to_check:

    total = population_totals[ct]

    for code, name in pattern_names.items():

        count = pattern_counts[ct].get(code, 0)

        rows.append({
            "cell_type": ct,
            "marker_pattern": name,
            "cell_count": count,
            "percent_of_population": round(
                100 * count / total, 2
            ) if total else np.nan
        })

pattern_summary = pd.DataFrame(rows)

print("\nEXACT CORE MARKER COMBINATIONS")
display(
    pattern_summary[
        pattern_summary["marker_pattern"] != "None"
    ].sort_values(
        ["cell_type", "cell_count"],
        ascending=[True, False]
    )
)

print("\nRETIC FIB III ONLY")
display(
    pattern_summary[
        (pattern_summary["cell_type"] == "Retic Fib III") &
        (pattern_summary["marker_pattern"] != "None")
    ].sort_values("cell_count", ascending=False)
)
```

```python

import h5py
import numpy as np
import pandas as pd
from scipy.sparse import csr_matrix

chunk_size = 25000

with h5py.File(file_path, "r") as f:

    obs = f["obs"]

    def read_cat(group):
        categories = [
            x.decode("utf-8") if isinstance(x, bytes) else str(x)
            for x in group["categories"][:]
        ]
        return categories, group["codes"][:]

    sample_categories, sample_codes = read_cat(obs["sample_id"])
    donor_categories, donor_codes = read_cat(obs["donor_id"])
    site_categories, site_codes = read_cat(obs["anatomic_site"])
    compartment_categories, compartment_codes = read_cat(
        obs["tissue_compartment"]
    )
    detailed_categories, detailed_codes = read_cat(
        obs["cell_type.detailed"]
    )

    genes = [
        x.decode("utf-8") if isinstance(x, bytes) else str(x)
        for x in f["var"]["_index"][:]
    ]

    marker_indices = [
        genes.index(g) for g in ["ANGPTL7", "ITGA6", "KLF5"]
    ]

    dermis_code = compartment_categories.index("dermis")
    retic_iii_code = detailed_categories.index("Retic Fib III")

    layer = f["layers"]["counts"]
    n_cells, n_genes = map(int, layer.attrs["shape"])

    sample_results = {}

    for start in range(0, n_cells, chunk_size):

        stop = min(start + chunk_size, n_cells)

        indptr = layer["indptr"][start:stop + 1]
        data_start = int(indptr[0])
        data_end = int(indptr[-1])

        matrix = csr_matrix(
            (
                layer["data"][data_start:data_end],
                layer["indices"][data_start:data_end],
                indptr - data_start
            ),
            shape=(stop - start, n_genes)
        )

        expression = (matrix[:, marker_indices] > 0).toarray()

        selected = (
            (compartment_codes[start:stop] == dermis_code) &
            (detailed_codes[start:stop] == retic_iii_code)
        )

        local_rows = np.where(selected)[0]

        for row in local_rows:

            global_row = start + row
            sample_code = int(sample_codes[global_row])

            if sample_code < 0:
                continue

            if sample_code not in sample_results:
                sample_results[sample_code] = {
                    "cells": 0,
                    "two_or_more": 0,
                    "all_three": 0,
                    "donor_code": int(donor_codes[global_row]),
                    "site_code": int(site_codes[global_row])
                }

            values = expression[row]

            sample_results[sample_code]["cells"] += 1
            sample_results[sample_code]["two_or_more"] += int(
                values.sum() >= 2
            )
            sample_results[sample_code]["all_three"] += int(
                values.sum() == 3
            )

        if stop % 100000 < chunk_size or stop == n_cells:
            print(f"Processed {stop:,} of {n_cells:,} cells")

rows = []

for sample_code, values in sample_results.items():

    n = values["cells"]
    donor_code = values["donor_code"]
    site_code = values["site_code"]

    rows.append({
        "sample_id": sample_categories[sample_code],
        "donor_id": (
            donor_categories[donor_code]
            if donor_code >= 0 else "Unknown"
        ),
        "anatomic_site": (
            site_categories[site_code]
            if site_code >= 0 else "Unknown"
        ),
        "retic_fib_III_cells": n,
        "two_or_more_markers_pct": round(
            100 * values["two_or_more"] / n, 2
        ),
        "all_three_markers_pct": round(
            100 * values["all_three"] / n, 2
        )
    })

retic_iii_sample_summary = pd.DataFrame(rows)

retic_iii_sample_summary = retic_iii_sample_summary.sort_values(
    "two_or_more_markers_pct",
    ascending=False
).reset_index(drop=True)

print("SAMPLES CONTAINING DERMAL RETIC FIB III:", len(rows))
print("\nSAMPLE-LEVEL MARKER CO-EXPRESSION")
display(retic_iii_sample_summary)

print("\nSUMMARY ACROSS SAMPLES")
display(
    retic_iii_sample_summary[
        ["retic_fib_III_cells",
         "two_or_more_markers_pct",
         "all_three_markers_pct"]
    ].describe().round(2)
)
```

```python

# ANATOMICAL-SITE ROBUSTNESS ANALYSIS

df = retic_iii_sample_summary.copy()

site_summary = (
    df.groupby("anatomic_site")
    .agg(
        sample_count=("sample_id", "count"),
        donor_count=("donor_id", "nunique"),
        median_retic_fib_III_cells=("retic_fib_III_cells", "median"),
        median_coexpression=("two_or_more_markers_pct", "median"),
        mean_coexpression=("two_or_more_markers_pct", "mean"),
        minimum_coexpression=("two_or_more_markers_pct", "min"),
        maximum_coexpression=("two_or_more_markers_pct", "max"),
        median_all_three=("all_three_markers_pct", "median")
    )
    .reset_index()
    .sort_values("median_coexpression", ascending=False)
)

print("ANATOMICAL-SITE SUMMARY: ALL SAMPLES")
display(site_summary.round(2))

# Repeat after excluding samples with fewer than 20 Retic Fib III cells.
filtered = df[df["retic_fib_III_cells"] >= 20]

site_summary_filtered = (
    filtered.groupby("anatomic_site")
    .agg(
        sample_count=("sample_id", "count"),
        donor_count=("donor_id", "nunique"),
        median_retic_fib_III_cells=("retic_fib_III_cells", "median"),
        median_coexpression=("two_or_more_markers_pct", "median"),
        mean_coexpression=("two_or_more_markers_pct", "mean"),
        minimum_coexpression=("two_or_more_markers_pct", "min"),
        maximum_coexpression=("two_or_more_markers_pct", "max")
    )
    .reset_index()
    .sort_values("median_coexpression", ascending=False)
)

print("\nANATOMICAL-SITE SUMMARY: SAMPLES WITH AT LEAST 20 CELLS")
display(site_summary_filtered.round(2))

# Identify sites represented by multiple donors.
print("\nSITES WITH AT LEAST 3 DISTINCT DONORS")
display(
    site_summary_filtered[
        site_summary_filtered["donor_count"] >= 3
    ].round(2)
)
```

```python
# DONOR-MATCHED ANATOMICAL-SITE COVERAGE

df = retic_iii_sample_summary.copy()

# Summarize the anatomical sites represented for each donor.
donor_site_summary = (
    df.groupby("donor_id")
    .agg(
        sample_count=("sample_id", "count"),
        site_count=("anatomic_site", "nunique"),
        anatomical_sites=(
            "anatomic_site",
            lambda x: ", ".join(sorted(x.unique()))
        )
    )
    .reset_index()
    .sort_values(["site_count", "sample_count"], ascending=False)
)

print("DONORS REPRESENTED AT MULTIPLE ANATOMICAL SITES")
display(
    donor_site_summary[
        donor_site_summary["site_count"] >= 2
    ].round(2)
)

print("\nOVERALL DONOR COVERAGE")
print("Total donors:", df["donor_id"].nunique())
print("Donors with 2+ sites:", (donor_site_summary["site_count"] >= 2).sum())
print("Donors with 5+ sites:", (donor_site_summary["site_count"] >= 5).sum())

# Count how many donors contribute to each donor-site combination.
donor_site_counts = (
    df.groupby("anatomic_site")["donor_id"]
    .nunique()
    .sort_values(ascending=False)
)

print("\nDONOR COUNTS BY SITE")
display(donor_site_counts.to_frame("distinct_donors"))

# Identify site pairs that can be compared within the same donors.
site_donor_matrix = pd.crosstab(
    df["donor_id"],
    df["anatomic_site"]
)

site_pair_counts = site_donor_matrix.T.dot(site_donor_matrix)

print("\nNUMBER OF SHARED DONORS FOR EACH SITE PAIR")
display(site_pair_counts)
```

```python
# CORRECTED SHARED-DONOR MATRIX
# Each donor counts once per anatomical site, regardless of sample count.

df = retic_iii_sample_summary.copy()

donor_site_presence = pd.crosstab(
    df["donor_id"],
    df["anatomic_site"]
).gt(0).astype(int)

shared_donor_counts = donor_site_presence.T.dot(donor_site_presence)

print("CORRECTED NUMBER OF SHARED DONORS FOR EACH SITE PAIR")
display(shared_donor_counts)

print("\nCHECK: DISTINCT DONORS PER SITE")
display(
    donor_site_presence.sum()
    .sort_values(ascending=False)
    .to_frame("distinct_donors")
)

print("\nDONORS WITH THE BROADEST SITE COVERAGE")
display(
    donor_site_presence.sum(axis=1)
    .sort_values(ascending=False)
    .to_frame("number_of_sites")
    .head(12)
)
```

```python
# WITHIN-DONOR COMPARISON: SOLE VS OTHER ANATOMICAL SITES

df = retic_iii_sample_summary.copy()

# Exclude low-cell-count samples for this comparison.
df = df[df["retic_fib_III_cells"] >= 20].copy()

# If a donor has multiple samples from the same site,
# summarize those samples using the donor-site median.
donor_site_values = (
    df.groupby(["donor_id", "anatomic_site"])
    ["two_or_more_markers_pct"]
    .median()
    .unstack()
)

results = []

if "sole" not in donor_site_values.columns:
    print("No eligible sole samples found.")
else:
    for site in donor_site_values.columns:
        if site == "sole":
            continue

        paired = donor_site_values[["sole", site]].dropna()

        if len(paired) < 3:
            continue

        differences = paired["sole"] - paired[site]

        results.append({
            "comparison_site": site,
            "matched_donors": len(paired),
            "median_sole_coexpression": paired["sole"].median(),
            "median_other_site_coexpression": paired[site].median(),
            "median_paired_difference": differences.median(),
            "mean_paired_difference": differences.mean(),
            "donors_sole_higher": (differences > 0).sum(),
            "donors_equal": (differences == 0).sum(),
            "donors_other_site_higher": (differences < 0).sum()
        })

paired_results = (
    pd.DataFrame(results)
    .sort_values("median_paired_difference", ascending=False)
)

print("WITHIN-DONOR COMPARISONS: SOLE MINUS OTHER SITE")
display(paired_results.round(2))

# Show the actual paired values for the most informative comparisons.
priority_sites = [
    site for site in ["popliteal fossae", "antecubital fossa", "abdomen",
                      "back", "central scalp", "occipital scalp"]
    if site in donor_site_values.columns
]

print("\nDONOR-LEVEL VALUES FOR PRIORITY SITE COMPARISONS")

for site in priority_sites:
    paired = donor_site_values[["sole", site]].dropna().copy()
    paired["sole_minus_other_site"] = paired["sole"] - paired[site]

    print(f"\nSOLE VS {site.upper()} (n={len(paired)} donors)")
    display(paired.round(2))
```

```python
# DONOR-MATCHED SITE COMPARISONS
# Compare all sites with at least 5 shared donors.
# Repeat after excluding samples with fewer than 20 Retic Fib III cells.

from itertools import combinations
from scipy.stats import binomtest
import pandas as pd
import numpy as np

def paired_site_tests(data, min_shared_donors=5):
    # One value per donor per site.
    donor_site = (
        data.groupby(["donor_id", "anatomic_site"])["two_or_more_markers_pct"]
        .median()
        .unstack()
    )

    results = []

    for site_a, site_b in combinations(donor_site.columns, 2):
        paired = donor_site[[site_a, site_b]].dropna()
        n = len(paired)

        if n < min_shared_donors:
            continue

        differences = paired[site_a] - paired[site_b]
        positive = int((differences > 0).sum())
        negative = int((differences < 0).sum())
        non_tied = positive + negative

        p_value = (
            binomtest(positive, non_tied, 0.5, alternative="two-sided").pvalue
            if non_tied > 0 else 1.0
        )

        results.append({
            "site_A": site_a,
            "site_B": site_b,
            "matched_donors": n,
            "median_paired_difference_A_minus_B": differences.median(),
            "mean_paired_difference_A_minus_B": differences.mean(),
            "site_A_higher": positive,
            "site_B_higher": negative,
            "ties": n - non_tied,
            "sign_test_p": p_value
        })

    result = pd.DataFrame(results)

    if len(result):
        # Holm correction for all tested site pairs.
        order = np.argsort(result["sign_test_p"].values)
        raw_p = result["sign_test_p"].values
        adjusted = np.empty(len(raw_p))
        running_max = 0

        for rank, index in enumerate(order):
            corrected = min(1.0, (len(raw_p) - rank) * raw_p[index])
            running_max = max(running_max, corrected)
            adjusted[index] = running_max

        result["holm_adjusted_p"] = adjusted
        result = result.sort_values("sign_test_p")

    return result


# All samples
all_site_tests = paired_site_tests(retic_iii_sample_summary)

print("ALL SAMPLES: DONOR-MATCHED SITE TESTS")
display(all_site_tests.round(4))

# Sensitivity analysis: at least 20 Retic Fib III cells per sample
filtered_samples = retic_iii_sample_summary[
    retic_iii_sample_summary["retic_fib_III_cells"] >= 20
].copy()

filtered_site_tests = paired_site_tests(filtered_samples)

print("\nSENSITIVITY ANALYSIS: AT LEAST 20 RETIC FIB III CELLS")
display(filtered_site_tests.round(4))

# Focus on the most informative sole comparisons
priority_sites = [
    "occipital scalp",
    "popliteal fossae",
    "antecubital fossa",
    "central scalp"
]

print("\nPRIORITY SOLE COMPARISONS: ALL SAMPLES")
display(
    all_site_tests[
        (all_site_tests["site_A"] == "sole") &
        (all_site_tests["site_B"].isin(priority_sites))
    ].round(4)
)

print("\nPRIORITY SOLE COMPARISONS: FILTERED SAMPLES")
display(
    filtered_site_tests[
        (filtered_site_tests["site_A"] == "sole") &
        (filtered_site_tests["site_B"].isin(priority_sites))
    ].round(4)
)
```

```python
# CORRECTED PRIORITY COMPARISON EXTRACTION
# Standardize every comparison as sole minus other site.

def extract_sole_comparisons(test_results, priority_sites):
    rows = []

    for site in priority_sites:
        match = test_results[
            (
                (test_results["site_A"] == "sole") &
                (test_results["site_B"] == site)
            ) |
            (
                (test_results["site_A"] == site) &
                (test_results["site_B"] == "sole")
            )
        ]

        if match.empty:
            continue

        r = match.iloc[0]
        sole_is_A = r["site_A"] == "sole"

        rows.append({
            "comparison": f"sole vs {site}",
            "matched_donors": r["matched_donors"],
            "median_sole_minus_other": (
                r["median_paired_difference_A_minus_B"]
                if sole_is_A else
                -r["median_paired_difference_A_minus_B"]
            ),
            "mean_sole_minus_other": (
                r["mean_paired_difference_A_minus_B"]
                if sole_is_A else
                -r["mean_paired_difference_A_minus_B"]
            ),
            "donors_sole_higher": (
                r["site_A_higher"] if sole_is_A
                else r["site_B_higher"]
            ),
            "donors_other_higher": (
                r["site_B_higher"] if sole_is_A
                else r["site_A_higher"]
            ),
            "sign_test_p": r["sign_test_p"],
            "holm_adjusted_p": r["holm_adjusted_p"]
        })

    return pd.DataFrame(rows)


priority_sites = [
    "occipital scalp",
    "central scalp",
    "popliteal fossae",
    "antecubital fossa",
    "knee",
    "abdomen"
]

print("ALL SAMPLES")
display(
    extract_sole_comparisons(all_site_tests, priority_sites).round(4)
)

print("\nSAMPLES WITH AT LEAST 20 RETIC FIB III CELLS")
display(
    extract_sole_comparisons(filtered_site_tests, priority_sites).round(4)
)
```

```python
import h5py
import numpy as np
import pandas as pd

with h5py.File(h5ad_path, "r") as f:

    def decode_categories(group):
        return [
            x.decode("utf-8") if isinstance(x, bytes) else str(x)
            for x in group["categories"][:]
        ]

    sample_cats = decode_categories(f["obs"]["sample_id"])
    donor_cats = decode_categories(f["obs"]["donor_id"])
    site_cats = decode_categories(f["obs"]["anatomic_site"])
    compartment_cats = decode_categories(f["obs"]["tissue_compartment"])
    celltype_cats = decode_categories(f["obs"]["cell_type.detailed"])

    sample_codes = f["obs"]["sample_id"]["codes"][:]
    donor_codes = f["obs"]["donor_id"]["codes"][:]
    site_codes = f["obs"]["anatomic_site"]["codes"][:]
    compartment_codes = f["obs"]["tissue_compartment"]["codes"][:]
    celltype_codes = f["obs"]["cell_type.detailed"]["codes"][:]

    dermis_code = compartment_cats.index("dermis")
    retic_iii_code = celltype_cats.index("Retic Fib III")

    target_mask = (
        (compartment_codes == dermis_code) &
        (celltype_codes == retic_iii_code)
    )

    print("Total dermal Retic Fib III cells:", target_mask.sum())

    target_sample_codes = sample_codes[target_mask]

    sample_counts = np.bincount(
        target_sample_codes,
        minlength=len(sample_cats)
    )

    target_summary = pd.DataFrame({
        "sample_id": sample_cats,
        "donor_id": [
            donor_cats[np.where((sample_codes == i) & (donor_codes >= 0))[0][0]]
            if np.any((sample_codes == i) & (donor_codes >= 0))
            else None
            for i in range(len(sample_cats))
        ],
        "anatomic_site": [
            site_cats[np.where((sample_codes == i) & (site_codes >= 0))[0][0]]
            if np.any((sample_codes == i) & (site_codes >= 0))
            else None
            for i in range(len(sample_cats))
        ],
        "retic_iii_cells": sample_counts
    })

    target_summary = target_summary[
        target_summary["retic_iii_cells"] > 0
    ].reset_index(drop=True)

    print("\nSamples with Retic Fib III cells:", len(target_summary))
    print("Total cells across samples:", target_summary["retic_iii_cells"].sum())
    print("\nCell-count summary:")
    print(target_summary["retic_iii_cells"].describe())
```

```python
import h5py
import numpy as np
import pandas as pd

marker_positions = {
    "ANGPTL7": 17,
    "ITGA6": 306,
    "KLF5": 325,
    "PRG4": 447,
    "COMP": 131,
    "EGFR": 181
}

target_rows = np.where(target_mask)[0]
chunk_size = 50000

sample_marker_results = []

with h5py.File(h5ad_path, "r") as f:

    counts = f["layers/counts"]

    data = counts["data"]
    indices = counts["indices"]
    indptr = counts["indptr"]

    for start in range(0, len(target_rows), chunk_size):

        rows = target_rows[start:start + chunk_size]

        marker_detected = np.zeros(
            (len(rows), len(marker_positions)),
            dtype=bool
        )

        for j, gene_col in enumerate(marker_positions.values()):

            for k, row in enumerate(rows):

                left = indptr[row]
                right = indptr[row + 1]

                row_indices = indices[left:right]
                row_data = data[left:right]

                matches = np.where(row_indices == gene_col)[0]

                if len(matches) > 0:
                    marker_detected[k, j] = row_data[matches[0]] > 0

        chunk_df = pd.DataFrame(
            marker_detected,
            columns=marker_positions.keys()
        )

        chunk_df["sample_code"] = sample_codes[rows]

        chunk_summary = chunk_df.groupby("sample_code").sum()
        sample_marker_results.append(chunk_summary)

        print(
            f"Processed {start + len(rows):,} / "
            f"{len(target_rows):,} cells"
        )

marker_counts = (
    pd.concat(sample_marker_results)
    .groupby(level=0)
    .sum()
)

print("\nMarker detection counts:")
display(marker_counts.head())
```

```python
import h5py
import numpy as np
import pandas as pd

core_positions = [17, 306, 325]  # ANGPTL7, ITGA6, KLF5
egfr_position = 181
prg4_position = 447
comp_position = 131

target_rows = np.where(target_mask)[0]

results = []

with h5py.File(h5ad_path, "r") as f:

    counts = f["layers/counts"]
    data = counts["data"]
    indices = counts["indices"]
    indptr = counts["indptr"]

    for row in target_rows:

        left, right = indptr[row], indptr[row + 1]
        row_indices = indices[left:right]
        row_data = data[left:right]

        detected = set(row_indices[row_data > 0])

        core_detected = [
            gene in detected for gene in core_positions
        ]

        core_count = sum(core_detected)

        results.append({
            "sample_code": sample_codes[row],
            "core_1plus": core_count >= 1,
            "core_2plus": core_count >= 2,
            "core_all3": core_count == 3,
            "core_2plus_EGFR": core_count >= 2 and egfr_position in detected,
            "PRG4_COMP": prg4_position in detected and comp_position in detected
        })

cell_coexpression = pd.DataFrame(results)

coexpression_summary = (
    cell_coexpression
    .groupby("sample_code")
    .agg(
        retic_iii_cells=("core_1plus", "size"),
        core_1plus=("core_1plus", "sum"),
        core_2plus=("core_2plus", "sum"),
        core_all3=("core_all3", "sum"),
        core_2plus_EGFR=("core_2plus_EGFR", "sum"),
        PRG4_COMP=("PRG4_COMP", "sum")
    )
    .reset_index()
)

for col in [
    "core_1plus", "core_2plus", "core_all3",
    "core_2plus_EGFR", "PRG4_COMP"
]:
    coexpression_summary[col + "_pct"] = (
        coexpression_summary[col] /
        coexpression_summary["retic_iii_cells"] * 100
    )

print("OVERALL CO-EXPRESSION COUNTS")
print(coexpression_summary[
    ["retic_iii_cells", "core_1plus", "core_2plus",
     "core_all3", "core_2plus_EGFR", "PRG4_COMP"]
].sum())

print("\nOVERALL CO-EXPRESSION PERCENTAGES")
display(
    coexpression_summary[
        ["core_1plus_pct", "core_2plus_pct", "core_all3_pct",
         "core_2plus_EGFR_pct", "PRG4_COMP_pct"]
    ].mean()
)
```

```python
import h5py
import numpy as np
import pandas as pd
from scipy.sparse import csr_matrix

h5ad_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"
output_folder = "/content/drive/MyDrive/Neurothesis Resoure Library"

markers = {
    "ANGPTL7": 17,
    "ITGA6": 306,
    "KLF5": 325,
    "PRG4": 447,
    "COMP": 131,
    "EGFR": 181
}

with h5py.File(h5ad_path, "r") as f:

    def categories(name):
        return [
            x.decode("utf-8") if isinstance(x, bytes) else str(x)
            for x in f[f"obs/{name}/categories"][:]
        ]

    compartment_cats = categories("tissue_compartment")
    celltype_cats = categories("cell_type.detailed")
    sample_cats = categories("sample_id")
    donor_cats = categories("donor_id")
    site_cats = categories("anatomic_site")

    compartment = f["obs/tissue_compartment/codes"][:]
    celltype = f["obs/cell_type.detailed/codes"][:]
    sample = f["obs/sample_id/codes"][:]
    donor = f["obs/donor_id/codes"][:]
    site = f["obs/anatomic_site/codes"][:]

    target_rows = np.where(
        (compartment == compartment_cats.index("dermis")) &
        (celltype == celltype_cats.index("Retic Fib III"))
    )[0]

    print("Target cells:", len(target_rows))

    matrix = f["layers/counts"]
    indptr = matrix["indptr"][:]
    indices = matrix["indices"][:]
    data = matrix["data"][:]

    n_cells = len(target_rows)
    detected = np.zeros((n_cells, len(markers)), dtype=bool)

    for i, row in enumerate(target_rows):
        left, right = indptr[row], indptr[row + 1]
        row_indices = indices[left:right]
        row_data = data[left:right]

        for j, gene_col in enumerate(markers.values()):
            matches = np.where(row_indices == gene_col)[0]
            if len(matches):
                detected[i, j] = row_data[matches[0]] > 0

        if (i + 1) % 2000 == 0:
            print(f"Processed {i + 1:,} / {n_cells:,}")

    cell_df = pd.DataFrame(
        detected,
        columns=markers.keys()
    )

    core_count = cell_df[["ANGPTL7", "ITGA6", "KLF5"]].sum(axis=1)

    cell_df["core_1plus"] = core_count >= 1
    cell_df["core_2plus"] = core_count >= 2
    cell_df["core_all3"] = core_count == 3
    cell_df["core_2plus_EGFR"] = (
        (core_count >= 2) & cell_df["EGFR"]
    )
    cell_df["PRG4_COMP"] = (
        cell_df["PRG4"] & cell_df["COMP"]
    )

    cell_df["sample_code"] = sample[target_rows]

    summary = (
        cell_df.groupby("sample_code")
        .agg(
            retic_iii_cells=("core_1plus", "size"),
            ANGPTL7=("ANGPTL7", "sum"),
            ITGA6=("ITGA6", "sum"),
            KLF5=("KLF5", "sum"),
            PRG4=("PRG4", "sum"),
            COMP=("COMP", "sum"),
            EGFR=("EGFR", "sum"),
            core_1plus=("core_1plus", "sum"),
            core_2plus=("core_2plus", "sum"),
            core_all3=("core_all3", "sum"),
            core_2plus_EGFR=("core_2plus_EGFR", "sum"),
            PRG4_COMP=("PRG4_COMP", "sum")
        )
        .reset_index()
    )

    summary["sample_id"] = summary["sample_code"].map(
        lambda x: sample_cats[x]
    )

    summary["donor_id"] = summary["sample_code"].map(
        lambda x: donor_cats[donor[np.where(sample == x)[0][0]]]
    )

    summary["anatomic_site"] = summary["sample_code"].map(
        lambda x: site_cats[site[np.where(sample == x)[0][0]]]
    )

    for col in [
        "ANGPTL7", "ITGA6", "KLF5", "PRG4", "COMP", "EGFR",
        "core_1plus", "core_2plus", "core_all3",
        "core_2plus_EGFR", "PRG4_COMP"
    ]:
        summary[col + "_pct"] = (
            summary[col] / summary["retic_iii_cells"] * 100
        )

    output_path = output_folder + "/Retic_Fib_III_Marker_Coexpression_By_Sample.csv"
    summary.to_csv(output_path, index=False)

    print("\nSAVED:", output_path)
    print("Samples:", len(summary))
    print("Total cells:", summary["retic_iii_cells"].sum())

    print("\nOVERALL POOLED CO-EXPRESSION")
    for col in [
        "core_1plus", "core_2plus", "core_all3",
        "core_2plus_EGFR", "PRG4_COMP"
    ]:
        print(
            col,
            round(summary[col].sum() /
                  summary["retic_iii_cells"].sum() * 100, 2),
            "%"
        )

    print("\nSAMPLE-LEVEL SUMMARY")
    display(summary.head())
```

```python
import numpy as np
import pandas as pd

# 1. Overall sample-level distribution
sample_distribution = summary[
    ["core_2plus_pct", "core_all3_pct", "retic_iii_cells"]
].describe(percentiles=[0.25, 0.50, 0.75])

print("SAMPLE-LEVEL DISTRIBUTION")
display(sample_distribution.round(2))

# 2. Sensitivity analyses by minimum sample cell count
print("\nSENSITIVITY ANALYSIS")

for threshold in [20, 50]:
    subset = summary[summary["retic_iii_cells"] >= threshold]

    print(f"\nMinimum {threshold} cells per sample")
    print("Samples:", len(subset))
    print("Donors:", subset["donor_id"].nunique())
    print("Median 2+ core markers:",
          round(subset["core_2plus_pct"].median(), 2), "%")
    print("Mean 2+ core markers:",
          round(subset["core_2plus_pct"].mean(), 2), "%")
    print("Median all 3 markers:",
          round(subset["core_all3_pct"].median(), 2), "%")

# 3. Donor-level summary
donor_summary = (
    summary.groupby("donor_id")
    .agg(
        samples=("sample_id", "nunique"),
        total_cells=("retic_iii_cells", "sum"),
        median_core_2plus=("core_2plus_pct", "median"),
        mean_core_2plus=("core_2plus_pct", "mean"),
        median_all3=("core_all3_pct", "median")
    )
    .reset_index()
    .sort_values("median_core_2plus", ascending=False)
)

print("\nDONOR-LEVEL SUMMARY")
display(donor_summary.round(2))

# 4. Anatomical-site summary
site_summary = (
    summary.groupby("anatomic_site")
    .agg(
        samples=("sample_id", "nunique"),
        donors=("donor_id", "nunique"),
        total_cells=("retic_iii_cells", "sum"),
        median_core_2plus=("core_2plus_pct", "median"),
        mean_core_2plus=("core_2plus_pct", "mean"),
        median_all3=("core_all3_pct", "median")
    )
    .reset_index()
    .sort_values("median_core_2plus", ascending=False)
)

print("\nANATOMICAL-SITE SUMMARY")
display(site_summary.round(2))

# 5. Save outputs
folder = "/content/drive/MyDrive/Neurothesis Resoure Library"

donor_summary.to_csv(
    folder + "/Retic_Fib_III_Marker_Coexpression_By_Donor.csv",
    index=False
)

site_summary.to_csv(
    folder + "/Retic_Fib_III_Marker_Coexpression_By_Site.csv",
    index=False
)

print("\nSaved donor-level and site-level summaries.")
```

```python
import h5py
import numpy as np
import pandas as pd
from collections import defaultdict

h5ad_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"
output_folder = "/content/drive/MyDrive/Neurothesis Resoure Library"

counts = defaultdict(lambda: {
    "Mast": 0,
    "Schwann": 0,
    "Retic Fib III": 0
})

with h5py.File(h5ad_path, "r") as f:

    def get_categories(path):
        return [
            x.decode("utf-8") if isinstance(x, bytes) else str(x)
            for x in f[path + "/categories"][:]
        ]

    compartment_cats = get_categories("obs/tissue_compartment")
    celltype_cats = get_categories("obs/cell_type.detailed")
    sample_cats = get_categories("obs/sample_id")

    dermis_code = compartment_cats.index("dermis")

    target_codes = {
        name: celltype_cats.index(name)
        for name in ["Mast", "Schwann", "Retic Fib III"]
    }

    compartment_ds = f["obs/tissue_compartment/codes"]
    celltype_ds = f["obs/cell_type.detailed/codes"]
    sample_ds = f["obs/sample_id/codes"]

    total = compartment_ds.shape[0]
    chunk_size = 100_000

    for start in range(0, total, chunk_size):
        end = min(start + chunk_size, total)

        compartments = compartment_ds[start:end]
        celltypes = celltype_ds[start:end]
        samples = sample_ds[start:end]

        mask = compartments == dermis_code

        for name, code in target_codes.items():
            selected = mask & (celltypes == code)

            if selected.any():
                sample_counts = np.bincount(
                    samples[selected],
                    minlength=len(sample_cats)
                )

                for sample_code in np.where(sample_counts > 0)[0]:
                    counts[sample_code][name] += int(
                        sample_counts[sample_code]
                    )

        print(f"Processed {end:,} / {total:,}")

    rows = []

    for sample_code, values in counts.items():
        rows.append({
            "sample_id": sample_cats[sample_code],
            "Mast": values["Mast"],
            "Schwann": values["Schwann"],
            "Retic_Fib_III": values["Retic Fib III"]
        })

    sample_counts_df = pd.DataFrame(rows)

    sample_counts_df["all_three_present"] = (
        (sample_counts_df["Mast"] > 0) &
        (sample_counts_df["Schwann"] > 0) &
        (sample_counts_df["Retic_Fib_III"] > 0)
    )

    sample_counts_df = sample_counts_df.sort_values(
        ["all_three_present", "Mast"],
        ascending=[False, False]
    )

    save_path = output_folder + "/Dermal_Three_Population_Counts_By_Sample.csv"
    sample_counts_df.to_csv(save_path, index=False)

    print("\nSAVED:", save_path)
    print("\nSamples represented:", len(sample_counts_df))
    print("Samples with all three populations:",
          sample_counts_df["all_three_present"].sum())

    display(sample_counts_df.head(15))
```

```python
import h5py
import numpy as np
import pandas as pd
from scipy.spatial import cKDTree

h5ad_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

target_sample = (
    "202304121534_MSSM-Ji-Skin3-VA227_VMSC02701_region_0"
)

populations = ["Mast", "Schwann", "Retic Fib III"]
coordinates = {name: [] for name in populations}

with h5py.File(h5ad_path, "r") as f:

    def get_categories(path):
        return [
            x.decode("utf-8") if isinstance(x, bytes) else str(x)
            for x in f[path + "/categories"][:]
        ]

    sample_cats = get_categories("obs/sample_id")
    compartment_cats = get_categories("obs/tissue_compartment")
    celltype_cats = get_categories("obs/cell_type.detailed")

    sample_code = sample_cats.index(target_sample)
    dermis_code = compartment_cats.index("dermis")

    population_codes = {
        name: celltype_cats.index(name)
        for name in populations
    }

    sample_ds = f["obs/sample_id/codes"]
    compartment_ds = f["obs/tissue_compartment/codes"]
    celltype_ds = f["obs/cell_type.detailed/codes"]
    spatial_ds = f["obsm/spatial"]

    total = sample_ds.shape[0]
    chunk_size = 100_000

    for start in range(0, total, chunk_size):
        end = min(start + chunk_size, total)

        samples = sample_ds[start:end]
        compartments = compartment_ds[start:end]
        celltypes = celltype_ds[start:end]

        base_mask = (
            (samples == sample_code) &
            (compartments == dermis_code)
        )

        if not base_mask.any():
            continue

        for name, code in population_codes.items():
            mask = base_mask & (celltypes == code)

            if mask.any():
                chunk_coords = spatial_ds[start:end][mask]
                valid = np.isfinite(chunk_coords).all(axis=1)

                coordinates[name].append(chunk_coords[valid])

    coordinates = {
        name: np.vstack(parts) if parts else np.empty((0, 2))
        for name, parts in coordinates.items()
    }

print("SAMPLE:", target_sample)

for name, coords in coordinates.items():
    print(name, "cells with valid coordinates:", len(coords))

if all(len(coordinates[name]) > 0 for name in populations):

    trees = {
        name: cKDTree(coordinates[name])
        for name in populations
    }

    comparisons = [
        ("Mast", "Schwann"),
        ("Mast", "Retic Fib III"),
        ("Retic Fib III", "Schwann")
    ]

    results = []

    for source, target in comparisons:
        distances, _ = trees[target].query(
            coordinates[source],
            k=1
        )

        results.append({
            "comparison": source + " to " + target,
            "source_cells": len(distances),
            "median_distance": np.median(distances),
            "mean_distance": np.mean(distances),
            "25th_percentile": np.percentile(distances, 25),
            "75th_percentile": np.percentile(distances, 75)
        })

    pilot_results = pd.DataFrame(results)

    print("\nNEAREST-NEIGHBOR DISTANCES")
    display(pilot_results)

    print(
        "\nReminder: distances are in original coordinate units, "
        "not confirmed micrometers."
    )

else:
    print("At least one population has no valid coordinates.")
```

```python
import h5py
import numpy as np

file_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

def inspect_h5_group(group, depth=0, max_depth=4):
    for name, item in group.items():
        indent = "  " * depth

        if isinstance(item, h5py.Group):
            print(f"{indent}GROUP: {name}")

            for key, value in item.attrs.items():
                print(f"{indent}  ATTRIBUTE {key}: {value}")

            if depth < max_depth:
                inspect_h5_group(item, depth + 1, max_depth)

        else:
            print(
                f"{indent}DATASET: {name} | "
                f"shape={item.shape} | dtype={item.dtype}"
            )

            for key, value in item.attrs.items():
                print(f"{indent}  ATTRIBUTE {key}: {value}")

            if item.size <= 30:
                print(f"{indent}  VALUES: {item[()]}")

with h5py.File(file_path, "r") as f:
    print("INSPECTING UNS/SPATIAL_NEIGHBORS")
    inspect_h5_group(f["uns/spatial_neighbors"])
```

```python
import h5py
import numpy as np

file_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

rng = np.random.default_rng(42)

with h5py.File(file_path, "r") as f:

    coords = f["obsm/spatial"]
    graph = f["obsp/spatial_distances"]

    indptr = graph["indptr"]
    indices = graph["indices"]
    distances = graph["data"]

    # Examine how many neighbors each cell has
    degrees = np.diff(indptr[:])

    print("NEIGHBOR COUNT DISTRIBUTION")
    print("Minimum:", degrees.min())
    print("Median:", np.median(degrees))
    print("Mean:", degrees.mean())
    print("Maximum:", degrees.max())

    # Randomly select 100 cells to validate distances
    test_rows = rng.choice(len(indptr) - 1, size=100, replace=False)

    differences = []

    for row in test_rows:

        start = indptr[row]
        end = indptr[row + 1]

        neighbor_ids = indices[start:end]
        stored_distances = distances[start:end]

        if len(neighbor_ids) == 0:
            continue

        center = coords[row]
        neighbor_coords = coords[neighbor_ids]

        calculated = np.linalg.norm(
            neighbor_coords - center,
            axis=1
        )

        differences.extend(
            np.abs(calculated - stored_distances)
        )

    differences = np.array(differences)

    print("\nDISTANCE VALIDATION")
    print("Edges tested:", len(differences))
    print("Median absolute difference:", np.median(differences))
    print("Maximum absolute difference:", np.max(differences))
```

```python
import h5py
import numpy as np

file_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

with h5py.File(file_path, "r") as f:

    sample_codes = f["obs/sample_id/codes"][:]
    compartment_codes = f["obs/tissue_compartment/codes"][:]

    graph = f["obsp/spatial_connectivities"]
    indices = graph["indices"]
    indptr = graph["indptr"]

    total_edges = len(indices)

    cross_sample = 0
    cross_compartment = 0
    same_sample_same_compartment = 0

    chunk_size = 100_000

    for start in range(0, len(sample_codes), chunk_size):

        end = min(start + chunk_size, len(sample_codes))

        row_counts = np.diff(indptr[start:end + 1])

        if row_counts.sum() == 0:
            continue

        row_ids = np.repeat(
            np.arange(start, end),
            row_counts
        )

        neighbor_ids = indices[indptr[start]:indptr[end]]

        same_sample = (
            sample_codes[row_ids] == sample_codes[neighbor_ids]
        )

        same_compartment = (
            compartment_codes[row_ids] ==
            compartment_codes[neighbor_ids]
        )

        cross_sample += np.sum(~same_sample)

        cross_compartment += np.sum(
            same_sample & ~same_compartment
        )

        same_sample_same_compartment += np.sum(
            same_sample & same_compartment
        )

print("SPATIAL GRAPH BOUNDARY VALIDATION")
print("Total directed edges:", total_edges)

print(
    "Cross-sample edges:",
    cross_sample,
    f"({100 * cross_sample / total_edges:.4f}%)"
)

print(
    "Same-sample, cross-compartment edges:",
    cross_compartment,
    f"({100 * cross_compartment / total_edges:.4f}%)"
)

print(
    "Same-sample, same-compartment edges:",
    same_sample_same_compartment,
    f"({100 * same_sample_same_compartment / total_edges:.4f}%)"
)
```

```python
import h5py
import numpy as np
import pandas as pd

file_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

populations = ["Mast", "Schwann", "Retic Fib III"]

with h5py.File(file_path, "r") as f:

    # Read annotation categories and identify their integer codes
    compartment_categories = f["obs/tissue_compartment/categories"].asstr()[:]
    cell_categories = f["obs/cell_type.detailed/categories"].asstr()[:]

    dermis_code = np.where(compartment_categories == "dermis")[0][0]

    population_codes = {
        name: int(np.where(cell_categories == name)[0][0])
        for name in populations
    }

    compartment_codes = f["obs/tissue_compartment/codes"][:]
    cell_codes = f["obs/cell_type.detailed/codes"][:]

    graph = f["obsp/spatial_connectivities"]
    indices = graph["indices"]
    indptr = graph["indptr"]

    # Rows = source population; columns = neighbor population
    population_matrix = np.zeros((3, 3), dtype=np.int64)
    total_dermal_neighbors = np.zeros(3, dtype=np.int64)

    total_dermal_edges = 0
    total_edges_from_selected_sources = 0

    chunk_size = 100_000

    for start in range(0, len(cell_codes), chunk_size):

        end = min(start + chunk_size, len(cell_codes))

        row_counts = np.diff(indptr[start:end + 1])

        if row_counts.sum() == 0:
            continue

        row_ids = np.repeat(
            np.arange(start, end),
            row_counts
        )

        neighbor_ids = indices[indptr[start]:indptr[end]]

        # Keep only edges where BOTH cells are in the dermis
        valid = (
            (compartment_codes[row_ids] == dermis_code) &
            (compartment_codes[neighbor_ids] == dermis_code)
        )

        source_ids = row_ids[valid]
        target_ids = neighbor_ids[valid]

        source_types = cell_codes[source_ids]
        target_types = cell_codes[target_ids]

        total_dermal_edges += len(source_ids)

        for i, source_name in enumerate(populations):

            source_code = population_codes[source_name]
            source_mask = source_types == source_code

            total_neighbors = np.sum(source_mask)
            total_dermal_neighbors[i] += total_neighbors

            total_edges_from_selected_sources += total_neighbors

            for j, target_name in enumerate(populations):

                target_code = population_codes[target_name]

                population_matrix[i, j] += np.sum(
                    source_mask & (target_types == target_code)
                )

# Build results table
results = []

for i, source_name in enumerate(populations):

    denominator = total_dermal_neighbors[i]

    for j, target_name in enumerate(populations):

        count = population_matrix[i, j]

        results.append({
            "source_population": source_name,
            "neighbor_population": target_name,
            "neighbor_edges": count,
            "all_dermal_neighbors_of_source": denominator,
            "percentage_of_dermal_neighbors": (
                100 * count / denominator if denominator else np.nan
            )
        })

results_df = pd.DataFrame(results)

print("DERMAL GRAPH NEIGHBORHOOD SUMMARY")
print("Total dermal-to-dermal edges:", total_dermal_edges)
print("Edges originating from selected populations:",
      total_edges_from_selected_sources)

display(results_df.round(3))
```

```python
import h5py
import numpy as np
import pandas as pd

file_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

populations = ["Mast", "Schwann", "Retic Fib III"]
n_pop = len(populations)

with h5py.File(file_path, "r") as f:

    # Category definitions
    compartment_categories = f["obs/tissue_compartment/categories"].asstr()[:]
    cell_categories = f["obs/cell_type.detailed/categories"].asstr()[:]
    sample_categories = f["obs/sample_id/categories"].asstr()[:]

    dermis_code = int(np.where(compartment_categories == "dermis")[0][0])

    population_codes = {
        name: int(np.where(cell_categories == name)[0][0])
        for name in populations
    }

    n_samples = len(sample_categories)

    # Read categorical codes
    compartment_codes = f["obs/tissue_compartment/codes"][:]
    cell_codes = f["obs/cell_type.detailed/codes"][:]
    sample_codes = f["obs/sample_id/codes"][:]

    # Per-sample population cell counts
    dermal_cell_counts = np.zeros(n_samples, dtype=np.int64)
    population_cell_counts = np.zeros(
        (n_samples, n_pop), dtype=np.int64
    )

    for start in range(0, len(cell_codes), 100_000):

        end = min(start + 100_000, len(cell_codes))

        compartments = compartment_codes[start:end]
        cells = cell_codes[start:end]
        samples = sample_codes[start:end]

        dermal = compartments == dermis_code
        valid = dermal & (samples >= 0)

        dermal_cell_counts += np.bincount(
            samples[valid],
            minlength=n_samples
        )

        for j, name in enumerate(populations):

            mask = valid & (cells == population_codes[name])

            population_cell_counts[:, j] += np.bincount(
                samples[mask],
                minlength=n_samples
            )

    # Graph neighborhood counts
    graph = f["obsp/spatial_connectivities"]
    indices = graph["indices"]
    indptr = graph["indptr"]

    edge_counts = np.zeros(
        (n_samples, n_pop, n_pop), dtype=np.int64
    )

    source_neighbor_totals = np.zeros(
        (n_samples, n_pop), dtype=np.int64
    )

    for start in range(0, len(cell_codes), 100_000):

        end = min(start + 100_000, len(cell_codes))

        row_counts = np.diff(indptr[start:end + 1])

        if row_counts.sum() == 0:
            continue

        row_ids = np.repeat(
            np.arange(start, end),
            row_counts
        )

        neighbor_ids = indices[indptr[start]:indptr[end]]

        valid = (
            (compartment_codes[row_ids] == dermis_code) &
            (compartment_codes[neighbor_ids] == dermis_code) &
            (sample_codes[row_ids] >= 0) &
            (sample_codes[row_ids] == sample_codes[neighbor_ids])
        )

        source_ids = row_ids[valid]
        target_ids = neighbor_ids[valid]

        source_samples = sample_codes[source_ids]
        source_types = cell_codes[source_ids]
        target_types = cell_codes[target_ids]

        source_pop_idx = np.full(len(source_ids), -1, dtype=np.int8)
        target_pop_idx = np.full(len(target_ids), -1, dtype=np.int8)

        for j, name in enumerate(populations):
            source_pop_idx[source_types == population_codes[name]] = j
            target_pop_idx[target_types == population_codes[name]] = j

        selected_sources = source_pop_idx >= 0

        # Count all dermal neighbors for each selected source population
        denominator_keys = (
            source_samples[selected_sources] * n_pop +
            source_pop_idx[selected_sources]
        )

        source_neighbor_totals += np.bincount(
            denominator_keys,
            minlength=n_samples * n_pop
        ).reshape(n_samples, n_pop)

        # Count edges between the three selected populations
        selected_edges = selected_sources & (target_pop_idx >= 0)

        edge_keys = (
            source_samples[selected_edges] * n_pop * n_pop +
            source_pop_idx[selected_edges] * n_pop +
            target_pop_idx[selected_edges]
        )

        edge_counts += np.bincount(
            edge_keys,
            minlength=n_samples * n_pop * n_pop
        ).reshape(n_samples, n_pop, n_pop)

# Assemble sample-level results
records = []

for s, sample_id in enumerate(sample_categories):

    for i, source_name in enumerate(populations):

        denominator = source_neighbor_totals[s, i]

        if denominator == 0:
            continue

        for j, target_name in enumerate(populations):

            observed_edges = edge_counts[s, i, j]
            observed_pct = 100 * observed_edges / denominator

            target_count = population_cell_counts[s, j]
            total_cells = dermal_cell_counts[s]

            abundance_pct = (
                100 * target_count / total_cells
                if total_cells > 0 else np.nan
            )

            enrichment = (
                observed_pct / abundance_pct
                if abundance_pct > 0 else np.nan
            )

            records.append({
                "sample_id": sample_id,
                "source_population": source_name,
                "target_population": target_name,
                "source_neighbor_edges": denominator,
                "observed_target_edges": observed_edges,
                "observed_neighbor_pct": observed_pct,
                "target_dermal_cells": target_count,
                "total_dermal_cells": total_cells,
                "target_abundance_pct": abundance_pct,
                "sample_enrichment": enrichment
            })

sample_enrichment_df = pd.DataFrame(records)

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Sample_Specific_Dermal_Neighborhood_Enrichment.csv"
)

sample_enrichment_df.to_csv(output_path, index=False)

print("SAVED:", output_path)
print("Samples represented:", sample_enrichment_df["sample_id"].nunique())
print("Sample-level comparisons:", len(sample_enrichment_df))

print("\nSAMPLE-SPECIFIC ENRICHMENT SUMMARY")

summary = (
    sample_enrichment_df
    .groupby(["source_population", "target_population"])
    ["sample_enrichment"]
    .agg(["count", "median", "mean", "std"])
    .reset_index()
)

display(summary.round(3))
```

```python
import h5py
import numpy as np
import pandas as pd

file_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

populations = ["Mast", "Schwann", "Retic Fib III"]
n_pop = len(populations)

with h5py.File(file_path, "r") as f:

    compartment_categories = f["obs/tissue_compartment/categories"].asstr()[:]
    cell_categories = f["obs/cell_type.detailed/categories"].asstr()[:]
    sample_categories = f["obs/sample_id/categories"].asstr()[:]

    dermis_code = int(np.where(compartment_categories == "dermis")[0][0])

    population_codes = {
        name: int(np.where(cell_categories == name)[0][0])
        for name in populations
    }

    n_samples = len(sample_categories)

    compartment_codes = f["obs/tissue_compartment/codes"][:]
    cell_codes = f["obs/cell_type.detailed/codes"][:]
    sample_codes = f["obs/sample_id/codes"][:]

    # Count dermal cells and population abundance per sample
    dermal_cell_counts = np.zeros(n_samples, dtype=np.int64)
    population_cell_counts = np.zeros((n_samples, n_pop), dtype=np.int64)

    for start in range(0, len(cell_codes), 100_000):

        end = min(start + 100_000, len(cell_codes))

        compartments = compartment_codes[start:end]
        cells = cell_codes[start:end]
        samples = sample_codes[start:end]

        valid = (
            (compartments == dermis_code) &
            (samples >= 0)
        )

        dermal_cell_counts += np.bincount(
            samples[valid],
            minlength=n_samples
        )

        for j, name in enumerate(populations):

            mask = valid & (cells == population_codes[name])

            population_cell_counts[:, j] += np.bincount(
                samples[mask],
                minlength=n_samples
            )

    # Count graph edges
    graph = f["obsp/spatial_connectivities"]
    indices = graph["indices"]
    indptr = graph["indptr"]

    edge_counts = np.zeros((n_samples, n_pop, n_pop), dtype=np.int64)
    source_neighbor_totals = np.zeros((n_samples, n_pop), dtype=np.int64)

    for start in range(0, len(cell_codes), 100_000):

        end = min(start + 100_000, len(cell_codes))

        row_counts = np.diff(indptr[start:end + 1])

        if row_counts.sum() == 0:
            continue

        row_ids = np.repeat(np.arange(start, end), row_counts)
        neighbor_ids = indices[indptr[start]:indptr[end]]

        valid = (
            (compartment_codes[row_ids] == dermis_code) &
            (compartment_codes[neighbor_ids] == dermis_code) &
            (sample_codes[row_ids] >= 0) &
            (sample_codes[neighbor_ids] >= 0) &
            (sample_codes[row_ids] == sample_codes[neighbor_ids]) &
            (cell_codes[row_ids] >= 0) &
            (cell_codes[neighbor_ids] >= 0)
        )

        source_ids = row_ids[valid]
        target_ids = neighbor_ids[valid]

        source_samples = sample_codes[source_ids]
        source_types = cell_codes[source_ids]
        target_types = cell_codes[target_ids]

        source_pop_idx = np.full(len(source_ids), -1, dtype=np.int8)
        target_pop_idx = np.full(len(target_ids), -1, dtype=np.int8)

        for j, name in enumerate(populations):
            source_pop_idx[source_types == population_codes[name]] = j
            target_pop_idx[target_types == population_codes[name]] = j

        selected_sources = source_pop_idx >= 0

        denominator_keys = (
            source_samples[selected_sources] * n_pop +
            source_pop_idx[selected_sources]
        )

        source_neighbor_totals += np.bincount(
            denominator_keys,
            minlength=n_samples * n_pop
        ).reshape(n_samples, n_pop)

        selected_edges = selected_sources & (target_pop_idx >= 0)

        edge_keys = (
            source_samples[selected_edges] * n_pop * n_pop +
            source_pop_idx[selected_edges] * n_pop +
            target_pop_idx[selected_edges]
        )

        edge_counts += np.bincount(
            edge_keys,
            minlength=n_samples * n_pop * n_pop
        ).reshape(n_samples, n_pop, n_pop)

# Create sample-level output
records = []

for s, sample_id in enumerate(sample_categories):

    for i, source_name in enumerate(populations):

        denominator = source_neighbor_totals[s, i]

        if denominator == 0:
            continue

        for j, target_name in enumerate(populations):

            observed_edges = edge_counts[s, i, j]

            observed_pct = 100 * observed_edges / denominator

            target_count = population_cell_counts[s, j]
            total_cells = dermal_cell_counts[s]

            abundance_pct = (
                100 * target_count / total_cells
                if total_cells > 0 else np.nan
            )

            enrichment = (
                observed_pct / abundance_pct
                if abundance_pct > 0 else np.nan
            )

            records.append({
                "sample_id": sample_id,
                "source_population": source_name,
                "target_population": target_name,
                "source_neighbor_edges": denominator,
                "observed_target_edges": observed_edges,
                "observed_neighbor_pct": observed_pct,
                "target_dermal_cells": target_count,
                "total_dermal_cells": total_cells,
                "target_abundance_pct": abundance_pct,
                "sample_enrichment": enrichment
            })

sample_enrichment_df = pd.DataFrame(records)

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Sample_Specific_Dermal_Neighborhood_Enrichment.csv"
)

sample_enrichment_df.to_csv(output_path, index=False)

print("SAVED:", output_path)
print("Samples represented:", sample_enrichment_df["sample_id"].nunique())
print("Sample-level comparisons:", len(sample_enrichment_df))

summary = (
    sample_enrichment_df
    .groupby(["source_population", "target_population"])
    ["sample_enrichment"]
    .agg(["count", "median", "mean", "std"])
    .reset_index()
)

print("\nSAMPLE-SPECIFIC ENRICHMENT SUMMARY")
display(summary.round(3))
```

```python
import pandas as pd
import numpy as np

folder = "/content/drive/MyDrive/Neurothesis Resoure Library/"

enrichment = pd.read_csv(
    folder + "Sample_Specific_Dermal_Neighborhood_Enrichment.csv"
)

counts = pd.read_csv(
    folder + "Dermal_Three_Population_Counts_By_Sample.csv"
)

# Confirm one row per sample in the count file
counts = counts.drop_duplicates(subset="sample_id").copy()

# Define sample eligibility
counts["minimum_population_count"] = counts[
    ["Mast", "Schwann", "Retic_Fib_III"]
].min(axis=1)

thresholds = {
    "All samples": 0,
    "At least 20 cells each": 20,
    "At least 50 cells each": 50
}

pairs = [
    ("Schwann", "Retic Fib III"),
    ("Retic Fib III", "Schwann"),
    ("Mast", "Schwann"),
    ("Mast", "Retic Fib III"),
    ("Schwann", "Mast"),
    ("Retic Fib III", "Mast")
]

records = []

for group_name, threshold in thresholds.items():

    eligible_ids = counts.loc[
        counts["minimum_population_count"] >= threshold,
        "sample_id"
    ]

    subset = enrichment[
        enrichment["sample_id"].isin(eligible_ids)
    ]

    for source, target in pairs:

        values = subset.loc[
            (subset["source_population"] == source) &
            (subset["target_population"] == target),
            "sample_enrichment"
        ].dropna()

        records.append({
            "sample_group": group_name,
            "source": source,
            "target": target,
            "sample_count": len(values),
            "median_enrichment": values.median(),
            "mean_enrichment": values.mean(),
            "standard_deviation": values.std()
        })

sensitivity_df = pd.DataFrame(records)

print("SAMPLE-SIZE SENSITIVITY ANALYSIS")
display(sensitivity_df.round(3))

output_path = folder + "Neighborhood_Enrichment_Sample_Size_Sensitivity.csv"
sensitivity_df.to_csv(output_path, index=False)

print("\nSAVED:", output_path)
```

```python
import h5py
import numpy as np
import pandas as pd

file_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

with h5py.File(file_path, "r") as f:

    sample_categories = f["obs/sample_id/categories"].asstr()[:]
    donor_categories = f["obs/donor_id/categories"].asstr()[:]
    site_categories = f["obs/anatomic_site/categories"].asstr()[:]

    sample_codes = f["obs/sample_id/codes"][:]
    donor_codes = f["obs/donor_id/codes"][:]
    site_codes = f["obs/anatomic_site/codes"][:]

    n_samples = len(sample_categories)

    # Store the first observed donor and site for each sample
    sample_donor = np.full(n_samples, -1, dtype=np.int64)
    sample_site = np.full(n_samples, -1, dtype=np.int64)

    donor_mismatches = 0
    site_mismatches = 0

    for start in range(0, len(sample_codes), 100_000):

        end = min(start + 100_000, len(sample_codes))

        s = sample_codes[start:end]
        d = donor_codes[start:end]
        a = site_codes[start:end]

        for sample_index in np.unique(s[s >= 0]):

            mask = s == sample_index

            observed_donors = np.unique(d[mask])
            observed_sites = np.unique(a[mask])

            if sample_donor[sample_index] == -1:
                sample_donor[sample_index] = observed_donors[0]

            if sample_site[sample_index] == -1:
                sample_site[sample_index] = observed_sites[0]

            donor_mismatches += np.sum(
                observed_donors != sample_donor[sample_index]
            )

            site_mismatches += np.sum(
                observed_sites != sample_site[sample_index]
            )

    mapping = pd.DataFrame({
        "sample_id": sample_categories,
        "donor_id": [
            donor_categories[x] if x >= 0 else None
            for x in sample_donor
        ],
        "anatomic_site": [
            site_categories[x] if x >= 0 else None
            for x in sample_site
        ]
    })

    output_path = (
        "/content/drive/MyDrive/Neurothesis Resoure Library/"
        "Verified_Sample_Donor_Site_Mapping.csv"
    )

    mapping.to_csv(output_path, index=False)

print("SAVED:", output_path)
print("Samples mapped:", mapping["sample_id"].nunique())
print("Unique donors:", mapping["donor_id"].nunique())
print("Unique anatomical sites:", mapping["anatomic_site"].nunique())
print("Samples with missing donor/site:", mapping[["donor_id", "anatomic_site"]].isna().any(axis=1).sum())
print("Donor consistency mismatches:", donor_mismatches)
print("Anatomical-site consistency mismatches:", site_mismatches)

print("\nSAMPLES PER DONOR:")
display(
    mapping.groupby("donor_id")
    .size()
    .sort_values(ascending=False)
    .rename("sample_count")
    .reset_index()
)

print("\nSAMPLES PER ANATOMICAL SITE:")
display(
    mapping.groupby("anatomic_site")
    .size()
    .sort_values(ascending=False)
    .rename("sample_count")
    .reset_index()
)
```

```python
import pandas as pd

folder = "/content/drive/MyDrive/Neurothesis Resoure Library/"

mapping = pd.read_csv(
    folder + "Verified_Sample_Donor_Site_Mapping.csv"
)

enrichment = pd.read_csv(
    folder + "Sample_Specific_Dermal_Neighborhood_Enrichment.csv"
)

# Attach verified donor and anatomical-site information
merged = enrichment.merge(
    mapping,
    on="sample_id",
    how="left",
    validate="many_to_one"
)

# Confirm the merge was complete
print("Rows before merge:", len(enrichment))
print("Rows after merge:", len(merged))
print("Rows missing donor:", merged["donor_id"].isna().sum())
print("Unique donors represented:", merged["donor_id"].nunique())

# One summary per donor and population pair
donor_summary = (
    merged.groupby(
        ["source_population", "target_population", "donor_id"],
        as_index=False
    )
    .agg(
        sample_count=("sample_id", "nunique"),
        median_sample_enrichment=("sample_enrichment", "median"),
        mean_sample_enrichment=("sample_enrichment", "mean")
    )
)

# Summarize across donors, giving each donor equal weight
pair_summary = (
    donor_summary.groupby(
        ["source_population", "target_population"]
    )
    .agg(
        donor_count=("donor_id", "nunique"),
        total_samples=("sample_count", "sum"),
        median_donor_enrichment=("median_sample_enrichment", "median"),
        mean_donor_enrichment=("median_sample_enrichment", "mean"),
        donor_enrichment_sd=("median_sample_enrichment", "std"),
        donor_enrichment_q25=("median_sample_enrichment", lambda x: x.quantile(0.25)),
        donor_enrichment_q75=("median_sample_enrichment", lambda x: x.quantile(0.75))
    )
    .reset_index()
)

# Save both levels of results
donor_path = folder + "Donor_Level_Neighborhood_Enrichment.csv"
pair_path = folder + "Donor_Level_Neighborhood_Enrichment_Summary.csv"

donor_summary.to_csv(donor_path, index=False)
pair_summary.to_csv(pair_path, index=False)

print("\nSAVED DONOR-LEVEL DATA:", donor_path)
print("SAVED POPULATION-PAIR SUMMARY:", pair_path)

print("\nDONOR-LEVEL POPULATION-PAIR SUMMARY:")
display(
    pair_summary.sort_values(
        ["source_population", "target_population"]
    ).round(3)
)

print("\nNUMBER OF SAMPLES CONTRIBUTED PER DONOR:")
display(
    donor_summary.groupby("donor_id")["sample_count"]
    .agg(["count", "sum", "median", "max"])
    .rename(columns={
        "count": "population_pairs",
        "sum": "total_pair_sample_records",
        "median": "median_samples_per_pair",
        "max": "max_samples_per_pair"
    })
    .sort_values("total_pair_sample_records", ascending=False)
)
```

```python
import pandas as pd
import numpy as np

folder = "/content/drive/MyDrive/Neurothesis Resoure Library/"

# Load verified data
mapping = pd.read_csv(
    folder + "Verified_Sample_Donor_Site_Mapping.csv"
)

enrichment = pd.read_csv(
    folder + "Sample_Specific_Dermal_Neighborhood_Enrichment.csv"
)

# Attach donor and anatomical-site information
merged = enrichment.merge(
    mapping,
    on="sample_id",
    how="left",
    validate="many_to_one"
)

print("Rows after merge:", len(merged))
print("Missing donor assignments:", merged["donor_id"].isna().sum())
print("Missing site assignments:", merged["anatomic_site"].isna().sum())

# First calculate one median per donor, site, and population pair.
# This prevents donors with multiple samples at the same site
# from receiving extra weight in the site-level summary.

donor_site = (
    merged.groupby(
        [
            "donor_id",
            "anatomic_site",
            "source_population",
            "target_population"
        ],
        observed=True
    )
    .agg(
        donor_site_median_enrichment=("sample_enrichment", "median"),
        sample_count=("sample_id", "nunique")
    )
    .reset_index()
)

# Summarize across donors within each anatomical site.
site_summary = (
    donor_site.groupby(
        [
            "anatomic_site",
            "source_population",
            "target_population"
        ],
        observed=True
    )
    .agg(
        donor_count=("donor_id", "nunique"),
        total_samples=("sample_count", "sum"),
        median_donor_enrichment=(
            "donor_site_median_enrichment", "median"
        ),
        mean_donor_enrichment=(
            "donor_site_median_enrichment", "mean"
        ),
        donor_enrichment_q25=(
            "donor_site_median_enrichment",
            lambda x: x.quantile(0.25)
        ),
        donor_enrichment_q75=(
            "donor_site_median_enrichment",
            lambda x: x.quantile(0.75)
        )
    )
    .reset_index()
)

output_path = folder + "Donor_Balanced_Site_Specific_Enrichment.csv"
site_summary.to_csv(output_path, index=False)

# Focus on cross-population relationships, excluding self-neighbor pairs.
cross_population = site_summary[
    site_summary["source_population"]
    != site_summary["target_population"]
].copy()

print("\nSAVED:", output_path)
print("Donor-site observations:", len(donor_site))
print("Anatomical-site population-pair summaries:", len(site_summary))

print("\nCROSS-POPULATION ENRICHMENT BY ANATOMICAL SITE:")
display(
    cross_population.sort_values(
        ["source_population", "target_population", "anatomic_site"]
    ).round(3).reset_index(drop=True)
)
```

```python
import h5py
import numpy as np

file_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

with h5py.File(file_path, "r") as f:

    indptr = f["obsp/spatial_connectivities/indptr"]
    indices = f["obsp/spatial_connectivities/indices"]

    n_cells = len(indptr) - 1
    n_edges = len(indices)

    print("CELLS:", n_cells)
    print("DIRECTED EDGES:", n_edges)
    print("MEAN OUT-DEGREE:", n_edges / n_cells)

    # Inspect degree distribution
    degrees = np.diff(indptr[:])

    print("\nDEGREE DISTRIBUTION")
    print("Minimum:", degrees.min())
    print("Median:", np.median(degrees))
    print("Mean:", degrees.mean())
    print("Maximum:", degrees.max())

    # Check whether sampled directed edges have reciprocal edges.
    rng = np.random.default_rng(42)

    sampled_rows = rng.choice(
        n_cells,
        size=min(1000, n_cells),
        replace=False
    )

    reciprocal = 0
    sampled_edge_count = 0

    for row in sampled_rows:

        start = indptr[row]
        end = indptr[row + 1]

        neighbors = indices[start:end]

        for neighbor in neighbors:

            neighbor_start = indptr[neighbor]
            neighbor_end = indptr[neighbor + 1]

            reverse_neighbors = indices[
                neighbor_start:neighbor_end
            ]

            sampled_edge_count += 1

            if row in reverse_neighbors:
                reciprocal += 1

    print("\nRECIPROCITY CHECK")
    print("Sampled directed edges:", sampled_edge_count)
    print("Reciprocal edges:", reciprocal)
    print(
        "Reciprocity percentage:",
        round(100 * reciprocal / sampled_edge_count, 3)
    )
```

```python
import h5py
import numpy as np
from scipy.sparse import csr_matrix

file_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

with h5py.File(file_path, "r") as f:

    indptr = f["obsp/spatial_connectivities/indptr"][:]
    indices = f["obsp/spatial_connectivities/indices"][:]

n_cells = len(indptr) - 1

# Reconstruct the sparse connectivity graph.
graph = csr_matrix(
    (
        np.ones(len(indices), dtype=np.uint8),
        indices,
        indptr
    ),
    shape=(n_cells, n_cells)
)

print("GRAPH STRUCTURE")
print("Shape:", graph.shape)
print("Stored edges:", graph.nnz)
print("Sorted indices:", graph.has_sorted_indices)
print("Canonical format:", graph.has_canonical_format)

# Compare every edge with its reverse.
asymmetric_edges = (graph != graph.T).nnz

print("\nFULL GRAPH RECIPROCITY")
print("Asymmetric directed edges:", asymmetric_edges)
print(
    "Reciprocity:",
    round(100 * (1 - asymmetric_edges / graph.nnz), 6),
    "%"
)

print("\nSELF-CONNECTIONS")
print("Self-connections:", graph.diagonal().sum())

print("\nDUPLICATE EDGE CHECK")
print("Duplicate entries:", len(indices) - graph.nnz)
```

```python
import h5py
import numpy as np
import pandas as pd
from scipy.sparse import csr_matrix

file_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

with h5py.File(file_path, "r") as f:

    # Read categorical annotations
    cell_categories = f["obs/cell_type.detailed/categories"].asstr()[:]
    cell_codes = f["obs/cell_type.detailed/codes"][:]

    compartment_categories = f["obs/tissue_compartment/categories"].asstr()[:]
    compartment_codes = f["obs/tissue_compartment/codes"][:]

    # Reconstruct graph
    indptr = f["obsp/spatial_connectivities/indptr"][:]
    indices = f["obsp/spatial_connectivities/indices"][:]

n_cells = len(cell_codes)

graph = csr_matrix(
    (
        np.ones(len(indices), dtype=np.uint8),
        indices,
        indptr
    ),
    shape=(n_cells, n_cells)
)

# Select dermal cells belonging to the three populations.
selected_populations = [
    "Mast",
    "Schwann",
    "Retic Fib III"
]

cell_names = np.array([
    cell_categories[x] if x >= 0 else "Unknown"
    for x in cell_codes
])

compartment_names = np.array([
    compartment_categories[x] if x >= 0 else "Unknown"
    for x in compartment_codes
])

dermal = compartment_names == "dermis"
selected = dermal & np.isin(cell_names, selected_populations)

# Count undirected edges once by retaining only upper-triangle entries.
upper = graph.tocsr().maximum(graph.T.tocsr())
upper = upper.tocoo()

source_selected = selected[upper.row]
target_selected = selected[upper.col]

keep = source_selected & target_selected

edge_rows = upper.row[keep]
edge_cols = upper.col[keep]

edge_source = cell_names[edge_rows]
edge_target = cell_names[edge_cols]

# Sort each population pair alphabetically so that
# each undirected pair is counted exactly once.
pair_counts = {}

for a, b in zip(edge_source, edge_target):

    pair = tuple(sorted((a, b)))
    pair_counts[pair] = pair_counts.get(pair, 0) + 1

results = []

for i, source in enumerate(selected_populations):

    for target in selected_populations[i:]:

        pair = tuple(sorted((source, target)))

        results.append({
            "population_1": pair[0],
            "population_2": pair[1],
            "undirected_edge_count": pair_counts.get(pair, 0)
        })

edge_summary = pd.DataFrame(results)

print("SELECTED DERMAL CELLS:", selected.sum())
print("SELECTED UNDIRECTED EDGES:", len(edge_rows))

print("\nPOPULATION-PAIR EDGE COUNTS:")
display(edge_summary)

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Dermal_Undirected_Three_Population_Edges.csv"
)

edge_summary.to_csv(output_path, index=False)

print("\nSAVED:", output_path)
```

```python
import h5py
import numpy as np
import pandas as pd

file_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

with h5py.File(file_path, "r") as f:

    cell_categories = f["obs/cell_type.detailed/categories"].asstr()[:]
    cell_codes = f["obs/cell_type.detailed/codes"][:]

    compartment_categories = f["obs/tissue_compartment/categories"].asstr()[:]
    compartment_codes = f["obs/tissue_compartment/codes"][:]

    indptr = f["obsp/spatial_connectivities/indptr"][:]

# Convert categorical codes to labels
cell_names = np.array([
    cell_categories[x] if x >= 0 else "Unknown"
    for x in cell_codes
])

compartment_names = np.array([
    compartment_categories[x] if x >= 0 else "Unknown"
    for x in compartment_codes
])

# Degree of each cell in the full spatial graph
degrees = np.diff(indptr)

selected_populations = [
    "Mast",
    "Schwann",
    "Retic Fib III"
]

results = []

for population in selected_populations:

    mask = (
        (cell_names == population) &
        (compartment_names == "dermis")
    )

    population_degrees = degrees[mask]

    results.append({
        "population": population,
        "cell_count": int(mask.sum()),
        "mean_degree": population_degrees.mean(),
        "median_degree": np.median(population_degrees),
        "degree_q25": np.percentile(population_degrees, 25),
        "degree_q75": np.percentile(population_degrees, 75),
        "zero_degree_cells": int((population_degrees == 0).sum())
    })

degree_summary = pd.DataFrame(results)

print("DERMAL POPULATION DEGREE DISTRIBUTIONS")
display(degree_summary.round(3))

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Dermal_Population_Degree_Summary.csv"
)

degree_summary.to_csv(output_path, index=False)

print("\nSAVED:", output_path)
```

```python
import h5py
import numpy as np
import pandas as pd

file_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

with h5py.File(file_path, "r") as f:

    cell_categories = f["obs/cell_type.detailed/categories"].asstr()[:]
    cell_codes = f["obs/cell_type.detailed/codes"][:]

    compartment_categories = f["obs/tissue_compartment/categories"].asstr()[:]
    compartment_codes = f["obs/tissue_compartment/codes"][:]

    indptr = f["obsp/spatial_connectivities/indptr"][:]
    indices = f["obsp/spatial_connectivities/indices"][:]

cell_names = np.array([
    cell_categories[x] if x >= 0 else "Unknown"
    for x in cell_codes
])

compartment_names = np.array([
    compartment_categories[x] if x >= 0 else "Unknown"
    for x in compartment_codes
])

dermal = compartment_names == "dermis"

# Count only graph edges whose two endpoints are dermal.
dermal_degrees = np.zeros(len(cell_names), dtype=np.int32)

for i in np.where(dermal)[0]:

    neighbors = indices[indptr[i]:indptr[i+1]]

    dermal_degrees[i] = np.sum(dermal[neighbors])

selected_populations = [
    "Mast",
    "Schwann",
    "Retic Fib III"
]

results = []

for population in selected_populations:

    mask = dermal & (cell_names == population)
    values = dermal_degrees[mask]

    results.append({
        "population": population,
        "cell_count": int(mask.sum()),
        "mean_dermal_degree": values.mean(),
        "median_dermal_degree": np.median(values),
        "degree_q25": np.percentile(values, 25),
        "degree_q75": np.percentile(values, 75),
        "zero_dermal_degree_cells": int((values == 0).sum())
    })

dermal_degree_summary = pd.DataFrame(results)

print("DERMAL-ONLY DEGREE DISTRIBUTIONS")
display(dermal_degree_summary.round(3))

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Dermal_Only_Population_Degree_Summary.csv"
)

dermal_degree_summary.to_csv(output_path, index=False)

print("\nSAVED:", output_path)
```

```python
import h5py
import numpy as np
import pandas as pd

file_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

with h5py.File(file_path, "r") as f:

    sample_categories = f["obs/sample_id/categories"].asstr()[:]
    sample_codes = f["obs/sample_id/codes"][:]

    cell_categories = f["obs/cell_type.detailed/categories"].asstr()[:]
    cell_codes = f["obs/cell_type.detailed/codes"][:]

    compartment_categories = f["obs/tissue_compartment/categories"].asstr()[:]
    compartment_codes = f["obs/tissue_compartment/codes"][:]

    coordinates = f["obsm/spatial"][:]

# Identify the three populations in the dermis.
population_names = np.array([
    cell_categories[c] if c >= 0 else "Unknown"
    for c in cell_codes
])

compartment_names = np.array([
    compartment_categories[c] if c >= 0 else "Unknown"
    for c in compartment_codes
])

selected_populations = ["Mast", "Schwann", "Retic Fib III"]

mask = (
    (compartment_names == "dermis") &
    np.isin(population_names, selected_populations)
)

# Keep only selected cells to minimize memory use.
df = pd.DataFrame({
    "sample_code": sample_codes[mask],
    "population": population_names[mask],
    "x": coordinates[mask, 0],
    "y": coordinates[mask, 1]
})

# Calculate sample-relative spatial coordinates.
df["x_min"] = df.groupby("sample_code")["x"].transform("min")
df["y_min"] = df.groupby("sample_code")["y"].transform("min")

block_widths = [500, 1000, 2000]

summaries = []

for width in block_widths:

    df["block_x"] = np.floor(
        (df["x"] - df["x_min"]) / width
    ).astype(int)

    df["block_y"] = np.floor(
        (df["y"] - df["y_min"]) / width
    ).astype(int)

    counts = (
        df.groupby(
            ["sample_code", "block_x", "block_y", "population"]
        )
        .size()
        .unstack(fill_value=0)
    )

    for population in selected_populations:
        if population not in counts.columns:
            counts[population] = 0

    counts = counts[selected_populations]

    occupied = counts.sum(axis=1) > 0
    multi_population = (counts > 0).sum(axis=1) >= 2
    all_three = (counts > 0).all(axis=1)

    summaries.append({
        "block_width": width,
        "occupied_blocks": int(occupied.sum()),
        "blocks_with_at_least_2_populations": int(multi_population.sum()),
        "blocks_with_all_3_populations": int(all_three.sum()),
        "median_selected_cells_per_block": counts.sum(axis=1).median(),
        "blocks_with_at_least_5_cells_of_each_population": int(
            (counts >= 5).all(axis=1).sum()
        ),
        "blocks_with_at_least_10_cells_of_each_population": int(
            (counts >= 10).all(axis=1).sum()
        )
    })

summary = pd.DataFrame(summaries)

print("SPATIAL BLOCK OCCUPANCY")
display(summary)

print("\nSAVED BLOCK-SCALE DIAGNOSTIC")
output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Spatial_Block_Occupancy_Diagnostic.csv"
)

summary.to_csv(output_path, index=False)
print(output_path)
```

```python
import h5py
import numpy as np
import pandas as pd
from scipy.sparse import csr_matrix, triu

file_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

# Load the required data.
with h5py.File(file_path, "r") as f:

    cell_categories = f["obs/cell_type.detailed/categories"].asstr()[:]
    cell_codes = f["obs/cell_type.detailed/codes"][:]

    compartment_categories = f["obs/tissue_compartment/categories"].asstr()[:]
    compartment_codes = f["obs/tissue_compartment/codes"][:]

    sample_codes = f["obs/sample_id/codes"][:]
    coordinates = f["obsm/spatial"][:]

    indptr = f["obsp/spatial_connectivities/indptr"][:]
    indices = f["obsp/spatial_connectivities/indices"][:]

n_cells = len(cell_codes)

# Reconstruct cell labels.
cell_names = np.array([
    cell_categories[c] if c >= 0 else "Unknown"
    for c in cell_codes
])

compartment_names = np.array([
    compartment_categories[c] if c >= 0 else "Unknown"
    for c in compartment_codes
])

# Select the three populations in the dermis.
population_names = ["Mast", "Schwann", "Retic Fib III"]

selected_mask = (
    (compartment_names == "dermis") &
    np.isin(cell_names, population_names)
)

selected_indices = np.where(selected_mask)[0]

# Encode population labels as integers.
label_map = {
    "Mast": 0,
    "Schwann": 1,
    "Retic Fib III": 2
}

labels = np.array([
    label_map[cell_names[i]]
    for i in selected_indices
], dtype=np.int8)

# Construct the symmetric graph.
graph = csr_matrix(
    (np.ones(len(indices), dtype=np.uint8), indices, indptr),
    shape=(n_cells, n_cells)
)

# Extract each undirected edge exactly once.
edges = triu(graph, k=1, format="coo")

# Keep edges with both endpoints in the selected populations and dermis.
inverse = np.full(n_cells, -1, dtype=np.int32)
inverse[selected_indices] = np.arange(len(selected_indices))

edge_a = inverse[edges.row]
edge_b = inverse[edges.col]

valid = (edge_a >= 0) & (edge_b >= 0)

edge_a = edge_a[valid]
edge_b = edge_b[valid]

# Create spatial blocks within each sample.
selected_coords = coordinates[selected_indices]
selected_samples = sample_codes[selected_indices]

width = 1000

block_x = np.zeros(len(selected_indices), dtype=np.int32)
block_y = np.zeros(len(selected_indices), dtype=np.int32)

for sample in np.unique(selected_samples):

    sample_mask = selected_samples == sample

    x_min = selected_coords[sample_mask, 0].min()
    y_min = selected_coords[sample_mask, 1].min()

    block_x[sample_mask] = np.floor(
        (selected_coords[sample_mask, 0] - x_min) / width
    ).astype(int)

    block_y[sample_mask] = np.floor(
        (selected_coords[sample_mask, 1] - y_min) / width
    ).astype(int)

# Identify cells belonging to the same sample and spatial block.
block_table = pd.DataFrame({
    "sample": selected_samples,
    "block_x": block_x,
    "block_y": block_y
})

block_groups = [
    group.index.to_numpy()
    for _, group in block_table.groupby(
        ["sample", "block_x", "block_y"],
        sort=False
    )
]

# Observed Schwann-Retic Fib III undirected edge count.
observed = np.sum(
    ((labels[edge_a] == 1) & (labels[edge_b] == 2)) |
    ((labels[edge_a] == 2) & (labels[edge_b] == 1))
)

print("SELECTED CELLS:", len(labels))
print("UNDIRECTED SELECTED EDGES:", len(edge_a))
print("OBSERVED SCHWANN-RETIC FIB III EDGES:", observed)
print("SPATIAL BLOCKS:", len(block_groups))

# Pilot permutations.
n_permutations = 199
rng = np.random.default_rng(20261108)

null_counts = np.zeros(n_permutations, dtype=np.int32)

for p in range(n_permutations):

    shuffled = labels.copy()

    for group in block_groups:
        shuffled[group] = rng.permutation(shuffled[group])

    null_counts[p] = np.sum(
        ((shuffled[edge_a] == 1) & (shuffled[edge_b] == 2)) |
        ((shuffled[edge_a] == 2) & (shuffled[edge_b] == 1))
    )

# Summarize the null distribution.
null_summary = pd.DataFrame([{
    "observed_edges": observed,
    "null_mean": null_counts.mean(),
    "null_sd": null_counts.std(ddof=1),
    "null_median": np.median(null_counts),
    "null_2_5_percentile": np.percentile(null_counts, 2.5),
    "null_97_5_percentile": np.percentile(null_counts, 97.5),
    "permutations_at_least_observed": int(
        np.sum(null_counts >= observed)
    ),
    "empirical_upper_tail_p": (
        1 + np.sum(null_counts >= observed)
    ) / (n_permutations + 1)
}])

print("\nPILOT NULL MODEL RESULTS")
display(null_summary.round(4))

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Schwann_ReticFibIII_Pilot_Block_Permutation.csv"
)

pd.DataFrame({"null_edge_count": null_counts}).to_csv(
    output_path, index=False
)

print("\nSAVED:", output_path)
```

```python
import numpy as np
import pandas as pd

# Retain the original 199 pilot permutations.
pilot_counts = null_counts.copy()

# Add 1,800 independent permutations.
additional_permutations = 1800
rng = np.random.default_rng(20261109)

additional_counts = np.zeros(
    additional_permutations,
    dtype=np.int32
)

for p in range(additional_permutations):

    shuffled = labels.copy()

    for group in block_groups:
        shuffled[group] = rng.permutation(shuffled[group])

    additional_counts[p] = np.sum(
        ((shuffled[edge_a] == 1) & (shuffled[edge_b] == 2)) |
        ((shuffled[edge_a] == 2) & (shuffled[edge_b] == 1))
    )

# Combine all 1,999 permutations.
all_null_counts = np.concatenate([
    pilot_counts,
    additional_counts
])

n_permutations = len(all_null_counts)

exceedances = int(np.sum(all_null_counts >= observed))

summary = pd.DataFrame([{
    "observed_edges": observed,
    "permutations": n_permutations,
    "null_mean": all_null_counts.mean(),
    "null_sd": all_null_counts.std(ddof=1),
    "null_median": np.median(all_null_counts),
    "null_2_5_percentile": np.percentile(all_null_counts, 2.5),
    "null_97_5_percentile": np.percentile(all_null_counts, 97.5),
    "observed_excess_over_null_pct": (
        (observed / all_null_counts.mean()) - 1
    ) * 100,
    "permutations_at_least_observed": exceedances,
    "empirical_upper_tail_p": (
        1 + exceedances
    ) / (n_permutations + 1)
}])

print("FINAL-SIZE PILOT NULL MODEL")
display(summary.round(5))

# Save permutation distribution.
distribution_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Schwann_ReticFibIII_1999_Block_Permutations.csv"
)

pd.DataFrame({
    "null_edge_count": all_null_counts
}).to_csv(distribution_path, index=False)

# Save summary.
summary_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Schwann_ReticFibIII_1999_Block_Permutation_Summary.csv"
)

summary.to_csv(summary_path, index=False)

print("\nSAVED DISTRIBUTION:", distribution_path)
print("SAVED SUMMARY:", summary_path)
```

```python
import numpy as np
import pandas as pd

block_widths = [500, 1000, 2000]
offset_fractions = [0, 0.5]
n_permutations = 999

rng = np.random.default_rng(20261110)

sensitivity_results = []

for width in block_widths:

    for offset_fraction in offset_fractions:

        offset = width * offset_fraction

        block_x = np.floor(
            (selected_coords[:, 0] -
             np.array([
                 selected_coords[selected_samples == s, 0].min()
                 for s in selected_samples
             ]) + offset) / width
        ).astype(int)

        block_y = np.floor(
            (selected_coords[:, 1] -
             np.array([
                 selected_coords[selected_samples == s, 1].min()
                 for s in selected_samples
             ]) + offset) / width
        ).astype(int)

        block_table = pd.DataFrame({
            "sample": selected_samples,
            "block_x": block_x,
            "block_y": block_y
        })

        groups = [
            group.index.to_numpy()
            for _, group in block_table.groupby(
                ["sample", "block_x", "block_y"],
                sort=False
            )
        ]

        null_counts = np.zeros(n_permutations, dtype=np.int32)

        for p in range(n_permutations):

            shuffled = labels.copy()

            for group in groups:
                shuffled[group] = rng.permutation(shuffled[group])

            null_counts[p] = np.sum(
                ((shuffled[edge_a] == 1) &
                 (shuffled[edge_b] == 2)) |
                ((shuffled[edge_a] == 2) &
                 (shuffled[edge_b] == 1))
            )

        exceedances = int(np.sum(null_counts >= observed))

        sensitivity_results.append({
            "block_width": width,
            "grid_offset_fraction": offset_fraction,
            "occupied_blocks": len(groups),
            "observed_edges": int(observed),
            "null_mean": null_counts.mean(),
            "null_sd": null_counts.std(ddof=1),
            "observed_excess_pct": (
                observed / null_counts.mean() - 1
            ) * 100,
            "permutations_at_least_observed": exceedances,
            "empirical_upper_tail_p": (
                1 + exceedances
            ) / (n_permutations + 1)
        })

        print(
            f"Completed width={width}, "
            f"offset={offset_fraction}"
        )

sensitivity_df = pd.DataFrame(sensitivity_results)

print("\nSPATIAL SCALE AND GRID SENSITIVITY")
display(sensitivity_df.round(5))

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Schwann_ReticFibIII_Spatial_Sensitivity.csv"
)

sensitivity_df.to_csv(output_path, index=False)

print("\nSAVED:", output_path)
```

```python
import numpy as np
import pandas as pd

# Confirm the three population labels.
print("Population label counts:")
print(pd.Series(labels).value_counts().sort_index())

# Calculate observed unique undirected edge counts.
pair_definitions = {
    "Mast–Schwann": (0, 1),
    "Mast–Retic Fib III": (0, 2),
    "Schwann–Retic Fib III": (1, 2)
}

observed_pair_counts = {}

for pair_name, (a, b) in pair_definitions.items():
    observed_pair_counts[pair_name] = int(np.sum(
        ((labels[edge_a] == a) & (labels[edge_b] == b)) |
        ((labels[edge_a] == b) & (labels[edge_b] == a))
    ))

print("\nOBSERVED UNIQUE UNDIRECTED EDGES")
display(pd.DataFrame(
    list(observed_pair_counts.items()),
    columns=["population_pair", "observed_edges"]
))

# Generate a shared set of 1,999 block-constrained permutations.
n_permutations = 1999
rng = np.random.default_rng(20261111)

null_pair_counts = {
    pair_name: np.zeros(n_permutations, dtype=np.int32)
    for pair_name in pair_definitions
}

for p in range(n_permutations):

    shuffled = labels.copy()

    for group in block_groups:
        shuffled[group] = rng.permutation(shuffled[group])

    for pair_name, (a, b) in pair_definitions.items():

        null_pair_counts[pair_name][p] = np.sum(
            ((shuffled[edge_a] == a) & (shuffled[edge_b] == b)) |
            ((shuffled[edge_a] == b) & (shuffled[edge_b] == a))
        )

# Summarize each pair.
pair_results = []

for pair_name in pair_definitions:

    null = null_pair_counts[pair_name]
    observed_count = observed_pair_counts[pair_name]
    exceedances = int(np.sum(null >= observed_count))

    pair_results.append({
        "population_pair": pair_name,
        "observed_edges": observed_count,
        "null_mean": null.mean(),
        "null_sd": null.std(ddof=1),
        "observed_excess_pct": (
            observed_count / null.mean() - 1
        ) * 100,
        "permutations_at_least_observed": exceedances,
        "empirical_upper_tail_p": (
            1 + exceedances
        ) / (n_permutations + 1)
    })

pair_results_df = pd.DataFrame(pair_results)

print("\nTHREE-PAIR BLOCK-PERMUTATION RESULTS")
display(pair_results_df.round(5))

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Three_Population_Pair_Block_Permutation.csv"
)

pair_results_df.to_csv(output_path, index=False)

print("\nSAVED:", output_path)
```

```python
import numpy as np
import pandas as pd

# Benjamini-Hochberg correction.
p_values = donor_permutation_df[
    "empirical_upper_tail_p"
].to_numpy()

n_tests = len(p_values)

sorted_indices = np.argsort(p_values)
sorted_p = p_values[sorted_indices]

adjusted_sorted = (
    sorted_p * n_tests /
    np.arange(1, n_tests + 1)
)

# Enforce monotonicity.
adjusted_sorted = np.minimum.accumulate(
    adjusted_sorted[::-1]
)[::-1]

adjusted_sorted = np.clip(adjusted_sorted, 0, 1)

adjusted_p = np.empty(n_tests)
adjusted_p[sorted_indices] = adjusted_sorted

donor_permutation_df["BH_adjusted_p"] = adjusted_p

donor_permutation_df["BH_significant_0.05"] = (
    donor_permutation_df["BH_adjusted_p"] <= 0.05
)

print("DONOR RESULTS WITH MULTIPLE-TESTING CORRECTION")

display(
    donor_permutation_df.sort_values(
        "BH_adjusted_p"
    ).round(4)
)

print("\nMULTIPLE-TESTING SUMMARY")
print(
    "Donors significant after BH correction:",
    int(donor_permutation_df["BH_significant_0.05"].sum()),
    "of",
    n_tests
)

# Validate that graph edges remain within donor and sample boundaries.
edge_donors_a = selected_donors[edge_a]
edge_donors_b = selected_donors[edge_b]

edge_samples_a = selected_samples[edge_a]
edge_samples_b = selected_samples[edge_b]

cross_donor_edges = np.sum(
    edge_donors_a != edge_donors_b
)

cross_sample_edges = np.sum(
    edge_samples_a != edge_samples_b
)

print("\nGRAPH EDGE VALIDATION")
print("Total unique graph edges:", len(edge_a))
print("Edges crossing donor boundaries:", int(cross_donor_edges))
print("Edges crossing sample boundaries:", int(cross_sample_edges))

# Save corrected results.
output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Schwann_ReticFibIII_Donor_Permutations_BH_Corrected.csv"
)

donor_permutation_df.to_csv(output_path, index=False)

print("\nSAVED:", output_path)
```

```python
import numpy as np
import pandas as pd

n_permutations = 9999
rng = np.random.default_rng(20261112)

n_donors = len(donor_names)

null_by_donor = np.zeros(
    (n_permutations, n_donors),
    dtype=np.int32
)

for p in range(n_permutations):

    shuffled = labels.copy()

    for group in block_groups:
        shuffled[group] = rng.permutation(shuffled[group])

    pair_edges = (
        ((shuffled[edge_a] == 1) & (shuffled[edge_b] == 2)) |
        ((shuffled[edge_a] == 2) & (shuffled[edge_b] == 1))
    )

    null_by_donor[p] = np.bincount(
        edge_donor_codes[pair_edges],
        minlength=n_donors
    )

    if (p + 1) % 1000 == 0:
        print(f"Completed {p + 1} / {n_permutations}")

# Calculate donor-specific permutation statistics.
results = []

for i, donor in enumerate(donor_names):

    null_values = null_by_donor[:, i]
    observed_count = observed_by_donor[i]

    exceedances = int(
        np.sum(null_values >= observed_count)
    )

    null_mean = null_values.mean()

    results.append({
        "donor_id": donor,
        "observed_edges": int(observed_count),
        "null_mean": null_mean,
        "null_sd": null_values.std(ddof=1),
        "observed_excess_pct": (
            (observed_count / null_mean) - 1
        ) * 100 if null_mean > 0 else np.nan,
        "permutations_at_least_observed": exceedances,
        "empirical_upper_tail_p": (
            1 + exceedances
        ) / (n_permutations + 1)
    })

final_donor_df = pd.DataFrame(results)

# Benjamini-Hochberg correction.
p_values = final_donor_df["empirical_upper_tail_p"].to_numpy()

m = len(p_values)
order = np.argsort(p_values)
sorted_p = p_values[order]

adjusted_sorted = sorted_p * m / np.arange(1, m + 1)
adjusted_sorted = np.minimum.accumulate(adjusted_sorted[::-1])[::-1]
adjusted_sorted = np.minimum(adjusted_sorted, 1)

adjusted_p = np.empty(m)
adjusted_p[order] = adjusted_sorted

final_donor_df["BH_adjusted_p"] = adjusted_p
final_donor_df["BH_significant_0.05"] = adjusted_p < 0.05

final_donor_df = final_donor_df.sort_values(
    "empirical_upper_tail_p"
).reset_index(drop=True)

print("\nFINAL DONOR RESULTS: 9,999 PERMUTATIONS")
display(final_donor_df.round(5))

print("\nMULTIPLE-TESTING SUMMARY")
print(
    "Donors significant after BH correction:",
    int(final_donor_df["BH_significant_0.05"].sum()),
    "of",
    m
)

print(
    "Donors with positive observed excess:",
    int((final_donor_df["observed_excess_pct"] > 0).sum()),
    "of",
    m
)

# Save final results.
output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Schwann_ReticFibIII_Donor_Permutations_9999_BH.csv"
)

final_donor_df.to_csv(output_path, index=False)

print("\nSAVED:", output_path)
```

```python
import numpy as np
import pandas as pd
from pathlib import Path

# Match each donor's observed edge count to its null distribution
donors = final_donor_df["donor_id"].tolist()

observed_by_donor = (
    final_donor_df.set_index("donor_id")["observed_edges"].to_dict()
)

null_arrays = {
    donor: np.asarray(null_by_donor[donor])
    for donor in donors
}

# Confirm that all donors have the same number of permutations
permutation_counts = {len(values) for values in null_arrays.values()}
assert len(permutation_counts) == 1, (
    f"Unequal permutation counts: {permutation_counts}"
)

n_permutations = permutation_counts.pop()

results = []

# First calculate the full combined result, then leave out each donor
for excluded_donor in [None] + donors:

    included_donors = [
        donor for donor in donors
        if donor != excluded_donor
    ]

    observed_total = sum(
        observed_by_donor[donor]
        for donor in included_donors
    )

    combined_null = np.sum(
        np.vstack([
            null_arrays[donor]
            for donor in included_donors
        ]),
        axis=0
    )

    exceedances = int(np.sum(combined_null >= observed_total))

    empirical_p = (
        exceedances + 1
    ) / (
        n_permutations + 1
    )

    null_mean = float(np.mean(combined_null))

    excess_pct = (
        (observed_total - null_mean) / null_mean
    ) * 100

    results.append({
        "excluded_donor": (
            "None" if excluded_donor is None
            else excluded_donor
        ),
        "donors_included": len(included_donors),
        "observed_edges": observed_total,
        "null_mean": null_mean,
        "null_sd": float(np.std(combined_null, ddof=1)),
        "observed_excess_pct": excess_pct,
        "permutations_at_least_observed": exceedances,
        "empirical_p": empirical_p
    })

leave_one_out_df = pd.DataFrame(results)

print("LEAVE-ONE-DONOR-OUT SENSITIVITY ANALYSIS")
display(
    leave_one_out_df.round(5)
)

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Schwann_ReticFibIII_Leave_One_Donor_Out.csv"
)

leave_one_out_df.to_csv(output_path, index=False)

print("\nSAVED:", output_path)
```

```python
import numpy as np
import pandas as pd

# Align observed counts and permutation columns
donor_names = [str(d) for d in donor_names]

observed = (
    final_donor_df
    .assign(donor_id=final_donor_df["donor_id"].astype(str))
    .set_index("donor_id")
    .reindex(donor_names)["observed_edges"]
    .to_numpy(dtype=float)
)

null_means = null_by_donor.mean(axis=0)

assert null_by_donor.shape[1] == len(donor_names)
assert np.all(null_means > 0)

# Each donor contributes equally to the observed statistic
observed_relative_excess = (
    (observed - null_means) / null_means
) * 100

donor_balanced_observed = observed_relative_excess.mean()

# Construct the equally weighted permutation distribution
permutation_relative_excess = (
    (null_by_donor - null_means) / null_means
) * 100

donor_balanced_null = permutation_relative_excess.mean(axis=1)

# Empirical upper-tail test
exceedances = int(
    np.sum(donor_balanced_null >= donor_balanced_observed)
)

empirical_p = (
    exceedances + 1
) / (
    len(donor_balanced_null) + 1
)

print("DONOR-BALANCED SPATIAL PERMUTATION TEST")
print("---------------------------------------")
print("Number of donors:", len(donor_names))
print("Number of permutations:", len(donor_balanced_null))
print(f"Observed mean donor excess: {donor_balanced_observed:.2f}%")
print(f"Null mean: {donor_balanced_null.mean():.4f}%")
print(f"Null SD: {donor_balanced_null.std(ddof=1):.4f}")
print(f"Null 2.5th percentile: {np.percentile(donor_balanced_null, 2.5):.4f}%")
print(f"Null 97.5th percentile: {np.percentile(donor_balanced_null, 97.5):.4f}%")
print("Permutation exceedances:", exceedances)
print(f"Empirical p-value: {empirical_p:.5f}")

print("\nINDIVIDUAL DONOR RELATIVE EXCESS")
display(
    pd.DataFrame({
        "donor_id": donor_names,
        "observed_relative_excess_pct": observed_relative_excess
    }).sort_values(
        "observed_relative_excess_pct",
        ascending=False
    ).round(3)
)

# Save results
summary = pd.DataFrame([{
    "donors": len(donor_names),
    "permutations": len(donor_balanced_null),
    "observed_mean_donor_excess_pct": donor_balanced_observed,
    "null_mean_pct": donor_balanced_null.mean(),
    "null_sd_pct": donor_balanced_null.std(ddof=1),
    "null_2.5_pct": np.percentile(donor_balanced_null, 2.5),
    "null_97.5_pct": np.percentile(donor_balanced_null, 97.5),
    "exceedances": exceedances,
    "empirical_p": empirical_p
}])

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Schwann_ReticFibIII_Donor_Balanced_Permutation.csv"
)

summary.to_csv(output_path, index=False)
print("\nSAVED:", output_path)
```

```python
import numpy as np
import pandas as pd

# Confirm alignment
selected_labels = labels.astype(np.int8)

assert len(selected_labels) == len(selected_samples)
assert len(selected_labels) == len(selected_coords)
assert len(edge_a) == len(edge_b)

# Encode sample identities
sample_codes, sample_names = pd.factorize(selected_samples)

# Confirm that graph edges do not cross sample boundaries
edge_sample_a = sample_codes[edge_a]
edge_sample_b = sample_codes[edge_b]

assert np.all(edge_sample_a == edge_sample_b), (
    "Some graph edges cross sample boundaries."
)

population_map = {
    0: "Mast",
    1: "Schwann",
    2: "Retic Fib III"
}

sample_results = []

for sample_code, sample_name in enumerate(sample_names):

    # Cells belonging to this sample
    cell_indices = np.flatnonzero(sample_codes == sample_code)
    n_sample = len(cell_indices)

    # Edges belonging to this sample
    edge_mask = edge_sample_a == sample_code
    a = edge_a[edge_mask]
    b = edge_b[edge_mask]
    n_edges = len(a)

    for population_id, population_name in population_map.items():

        x = (selected_labels[cell_indices] == population_id).astype(float)

        population_count = int(x.sum())
        other_count = n_sample - population_count

        # Avoid unstable estimates in very small groups
        valid = (
            n_sample >= 20
            and population_count >= 5
            and other_count >= 5
            and n_edges > 0
        )

        if valid:
            # Map global cell indices to sample-local indices
            local_index = np.full(len(selected_labels), -1, dtype=np.int32)
            local_index[cell_indices] = np.arange(n_sample)

            local_a = local_index[a]
            local_b = local_index[b]

            centered = x - x.mean()

            edge_products = (
                centered[local_a] * centered[local_b]
            )

            denominator = np.sum(centered ** 2)

            morans_i = (
                (n_sample / n_edges)
                * np.sum(edge_products)
                / denominator
            )
        else:
            morans_i = np.nan

        sample_results.append({
            "sample_id": str(sample_name),
            "population": population_name,
            "sample_cells": n_sample,
            "population_cells": population_count,
            "other_cells": other_count,
            "graph_edges": n_edges,
            "Moran_I": morans_i,
            "valid": valid
        })

sample_moran_df = pd.DataFrame(sample_results)

print("SAMPLE-LEVEL SPATIAL AUTOCORRELATION")
display(
    sample_moran_df.groupby("population").agg(
        total_samples=("sample_id", "nunique"),
        valid_samples=("valid", "sum"),
        median_Moran_I=("Moran_I", "median"),
        mean_Moran_I=("Moran_I", "mean"),
        q25=("Moran_I", lambda x: x.quantile(0.25)),
        q75=("Moran_I", lambda x: x.quantile(0.75)),
        positive_Moran_I=("Moran_I", lambda x: (x > 0).sum())
    ).round(4)
)

print("\nSAMPLE-LEVEL RESULTS")
display(
    sample_moran_df[
        sample_moran_df["valid"]
    ].sort_values(
        ["population", "Moran_I"],
        ascending=[True, False]
    ).head(20).round(4)
)

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Sample_Level_Spatial_Autocorrelation.csv"
)

sample_moran_df.to_csv(output_path, index=False)
print("\nSAVED:", output_path)
```

```python
import numpy as np
import pandas as pd
from scipy import sparse

# Original full-dataset graph
A = full_graph.tocsr()

# Symmetric union of the original directed graph
A_symmetric = A.maximum(A.T).tocsr()

n = A.shape[0]

population_map = {
    0: "Mast",
    1: "Schwann",
    2: "Retic Fib III"
}

results = []

for population_id, population_name in population_map.items():

    # Indicator across ALL 1,201,886 cells
    x = np.zeros(n, dtype=np.float64)

    target_indices = selected_indices[
        selected_labels == population_id
    ]

    x[target_indices] = 1.0

    centered = x - x.mean()
    denominator = np.sum(centered ** 2)

    # Original directed graph
    directed_I = (
        (n / A.sum())
        * (centered @ (A @ centered))
        / denominator
    )

    # Symmetric union graph
    symmetric_I = (
        (n / A_symmetric.sum())
        * (centered @ (A_symmetric @ centered))
        / denominator
    )

    results.append({
        "population": population_name,
        "cell_count": len(target_indices),
        "full_dataset_cells": n,
        "directed_Moran_I": float(directed_I),
        "symmetric_Moran_I": float(symmetric_I)
    })

full_moran_df = pd.DataFrame(results)

print("FULL-DATASET SPATIAL AUTOCORRELATION")
display(full_moran_df.round(6))

print("\nGRAPH COMPARISON")
print("Original directed edges:", A.nnz)
print("Symmetric union edges:", A_symmetric.nnz)
print("Original graph self-connections:",
      np.count_nonzero(A.diagonal()))
print("Symmetric graph self-connections:",
      np.count_nonzero(A_symmetric.diagonal()))
```

```python
import numpy as np
import pandas as pd
from scipy import sparse

print("VALIDATING FULL SAMPLE MAPPING")

n_full = len(sample_codes_full)

# Match selected-cell sample names to the H5AD category codes
category_lookup = {
    name: i for i, name in enumerate(sample_categories_full)
}

selected_sample_codes_full = np.array([
    category_lookup[str(name)] for name in sample_ids
], dtype=np.int16)

mapping_matches = np.array_equal(
    sample_codes_full[selected_indices],
    selected_sample_codes_full
)

print("Selected cells:", len(selected_indices))
print("Mapping matches:", mapping_matches)
print(
    "Mismatches:",
    int(np.sum(
        sample_codes_full[selected_indices]
        != selected_sample_codes_full
    ))
)

if not mapping_matches:
    raise ValueError(
        "Sample mapping mismatch. Stop before calculating Moran's I."
    )

# Reconstruct full-length labels.
# -1 means a cell outside the three selected populations.
full_labels_corrected = np.full(n_full, -1, dtype=np.int8)
full_labels_corrected[selected_indices] = selected_labels

# Use the symmetric union of the original full-dataset graph.
if "A_symmetric" not in globals():
    A_symmetric = full_graph.maximum(full_graph.T).tocsr()

A_symmetric = A_symmetric.tocsr()

print("\nFULL GRAPH")
print("Shape:", A_symmetric.shape)
print("Connections:", A_symmetric.nnz)

populations = {
    "Mast": 0,
    "Schwann": 1,
    "Retic Fib III": 2
}

results = []

print("\nCALCULATING SAMPLE-LEVEL MORAN'S I")

for sample_code, sample_name in enumerate(sample_categories_full):

    # All cells belonging to this sample
    sample_indices = np.flatnonzero(
        sample_codes_full == sample_code
    )

    # Restrict the full graph to this sample's cells
    sample_graph = A_symmetric[sample_indices, :][:, sample_indices]

    n = len(sample_indices)
    S0 = float(sample_graph.sum())

    if n < 2 or S0 == 0:
        continue

    sample_labels = full_labels_corrected[sample_indices]

    for population, population_code in populations.items():

        y = (sample_labels == population_code).astype(np.float64)

        population_count = int(y.sum())

        # Moran's I is undefined for a constant indicator
        if population_count == 0 or population_count == n:
            continue

        z = y - y.mean()
        denominator = float(np.dot(z, z))

        if denominator == 0:
            continue

        numerator = float(z @ (sample_graph @ z))

        moran_i = (n / S0) * (numerator / denominator)

        results.append({
            "sample_id": sample_name,
            "sample_code": sample_code,
            "population": population,
            "total_sample_cells": n,
            "population_cells": population_count,
            "population_fraction": population_count / n,
            "graph_edges_directed": int(sample_graph.nnz),
            "moran_i": moran_i
        })

    if (sample_code + 1) % 20 == 0:
        print("Processed", sample_code + 1, "/", len(sample_categories_full))

moran_full_sample_df = pd.DataFrame(results)

print("\nFULL-TISSUE SAMPLE-LEVEL MORAN'S I")
print(
    moran_full_sample_df
    .groupby("population")["moran_i"]
    .agg(
        sample_count="count",
        mean="mean",
        median="median",
        q25=lambda x: x.quantile(0.25),
        q75=lambda x: x.quantile(0.75),
        positive_samples=lambda x: int((x > 0).sum()),
        negative_samples=lambda x: int((x < 0).sum())
    )
    .round(4)
)

print("\nPOPULATION COUNTS BY SAMPLE")
print(
    moran_full_sample_df
    .groupby("population")["population_cells"]
    .agg(["count", "sum", "min", "median", "max"])
)

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Full_Tissue_Sample_Level_Moran.csv"
)

moran_full_sample_df.to_csv(output_path, index=False)

print("\nSAVED:", output_path)
print("Rows:", len(moran_full_sample_df))
```

```python
import numpy as np
import pandas as pd
from scipy import sparse

print("FULL-TISSUE NEIGHBOR COMPOSITION ANALYSIS")

# Confirm the graph and labels are aligned
assert A_symmetric.shape == (len(sample_codes_full), len(sample_codes_full))
assert len(full_labels_corrected) == len(sample_codes_full)

population_names = {
    0: "Mast",
    1: "Schwann",
    2: "Retic Fib III"
}

results = []

for sample_code, sample_name in enumerate(sample_categories_full):

    sample_indices = np.flatnonzero(
        sample_codes_full == sample_code
    )

    # Full graph restricted to this sample
    sample_graph = A_symmetric[sample_indices, :][:, sample_indices]

    sample_labels = full_labels_corrected[sample_indices]

    # Number of graph neighbors for every cell
    degrees = np.asarray(sample_graph.sum(axis=1)).ravel()

    for population_code, population_name in population_names.items():

        source_mask = sample_labels == population_code
        source_positions = np.flatnonzero(source_mask)

        if len(source_positions) == 0:
            continue

        # Rows: source cells; columns: all cells in this sample
        source_graph = sample_graph[source_positions, :]

        source_degrees = degrees[source_positions]
        total_edges = float(source_degrees.sum())

        # Count edges to each target population
        target_counts = {}

        for target_code, target_name in population_names.items():

            target_mask = (
                sample_labels == target_code
            ).astype(np.float64)

            target_counts[target_name] = np.asarray(
                source_graph @ target_mask
            ).ravel().sum()

        # All other annotated cell types combined
        other_count = total_edges - sum(target_counts.values())

        row = {
            "sample_id": sample_name,
            "sample_code": sample_code,
            "source_population": population_name,
            "source_cells": len(source_positions),
            "mean_neighbors_per_cell": float(source_degrees.mean()),
            "median_neighbors_per_cell": float(np.median(source_degrees)),
            "total_neighbor_edges": int(total_edges)
        }

        for target_name, count in target_counts.items():
            row[f"{target_name}_neighbor_edges"] = int(count)
            row[f"{target_name}_neighbor_pct"] = (
                100 * count / total_edges if total_edges else np.nan
            )

        row["Other_neighbor_edges"] = int(round(other_count))
        row["Other_neighbor_pct"] = (
            100 * other_count / total_edges if total_edges else np.nan
        )

        results.append(row)

    if (sample_code + 1) % 20 == 0:
        print("Processed", sample_code + 1, "/", len(sample_categories_full))

neighbor_composition_df = pd.DataFrame(results)

print("\nOVERALL SUMMARY BY SOURCE POPULATION")

summary = neighbor_composition_df.groupby(
    "source_population"
).agg(
    samples=("sample_id", "count"),
    mean_neighbors=("mean_neighbors_per_cell", "mean"),
    median_neighbors=("mean_neighbors_per_cell", "median"),
    Mast_neighbor_pct=("Mast_neighbor_pct", "median"),
    Schwann_neighbor_pct=("Schwann_neighbor_pct", "median"),
    Retic_neighbor_pct=("Retic Fib III_neighbor_pct", "median"),
    Other_neighbor_pct=("Other_neighbor_pct", "median")
).round(3)

print(summary)

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Full_Tissue_Neighbor_Composition_By_Sample.csv"
)

neighbor_composition_df.to_csv(output_path, index=False)

print("\nSAVED:", output_path)
print("Rows:", len(neighbor_composition_df))
```

```python
import numpy as np
import pandas as pd

print("FULL-TISSUE ABUNDANCE-ADJUSTED NEIGHBOR ENRICHMENT")

population_names = ["Mast", "Schwann", "Retic Fib III"]

population_codes = {
    "Mast": 0,
    "Schwann": 1,
    "Retic Fib III": 2
}

# Build a lookup of total cells per sample
sample_sizes = np.bincount(
    sample_codes_full.astype(int),
    minlength=len(sample_categories_full)
)

# Build population counts per sample
population_counts_by_sample = {}

for code, name in population_codes.items():
    counts = np.zeros(len(sample_categories_full), dtype=int)

    for sample_code in range(len(sample_categories_full)):
        mask = (
            (sample_codes_full == sample_code) &
            (full_labels_corrected == code)
        )
        counts[sample_code] = int(mask.sum())

    population_counts_by_sample[name] = counts

results = []

for _, row in neighbor_composition_df.iterrows():

    sample_code = int(row["sample_code"])
    source = row["source_population"]
    total_cells = int(sample_sizes[sample_code])

    # Abundance of each target population in this sample
    target_abundances = {
        name: population_counts_by_sample[name][sample_code] / total_cells
        for name in population_names
    }

    other_abundance = (
        total_cells -
        sum(
            population_counts_by_sample[name][sample_code]
            for name in population_names
        )
    ) / total_cells

    target_abundances["Other"] = other_abundance

    # Observed neighbor composition
    observed_fractions = {
        name: float(row[f"{name}_neighbor_pct"]) / 100
        for name in population_names
    }

    observed_fractions["Other"] = (
        float(row["Other_neighbor_pct"]) / 100
    )

    for target, abundance in target_abundances.items():

        observed = observed_fractions[target]

        enrichment = (
            observed / abundance
            if abundance > 0 else np.nan
        )

        results.append({
            "sample_id": row["sample_id"],
            "sample_code": sample_code,
            "source_population": source,
            "target_population": target,
            "source_cells": int(row["source_cells"]),
            "sample_total_cells": total_cells,
            "target_cell_fraction": abundance,
            "observed_neighbor_fraction": observed,
            "enrichment": enrichment
        })

enrichment_df = pd.DataFrame(results)

print("\nVALIDATION")

composition_sums = (
    neighbor_composition_df[
        [
            "Mast_neighbor_pct",
            "Schwann_neighbor_pct",
            "Retic Fib III_neighbor_pct",
            "Other_neighbor_pct"
        ]
    ].sum(axis=1)
)

print("Neighbor composition rows:", len(composition_sums))
print("Minimum percentage sum:", round(composition_sums.min(), 4))
print("Maximum percentage sum:", round(composition_sums.max(), 4))
print("Rows summing to approximately 100%:",
      int(np.isclose(composition_sums, 100, atol=0.1).sum()))

print("\nABUNDANCE-ADJUSTED ENRICHMENT SUMMARY")

summary = enrichment_df.groupby(
    ["source_population", "target_population"]
).agg(
    samples=("enrichment", "count"),
    median_enrichment=("enrichment", "median"),
    mean_enrichment=("enrichment", "mean"),
    q25=("enrichment", lambda x: x.quantile(0.25)),
    q75=("enrichment", lambda x: x.quantile(0.75)),
    samples_above_1=("enrichment", lambda x: int((x > 1).sum()))
).round(3)

print(summary.to_string())

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Full_Tissue_Abundance_Adjusted_Neighbor_Enrichment.csv"
)

enrichment_df.to_csv(output_path, index=False)

summary_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Full_Tissue_Abundance_Adjusted_Neighbor_Enrichment_Summary.csv"
)

summary.to_csv(summary_path)

print("\nSAVED:", output_path)
print("SAVED:", summary_path)
print("Total rows:", len(enrichment_df))
```

```python
import numpy as np
import pandas as pd

print("REBUILDING POPULATION COUNTS AND NEIGHBOR ENRICHMENT")

population_codes = {
    "Mast": 0,
    "Schwann": 1,
    "Retic Fib III": 2
}

n_samples = len(sample_categories_full)

# Rebuild total cell counts per sample
sample_sizes = np.bincount(
    sample_codes_full.astype(int),
    minlength=n_samples
)

# Rebuild population counts per sample
population_counts_by_sample = {}

for name, code in population_codes.items():

    valid = full_labels_corrected == code

    counts = np.bincount(
        sample_codes_full[valid].astype(int),
        minlength=n_samples
    )

    population_counts_by_sample[name] = counts

print("Population count arrays:")
for name, counts in population_counts_by_sample.items():
    print(name, "| total:", int(counts.sum()))

results = []

for _, row in neighbor_composition_df.iterrows():

    sample_code = int(row["sample_code"])
    source = row["source_population"]
    total_cells = int(sample_sizes[sample_code])

    target_abundances = {}

    for name in population_codes:

        target_abundances[name] = (
            population_counts_by_sample[name][sample_code]
            / total_cells
        )

    other_cells = total_cells - sum(
        population_counts_by_sample[name][sample_code]
        for name in population_codes
    )

    target_abundances["Other"] = other_cells / total_cells

    observed_fractions = {
        name: float(row[f"{name}_neighbor_pct"]) / 100
        for name in population_codes
    }

    observed_fractions["Other"] = (
        float(row["Other_neighbor_pct"]) / 100
    )

    for target, abundance in target_abundances.items():

        observed = observed_fractions[target]

        enrichment = (
            observed / abundance
            if abundance > 0 else np.nan
        )

        results.append({
            "sample_id": row["sample_id"],
            "sample_code": sample_code,
            "source_population": source,
            "target_population": target,
            "source_cells": int(row["source_cells"]),
            "sample_total_cells": total_cells,
            "target_cell_fraction": abundance,
            "observed_neighbor_fraction": observed,
            "enrichment": enrichment
        })

enrichment_df = pd.DataFrame(results)

print("\nABUNDANCE-ADJUSTED ENRICHMENT SUMMARY")

summary = enrichment_df.groupby(
    ["source_population", "target_population"]
).agg(
    samples=("enrichment", "count"),
    median_enrichment=("enrichment", "median"),
    mean_enrichment=("enrichment", "mean"),
    q25=("enrichment", lambda x: x.quantile(0.25)),
    q75=("enrichment", lambda x: x.quantile(0.75)),
    samples_above_1=("enrichment", lambda x: int((x > 1).sum()))
).round(3)

print(summary.to_string())

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Full_Tissue_Abundance_Adjusted_Neighbor_Enrichment.csv"
)

enrichment_df.to_csv(output_path, index=False)

summary_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Full_Tissue_Abundance_Adjusted_Neighbor_Enrichment_Summary.csv"
)

summary.to_csv(summary_path)

print("\nSAVED:", output_path)
print("SAVED:", summary_path)
print("Total rows:", len(enrichment_df))
```

```python
import numpy as np
import pandas as pd
from scipy.stats import wilcoxon

print("DONOR-LEVEL INFERENCE AND BOOTSTRAP CONFIDENCE INTERVALS")

rng = np.random.default_rng(42)
results = []

for (source, target), group in donor_enrichment.groupby(
    ["source_population", "target_population"]
):

    values = group["donor_median_enrichment"].dropna().to_numpy()
    n = len(values)

    # One-sided Wilcoxon signed-rank test against enrichment = 1
    test = wilcoxon(
        values - 1,
        alternative="greater",
        method="auto"
    )

    # Bootstrap confidence interval for the median donor enrichment
    bootstrap_medians = np.array([
        np.median(rng.choice(values, size=n, replace=True))
        for _ in range(10000)
    ])

    ci_low, ci_high = np.percentile(
        bootstrap_medians, [2.5, 97.5]
    )

    results.append({
        "source_population": source,
        "target_population": target,
        "donors": n,
        "median_enrichment": np.median(values),
        "mean_enrichment": np.mean(values),
        "ci_95_lower": ci_low,
        "ci_95_upper": ci_high,
        "donors_above_1": int((values > 1).sum()),
        "wilcoxon_statistic": test.statistic,
        "p_value": test.pvalue
    })

inference_df = pd.DataFrame(results)

# Benjamini-Hochberg correction across all tested relationships
p = inference_df["p_value"].to_numpy()
m = len(p)
order = np.argsort(p)
ranked_p = p[order]

adjusted = ranked_p * m / np.arange(1, m + 1)
adjusted = np.minimum.accumulate(adjusted[::-1])[::-1]
adjusted = np.clip(adjusted, 0, 1)

inference_df["p_adjusted_BH"] = np.empty(m)
inference_df.loc[order, "p_adjusted_BH"] = adjusted

inference_df = inference_df.sort_values(
    ["source_population", "target_population"]
)

display_columns = [
    "source_population",
    "target_population",
    "donors",
    "median_enrichment",
    "ci_95_lower",
    "ci_95_upper",
    "donors_above_1",
    "p_value",
    "p_adjusted_BH"
]

print("\nINFERENCE RESULTS")
print(
    inference_df[display_columns]
    .round(5)
    .to_string(index=False)
)

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Full_Tissue_Donor_Level_Enrichment_Inference.csv"
)

inference_df.to_csv(output_path, index=False)

print("\nSAVED:", output_path)
```

```python
import numpy as np
import pandas as pd

print("LEAVE-ONE-ANATOMICAL-SITE-OUT SENSITIVITY")

pairs = [
    ("Schwann", "Retic Fib III"),
    ("Retic Fib III", "Schwann")
]

sensitivity_results = []

for source, target in pairs:

    subset = site_summary[
        (site_summary["source_population"] == source) &
        (site_summary["target_population"] == target)
    ].copy()

    all_site_medians = subset[
        "median_donor_enrichment"
    ].to_numpy()

    overall_median = np.median(all_site_medians)

    for site in subset["anatomic_site"].unique():

        remaining = subset[
            subset["anatomic_site"] != site
        ]

        remaining_values = remaining[
            "median_donor_enrichment"
        ].to_numpy()

        sensitivity_results.append({
            "source_population": source,
            "target_population": target,
            "excluded_site": site,
            "remaining_sites": len(remaining_values),
            "overall_site_median": overall_median,
            "leave_one_out_median": np.median(remaining_values),
            "change_from_overall": (
                np.median(remaining_values) - overall_median
            ),
            "minimum_remaining_site": remaining_values.min(),
            "sites_remaining_above_1": int(
                (remaining_values > 1).sum()
            )
        })

sensitivity_df = pd.DataFrame(sensitivity_results)

print("\nRESULTS")
print(
    sensitivity_df.round(3).to_string(index=False)
)

print("\nOVERALL SENSITIVITY RANGE")

summary = sensitivity_df.groupby(
    ["source_population", "target_population"]
).agg(
    lowest_leave_one_out=("leave_one_out_median", "min"),
    highest_leave_one_out=("leave_one_out_median", "max"),
    minimum_site_after_exclusion=("minimum_remaining_site", "min"),
    minimum_sites_above_1=("sites_remaining_above_1", "min")
).round(3)

print(summary.to_string())

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Anatomical_Site_Leave_One_Out_Sensitivity.csv"
)

sensitivity_df.to_csv(output_path, index=False)

print("\nSAVED:", output_path)
```

```python
import numpy as np
import pandas as pd

print("SPATIAL BLOCK OCCUPANCY AUDIT")

block_widths = [250, 500, 1000, 2000]

results = []

for sample_code in range(len(sample_categories_full)):

    sample_mask = sample_codes == sample_code

    if not sample_mask.any():
        continue

    coords = selected_coords[sample_mask]
    sample_labels = selected_labels[sample_mask]

    x = coords[:, 0]
    y = coords[:, 1]

    for width in block_widths:

        # Anchor blocks to the minimum sample coordinates
        bx = np.floor((x - x.min()) / width).astype(int)
        by = np.floor((y - y.min()) / width).astype(int)

        block_ids = pd.factorize(
            list(zip(bx, by))
        )[0]

        block_df = pd.DataFrame({
            "block": block_ids,
            "label": sample_labels
        })

        counts = block_df.pivot_table(
            index="block",
            columns="label",
            aggfunc="size",
            fill_value=0
        )

        for code, name in enumerate(
            ["Mast", "Schwann", "Retic Fib III"]
        ):
            if code not in counts.columns:
                counts[code] = 0

        mast = counts[0].to_numpy()
        schwann = counts[1].to_numpy()
        retic = counts[2].to_numpy()

        results.append({
            "sample_id": sample_ids[sample_mask][0],
            "sample_code": sample_code,
            "block_width": width,
            "occupied_blocks": len(counts),
            "blocks_with_mast": int((mast > 0).sum()),
            "blocks_with_schwann": int((schwann > 0).sum()),
            "blocks_with_retic": int((retic > 0).sum()),
            "blocks_with_both": int(
                ((schwann > 0) & (retic > 0)).sum()
            ),
            "schwann_cells": int(schwann.sum()),
            "retic_cells": int(retic.sum())
        })

block_occupancy_df = pd.DataFrame(results)

print("\nSUMMARY BY BLOCK WIDTH")

summary = block_occupancy_df.groupby("block_width").agg(
    samples=("sample_code", "nunique"),
    median_occupied_blocks=("occupied_blocks", "median"),
    median_schwann_blocks=("blocks_with_schwann", "median"),
    median_retic_blocks=("blocks_with_retic", "median"),
    median_shared_blocks=("blocks_with_both", "median")
)

print(summary.to_string())

print("\nSAVED:")
output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Spatial_Block_Occupancy_Audit.csv"
)

block_occupancy_df.to_csv(output_path, index=False)
print(output_path)
```

```python
import numpy as np
import pandas as pd

print("SHARED-BLOCK CELL COMPOSITION AUDIT")

block_widths = [250, 500]
detail_results = []

for sample_code in range(len(sample_categories_full)):

    sample_mask = sample_codes == sample_code

    if not sample_mask.any():
        continue

    coords = selected_coords[sample_mask]
    sample_labels = selected_labels[sample_mask]

    x = coords[:, 0]
    y = coords[:, 1]

    for width in block_widths:

        bx = np.floor((x - x.min()) / width).astype(int)
        by = np.floor((y - y.min()) / width).astype(int)

        block_keys = pd.MultiIndex.from_arrays([bx, by])
        block_ids, _ = pd.factorize(block_keys)

        block_df = pd.DataFrame({
            "block": block_ids,
            "label": sample_labels
        })

        counts = block_df.pivot_table(
            index="block",
            columns="label",
            aggfunc="size",
            fill_value=0
        )

        for code in [0, 1, 2]:
            if code not in counts.columns:
                counts[code] = 0

        shared = counts[
            (counts[1] > 0) & (counts[2] > 0)
        ].copy()

        if len(shared) == 0:
            continue

        detail_results.append({
            "sample_id": sample_ids[sample_mask][0],
            "block_width": width,
            "shared_blocks": len(shared),
            "median_schwann_per_shared_block": shared[1].median(),
            "median_retic_per_shared_block": shared[2].median(),
            "median_total_cells_per_shared_block": (
                shared[[0, 1, 2]].sum(axis=1).median()
            ),
            "shared_blocks_with_5plus_each": int(
                ((shared[1] >= 5) & (shared[2] >= 5)).sum()
            ),
            "shared_blocks_with_10plus_each": int(
                ((shared[1] >= 10) & (shared[2] >= 10)).sum()
            )
        })

shared_block_detail_df = pd.DataFrame(detail_results)

print("\nSUMMARY BY BLOCK WIDTH")

summary = shared_block_detail_df.groupby("block_width").agg(
    samples=("sample_id", "nunique"),
    median_shared_blocks=("shared_blocks", "median"),
    median_schwann_cells=("median_schwann_per_shared_block", "median"),
    median_retic_cells=("median_retic_per_shared_block", "median"),
    median_total_cells=("median_total_cells_per_shared_block", "median"),
    median_blocks_5plus_each=("shared_blocks_with_5plus_each", "median"),
    median_blocks_10plus_each=("shared_blocks_with_10plus_each", "median")
)

print(summary.to_string())

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Shared_Block_Cell_Composition_Audit.csv"
)

shared_block_detail_df.to_csv(output_path, index=False)

print("\nSAVED:")
print(output_path)
```

```python
import numpy as np
import pandas as pd
from scipy.spatial import cKDTree

print("SCHWANN–RETIC FIB III DISTANCE-THRESHOLD AUDIT")

distance_thresholds = [25, 50, 100, 200, 400]

distance_results = []

for sample_code in range(len(sample_categories_full)):

    sample_mask = sample_codes == sample_code

    if not sample_mask.any():
        continue

    coords = selected_coords[sample_mask]
    sample_labels = selected_labels[sample_mask]

    schwann_coords = coords[sample_labels == 1]
    retic_coords = coords[sample_labels == 2]

    if len(schwann_coords) == 0 or len(retic_coords) == 0:
        continue

    tree = cKDTree(retic_coords)
    distances, _ = tree.query(schwann_coords, k=1)

    for threshold in distance_thresholds:

        distance_results.append({
            "sample_id": sample_ids[sample_mask][0],
            "sample_code": sample_code,
            "threshold": threshold,
            "schwann_cells": len(schwann_coords),
            "retic_cells": len(retic_coords),
            "schwann_within_threshold": int(
                (distances <= threshold).sum()
            ),
            "schwann_fraction_within_threshold": float(
                (distances <= threshold).mean()
            ),
            "median_nearest_distance": float(
                np.median(distances)
            )
        })

distance_threshold_df = pd.DataFrame(distance_results)

summary = distance_threshold_df.groupby("threshold").agg(
    samples=("sample_code", "nunique"),
    median_fraction=("schwann_fraction_within_threshold", "median"),
    mean_fraction=("schwann_fraction_within_threshold", "mean"),
    q25_fraction=("schwann_fraction_within_threshold",
                  lambda x: x.quantile(0.25)),
    q75_fraction=("schwann_fraction_within_threshold",
                  lambda x: x.quantile(0.75)),
    median_nearest_distance=("median_nearest_distance", "median")
)

print("\nDISTANCE-THRESHOLD SUMMARY")
print(summary.round(4).to_string())

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Schwann_Retic_Distance_Threshold_Audit.csv"
)

distance_threshold_df.to_csv(output_path, index=False)

print("\nSAVED:")
print(output_path)
```

```python
import numpy as np
import pandas as pd
from scipy.spatial import cKDTree

print("WITHIN-SAMPLE RANDOM-LABELING NULL MODEL")

rng = np.random.default_rng(42)
n_permutations = 999
thresholds = [100, 200]

sample_records = []

for sample_code in range(len(sample_categories_full)):

    sample_mask = sample_codes == sample_code

    if not sample_mask.any():
        continue

    coords = selected_coords[sample_mask]
    labels_sample = selected_labels[sample_mask]
    sample_id = sample_ids[sample_mask][0]
    donor_id = sample_to_donor[sample_id]

    schwann_coords = coords[labels_sample == 1]
    retic_coords = coords[labels_sample == 2]

    if len(schwann_coords) == 0 or len(retic_coords) == 0:
        continue

    observed_distances, _ = cKDTree(retic_coords).query(
        schwann_coords, k=1
    )

    observed = np.array([
        np.mean(observed_distances <= threshold)
        for threshold in thresholds
    ])

    null_values = np.zeros((n_permutations, len(thresholds)))

    for permutation in range(n_permutations):

        permuted_labels = rng.permutation(labels_sample)

        perm_schwann = coords[permuted_labels == 1]
        perm_retic = coords[permuted_labels == 2]

        perm_distances, _ = cKDTree(perm_retic).query(
            perm_schwann, k=1
        )

        null_values[permutation] = [
            np.mean(perm_distances <= threshold)
            for threshold in thresholds
        ]

    sample_records.append({
        "sample_id": sample_id,
        "donor_id": donor_id,
        "observed": observed,
        "null_values": null_values
    })

    if len(sample_records) % 20 == 0:
        print("Completed samples:", len(sample_records))


# Build sample-level observed results
observed_rows = []

for record in sample_records:
    for j, threshold in enumerate(thresholds):
        observed_rows.append({
            "sample_id": record["sample_id"],
            "donor_id": record["donor_id"],
            "threshold": threshold,
            "observed_fraction": record["observed"][j],
            "null_mean": record["null_values"][:, j].mean(),
            "null_sd": record["null_values"][:, j].std()
        })

random_label_sample_df = pd.DataFrame(observed_rows)


# Donor-balanced observed statistics
donor_observed = (
    random_label_sample_df
    .groupby(["donor_id", "threshold"])["observed_fraction"]
    .median()
    .reset_index()
)

summary_rows = []

for threshold in thresholds:

    observed_donor_values = donor_observed.loc[
        donor_observed["threshold"] == threshold,
        "observed_fraction"
    ].to_numpy()

    observed_statistic = np.median(observed_donor_values)

    null_donor_statistics = []

    for permutation in range(n_permutations):

        permutation_rows = []

        for record in sample_records:
            j = thresholds.index(threshold)

            permutation_rows.append({
                "donor_id": record["donor_id"],
                "value": record["null_values"][permutation, j]
            })

        permutation_df = pd.DataFrame(permutation_rows)

        donor_null = (
            permutation_df.groupby("donor_id")["value"]
            .median()
            .to_numpy()
        )

        null_donor_statistics.append(np.median(donor_null))

    null_donor_statistics = np.array(null_donor_statistics)

    p_value = (
        1 + np.sum(null_donor_statistics >= observed_statistic)
    ) / (n_permutations + 1)

    null_mean = null_donor_statistics.mean()

    summary_rows.append({
        "threshold": threshold,
        "donors": len(observed_donor_values),
        "observed_donor_median": observed_statistic,
        "null_donor_median_mean": null_mean,
        "null_donor_median_sd": null_donor_statistics.std(),
        "observed_excess_percentage_points": (
            observed_statistic - null_mean
        ) * 100,
        "empirical_p_value": p_value
    })


random_label_summary_df = pd.DataFrame(summary_rows)

print("\nDONOR-BALANCED RANDOM-LABELING RESULTS")
print(random_label_summary_df.round(5).to_string(index=False))


output_folder = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
)

random_label_sample_df.to_csv(
    output_folder + "Schwann_Retic_Random_Labeling_Sample_Results.csv",
    index=False
)

random_label_summary_df.to_csv(
    output_folder + "Schwann_Retic_Random_Labeling_Summary.csv",
    index=False
)

print("\nSAVED:")
print(output_folder + "Schwann_Retic_Random_Labeling_Sample_Results.csv")
print(output_folder + "Schwann_Retic_Random_Labeling_Summary.csv")
```

```python
import numpy as np
import pandas as pd

print("FULL-TISSUE DIRECTED GRAPH RANDOM-LABELING NULL")

rng = np.random.default_rng(20261108)
n_permutations = 1999

relationships = [
    (1, 2, "Schwann", "Retic Fib III"),
    (2, 1, "Retic Fib III", "Schwann")
]

sample_data = []

for sample_code in np.unique(sample_codes):

    selected_local = np.flatnonzero(sample_codes == sample_code)
    selected_global = selected_indices[selected_local]

    full_global = np.flatnonzero(
        sample_codes_full == sample_code
    )

    # Map selected cells to their positions in the full sample.
    selected_positions = np.searchsorted(
        full_global, selected_global
    )

    if not np.array_equal(
        full_global[selected_positions], selected_global
    ):
        raise ValueError("Selected cells do not map to full sample.")

    sample_id = sample_ids[selected_local[0]]

    # Original directed graph: selected source cells to ALL sample cells.
    sample_graph = full_graph[
        selected_global, :
    ][:, full_global].tocoo()

    sample_data.append({
        "sample_id": sample_id,
        "donor_id": sample_to_donor[sample_id],
        "selected_labels": selected_labels[selected_local].copy(),
        "full_labels": full_labels_corrected[full_global].copy(),
        "selected_positions": selected_positions,
        "edge_src": sample_graph.row,
        "edge_tgt": sample_graph.col
    })


def full_tissue_enrichment(
    selected_labels,
    full_labels,
    edge_src,
    edge_tgt,
    source_code,
    target_code
):

    source_edges = selected_labels[edge_src] == source_code
    total_edges = int(source_edges.sum())

    if total_edges == 0:
        return np.nan

    target_edges = int(
        (
            source_edges
            & (full_labels[edge_tgt] == target_code)
        ).sum()
    )

    target_abundance = np.mean(full_labels == target_code)

    if target_abundance == 0:
        return np.nan

    return (target_edges / total_edges) / target_abundance


# Observed sample-level enrichment
observed_rows = []

for item in sample_data:

    for source, target, source_name, target_name in relationships:

        value = full_tissue_enrichment(
            item["selected_labels"],
            item["full_labels"],
            item["edge_src"],
            item["edge_tgt"],
            source,
            target
        )

        if np.isfinite(value):
            observed_rows.append({
                "sample_id": item["sample_id"],
                "donor_id": item["donor_id"],
                "source_population": source_name,
                "target_population": target_name,
                "enrichment": value
            })

observed_df = pd.DataFrame(observed_rows)

observed_donor_df = (
    observed_df.groupby(
        ["donor_id", "source_population", "target_population"]
    )["enrichment"]
    .median()
    .reset_index()
)

observed_stats = observed_donor_df.groupby(
    ["source_population", "target_population"]
)["enrichment"].median()


# Randomization
null_statistics = {
    (source_name, target_name): []
    for _, _, source_name, target_name in relationships
}

for permutation in range(n_permutations):

    permutation_rows = []

    for item in sample_data:

        perm_selected = rng.permutation(item["selected_labels"])

        perm_full = item["full_labels"].copy()
        perm_full[item["selected_positions"]] = perm_selected

        for source, target, source_name, target_name in relationships:

            value = full_tissue_enrichment(
                perm_selected,
                perm_full,
                item["edge_src"],
                item["edge_tgt"],
                source,
                target
            )

            if np.isfinite(value):
                permutation_rows.append({
                    "donor_id": item["donor_id"],
                    "source_population": source_name,
                    "target_population": target_name,
                    "enrichment": value
                })

    permutation_df = pd.DataFrame(permutation_rows)

    donor_values = (
        permutation_df.groupby(
            ["donor_id", "source_population", "target_population"]
        )["enrichment"]
        .median()
    )

    permutation_stats = donor_values.groupby(
        ["source_population", "target_population"]
    ).median()

    for relationship in null_statistics:
        null_statistics[relationship].append(
            permutation_stats.loc[relationship]
        )

    if (permutation + 1) % 250 == 0:
        print("Completed permutations:", permutation + 1)


# Final summary
summary_rows = []

for relationship, values in null_statistics.items():

    null_values = np.asarray(values)
    observed_value = observed_stats.loc[relationship]
    null_mean = null_values.mean()

    p_value = (
        1 + np.sum(null_values >= observed_value)
    ) / (n_permutations + 1)

    summary_rows.append({
        "source_population": relationship[0],
        "target_population": relationship[1],
        "donors": observed_donor_df[
            (observed_donor_df["source_population"] == relationship[0])
            & (observed_donor_df["target_population"] == relationship[1])
        ]["donor_id"].nunique(),
        "observed_donor_median": observed_value,
        "null_mean": null_mean,
        "null_sd": null_values.std(),
        "observed_excess_percentage": (
            (observed_value / null_mean) - 1
        ) * 100,
        "empirical_p_value": p_value
    })

summary_df = pd.DataFrame(summary_rows)

# Benjamini-Hochberg correction across the two tested directions.
p = summary_df["empirical_p_value"].to_numpy()
order = np.argsort(p)
adjusted = np.empty(len(p))
adjusted[order] = np.minimum.accumulate(
    (p[order] * len(p) / np.arange(1, len(p) + 1))[::-1]
)[::-1]
summary_df["BH_adjusted_p"] = np.minimum(adjusted, 1.0)

print("\nOBSERVED ENRICHMENT CHECK")
print(
    observed_donor_df.groupby(
        ["source_population", "target_population"]
    )["enrichment"].median().round(4).to_string()
)

print("\nFULL-TISSUE RANDOM-LABELING RESULTS")
print(summary_df.round(5).to_string(index=False))


folder = "/content/drive/MyDrive/Neurothesis Resoure Library/"

observed_df.to_csv(
    folder + "Full_Tissue_Directed_Random_Labeling_Observed.csv",
    index=False
)

summary_df.to_csv(
    folder + "Full_Tissue_Directed_Random_Labeling_Summary.csv",
    index=False
)

print("\nSAVED:")
print(folder + "Full_Tissue_Directed_Random_Labeling_Observed.csv")
print(folder + "Full_Tissue_Directed_Random_Labeling_Summary.csv")
```

```python
import numpy as np
import pandas as pd
from scipy import sparse

print("VALIDATING FULL-TISSUE DIRECTED ENRICHMENT")

full_sample_lookup = {
    name: code
    for code, name in enumerate(sample_categories_full)
}

validation_rows = []

for sample_name in np.unique(sample_ids):

    selected_local = np.flatnonzero(sample_ids == sample_name)
    selected_global = selected_indices[selected_local]

    full_code = full_sample_lookup[sample_name]
    full_global = np.flatnonzero(sample_codes_full == full_code)

    selected_positions = np.searchsorted(
        full_global,
        selected_global
    )

    sample_graph = (
        full_graph[selected_global, :]
        [:, full_global]
        .tocoo()
    )

    source_labels = selected_labels[selected_local]
    target_labels = full_labels_corrected[full_global]

    edge_src = sample_graph.row
    edge_tgt = sample_graph.col

    for source_code, source_name, target_code, target_name in [
        (1, "Schwann", 2, "Retic Fib III"),
        (2, "Retic Fib III", 1, "Schwann")
    ]:

        source_mask = source_labels[edge_src] == source_code
        source_edge_count = source_mask.sum()

        target_edge_count = np.sum(
            source_mask &
            (target_labels[edge_tgt] == target_code)
        )

        target_abundance = np.mean(
            target_labels == target_code
        )

        if source_edge_count == 0 or target_abundance == 0:
            continue

        enrichment = (
            target_edge_count / source_edge_count
        ) / target_abundance

        validation_rows.append({
            "sample_id": sample_name,
            "donor_id": sample_to_donor[sample_name],
            "source_population": source_name,
            "target_population": target_name,
            "source_edges": source_edge_count,
            "target_edges": target_edge_count,
            "target_abundance": target_abundance,
            "enrichment": enrichment
        })

validation_df = pd.DataFrame(validation_rows)

donor_validation = (
    validation_df
    .groupby([
        "donor_id",
        "source_population",
        "target_population"
    ])["enrichment"]
    .median()
    .reset_index()
)

print("\nSAMPLE COUNTS")
print(
    validation_df.groupby(
        ["source_population", "target_population"]
    ).size()
)

print("\nDONOR COUNTS")
print(
    donor_validation.groupby(
        ["source_population", "target_population"]
    )["donor_id"].nunique()
)

print("\nVALIDATED DONOR-LEVEL ENRICHMENT")

print(
    donor_validation.groupby(
        ["source_population", "target_population"]
    )["enrichment"]
    .agg(["count", "median", "mean", "min", "max"])
    .round(4)
)

print("\nEXPECTED APPROXIMATE MEDIANS:")
print("Schwann -> Retic Fib III: 8.50")
print("Retic Fib III -> Schwann: 8.57")
```

```python
import numpy as np
import pandas as pd
from scipy.stats import rankdata

print("FULL-TISSUE DIRECTED GRAPH RANDOM-LABELING NULL")
print("Preparing sample-specific graph data...")

rng = np.random.default_rng(42)
N_PERMUTATIONS = 1999

full_sample_lookup = {
    name: code
    for code, name in enumerate(sample_categories_full)
}

sample_data = []

for sample_name in np.unique(sample_ids):

    selected_local = np.flatnonzero(sample_ids == sample_name)
    selected_global = selected_indices[selected_local]

    full_code = full_sample_lookup[sample_name]
    full_global = np.flatnonzero(sample_codes_full == full_code)

    selected_positions = np.searchsorted(
        full_global,
        selected_global
    )

    sample_graph = (
        full_graph[selected_global, :]
        [:, full_global]
        .tocoo()
    )

    sample_data.append({
        "sample_id": sample_name,
        "donor_id": sample_to_donor[sample_name],
        "source_labels": selected_labels[selected_local].copy(),
        "full_labels": full_labels_corrected[full_global].copy(),
        "selected_positions": selected_positions,
        "edge_src": sample_graph.row,
        "edge_tgt": sample_graph.col
    })

print("Prepared samples:", len(sample_data))

comparisons = [
    (1, "Schwann", 2, "Retic Fib III"),
    (2, "Retic Fib III", 1, "Schwann")
]

def calculate_enrichment(
    source_labels,
    full_labels,
    edge_src,
    edge_tgt,
    source_code,
    target_code
):
    source_mask = source_labels[edge_src] == source_code

    source_edges = source_mask.sum()

    if source_edges == 0:
        return np.nan

    target_edges = np.sum(
        source_mask &
        (full_labels[edge_tgt] == target_code)
    )

    target_abundance = np.mean(full_labels == target_code)

    if target_abundance == 0:
        return np.nan

    return (
        target_edges / source_edges
    ) / target_abundance


observed_donor = (
    validation_df
    .groupby([
        "donor_id",
        "source_population",
        "target_population"
    ])["enrichment"]
    .median()
)

observed_stats = {}

for source_code, source_name, target_code, target_name in comparisons:

    values = observed_donor.loc[
        (source_name, target_name)
    ].values

    observed_stats[(source_name, target_name)] = np.median(values)

null_distributions = {
    (source_name, target_name): []
    for _, source_name, _, target_name in comparisons
}

print("\nRunning permutations...")

for permutation in range(N_PERMUTATIONS):

    permutation_rows = []

    for item in sample_data:

        perm_source_labels = rng.permutation(
            item["source_labels"]
        )

        perm_full_labels = item["full_labels"].copy()

        perm_full_labels[
            item["selected_positions"]
        ] = perm_source_labels

        for source_code, source_name, target_code, target_name in comparisons:

            enrichment = calculate_enrichment(
                perm_source_labels,
                perm_full_labels,
                item["edge_src"],
                item["edge_tgt"],
                source_code,
                target_code
            )

            permutation_rows.append({
                "donor_id": item["donor_id"],
                "source_population": source_name,
                "target_population": target_name,
                "enrichment": enrichment
            })

    permutation_df = pd.DataFrame(permutation_rows)

    donor_null = (
        permutation_df
        .groupby([
            "donor_id",
            "source_population",
            "target_population"
        ])["enrichment"]
        .median()
    )

    for _, source_name, _, target_name in comparisons:

        values = donor_null.loc[
            (source_name, target_name)
        ].dropna().values

        null_distributions[
            (source_name, target_name)
        ].append(np.median(values))

    if (permutation + 1) % 250 == 0:
        print(
            "Completed:",
            permutation + 1,
            "/",
            N_PERMUTATIONS
        )

print("\nCALCULATING NULL-MODEL RESULTS")

results = []

for _, source_name, _, target_name in comparisons:

    key = (source_name, target_name)

    observed = observed_stats[key]
    null_values = np.asarray(null_distributions[key])

    null_mean = np.mean(null_values)
    null_sd = np.std(null_values, ddof=1)

    empirical_p = (
        1 + np.sum(null_values >= observed)
    ) / (N_PERMUTATIONS + 1)

    results.append({
        "source_population": source_name,
        "target_population": target_name,
        "observed_donor_median": observed,
        "null_mean": null_mean,
        "null_sd": null_sd,
        "observed_excess_pct": (
            (observed - null_mean) / null_mean
        ) * 100 if null_mean > 0 else np.nan,
        "empirical_p_value": empirical_p
    })

results_df = pd.DataFrame(results)

# Benjamini-Hochberg correction across the two directional tests
p_values = results_df["empirical_p_value"].values
order = np.argsort(p_values)
adjusted = np.empty(len(p_values))
running_min = 1.0

for rank in range(len(p_values), 0, -1):
    index = order[rank - 1]
    adjusted_value = (
        p_values[index] * len(p_values) / rank
    )
    running_min = min(running_min, adjusted_value)
    adjusted[index] = min(running_min, 1.0)

results_df["BH_adjusted_p"] = adjusted

print("\nFULL-TISSUE DIRECTED RANDOM-LABELING RESULTS")
print(results_df.round(5).to_string(index=False))

results_df.to_csv(
    "/content/drive/MyDrive/Neurothesis Resoure Library/Full_Tissue_Directed_Random_Labeling_Null.csv",
    index=False
)

print("\nSaved full-tissue directed random-labeling results.")
```

```python
import h5py
import numpy as np
import pandas as pd

print("CELL-TYPE MARKER EXPRESSION ANALYSIS")

genes_to_check = [
    "PDGFRB", "RGS5", "MCAM", "NOTCH3", "DES",
    "SOX10", "S100B", "MPZ",
    "COL1A1", "COL3A1", "LUM", "PDGFRA",
    "KIT", "CPA3", "HDC",
    "VWF", "KDR"
]

populations = [
    "Peri",
    "SM",
    "Schwann",
    "Mast",
    "Papil Fib",
    "DS",
    "Retic Fib I",
    "Retic Fib II",
    "Retic Fib III",
    "Perivasc Fib I",
    "Perivasc Fib II",
    "HEC",
    "VEC",
    "LEC",
    "Melano",
    "LC",
    "Mac",
    "Mono",
    "Plasma"
]

def decode_value(value):
    if isinstance(value, bytes):
        return value.decode("utf-8")
    return value

with h5py.File(h5ad_path, "r") as f:

    # Cell annotations
    label_group = f["obs"]["cell_type.detailed"]

    categories = [
        decode_value(x)
        for x in label_group["categories"][:]
    ]

    cell_codes = label_group["codes"][:]

    category_lookup = {
        name: i for i, name in enumerate(categories)
    }

    # Gene indices
    gene_names = [
        decode_value(x)
        for x in f["var"]["_index"][:]
    ]

    gene_lookup = {
        str(g).upper(): i
        for i, g in enumerate(gene_names)
    }

    gene_columns = {
        gene: gene_lookup[gene.upper()]
        for gene in genes_to_check
        if gene.upper() in gene_lookup
    }

    column_to_gene = {
        col: gene for gene, col in gene_columns.items()
    }

    # Sparse log1p expression matrix
    X = f["layers"]["log1p"]

    indptr = X["indptr"]
    indices = X["indices"]
    data = X["data"]

    results = []

    for population in populations:

        if population not in category_lookup:
            continue

        population_code = category_lookup[population]

        rows = np.flatnonzero(cell_codes == population_code)

        if len(rows) == 0:
            continue

        sums = {gene: 0.0 for gene in gene_columns}
        expressing = {gene: 0 for gene in gene_columns}

        # Process rows in manageable chunks
        for start in range(0, len(rows), 10000):

            chunk_rows = rows[start:start + 10000]

            for row_idx in chunk_rows:

                a = int(indptr[row_idx])
                b = int(indptr[row_idx + 1])

                row_cols = indices[a:b]
                row_values = data[a:b]

                if len(row_cols) == 0:
                    continue

                mask = np.isin(
                    row_cols,
                    list(column_to_gene.keys())
                )

                if not np.any(mask):
                    continue

                matched_cols = row_cols[mask]
                matched_values = row_values[mask]

                for col, value in zip(matched_cols, matched_values):

                    gene = column_to_gene[int(col)]

                    sums[gene] += float(value)

                    if value > 0:
                        expressing[gene] += 1

        for gene in gene_columns:

            results.append({
                "population": population,
                "cell_count": len(rows),
                "gene": gene,
                "mean_log1p": sums[gene] / len(rows),
                "pct_expressing": (
                    100 * expressing[gene] / len(rows)
                )
            })

expression_results = pd.DataFrame(results)

expression_results = expression_results.sort_values(
    ["gene", "mean_log1p"],
    ascending=[True, False]
)

print("\nMARKER EXPRESSION RESULTS")

print(
    expression_results.round(4).to_string(index=False)
)

print("\nAUDIT COMPLETE")
```

```python
import h5py
import numpy as np
import pandas as pd

print("PERI CELL-LEVEL MARKER CO-EXPRESSION AUDIT")

populations = [
    "Peri",
    "Perivasc Fib I",
    "Perivasc Fib II",
    "SM",
    "VEC",
    "HEC",
    "LEC",
    "Retic Fib III"
]

pericyte_markers = ["PDGFRB", "RGS5", "MCAM", "NOTCH3"]
vascular_markers = ["VWF", "KDR"]
contractile_marker = "DES"

def decode_value(value):
    if isinstance(value, bytes):
        return value.decode("utf-8")
    return value

with h5py.File(h5ad_path, "r") as f:

    # Detailed cell labels
    label_group = f["obs"]["cell_type.detailed"]

    categories = [
        decode_value(x)
        for x in label_group["categories"][:]
    ]

    cell_codes = label_group["codes"][:]

    category_lookup = {
        name: i for i, name in enumerate(categories)
    }

    # Gene indices
    gene_names = [
        decode_value(x)
        for x in f["var"]["_index"][:]
    ]

    gene_lookup = {
        str(g).upper(): i
        for i, g in enumerate(gene_names)
    }

    marker_columns = {
        gene: gene_lookup[gene]
        for gene in (
            pericyte_markers
            + vascular_markers
            + [contractile_marker]
        )
    }

    # Sparse log1p matrix
    X = f["layers"]["log1p"]

    indptr = X["indptr"]
    indices = X["indices"]

    results = []

    for population in populations:

        if population not in category_lookup:
            continue

        population_code = category_lookup[population]
        rows = np.flatnonzero(cell_codes == population_code)

        core_counts = []
        vascular_counts = []
        des_positive = []

        for row_idx in rows:

            a = int(indptr[row_idx])
            b = int(indptr[row_idx + 1])

            expressed_columns = set(
                indices[a:b].tolist()
            )

            core_count = sum(
                marker_columns[g] in expressed_columns
                for g in pericyte_markers
            )

            vascular_count = sum(
                marker_columns[g] in expressed_columns
                for g in vascular_markers
            )

            des_positive.append(
                marker_columns[contractile_marker]
                in expressed_columns
            )

            core_counts.append(core_count)
            vascular_counts.append(vascular_count)

        core_counts = np.asarray(core_counts)
        vascular_counts = np.asarray(vascular_counts)
        des_positive = np.asarray(des_positive)

        results.append({
            "population": population,
            "cell_count": len(rows),
            "pericyte_markers_ge2_pct":
                100 * np.mean(core_counts >= 2),
            "pericyte_markers_ge3_pct":
                100 * np.mean(core_counts >= 3),
            "vascular_markers_ge1_pct":
                100 * np.mean(vascular_counts >= 1),
            "pericyte_ge2_and_vascular_ge1_pct":
                100 * np.mean(
                    (core_counts >= 2) &
                    (vascular_counts >= 1)
                ),
            "DES_positive_pct":
                100 * np.mean(des_positive),
            "mean_pericyte_marker_count":
                np.mean(core_counts)
        })

result_df = pd.DataFrame(results)

print("\nCELL-LEVEL CO-EXPRESSION RESULTS")
print(result_df.round(2).to_string(index=False))

print("\nAUDIT COMPLETE")
```

```python
import numpy as np
import pandas as pd
from scipy import sparse

print("EXPANDED SIX-POPULATION NEIGHBORHOOD ANALYSIS")

# Confirm the existing full-tissue graph and labels
assert A_symmetric.shape[0] == len(full_labels_corrected)
assert len(sample_codes_full) == len(full_labels_corrected)

population_names = [
    "Mast",
    "Schwann",
    "Peri",
    "Perivasc Fib I",
    "Perivasc Fib II",
    "Retic Fib III"
]

# Recover detailed labels from the H5AD file
import h5py

def decode_value(value):
    if isinstance(value, bytes):
        return value.decode("utf-8")
    return value

with h5py.File(h5ad_path, "r") as f:

    label_group = f["obs"]["cell_type.detailed"]

    categories = [
        decode_value(x)
        for x in label_group["categories"][:]
    ]

    cell_codes = label_group["codes"][:]

category_lookup = {
    name: i for i, name in enumerate(categories)
}

# Create full-length population codes
expanded_labels = np.full(
    len(cell_codes),
    -1,
    dtype=np.int8
)

for i, population in enumerate(population_names):
    expanded_labels[
        cell_codes == category_lookup[population]
    ] = i

print("Population counts:")

for i, population in enumerate(population_names):
    print(
        population,
        int(np.sum(expanded_labels == i))
    )

# Calculate sample-specific neighbor enrichment
results = []

for sample_code in np.unique(sample_codes_full):

    sample_mask = sample_codes_full == sample_code
    sample_indices = np.flatnonzero(sample_mask)

    if len(sample_indices) == 0:
        continue

    graph = A_symmetric[
        sample_indices, :
    ][:, sample_indices].tocsr()

    labels = expanded_labels[sample_indices]

    population_counts = np.array([
        np.sum(labels == i)
        for i in range(len(population_names))
    ])

    total_cells = len(labels)

    if np.any(population_counts == 0):
        continue

    for source_idx, source_name in enumerate(population_names):

        source_rows = np.flatnonzero(labels == source_idx)

        if len(source_rows) == 0:
            continue

        neighbor_counts = np.asarray(
            graph[source_rows].sum(axis=0)
        ).ravel()

        total_edges = neighbor_counts.sum()

        if total_edges == 0:
            continue

        for target_idx, target_name in enumerate(population_names):

            observed_fraction = (
                neighbor_counts[labels == target_idx].sum()
                / total_edges
            )

            target_abundance = (
                population_counts[target_idx] / total_cells
            )

            enrichment = (
                observed_fraction / target_abundance
            )

            results.append({
                "sample_code": int(sample_code),
                "source_population": source_name,
                "target_population": target_name,
                "source_cell_count": len(source_rows),
                "target_cell_count": int(
                    population_counts[target_idx]
                ),
                "observed_neighbor_fraction": observed_fraction,
                "target_abundance_fraction": target_abundance,
                "enrichment": enrichment
            })

expanded_enrichment = pd.DataFrame(results)

summary = (
    expanded_enrichment
    .groupby(
        ["source_population", "target_population"]
    )["enrichment"]
    .agg(
        sample_count="count",
        median_enrichment="median",
        mean_enrichment="mean",
        q25=lambda x: x.quantile(0.25),
        q75=lambda x: x.quantile(0.75)
    )
    .reset_index()
)

print("\nEXPANDED NEIGHBOR ENRICHMENT SUMMARY")

print(
    summary.round(3).to_string(index=False)
)

print("\nAUDIT COMPLETE")
```

```python
# DONOR-LEVEL INFERENCE: MAST-CELL NEIGHBORHOODS

import numpy as np
import pandas as pd
from scipy.stats import wilcoxon

rng = np.random.default_rng(42)
N_BOOT = 10000

# Wide donor-level matrix
wide = mast_donor_summary.pivot(
    index="donor_id",
    columns="target_population",
    values="donor_median_enrichment"
)

# Bootstrap confidence interval for the median
def bootstrap_median_ci(values, n_boot=N_BOOT):
    values = np.asarray(values, dtype=float)
    samples = rng.choice(
        values,
        size=(n_boot, len(values)),
        replace=True
    )
    medians = np.median(samples, axis=1)
    return np.percentile(medians, [2.5, 97.5])

# Multiple-testing corrections
def bh_adjust(p_values):
    p = np.asarray(p_values, dtype=float)
    n = len(p)
    order = np.argsort(p)
    ranked = p[order]
    adjusted = ranked * n / np.arange(1, n + 1)
    adjusted = np.minimum.accumulate(adjusted[::-1])[::-1]
    result = np.empty(n)
    result[order] = np.clip(adjusted, 0, 1)
    return result

def holm_adjust(p_values):
    p = np.asarray(p_values, dtype=float)
    n = len(p)
    order = np.argsort(p)
    adjusted = np.maximum.accumulate(
        (n - np.arange(n)) * p[order]
    )
    result = np.empty(n)
    result[order] = np.clip(adjusted, 0, 1)
    return result

# PART 1: Each target compared with enrichment baseline = 1
baseline_results = []

for target in wide.columns:
    values = wide[target].dropna().values
    ci = bootstrap_median_ci(values)

    test = wilcoxon(
        values - 1,
        alternative="greater",
        method="approx"
    )

    baseline_results.append({
        "target": target,
        "donors": len(values),
        "median_enrichment": np.median(values),
        "CI_lower": ci[0],
        "CI_upper": ci[1],
        "p_value": test.pvalue
    })

baseline_results = pd.DataFrame(baseline_results)
baseline_results["BH_adjusted_p"] = bh_adjust(
    baseline_results["p_value"]
)

print("MAST-CELL ENRICHMENT VS BASELINE OF 1")
print(
    baseline_results.sort_values("p_value")
    .round(5).to_string(index=False)
)

# PART 2: Paired donor-level comparisons
comparisons = [
    ("Perivasc Fib I", "Schwann"),
    ("Perivasc Fib I", "Retic Fib III"),
    ("Perivasc Fib I", "Peri"),
    ("Perivasc Fib II", "Schwann"),
    ("Perivasc Fib II", "Retic Fib III"),
    ("Perivasc Fib II", "Peri")
]

paired_results = []

for first, second in comparisons:
    differences = (wide[first] - wide[second]).dropna().values
    ci = bootstrap_median_ci(differences)

    test = wilcoxon(
        differences,
        alternative="greater",
        method="approx"
    )

    paired_results.append({
        "comparison": f"{first} > {second}",
        "donors": len(differences),
        "median_difference": np.median(differences),
        "CI_lower": ci[0],
        "CI_upper": ci[1],
        "p_value": test.pvalue
    })

paired_results = pd.DataFrame(paired_results)
paired_results["Holm_adjusted_p"] = holm_adjust(
    paired_results["p_value"]
)

print("\nPAIRED DONOR COMPARISONS")
print(
    paired_results.sort_values("p_value")
    .round(5).to_string(index=False)
)
```

```python
# 10-POPULATION FULL-TISSUE NEIGHBOR ENRICHMENT

import pandas as pd
import numpy as np

population_count = len(population_names_10)

results_10 = []

for sample_code in np.unique(sample_codes_full):

    sample_indices = np.flatnonzero(
        sample_codes_full == sample_code
    )

    labels = expanded_labels_10[sample_indices]

    # Keep only selected populations
    valid = labels >= 0
    local_indices = sample_indices[valid]
    local_labels = labels[valid]

    # Restrict graph to this sample's selected cells
    A_local = A_symmetric[local_indices, :][:, local_indices]

    counts = np.bincount(
        local_labels,
        minlength=population_count
    )

    total_selected = counts.sum()

    # Neighbor composition for each source population
    for source in range(population_count):

        source_mask = local_labels == source
        source_count = counts[source]

        if source_count == 0:
            continue

        source_rows = A_local[source_mask]

        # Count edges to each target population
        target_edges = np.asarray(
            source_rows @ np.eye(population_count)[local_labels]
        ).sum(axis=0)

        total_edges = target_edges.sum()

        if total_edges == 0:
            continue

        observed_fraction = target_edges / total_edges

        for target in range(population_count):

            target_count = counts[target]

            # Cannot calculate abundance-adjusted enrichment
            # when the target population is absent
            if target_count == 0:
                continue

            abundance_fraction = target_count / total_selected

            enrichment = (
                observed_fraction[target] / abundance_fraction
            )

            results_10.append({
                "sample_code": int(sample_code),
                "source_population": population_names_10[source],
                "target_population": population_names_10[target],
                "source_cell_count": int(source_count),
                "target_cell_count": int(target_count),
                "observed_neighbor_fraction": observed_fraction[target],
                "target_abundance_fraction": abundance_fraction,
                "enrichment": enrichment
            })

expanded_enrichment_10 = pd.DataFrame(results_10)

print("RESULTS")
print("Rows:", len(expanded_enrichment_10))
print("Samples:", expanded_enrichment_10["sample_code"].nunique())
print("Source populations:",
      expanded_enrichment_10["source_population"].nunique())
print("Target populations:",
      expanded_enrichment_10["target_population"].nunique())

print("\nMAST SOURCE RESULTS")

print(
    expanded_enrichment_10[
        expanded_enrichment_10["source_population"] == "Mast"
    ].groupby("target_population")["enrichment"]
    .agg(["count", "median", "mean"])
    .round(4)
)

print("\nMAST TARGET SAMPLE COUNT")

print(
    expanded_enrichment_10[
        expanded_enrichment_10["target_population"] == "Mast"
    ].groupby("source_population")["sample_code"]
    .nunique()
)
```

```python
# FULL-TISSUE ENRICHMENT: 10 POPULATIONS

import numpy as np
import pandas as pd

results_full_10 = []

for sample_code in np.unique(sample_codes_full):

    sample_indices = np.flatnonzero(
        sample_codes_full == sample_code
    )

    labels = expanded_labels_10[sample_indices]

    # Full sample graph, retaining every cell type
    A_sample = A_symmetric[sample_indices, :][:, sample_indices]

    # Count all cells belonging to each selected population
    counts = np.bincount(
        labels[labels >= 0],
        minlength=len(population_names_10)
    )

    total_cells = len(sample_indices)

    # Map every cell to its selected population or -1
    valid = labels >= 0
    selected_local_indices = np.flatnonzero(valid)
    selected_labels = labels[valid]

    # Full graph rows for selected source cells
    for source in range(len(population_names_10)):

        source_positions = np.flatnonzero(labels == source)

        if len(source_positions) == 0:
            continue

        source_rows = A_sample[source_positions, :]

        # Count neighbors belonging to each selected target population
        target_edges = np.asarray(
            source_rows[:, selected_local_indices]
            @ np.eye(len(population_names_10))[selected_labels]
        ).sum(axis=0)

        total_edges = float(source_rows.sum())

        if total_edges == 0:
            continue

        for target in range(len(population_names_10)):

            target_count = counts[target]

            if target_count == 0:
                continue

            observed_fraction = target_edges[target] / total_edges
            abundance_fraction = target_count / total_cells

            results_full_10.append({
                "sample_code": int(sample_code),
                "source_population": population_names_10[source],
                "target_population": population_names_10[target],
                "source_cell_count": int(counts[source]),
                "target_cell_count": int(target_count),
                "observed_neighbor_fraction": observed_fraction,
                "target_abundance_fraction": abundance_fraction,
                "enrichment": observed_fraction / abundance_fraction
            })

full_tissue_enrichment_10 = pd.DataFrame(results_full_10)

print("FULL-TISSUE RESULTS")
print("Rows:", len(full_tissue_enrichment_10))
print("Samples:", full_tissue_enrichment_10["sample_code"].nunique())

print("\nMAST SOURCE ENRICHMENT")

print(
    full_tissue_enrichment_10[
        full_tissue_enrichment_10["source_population"] == "Mast"
    ].groupby("target_population")["enrichment"]
    .agg(["count", "median", "mean"])
    .round(4)
)

print("\nMAST SOURCE SAMPLE COUNT:",
      full_tissue_enrichment_10[
          full_tissue_enrichment_10["source_population"] == "Mast"
      ]["sample_code"].nunique())

print("\nMAST TARGET SAMPLE COUNT:",
      full_tissue_enrichment_10[
          full_tissue_enrichment_10["target_population"] == "Mast"
      ]["sample_code"].nunique())
```

```python
# DONOR-LEVEL MAST NEIGHBORHOOD COMPARISON: 10 POPULATIONS

full_10 = full_tissue_enrichment_10.copy()

full_10["sample_id"] = full_10["sample_code"].map(
    lambda x: sample_categories_full[int(x)]
)

full_10 = full_10.merge(
    mapping[["sample_id", "donor_id"]],
    on="sample_id",
    how="left",
    validate="many_to_one"
)

mast_full_10 = full_10[
    full_10["source_population"] == "Mast"
].copy()

mast_donor_10 = (
    mast_full_10.groupby(
        ["donor_id", "target_population"]
    )["enrichment"]
    .median()
    .reset_index(name="donor_median_enrichment")
)

wide_10 = mast_donor_10.pivot(
    index="donor_id",
    columns="target_population",
    values="donor_median_enrichment"
)

print("DONOR COVERAGE")
print("Donors:", wide_10.shape[0])
print("Missing donor-target values:", wide_10.isna().sum().sum())

# Compare each perivascular fibroblast population
# with all other non-Mast targets
comparisons_10 = []

for perivasc in ["Perivasc Fib I", "Perivasc Fib II"]:
    for comparator in [
        "Schwann", "Retic Fib III", "Peri",
        "VEC", "HEC", "LEC", "SM"
    ]:
        differences = (
            wide_10[perivasc] - wide_10[comparator]
        ).dropna().values

        test = wilcoxon(
            differences,
            alternative="greater",
            method="approx"
        )

        comparisons_10.append({
            "comparison": f"{perivasc} > {comparator}",
            "donors": len(differences),
            "median_difference": np.median(differences),
            "p_value": test.pvalue
        })

comparisons_10 = pd.DataFrame(comparisons_10)

# Holm adjustment across all 14 comparisons
p = comparisons_10["p_value"].values
order = np.argsort(p)
adjusted = np.maximum.accumulate(
    (len(p) - np.arange(len(p))) * p[order]
)

holm = np.empty(len(p))
holm[order] = np.clip(adjusted, 0, 1)

comparisons_10["Holm_adjusted_p"] = holm

print("\nPAIRED DONOR-LEVEL COMPARISONS")
print(
    comparisons_10.sort_values("p_value")
    .round(5)
    .to_string(index=False)
)
```

```python
# ANATOMICAL-SITE CONSISTENCY OF MAST-CELL NEIGHBOR ENRICHMENT

site_data = mast_full_10.merge(
    mapping[["sample_id", "anatomic_site"]],
    on="sample_id",
    how="left",
    validate="many_to_one"
)

# One median per donor, anatomical site, and target
donor_site_mast = (
    site_data.groupby(
        ["donor_id", "anatomic_site", "target_population"]
    )["enrichment"]
    .median()
    .reset_index(name="donor_site_enrichment")
)

targets_to_compare = [
    "Perivasc Fib I",
    "Perivasc Fib II",
    "Schwann",
    "Retic Fib III",
    "Peri",
    "VEC",
    "HEC",
    "LEC"
]

# Summarize across donors within each anatomical site
site_summary = (
    donor_site_mast[
        donor_site_mast["target_population"].isin(targets_to_compare)
    ]
    .groupby(["anatomic_site", "target_population"])
    .agg(
        donor_count=("donor_id", "nunique"),
        median_enrichment=("donor_site_enrichment", "median"),
        mean_enrichment=("donor_site_enrichment", "mean")
    )
    .reset_index()
)

site_pivot = site_summary.pivot(
    index="anatomic_site",
    columns="target_population",
    values="median_enrichment"
)

count_pivot = site_summary.pivot(
    index="anatomic_site",
    columns="target_population",
    values="donor_count"
)

print("SITE-LEVEL MEDIAN ENRICHMENT")
print(site_pivot.round(3).to_string())

print("\nDONOR COUNTS PER SITE")
print(count_pivot.to_string())

# Count sites where each perivascular population exceeds comparators
print("\nSITE CONSISTENCY")

for perivasc in ["Perivasc Fib I", "Perivasc Fib II"]:
    print(f"\n{perivasc}")

    for comparator in [
        "Schwann", "Retic Fib III", "Peri",
        "VEC", "HEC", "LEC"
    ]:
        matched = site_pivot[[perivasc, comparator]].dropna()

        greater = (
            matched[perivasc] > matched[comparator]
        ).sum()

        print(
            f"Greater than {comparator}: "
            f"{greater}/{len(matched)} sites"
        )
```

```python
# MAST ENRICHMENT: CELL-COUNT SENSITIVITY ANALYSIS

import numpy as np
import pandas as pd
from scipy.stats import wilcoxon
from statsmodels.stats.multitest import multipletests

# Identify the enrichment column
value_candidates = ["enrichment", "sample_enrichment"]
value_col = next(
    (c for c in value_candidates if c in mast_full_10.columns),
    None
)

if value_col is None:
    raise ValueError(
        f"Could not identify enrichment column. "
        f"Available columns: {mast_full_10.columns.tolist()}"
    )

# Attach donor IDs if they are not already present
sensitivity_data = mast_full_10.copy()

if "donor_id" not in sensitivity_data.columns:
    sensitivity_data = sensitivity_data.merge(
        mapping[["sample_id", "donor_id"]],
        on="sample_id",
        how="left"
    )

if sensitivity_data["donor_id"].isna().any():
    raise ValueError("Some samples could not be matched to donor IDs.")

# Create sample-level matrices
enrichment_wide = sensitivity_data.pivot_table(
    index=["sample_id", "donor_id"],
    columns="target_population",
    values=value_col,
    aggfunc="first"
)

count_wide = sensitivity_data.pivot_table(
    index=["sample_id", "donor_id"],
    columns="target_population",
    values="target_cell_count",
    aggfunc="first"
)

source_counts = (
    sensitivity_data.groupby("sample_id")["source_cell_count"]
    .first()
)

comparators = [
    "Schwann",
    "Retic Fib III",
    "Peri",
    "VEC",
    "HEC",
    "LEC",
    "SM"
]

perivascular_targets = [
    "Perivasc Fib I",
    "Perivasc Fib II"
]

results = []

for threshold in [10, 20, 50]:

    threshold_results = []

    for target in perivascular_targets:
        for comparator in comparators:

            # Require sufficient Mast cells and both target populations
            eligible = (
                (source_counts >= threshold)
                & (count_wide[target] >= threshold)
                & (count_wide[comparator] >= threshold)
            )

            eligible_samples = source_counts.index[eligible]

            pair_enrichment = enrichment_wide.loc[
                enrichment_wide.index.get_level_values("sample_id").isin(
                    eligible_samples
                ),
                [target, comparator]
            ].dropna()

            # Calculate donor-level medians using the same eligible samples
            donor_values = (
                pair_enrichment
                .reset_index()
                .groupby("donor_id")[[target, comparator]]
                .median()
            )

            differences = (
                donor_values[target] - donor_values[comparator]
            ).dropna()

            if len(differences) < 5:
                p_value = np.nan
            else:
                try:
                    p_value = wilcoxon(
                        differences,
                        alternative="greater",
                        method="auto"
                    ).pvalue
                except ValueError:
                    p_value = np.nan

            threshold_results.append({
                "minimum_cells": threshold,
                "perivascular_target": target,
                "comparator": comparator,
                "eligible_samples": len(eligible_samples),
                "donors": len(differences),
                "median_paired_difference": differences.median(),
                "target_donor_median": donor_values[target].median(),
                "comparator_donor_median": donor_values[comparator].median(),
                "p_value": p_value
            })

    # Holm correction across the 14 comparisons at this threshold
    valid_indices = [
        i for i, row in enumerate(threshold_results)
        if pd.notna(row["p_value"])
    ]

    if valid_indices:
        adjusted = multipletests(
            [threshold_results[i]["p_value"] for i in valid_indices],
            method="holm"
        )[1]

        for i, adjusted_p in zip(valid_indices, adjusted):
            threshold_results[i]["holm_p"] = adjusted_p

    for row in threshold_results:
        row.setdefault("holm_p", np.nan)
        results.append(row)

sensitivity_results = pd.DataFrame(results)

print("CELL-COUNT SENSITIVITY RESULTS")
print(
    sensitivity_results.sort_values(
        ["minimum_cells", "perivascular_target", "holm_p"]
    ).round(4).to_string(index=False)
)

print("\nDONOR COVERAGE BY THRESHOLD")
print(
    sensitivity_results.groupby("minimum_cells")[
        ["eligible_samples", "donors"]
    ].agg(["min", "median", "max"]).to_string()
)
```

```python
# LEAVE-ONE-DONOR-OUT ROBUSTNESS
# Uses the same eligible samples for each paired comparison.

import numpy as np
import pandas as pd

comparators = [
    "Schwann", "Retic Fib III", "Peri",
    "VEC", "HEC", "LEC", "SM"
]

perivascular_targets = [
    "Perivasc Fib I", "Perivasc Fib II"
]

results = []

data = mast_full_10.merge(
    mapping[["sample_id", "donor_id"]],
    on="sample_id",
    how="left"
)

for threshold in [10, 20, 50]:

    for target in perivascular_targets:

        for comparator in comparators:

            pair = data[
                data["target_population"].isin([target, comparator])
            ].copy()

            # Require adequate Mast, target, and comparator counts
            source_ok = (
                pair.groupby("sample_id")["source_cell_count"]
                .first() >= threshold
            )

            target_counts = (
                pair[pair["target_population"] == target]
                .set_index("sample_id")["target_cell_count"]
            )

            comparator_counts = (
                pair[pair["target_population"] == comparator]
                .set_index("sample_id")["target_cell_count"]
            )

            eligible = (
                source_ok &
                (target_counts >= threshold) &
                (comparator_counts >= threshold)
            )

            eligible_samples = eligible[eligible].index

            pair = pair[
                pair["sample_id"].isin(eligible_samples)
            ]

            # Donor-level median enrichment for each population
            donor_values = (
                pair.groupby(["donor_id", "target_population"])["enrichment"]
                .median()
                .unstack()
            )

            if target not in donor_values or comparator not in donor_values:
                continue

            donor_values = donor_values.dropna(
                subset=[target, comparator]
            )

            differences = (
                donor_values[target] - donor_values[comparator]
            )

            if len(differences) < 3:
                continue

            # Recalculate the median paired difference after
            # removing each donor individually
            loo_medians = [
                differences.drop(index=donor).median()
                for donor in differences.index
            ]

            results.append({
                "minimum_cells": threshold,
                "target": target,
                "comparator": comparator,
                "donors": len(differences),
                "donors_with_positive_difference": int(
                    (differences > 0).sum()
                ),
                "full_median_difference": differences.median(),
                "minimum_LOO_median": min(loo_medians),
                "maximum_LOO_median": max(loo_medians),
                "all_LOO_medians_positive": all(
                    value > 0 for value in loo_medians
                )
            })

loo_results = pd.DataFrame(results)

print("LEAVE-ONE-DONOR-OUT RESULTS")
print(
    loo_results.sort_values(
        ["minimum_cells", "target", "comparator"]
    ).round(4).to_string(index=False)
)

print("\nOVERALL CHECK")
print(
    loo_results.groupby("minimum_cells").agg(
        comparisons=("comparator", "count"),
        all_leave_one_out_positive=(
            "all_LOO_medians_positive", "all"
        ),
        minimum_LOO_effect=("minimum_LOO_median", "min")
    ).round(4).to_string()
)
```

```python
# SINGLE-CELL MAST-CELL NEIGHBORHOOD ATLAS

import numpy as np
import pandas as pd
from scipy import sparse

# Verify core objects
assert A_symmetric.shape[0] == len(expanded_labels_10)
assert len(sample_codes_full) == len(expanded_labels_10)
assert len(population_names_10) == 10

n_cells = len(expanded_labels_10)

# Create a sparse one-hot annotation matrix
valid = (
    (expanded_labels_10 >= 0) &
    (expanded_labels_10 < len(population_names_10))
)

annotated_indices = np.flatnonzero(valid)
annotated_labels = expanded_labels_10[annotated_indices]

label_matrix = sparse.csr_matrix(
    (
        np.ones(len(annotated_indices), dtype=np.int8),
        (annotated_indices, annotated_labels)
    ),
    shape=(n_cells, len(population_names_10))
)

# Count each annotated population among every cell's neighbors
print("Calculating full-tissue neighborhood composition...")

neighbor_counts_all = A_symmetric @ label_matrix

# Extract Mast cells
mast_indices = np.flatnonzero(expanded_labels_10 == 0)

mast_neighbor_counts = neighbor_counts_all[
    mast_indices
].toarray()

mast_degrees = np.asarray(
    A_symmetric[mast_indices].sum(axis=1)
).ravel()

# Calculate unannotated/other neighbors
other_counts = (
    mast_degrees - mast_neighbor_counts.sum(axis=1)
)

other_counts = np.maximum(other_counts, 0)

# Construct cell-level table
neighborhood = pd.DataFrame(
    mast_neighbor_counts,
    columns=population_names_10
)

neighborhood["Other"] = other_counts
neighborhood["mast_cell_index"] = mast_indices
neighborhood["sample_code"] = sample_codes_full[mast_indices]
neighborhood["degree"] = mast_degrees

# Map sample codes to sample IDs
neighborhood["sample_id"] = [
    sample_categories_full[int(code)]
    for code in neighborhood["sample_code"]
]

# Add donor and anatomical site
neighborhood = neighborhood.merge(
    mapping[["sample_id", "donor_id", "anatomic_site"]],
    on="sample_id",
    how="left"
)

# Calculate neighbor fractions for each mast cell
population_columns = population_names_10 + ["Other"]

for pop in population_columns:
    neighborhood[f"{pop}_fraction"] = (
        neighborhood[pop] /
        neighborhood["degree"].replace(0, np.nan)
    )

print("\nMAST-CELL NEIGHBORHOOD ATLAS")
print("Mast cells:", len(neighborhood))
print("Samples:", neighborhood["sample_id"].nunique())
print("Donors:", neighborhood["donor_id"].nunique())
print("Cells with zero graph degree:", (neighborhood["degree"] == 0).sum())

print("\nMEAN NEIGHBOR COMPOSITION")
fraction_columns = [f"{p}_fraction" for p in population_columns]

print(
    neighborhood[fraction_columns]
    .mean()
    .sort_values(ascending=False)
    .round(4)
    .to_string()
)

print("\nMEDIAN NEIGHBOR COMPOSITION")
print(
    neighborhood[fraction_columns]
    .median()
    .sort_values(ascending=False)
    .round(4)
    .to_string()
)

# Save the cell-level dataset
output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Mast_Cell_Neighborhood_Atlas.csv"
)

neighborhood.to_csv(output_path, index=False)

print("\nSaved:", output_path)
```

```python
# MAST-CELL NEIGHBORHOOD PREVALENCE AND HETEROGENEITY

populations = population_names_10 + ["Other"]

summary_rows = []

for pop in populations:

    counts = neighborhood[pop]
    positive = counts > 0

    summary_rows.append({
        "population": pop,
        "mast_cells_with_neighbor": int(positive.sum()),
        "percent_mast_cells_with_neighbor": 100 * positive.mean(),
        "mean_neighbors_when_present": (
            counts[positive].mean() if positive.any() else np.nan
        ),
        "median_neighbors_when_present": (
            counts[positive].median() if positive.any() else np.nan
        ),
        "q90_when_present": (
            counts[positive].quantile(.90) if positive.any() else np.nan
        ),
        "total_neighbor_cells": int(counts.sum())
    })

prevalence_summary = pd.DataFrame(summary_rows)

print("POPULATION PREVALENCE ACROSS MAST-CELL NEIGHBORHOODS")
print(
    prevalence_summary.sort_values(
        "percent_mast_cells_with_neighbor",
        ascending=False
    ).round(2).to_string(index=False)
)

# Donor-level prevalence
donor_prevalence = (
    neighborhood.groupby("donor_id")[
        [pop for pop in populations]
    ]
    .agg(lambda x: (x > 0).mean() * 100)
)

print("\nDONOR-LEVEL PREVALENCE")
print(
    donor_prevalence.median()
    .sort_values(ascending=False)
    .round(2)
    .to_string()
)

print("\nDONOR CONSISTENCY")
for pop in populations:
    donor_values = donor_prevalence[pop]

    print(
        f"{pop}: "
        f"{(donor_values > 50).sum()}/"
        f"{len(donor_values)} donors have >50% of mast cells "
        f"with at least one such neighbor"
    )

# Graph degree distribution
print("\nMAST-CELL GRAPH DEGREE")
print(
    neighborhood["degree"]
    .describe(percentiles=[.10, .25, .50, .75, .90])
    .round(2)
)

# Save outputs
prevalence_summary.to_csv(
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Mast_Neighborhood_Prevalence.csv",
    index=False
)

donor_prevalence.to_csv(
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Mast_Neighborhood_Donor_Prevalence.csv"
)

print("\nSaved prevalence and donor-level tables.")
```

```python
# FIXED-MAST, WITHIN-SAMPLE RANDOM-LABELING NULL
# Preserve:
# 1. Mast-cell locations
# 2. Sample-specific counts of all 9 other annotated populations
# 3. Locations of unannotated cells
# 4. The original full-tissue graph
#
# Randomize:
# Identities of annotated non-Mast cells within each sample.

import numpy as np
import pandas as pd
from scipy import sparse
from statsmodels.stats.multitest import multipletests

N_PERM = 499
SEED = 20261108

rng = np.random.default_rng(SEED)

A_csr = A_symmetric.tocsr()

target_names = population_names_10[1:]
n_targets = len(target_names)

sample_records = []

for code in np.unique(sample_codes_full):

    sample_indices = np.flatnonzero(sample_codes_full == code)
    labels = expanded_labels_10[sample_indices].copy()

    mast_local = np.flatnonzero(labels == 0)

    if len(mast_local) == 0:
        continue

    mast_global = sample_indices[mast_local]

    # Obtain all graph neighbors of Mast cells
    mast_rows = A_csr[mast_global, :]
    neighbor_global = mast_rows.indices

    # Convert global cell indices to sample-local indices
    neighbor_local = np.searchsorted(
        sample_indices,
        neighbor_global
    )

    # Validate that graph edges remain within sample
    assert np.all(sample_indices[neighbor_local] == neighbor_global)

    # Only shuffle annotated non-Mast cells.
    # Mast cells and unannotated cells remain fixed.
    shuffle_positions = np.flatnonzero(labels > 0)

    # Observed target neighbor fractions
    observed_neighbor_labels = labels[neighbor_local]

    observed_counts = np.bincount(
        observed_neighbor_labels[
            observed_neighbor_labels >= 1
        ],
        minlength=10
    )[1:10]

    observed_fractions = (
        observed_counts / len(neighbor_local)
    )

    sample_id = sample_categories_full[int(code)]

    donor_id = mapping.loc[
        mapping["sample_id"] == sample_id,
        "donor_id"
    ].iloc[0]

    sample_records.append({
        "sample_code": int(code),
        "sample_id": sample_id,
        "donor_id": donor_id,
        "labels": labels,
        "shuffle_positions": shuffle_positions,
        "neighbor_local": neighbor_local,
        "observed_fractions": observed_fractions
    })

n_samples = len(sample_records)

observed_sample = np.vstack([
    record["observed_fractions"]
    for record in sample_records
])

null_sample = np.zeros(
    (N_PERM, n_samples, n_targets),
    dtype=np.float32
)

print("Samples included:", n_samples)
print("Mast cells represented:", sum(
    len(np.flatnonzero(r["labels"] == 0))
    for r in sample_records
))
print("Running", N_PERM, "permutations...")

for perm in range(N_PERM):

    for sample_i, record in enumerate(sample_records):

        permuted_labels = record["labels"].copy()

        positions = record["shuffle_positions"]

        permuted_labels[positions] = rng.permutation(
            permuted_labels[positions]
        )

        neighbor_labels = permuted_labels[
            record["neighbor_local"]
        ]

        counts = np.bincount(
            neighbor_labels[neighbor_labels >= 1],
            minlength=10
        )[1:10]

        null_sample[perm, sample_i, :] = (
            counts / len(neighbor_labels)
        )

    if (perm + 1) % 100 == 0:
        print("Completed", perm + 1, "/", N_PERM)

# Aggregate sample results to donor-level medians
donors = sorted(set(
    record["donor_id"] for record in sample_records
))

observed_donor = np.zeros((len(donors), n_targets))

null_donor = np.zeros(
    (N_PERM, len(donors), n_targets),
    dtype=np.float32
)

for donor_i, donor in enumerate(donors):

    sample_positions = [
        i for i, record in enumerate(sample_records)
        if record["donor_id"] == donor
    ]

    observed_donor[donor_i] = np.median(
        observed_sample[sample_positions],
        axis=0
    )

    null_donor[:, donor_i, :] = np.median(
        null_sample[:, sample_positions, :],
        axis=1
    )

# Overall statistic: median across donor-level medians
observed_overall = np.median(observed_donor, axis=0)
null_overall = np.median(null_donor, axis=1)

results = []

for j, target in enumerate(target_names):

    null_values = null_overall[:, j]
    observed = observed_overall[j]

    null_mean = null_values.mean()

    empirical_p = (
        1 + np.sum(null_values >= observed)
    ) / (N_PERM + 1)

    results.append({
        "target": target,
        "donors": len(donors),
        "observed_median_neighbor_fraction": observed,
        "null_mean": null_mean,
        "null_95th_percentile": np.quantile(null_values, .95),
        "observed_to_null_ratio": (
            observed / null_mean if null_mean > 0 else np.nan
        ),
        "empirical_p": empirical_p
    })

permutation_results = pd.DataFrame(results)

permutation_results["BH_adjusted_p"] = multipletests(
    permutation_results["empirical_p"],
    method="fdr_bh"
)[1]

permutation_results = permutation_results.sort_values(
    "observed_to_null_ratio",
    ascending=False
)

print("\nFIXED-MAST RANDOM-LABELING RESULTS")
print(permutation_results.round(5).to_string(index=False))

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Fixed_Mast_Random_Labeling_Null.csv"
)

permutation_results.to_csv(output_path, index=False)

print("\nSaved:", output_path)
```

```python
# BUILD DERMAL MAST-CELL AND MATCHED-CONTROL COHORTS

import numpy as np
import pandas as pd

# Recalculate degree directly from the graph used in the analysis
graph_degrees = np.asarray(
    A_symmetric.sum(axis=1)
).ravel().astype(np.int32)

# Define dermal mast cells
dermis_mask = compartment_codes == 0
mast_mask = expanded_labels_10 == 0

mast_global = np.flatnonzero(
    dermis_mask & mast_mask
)

# Non-mast dermal controls must have an annotated population
control_global = np.flatnonzero(
    dermis_mask &
    (expanded_labels_10 > 0)
)

print("Dermal mast cells:", len(mast_global))
print("Eligible annotated non-mast controls:", len(control_global))

# Build matching strata: sample + graph degree
mast_strata = pd.DataFrame({
    "sample_code": sample_codes_full[mast_global],
    "degree": graph_degrees[mast_global]
})

control_strata = pd.DataFrame({
    "sample_code": sample_codes_full[control_global],
    "degree": graph_degrees[control_global]
})

mast_strata_counts = mast_strata.value_counts().rename("mast_count")
control_strata_counts = control_strata.value_counts().rename("control_count")

matching_audit = pd.concat(
    [mast_strata_counts, control_strata_counts],
    axis=1
).fillna(0).astype(int)

matching_audit["matched"] = (
    matching_audit["mast_count"] > 0
) & (
    matching_audit["control_count"] > 0
)

print("\nMATCHING STRATA")
print("Total mast-containing strata:", len(matching_audit))
print("Strata with eligible controls:", matching_audit["matched"].sum())
print(
    "Mast cells in matchable strata:",
    matching_audit.loc[
        matching_audit["matched"], "mast_count"
    ].sum()
)
print(
    "Percentage of dermal mast cells matchable:",
    round(
        100 * matching_audit.loc[
            matching_audit["matched"], "mast_count"
        ].sum() / len(mast_global),
        2
    )
)

print("\nUNMATCHED MAST STRATA")
print(
    matching_audit.loc[
        ~matching_audit["matched"]
    ].head(20).to_string()
)

# Save exact cohort indices for the next analysis
matched_strata = matching_audit.index[
    matching_audit["matched"]
]

matched_mast_global = []
matched_control_pools = []

for sample_code, degree in matched_strata:
    mast_cells = mast_global[
        (sample_codes_full[mast_global] == sample_code) &
        (graph_degrees[mast_global] == degree)
    ]

    controls = control_global[
        (sample_codes_full[control_global] == sample_code) &
        (graph_degrees[control_global] == degree)
    ]

    matched_mast_global.extend(mast_cells.tolist())
    matched_control_pools.append({
        "sample_code": int(sample_code),
        "degree": int(degree),
        "mast_indices": mast_cells,
        "control_indices": controls
    })

matched_mast_global = np.asarray(
    matched_mast_global, dtype=np.int64
)

print("\nFINAL MATCHED COHORT")
print("Matched mast cells:", len(matched_mast_global))
print("Matching strata:", len(matched_control_pools))
print("Unique samples:", len(np.unique(
    sample_codes_full[matched_mast_global]
)))
```

```python
# SPATIAL-POSITION-PRESERVING MATCHED-CONTROL ANALYSIS

import numpy as np
import pandas as pd
from statsmodels.stats.multitest import multipletests

N_PERM_MATCH = 499
SEED_MATCH = 20261109
rng_match = np.random.default_rng(SEED_MATCH)

A_csr = A_symmetric.tocsr()
target_names = population_names_10[1:]
n_targets = len(target_names)

# Prepare graph-neighbor population counts for each candidate cell
candidate_global = np.concatenate([
    item["mast_indices"] for item in matched_control_pools
] + [
    item["control_indices"] for item in matched_control_pools
])

candidate_global = np.unique(candidate_global)

candidate_rows = A_csr[candidate_global, :]
candidate_neighbor_indices = candidate_rows.indices

# Map global cell indices to candidate-row positions
candidate_lookup = {
    int(global_idx): i
    for i, global_idx in enumerate(candidate_global)
}

candidate_neighbor_labels = expanded_labels_10[
    candidate_neighbor_indices
]

candidate_neighbor_counts = np.zeros(
    (len(candidate_global), n_targets),
    dtype=np.int16
)

row_lengths = np.diff(candidate_rows.indptr)

for i in range(len(candidate_global)):
    start = candidate_rows.indptr[i]
    end = candidate_rows.indptr[i + 1]
    labels_i = candidate_neighbor_labels[start:end]

    counts_i = np.bincount(
        labels_i[labels_i >= 1],
        minlength=10
    )[1:10]

    candidate_neighbor_counts[i] = counts_i

candidate_neighbor_fractions = (
    candidate_neighbor_counts /
    row_lengths[:, None]
)

print("Candidate cells:", len(candidate_global))
print("Candidate neighborhood matrix:", candidate_neighbor_fractions.shape)
print("All neighborhood fractions finite:",
      np.isfinite(candidate_neighbor_fractions).all())

# Observed mast-cell neighborhood fractions, matched cohort only
observed_mast_counts = []
observed_control_counts = []

for item in matched_control_pools:
    mast_positions = np.array([
        candidate_lookup[int(x)]
        for x in item["mast_indices"]
    ])

    control_positions = np.array([
        candidate_lookup[int(x)]
        for x in item["control_indices"]
    ])

    observed_mast_counts.append(
        candidate_neighbor_fractions[mast_positions].mean(axis=0)
    )

    observed_control_counts.append(
        candidate_neighbor_fractions[control_positions]
    )

# Calculate the observed matched difference by stratum,
# weighted by the number of mast cells in that stratum.
stratum_differences = []

for item, mast_mean in zip(
    matched_control_pools, observed_mast_counts
):
    mast_positions = np.array([
        candidate_lookup[int(x)]
        for x in item["mast_indices"]
    ])

    stratum_differences.append({
        "sample_code": item["sample_code"],
        "mast_count": len(item["mast_indices"]),
        "mast_mean": mast_mean
    })

# Donor identity for each sample
sample_to_donor_local = {
    row["sample_code"]: sample_to_donor[
        sample_categories_full[row["sample_code"]]
    ]
    for row in stratum_differences
}

# Aggregate mast observations to sample level
sample_mast = {}

for item, mast_mean in zip(
    matched_control_pools, observed_mast_counts
):
    code = item["sample_code"]
    sample_mast.setdefault(code, []).append(
        (len(item["mast_indices"]), mast_mean)
    )

sample_mast_fraction = {}

for code, entries in sample_mast.items():
    total_n = sum(n for n, _ in entries)
    sample_mast_fraction[code] = sum(
        n * values for n, values in entries
    ) / total_n

# Matched-control resampling
sample_codes_matched = sorted(sample_mast_fraction)
donors_matched = sorted(set(
    sample_to_donor_local[code]
    for code in sample_codes_matched
))

observed_donor_mast = np.zeros((len(donors_matched), n_targets))
null_donor_control = np.zeros(
    (N_PERM_MATCH, len(donors_matched), n_targets),
    dtype=np.float32
)

for donor_i, donor in enumerate(donors_matched):
    donor_samples = [
        code for code in sample_codes_matched
        if sample_to_donor_local[code] == donor
    ]

    observed_donor_mast[donor_i] = np.median(
        [sample_mast_fraction[code] for code in donor_samples],
        axis=0
    )

print("\nMatched donors:", len(donors_matched))
print("Matched samples:", len(sample_codes_matched))
print("Running matched-control resampling...")

for perm in range(N_PERM_MATCH):

    sample_control_results = {}

    for item in matched_control_pools:
        code = item["sample_code"]

        mast_n = len(item["mast_indices"])
        control_positions = np.array([
            candidate_lookup[int(x)]
            for x in item["control_indices"]
        ])

        selected = rng_match.choice(
            control_positions,
            size=mast_n,
            replace=False
        )

        sample_control_results.setdefault(code, []).append(
            (
                mast_n,
                candidate_neighbor_fractions[selected].mean(axis=0)
            )
        )

    sample_control_fraction = {}

    for code, entries in sample_control_results.items():
        total_n = sum(n for n, _ in entries)
        sample_control_fraction[code] = sum(
            n * values for n, values in entries
        ) / total_n

    for donor_i, donor in enumerate(donors_matched):
        donor_samples = [
            code for code in sample_codes_matched
            if sample_to_donor_local[code] == donor
        ]

        null_donor_control[perm, donor_i] = np.median(
            [sample_control_fraction[code] for code in donor_samples],
            axis=0
        )

    if (perm + 1) % 100 == 0:
        print("Completed", perm + 1, "/", N_PERM_MATCH)

# Compare donor-level mast-cell values with matched-control null
observed_overall_matched = np.median(
    observed_donor_mast, axis=0
)

null_overall_matched = np.median(
    null_donor_control, axis=1
)

results_matched = []

for j, target in enumerate(target_names):
    null_values = null_overall_matched[:, j]
    observed = observed_overall_matched[j]

    p_value = (
        1 + np.sum(null_values >= observed)
    ) / (N_PERM_MATCH + 1)

    results_matched.append({
        "target": target,
        "donors": len(donors_matched),
        "mast_observed_fraction": observed,
        "matched_control_null_mean": null_values.mean(),
        "matched_control_null_95th": np.quantile(
            null_values, .95
        ),
        "observed_to_control_ratio": (
            observed / null_values.mean()
            if null_values.mean() > 0 else np.nan
        ),
        "empirical_p": p_value
    })

results_matched = pd.DataFrame(results_matched)

results_matched["BH_adjusted_p"] = multipletests(
    results_matched["empirical_p"],
    method="fdr_bh"
)[1]

results_matched = results_matched.sort_values(
    "observed_to_control_ratio",
    ascending=False
)

print("\nMATCHED-CONTROL NEIGHBORHOOD RESULTS")
print(results_matched.round(5).to_string(index=False))

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Dermal_Mast_Matched_Control_Neighborhoods.csv"
)

results_matched.to_csv(output_path, index=False)

print("\nSaved:", output_path)
```

```python
# REBUILD CORE GRAPH AND METADATA WITHOUT LOADING EXPRESSION DATA

import h5py
import numpy as np
import pandas as pd
from scipy.sparse import csr_matrix

h5ad_path = "/content/drive/MyDrive/Neurothesis Resoure Library/merfish.integrated_annotated (2).h5ad"

def decode_categories(dataset):
    return [
        x.decode("utf-8") if isinstance(x, bytes) else str(x)
        for x in dataset[:]
    ]

with h5py.File(h5ad_path, "r") as f:

    # 1. Reconstruct the spatial connectivity graph
    g = f["obsp"]["spatial_connectivities"]
    graph_shape = tuple(g.attrs["shape"])

    A_original = csr_matrix(
        (g["data"][:], g["indices"][:], g["indptr"][:]),
        shape=graph_shape
    )

    # Ensure the graph is symmetric
    asymmetry_count = (A_original != A_original.T).nnz
    A_symmetric = A_original.maximum(A_original.T).tocsr()

    # 2. Load detailed cell annotations
    detailed_group = f["obs"]["cell_type.detailed"]
    detailed_categories = decode_categories(
        detailed_group["categories"]
    )
    detailed_codes = detailed_group["codes"][:]

    population_names_10 = [
        "Mast", "Schwann", "Peri",
        "Perivasc Fib I", "Perivasc Fib II",
        "Retic Fib III", "VEC", "HEC", "LEC", "SM"
    ]

    expanded_labels_10 = np.full(
        len(detailed_codes), -1, dtype=np.int8
    )

    for code, population in enumerate(population_names_10):
        category_code = detailed_categories.index(population)
        expanded_labels_10[
            detailed_codes == category_code
        ] = code

    # 3. Load sample, donor, site, and compartment metadata
    def load_categorical(field):
        group = f["obs"][field]
        categories = decode_categories(group["categories"])
        codes = group["codes"][:]
        return categories, codes

    sample_categories_full, sample_codes_full = load_categorical("sample_id")
    donor_categories, donor_codes_full = load_categorical("donor_id")
    anatomic_site_categories, anatomic_site_codes_full = load_categorical("anatomic_site")
    compartment_categories, compartment_codes = load_categorical("tissue_compartment")

    # 4. Load spatial coordinates
    coordinates = f["obsm"]["spatial"][:]

# 5. Reconstruct readable metadata mappings
sample_to_donor = {}
mapping_rows = []

for sample_code, sample_id in enumerate(sample_categories_full):

    sample_mask = sample_codes_full == sample_code

    donor_values = np.unique(donor_codes_full[sample_mask])
    site_values = np.unique(anatomic_site_codes_full[sample_mask])

    donor_values = donor_values[donor_values >= 0]
    site_values = site_values[site_values >= 0]

    if len(donor_values) != 1 or len(site_values) != 1:
        print("METADATA WARNING:", sample_id,
              "donors:", donor_values, "sites:", site_values)
        continue

    donor_id = donor_categories[int(donor_values[0])]
    site = anatomic_site_categories[int(site_values[0])]

    sample_to_donor[sample_id] = donor_id

    mapping_rows.append({
        "sample_id": sample_id,
        "donor_id": donor_id,
        "anatomic_site": site
    })

mapping = pd.DataFrame(mapping_rows)

# 6. Calculate graph degrees directly from reconstructed graph
graph_degrees = np.asarray(A_symmetric.sum(axis=1)).ravel()

# 7. Verify reconstruction
print("GRAPH AUDIT")
print("Shape:", A_symmetric.shape)
print("Nonzero edges:", A_symmetric.nnz)
print("Original asymmetry entries:", asymmetry_count)
print("Symmetric:", (A_symmetric != A_symmetric.T).nnz == 0)
print("Self-connections:", A_symmetric.diagonal().sum())

print("\nMETADATA AUDIT")
print("Cells:", len(expanded_labels_10))
print("Samples:", len(sample_categories_full))
print("Donors:", len(donor_categories))
print("Anatomical sites:", len(anatomic_site_categories))
print("Coordinates:", coordinates.shape)
print("Sample mapping rows:", len(mapping))

print("\nTEN-POPULATION COUNTS")
for code, population in enumerate(population_names_10):
    print(f"{population}: {(expanded_labels_10 == code).sum():,}")

print("\nDERMAL MAST AUDIT")
dermal_mask = compartment_codes == compartment_categories.index("dermis")
mast_mask = expanded_labels_10 == population_names_10.index("Mast")

print("Dermal Mast cells:", np.sum(dermal_mask & mast_mask))
print("Dermal annotated non-Mast controls:",
      np.sum(dermal_mask & (expanded_labels_10 >= 0) & ~mast_mask))

print("\nGRAPH DEGREE SUMMARY FOR DERMAL MAST CELLS")
mast_degrees = graph_degrees[dermal_mask & mast_mask]
print(pd.Series(mast_degrees).describe())

print("\nRECONSTRUCTION COMPLETE")
```

```python
# BUILD EXACT SAMPLE-AND-DEGREE MATCHED CONTROL POOLS

from collections import defaultdict
import numpy as np
import pandas as pd

mast_code = population_names_10.index("Mast")
dermis_code = compartment_categories.index("dermis")

# Eligible dermal Mast cells
dermal_mast_mask = (
    (compartment_codes == dermis_code) &
    (expanded_labels_10 == mast_code)
)

dermal_mast_indices = np.flatnonzero(dermal_mast_mask)

# Eligible controls: dermal, annotated, non-Mast cells
control_mask = (
    (compartment_codes == dermis_code) &
    (expanded_labels_10 >= 0) &
    (expanded_labels_10 != mast_code)
)

control_indices = np.flatnonzero(control_mask)

# Build pools keyed by exact sample code and graph degree
control_groups = defaultdict(list)

for idx in control_indices:
    key = (int(sample_codes_full[idx]), int(graph_degrees[idx]))
    control_groups[key].append(idx)

matched_control_pools = {
    key: np.asarray(indices, dtype=np.int32)
    for key, indices in control_groups.items()
}

# Count Mast cells in each exact matching stratum
mast_groups = defaultdict(list)

for idx in dermal_mast_indices:
    key = (int(sample_codes_full[idx]), int(graph_degrees[idx]))
    mast_groups[key].append(idx)

# Retain only strata with available controls
matched_mast_indices = []
matched_strata = []

unmatched_mast_count = 0
unmatched_strata = []

for key, mast_list in mast_groups.items():
    pool = matched_control_pools.get(key)

    if pool is None or len(pool) == 0:
        unmatched_mast_count += len(mast_list)
        unmatched_strata.append((key, len(mast_list), 0))
        continue

    if len(pool) < len(mast_list):
        unmatched_mast_count += len(mast_list)
        unmatched_strata.append((key, len(mast_list), len(pool)))
        continue

    matched_mast_indices.extend(mast_list)

    matched_strata.append({
        "sample_code": key[0],
        "degree": key[1],
        "mast_count": len(mast_list),
        "control_pool_size": len(pool),
        "control_to_mast_ratio": len(pool) / len(mast_list)
    })

matched_mast_indices = np.asarray(
    matched_mast_indices, dtype=np.int32
)

matched_strata_df = pd.DataFrame(matched_strata)

# Audit
print("MATCHED-CONTROL AUDIT")
print("Dermal Mast cells:", len(dermal_mast_indices))
print("Eligible dermal non-Mast controls:", len(control_indices))
print("Matched Mast cells:", len(matched_mast_indices))
print("Unmatched Mast cells:", unmatched_mast_count)
print("Matched strata:", len(matched_strata_df))
print("Samples represented:",
      matched_strata_df["sample_code"].nunique())

print("\nCONTROL-POOL SIZE")
print(matched_strata_df["control_pool_size"].describe())

print("\nCONTROL-TO-MAST RATIO")
print(matched_strata_df["control_to_mast_ratio"].describe())

print("\nPOOL SUFFICIENCY")
print(
    "Strata with enough controls:",
    (
        matched_strata_df["control_pool_size"] >=
        matched_strata_df["mast_count"]
    ).sum(),
    "/",
    len(matched_strata_df)
)

print("\nUNMATCHED STRATA")
print(unmatched_strata[:15])

print("\nMATCHING OBJECTS READY")
```

```python
# PRECOMPUTE NEIGHBORHOOD COMPOSITION FOR MAST CELLS AND MATCHED CONTROLS

import numpy as np
import pandas as pd

# Nine non-Mast target populations
target_names = [
    name for name in population_names_10
    if name != "Mast"
]
target_codes = [
    population_names_10.index(name)
    for name in target_names
]

# Unique set of all cells needed for the analysis
analysis_indices = np.unique(
    np.concatenate([matched_mast_indices, control_indices])
).astype(np.int32)

# Map original cell index to compact analysis-row index
candidate_row_lookup = np.full(
    len(expanded_labels_10), -1, dtype=np.int32
)
candidate_row_lookup[analysis_indices] = np.arange(
    len(analysis_indices), dtype=np.int32
)

matched_mast_rows = candidate_row_lookup[matched_mast_indices]
control_rows = candidate_row_lookup[control_indices]

# Metadata for candidate cells
candidate_sample_codes = sample_codes_full[analysis_indices]
candidate_donor_codes = donor_codes_full[analysis_indices]
candidate_degrees = graph_degrees[analysis_indices]
candidate_labels = expanded_labels_10[analysis_indices]

# Calculate neighborhood fractions
candidate_neighbor_fractions = np.zeros(
    (len(analysis_indices), len(target_names)),
    dtype=np.float32
)

for j, (target_name, target_code) in enumerate(
    zip(target_names, target_codes)
):
    print(f"Calculating neighborhood fractions: {target_name}")

    target_indicator = (
        expanded_labels_10 == target_code
    ).astype(np.float32)

    neighbor_counts = A_symmetric.dot(target_indicator)

    candidate_neighbor_fractions[:, j] = (
        neighbor_counts[analysis_indices] /
        candidate_degrees
    )

    del target_indicator, neighbor_counts

# Validation
print("\nNEIGHBORHOOD MATRIX AUDIT")
print("Candidate cells:", len(analysis_indices))
print("Matched Mast cells:", len(matched_mast_rows))
print("Control cells:", len(control_rows))
print("Targets:", target_names)
print("Matrix shape:", candidate_neighbor_fractions.shape)
print("All finite:",
      np.isfinite(candidate_neighbor_fractions).all())
print("Minimum fraction:",
      candidate_neighbor_fractions.min())
print("Maximum fraction:",
      candidate_neighbor_fractions.max())

print("\nMEAN NEIGHBORHOOD FRACTIONS")
print(
    pd.DataFrame({
        "population": target_names,
        "mast_mean": candidate_neighbor_fractions[
            matched_mast_rows
        ].mean(axis=0),
        "control_pool_mean": candidate_neighbor_fractions[
            control_rows
        ].mean(axis=0)
    }).to_string(index=False)
)

print("\nPRECOMPUTATION COMPLETE")
```

```python
# REBUILD NEIGHBORHOOD FRACTIONS USING MATCHABLE CONTROLS ONLY

import numpy as np
import pandas as pd

# Collect controls only from strata containing matched Mast cells
matched_control_candidate_indices = np.unique(
    np.concatenate([
        matched_control_pools[(int(row.sample_code), int(row.degree))]
        for row in matched_strata_df.itertuples()
    ])
).astype(np.int32)

# Combine matched Mast cells and their eligible control pools
analysis_indices = np.unique(
    np.concatenate([
        matched_mast_indices,
        matched_control_candidate_indices
    ])
).astype(np.int32)

# Verify all candidate cells have positive graph degree
candidate_degrees = graph_degrees[analysis_indices]

print("CANDIDATE AUDIT")
print("Matched Mast cells:", len(matched_mast_indices))
print("Eligible matched control candidates:",
      len(matched_control_candidate_indices))
print("Combined candidate cells:", len(analysis_indices))
print("Zero-degree candidates:", np.sum(candidate_degrees == 0))
print("Minimum degree:", candidate_degrees.min())

assert np.all(candidate_degrees > 0), (
    "Unexpected zero-degree candidates. Stop and inspect."
)

# Build original-index to compact-row lookup
candidate_row_lookup = np.full(
    len(expanded_labels_10), -1, dtype=np.int32
)
candidate_row_lookup[analysis_indices] = np.arange(
    len(analysis_indices), dtype=np.int32
)

matched_mast_rows = candidate_row_lookup[matched_mast_indices]
control_candidate_rows = candidate_row_lookup[
    matched_control_candidate_indices
]

candidate_neighbor_fractions = np.zeros(
    (len(analysis_indices), len(target_names)),
    dtype=np.float32
)

# Calculate fractions for each target population
for j, (target_name, target_code) in enumerate(
    zip(target_names, target_codes)
):
    print("Calculating:", target_name)

    target_indicator = (
        expanded_labels_10 == target_code
    ).astype(np.float32)

    neighbor_counts = A_symmetric.dot(target_indicator)

    candidate_neighbor_fractions[:, j] = (
        neighbor_counts[analysis_indices] /
        candidate_degrees
    )

    del target_indicator, neighbor_counts

# Final validation
print("\nFINAL NEIGHBORHOOD MATRIX AUDIT")
print("Shape:", candidate_neighbor_fractions.shape)
print("All finite:",
      np.isfinite(candidate_neighbor_fractions).all())
print("Minimum:", candidate_neighbor_fractions.min())
print("Maximum:", candidate_neighbor_fractions.max())

assert np.isfinite(candidate_neighbor_fractions).all()

print("\nMEAN NEIGHBORHOOD FRACTIONS")
print(pd.DataFrame({
    "population": target_names,
    "mast_mean": candidate_neighbor_fractions[
        matched_mast_rows
    ].mean(axis=0),
    "control_pool_mean": candidate_neighbor_fractions[
        control_candidate_rows
    ].mean(axis=0)
}).to_string(index=False))

print("\nREADY FOR MATCHED-CONTROL PERMUTATIONS")
```

```python
# MATCHED-CONTROL ANALYSIS: DONOR-LEVEL COMPARISON
# 999 matched-control resampling iterations

import numpy as np
import pandas as pd
from scipy.stats import wilcoxon
from statsmodels.stats.multitest import multipletests

from collections import defaultdict

N_PERMUTATIONS = 999
SEED = 20261110

rng = np.random.default_rng(SEED)
rng_boot = np.random.default_rng(SEED + 1)

# --------------------------------------------------
# 1. Reconstruct the corrected candidate set
# --------------------------------------------------

candidate_control_indices = np.unique(
    np.concatenate(list(matched_control_pools.values()))
).astype(np.int32)

candidate_indices = np.unique(
    np.concatenate([
        matched_mast_indices,
        candidate_control_indices
    ])
).astype(np.int32)

row_lookup = np.full(
    len(expanded_labels_10), -1, dtype=np.int32
)
row_lookup[candidate_indices] = np.arange(
    len(candidate_indices), dtype=np.int32
)

candidate_degrees_check = graph_degrees[candidate_indices]

assert np.all(candidate_degrees_check > 0)

print("CANDIDATE AUDIT")
print("Matched Mast cells:", len(matched_mast_indices))
print("Matched control candidates:", len(candidate_control_indices))
print("Combined candidates:", len(candidate_indices))
print("Zero-degree candidates:", np.sum(candidate_degrees_check == 0))

# --------------------------------------------------
# 2. Calculate neighborhood fractions
# --------------------------------------------------

target_names = [
    name for name in population_names_10
    if name != "Mast"
]

target_codes = [
    population_names_10.index(name)
    for name in target_names
]

candidate_neighbor_fractions = np.zeros(
    (len(candidate_indices), len(target_names)),
    dtype=np.float32
)

for j, (name, code) in enumerate(zip(target_names, target_codes)):

    print("Calculating:", name)

    indicator = (
        expanded_labels_10 == code
    ).astype(np.float32)

    counts = np.asarray(
        A_symmetric.dot(indicator)
    ).ravel()

    candidate_neighbor_fractions[:, j] = (
        counts[candidate_indices] /
        candidate_degrees_check
    )

    del indicator, counts

assert np.isfinite(candidate_neighbor_fractions).all()

# --------------------------------------------------
# 3. Prepare matched strata
# --------------------------------------------------

strata = []

for row in matched_strata_df.itertuples(index=False):

    key = (int(row.sample_code), int(row.degree))

    mast_idx = np.asarray(
        mast_groups[key], dtype=np.int32
    )

    control_idx = matched_control_pools[key]

    mast_rows = row_lookup[mast_idx]
    control_rows = row_lookup[control_idx]

    assert np.all(mast_rows >= 0)
    assert np.all(control_rows >= 0)

    sample_id = sample_categories_full[key[0]]
    donor_id = sample_to_donor[sample_id]
    donor_idx = donor_categories.index(donor_id)

    strata.append({
        "mast_rows": mast_rows,
        "control_rows": control_rows,
        "mast_count": len(mast_rows),
        "donor_idx": donor_idx
    })

n_donors = len(donor_categories)
n_targets = len(target_names)

# --------------------------------------------------
# 4. Calculate observed donor-level means
# --------------------------------------------------

mast_sums = np.zeros((n_donors, n_targets))
control_sums = np.zeros((n_donors, n_targets))
donor_cell_counts = np.zeros(n_donors)

for s in strata:

    m = s["mast_count"]
    d = s["donor_idx"]

    mast_mean = candidate_neighbor_fractions[
        s["mast_rows"]
    ].mean(axis=0)

    control_mean = candidate_neighbor_fractions[
        s["control_rows"]
    ].mean(axis=0)

    mast_sums[d] += mast_mean * m
    control_sums[d] += control_mean * m
    donor_cell_counts[d] += m

active_donors = donor_cell_counts > 0

mast_donor_means = (
    mast_sums[active_donors] /
    donor_cell_counts[active_donors, None]
)

control_donor_means = (
    control_sums[active_donors] /
    donor_cell_counts[active_donors, None]
)

donor_differences = mast_donor_means - control_donor_means

# --------------------------------------------------
# 5. Paired donor-level inference and bootstrap CI
# --------------------------------------------------

results = []

for j, name in enumerate(target_names):

    differences = donor_differences[:, j]

    observed_median = np.median(differences)

    if np.allclose(differences, 0):
        p_value = 1.0
    else:
        p_value = wilcoxon(
            differences,
            alternative="two-sided",
            zero_method="wilcox"
        ).pvalue

    boot_indices = rng_boot.integers(
        0,
        len(differences),
        size=(10000, len(differences))
    )

    boot_medians = np.median(
        differences[boot_indices],
        axis=1
    )

    ci_low, ci_high = np.percentile(
        boot_medians, [2.5, 97.5]
    )

    results.append({
        "population": name,
        "donors": len(differences),
        "mast_median_fraction": np.median(
            mast_donor_means[:, j]
        ),
        "control_median_fraction": np.median(
            control_donor_means[:, j]
        ),
        "median_difference_percentage_points":
            observed_median * 100,
        "bootstrap_95CI_low_pp": ci_low * 100,
        "bootstrap_95CI_high_pp": ci_high * 100,
        "wilcoxon_p": p_value
    })

# --------------------------------------------------
# 6. Multiple-testing correction
# --------------------------------------------------

results_df = pd.DataFrame(results)

results_df["BH_adjusted_p"] = multipletests(
    results_df["wilcoxon_p"],
    method="fdr_bh"
)[1]

# --------------------------------------------------
# 7. Matched-control resampling stability
# --------------------------------------------------

resampling_medians = np.zeros(
    (N_PERMUTATIONS, n_targets),
    dtype=np.float32
)

for iteration in range(N_PERMUTATIONS):

    sampled_control_sums = np.zeros(
        (n_donors, n_targets)
    )

    for s in strata:

        m = s["mast_count"]
        d = s["donor_idx"]

        selected = rng.choice(
            s["control_rows"],
            size=m,
            replace=False
        )

        selected_mean = candidate_neighbor_fractions[
            selected
        ].mean(axis=0)

        sampled_control_sums[d] += selected_mean * m

    sampled_control_means = (
        sampled_control_sums[active_donors] /
        donor_cell_counts[active_donors, None]
    )

    sampled_differences = (
        mast_donor_means - sampled_control_means
    )

    resampling_medians[iteration] = np.median(
        sampled_differences,
        axis=0
    )

    if (iteration + 1) % 100 == 0:
        print(
            f"Completed {iteration + 1}/{N_PERMUTATIONS}"
        )

# --------------------------------------------------
# 8. Summarize resampling distributions
# --------------------------------------------------

resampling_summary = []

for j, name in enumerate(target_names):

    low, high = np.percentile(
        resampling_medians[:, j],
        [2.5, 97.5]
    )

    resampling_summary.append({
        "population": name,
        "resampling_95_low_pp": low * 100,
        "resampling_95_high_pp": high * 100,
        "resampling_median_difference_pp":
            np.median(resampling_medians[:, j]) * 100
    })

resampling_df = pd.DataFrame(resampling_summary)

results_df = results_df.merge(
    resampling_df,
    on="population"
)

# --------------------------------------------------
# 9. Save results
# --------------------------------------------------

folder = "/content/drive/MyDrive/Neurothesis Resoure Library"

results_path = (
    folder + "/Matched_Control_Donor_Comparison.csv"
)

resampling_path = (
    folder + "/Matched_Control_Resampling_Distribution.csv"
)

results_df.to_csv(results_path, index=False)

pd.DataFrame(
    resampling_medians,
    columns=target_names
).to_csv(resampling_path, index=False)

# --------------------------------------------------
# 10. Final report
# --------------------------------------------------

print("\nMATCHED-CONTROL DONOR-LEVEL RESULTS")
print(
    results_df.round(5).to_string(index=False)
)

print("\nDONORS:", active_donors.sum())
print("MATCHED STRATA:", len(strata))
print("MATCHED MAST CELLS:", len(matched_mast_indices))
print("RESAMPLING ITERATIONS:", N_PERMUTATIONS)

print("\nSAVED:")
print(results_path)
print(resampling_path)

print("\nANALYSIS COMPLETE")
```

```python

import numpy as np
import pandas as pd

# Rebuild Mast-cell groups and matched non-Mast control pools.
# Matching is exact within each sample and graph-degree stratum.

dermis_mask = np.asarray(compartment_codes) == 0
labels = np.asarray(expanded_labels_10)
samples = np.asarray(sample_codes_full)
degrees = np.asarray(graph_degrees)

# Dermal Mast cells with a valid graph degree
mast_mask = (
    dermis_mask
    & (labels == 0)
    & (degrees > 0)
)

# Dermal cells belonging to one of the nine annotated
# non-Mast populations used in the analysis
control_mask = (
    dermis_mask
    & (labels >= 1)
    & (labels <= 9)
    & (degrees > 0)
)

mast_indices_all = np.flatnonzero(mast_mask)
control_indices_all = np.flatnonzero(control_mask)

def group_by_sample_and_degree(indices):
    indices = np.asarray(indices, dtype=np.int64)

    s = samples[indices]
    d = degrees[indices]

    order = np.lexsort((d, s))
    sorted_indices = indices[order]
    sorted_s = s[order]
    sorted_d = d[order]

    if len(sorted_indices) == 0:
        return {}

    starts = np.r_[
        0,
        np.flatnonzero(
            (sorted_s[1:] != sorted_s[:-1])
            | (sorted_d[1:] != sorted_d[:-1])
        ) + 1
    ]
    ends = np.r_[starts[1:], len(sorted_indices)]

    groups = {}
    for start, end in zip(starts, ends):
        key = (int(sorted_s[start]), int(sorted_d[start]))
        groups[key] = sorted_indices[start:end]

    return groups

all_mast_groups = group_by_sample_and_degree(mast_indices_all)
all_control_groups = group_by_sample_and_degree(control_indices_all)

# Retain only strata containing both Mast cells and eligible controls
matched_keys = sorted(
    key for key in all_mast_groups
    if key in all_control_groups
    and len(all_control_groups[key]) > 0
)

mast_groups = {
    key: all_mast_groups[key]
    for key in matched_keys
}

matched_control_pools = {
    key: all_control_groups[key]
    for key in matched_keys
}

matched_mast_indices = np.concatenate(
    list(mast_groups.values())
).astype(np.int32)

matched_strata_df = pd.DataFrame([
    {
        "sample_code": key[0],
        "degree": key[1],
        "mast_count": len(mast_groups[key]),
        "control_pool_size": len(matched_control_pools[key]),
        "control_to_mast_ratio": (
            len(matched_control_pools[key]) / len(mast_groups[key])
        )
    }
    for key in matched_keys
])

print("REBUILT MATCHED-CONTROL OBJECTS")
print("=" * 45)
print("Dermal Mast cells:", len(mast_indices_all))
print("Dermal eligible non-Mast controls:", len(control_indices_all))
print("Matched Mast cells:", len(matched_mast_indices))
print("Matched strata:", len(matched_keys))
print("Matched samples:", matched_strata_df["sample_code"].nunique())
print("Matched control candidates:",
      sum(len(v) for v in matched_control_pools.values()))
print("Unmatched Mast cells:",
      len(mast_indices_all) - len(matched_mast_indices))

print("\nCONTROL POOL AUDIT")
print(matched_strata_df["control_to_mast_ratio"].describe())

assert len(matched_control_pools) == len(mast_groups)
assert len(matched_mast_indices) > 0
assert all(len(v) > 0 for v in matched_control_pools.values())

print("\nSUCCESS: matched-control objects rebuilt.")
```

```python

import h5py
import numpy as np
import pandas as pd
from scipy.sparse import csr_matrix

h5ad_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "merfish.integrated_annotated (2).h5ad"
)

def decode_categories(dataset):
    return [
        x.decode("utf-8") if isinstance(x, bytes) else str(x)
        for x in dataset[:]
    ]

print("RESTORING METADATA AND SPATIAL GRAPH")
print("=" * 55)

with h5py.File(h5ad_path, "r") as f:

    # --------------------------------------------------
    # 1. CATEGORICAL METADATA
    # --------------------------------------------------

    def read_categorical(field):
        group = f["obs"][field]
        categories = decode_categories(group["categories"])
        codes = group["codes"][:].astype(np.int32)
        return codes, categories

    expanded_labels_original, detailed_categories = (
        read_categorical("cell_type.detailed")
    )

    broad_labels_original, broad_categories = (
        read_categorical("cell_type.broad")
    )

    compartment_codes, compartment_categories = (
        read_categorical("tissue_compartment")
    )

    sample_codes_full, sample_categories_full = (
        read_categorical("sample_id")
    )

    donor_codes_full, donor_categories = (
        read_categorical("donor_id")
    )

    anatomic_site_codes_full, anatomic_site_categories = (
        read_categorical("anatomic_site")
    )

    # --------------------------------------------------
    # 2. RECREATE THE TEN-POPULATION LABEL ARRAY
    # --------------------------------------------------

    population_names_10 = [
        "Mast",
        "Schwann",
        "Peri",
        "Perivasc Fib I",
        "Perivasc Fib II",
        "Retic Fib III",
        "VEC",
        "HEC",
        "LEC",
        "SM"
    ]

    detailed_code_map = {
        34: 0,  # Mast
        25: 1,  # Schwann
        20: 2,  # Peri
        32: 3,  # Perivasc Fib I
        33: 4,  # Perivasc Fib II
        31: 5,  # Retic Fib III
        23: 6,  # VEC
        22: 7,  # HEC
        24: 8,  # LEC
        21: 9   # SM
    }

    expanded_labels_10 = np.full(
        len(expanded_labels_original),
        -1,
        dtype=np.int8
    )

    for original_code, new_code in detailed_code_map.items():
        expanded_labels_10[
            expanded_labels_original == original_code
        ] = new_code

    # --------------------------------------------------
    # 3. SPATIAL COORDINATES
    # --------------------------------------------------

    coordinates = f["obsm"]["spatial"][:].astype(np.float64)

    # --------------------------------------------------
    # 4. RECONSTRUCT SPATIAL CONNECTIVITY GRAPH
    # --------------------------------------------------

    graph = f["obsp"]["spatial_connectivities"]

    graph_data = graph["data"][:]
    graph_indices = graph["indices"][:]
    graph_indptr = graph["indptr"][:]

    n_cells = len(expanded_labels_10)

    A_original = csr_matrix(
        (graph_data, graph_indices, graph_indptr),
        shape=(n_cells, n_cells)
    )

# --------------------------------------------------
# 5. CREATE THE SYMMETRIC BINARY GRAPH
# --------------------------------------------------

A_symmetric = A_original.maximum(A_original.T).tocsr()
A_symmetric.setdiag(0)
A_symmetric.eliminate_zeros()
A_symmetric.data[:] = 1

graph_degrees = np.asarray(
    A_symmetric.sum(axis=1)
).ravel().astype(np.int32)

# --------------------------------------------------
# 6. SAMPLE-TO-DONOR MAPPING
# --------------------------------------------------

sample_to_donor = {}

for sample_code in np.unique(sample_codes_full):
    donor_values = donor_codes_full[
        sample_codes_full == sample_code
    ]

    donor_values = donor_values[donor_values >= 0]

    if len(donor_values) > 0:
        values, counts = np.unique(
            donor_values,
            return_counts=True
        )

        sample_to_donor[int(sample_code)] = int(
            values[np.argmax(counts)]
        )

# Preserve the prior mapping variable name if needed
mapping = sample_to_donor

# --------------------------------------------------
# 7. VALIDATION
# --------------------------------------------------

print("\nRESTORATION AUDIT")
print("=" * 55)

print("Total cells:", n_cells)
print("Coordinates:", coordinates.shape)
print("Graph shape:", A_symmetric.shape)
print("Graph edges:", A_symmetric.nnz)
print("Graph symmetric:",
      (A_symmetric != A_symmetric.T).nnz == 0)

print("\nCELL POPULATIONS")
for i, name in enumerate(population_names_10):
    print(
        f"{name}:",
        int(np.sum(expanded_labels_10 == i))
    )

print("\nTISSUE COMPARTMENTS")
for i, name in enumerate(compartment_categories):
    print(
        name,
        int(np.sum(compartment_codes == i))
    )

print("\nMETADATA")
print("Samples:", len(sample_categories_full))
print("Donors:", len(donor_categories))
print("Anatomical sites:", len(anatomic_site_categories))
print("Sample-to-donor mappings:", len(sample_to_donor))

print("\nDERMAL MAST AUDIT")
dermis_mask = compartment_codes == 0
mast_mask = (
    dermis_mask
    & (expanded_labels_10 == 0)
)

print("Dermal Mast cells:", int(mast_mask.sum()))
print(
    "Dermal Mast cells with graph degree > 0:",
    int(np.sum(mast_mask & (graph_degrees > 0)))
)

assert n_cells == 1201886
assert A_symmetric.shape == (1201886, 1201886)
assert coordinates.shape == (1201886, 2)
assert (A_symmetric != A_symmetric.T).nnz == 0

print("\nSUCCESS: RESEARCH STATE RESTORED")
```

```python

import numpy as np
import pandas as pd
from scipy.stats import wilcoxon

# ============================================================
# 1. SETTINGS AND INPUT VALIDATION
# ============================================================

RNG_SEED = 20261110
N_BOOTSTRAP = 10000
N_RESAMPLES = 999

rng_boot = np.random.default_rng(RNG_SEED)
rng_resample = np.random.default_rng(RNG_SEED + 1)

output_dir = (
    "/content/drive/MyDrive/Neurothesis Resoure Library"
)

target_names = population_names_10[1:]

n_targets = len(target_names)

assert len(matched_control_pools) == len(mast_groups)
assert len(matched_mast_indices) > 0
assert A_symmetric.shape[0] == len(expanded_labels_10)

print("MATCHED-CONTROL NEIGHBORHOOD ANALYSIS")
print("=" * 65)
print("Matched Mast cells:", len(matched_mast_indices))
print("Matched strata:", len(matched_control_pools))
print("Target populations:", target_names)

# ============================================================
# 2. BUILD A SINGLE INDEX OF ALL MATCHED CONTROL CANDIDATES
# ============================================================

candidate_control_indices = np.unique(
    np.concatenate(list(matched_control_pools.values()))
).astype(np.int32)

print("\nUnique matched control candidates:",
      len(candidate_control_indices))

# Verify no Mast cells are included in the control pool.
assert np.all(
    expanded_labels_10[candidate_control_indices] != 0
)

# ============================================================
# 3. CALCULATE NEIGHBORHOOD FRACTIONS
# ============================================================

# Each row is a cell. Each column is its fraction of neighbors
# belonging to one of the nine target populations.

def calculate_neighborhood_fractions(indices):

    result = np.zeros(
        (len(indices), n_targets),
        dtype=np.float32
    )

    degrees = graph_degrees[indices].astype(np.float32)

    for j, target_code in enumerate(range(1, 10)):

        target_indicator = (
            expanded_labels_10 == target_code
        ).astype(np.float32)

        neighbor_counts = np.asarray(
            A_symmetric.dot(target_indicator)
        ).ravel()

        result[:, j] = (
            neighbor_counts[indices] / degrees
        )

        print(
            f"Calculated neighborhood fractions: "
            f"{target_names[j]}"
        )

    return result


# Calculate for Mast cells.
mast_neighbor_fractions = calculate_neighborhood_fractions(
    matched_mast_indices
)

# Calculate for all unique matched control candidates.
control_neighbor_fractions = calculate_neighborhood_fractions(
    candidate_control_indices
)

assert np.isfinite(mast_neighbor_fractions).all()
assert np.isfinite(control_neighbor_fractions).all()

# Map each global cell index to its row in the control matrix.
control_position = {
    int(cell_index): position
    for position, cell_index in enumerate(candidate_control_indices)
}

# ============================================================
# 4. STRATUM-LEVEL AND DONOR-LEVEL COMPARISONS
# ============================================================

stratum_records = []
donor_records = []

for key in sorted(mast_groups):

    sample_code, degree = key

    mast_indices = mast_groups[key]
    control_indices = matched_control_pools[key]

    mast_rows = np.searchsorted(
        matched_mast_indices,
        mast_indices
    )

    # matched_mast_indices is not necessarily sorted by key order.
    # Use an explicit lookup to ensure correct row alignment.
    mast_position = {
        int(cell_index): position
        for position, cell_index in enumerate(matched_mast_indices)
    }

    mast_rows = np.array(
        [mast_position[int(i)] for i in mast_indices],
        dtype=np.int64
    )

    control_rows = np.array(
        [control_position[int(i)] for i in control_indices],
        dtype=np.int64
    )

    mast_mean = mast_neighbor_fractions[mast_rows].mean(axis=0)

    control_mean = control_neighbor_fractions[
        control_rows
    ].mean(axis=0)

    n_mast = len(mast_indices)

    record = {
        "sample_code": sample_code,
        "degree": degree,
        "mast_count": n_mast
    }

    for j, name in enumerate(target_names):
        record[f"mast_{name}"] = mast_mean[j]
        record[f"control_{name}"] = control_mean[j]
        record[f"difference_{name}"] = (
            mast_mean[j] - control_mean[j]
        )

    stratum_records.append(record)


stratum_df = pd.DataFrame(stratum_records)

# Aggregate strata to samples, weighted by Mast-cell count.
sample_records = []

for sample_code, group in stratum_df.groupby("sample_code"):

    weights = group["mast_count"].to_numpy(dtype=float)
    weights = weights / weights.sum()

    record = {"sample_code": int(sample_code)}

    for name in target_names:
        record[f"mast_{name}"] = np.average(
            group[f"mast_{name}"],
            weights=weights
        )

        record[f"control_{name}"] = np.average(
            group[f"control_{name}"],
            weights=weights
        )

    sample_records.append(record)

sample_df = pd.DataFrame(sample_records)

# Aggregate samples to donors, weighting by the number of
# matched Mast cells represented by each sample.
sample_counts = (
    stratum_df.groupby("sample_code")["mast_count"]
    .sum()
    .to_dict()
)

sample_df["donor_code"] = (
    sample_df["sample_code"]
    .map(sample_to_donor)
)

donor_records = []

for donor_code, group in sample_df.groupby("donor_code"):

    weights = group["sample_code"].map(
        sample_counts
    ).to_numpy(dtype=float)

    weights = weights / weights.sum()

    record = {
        "donor_code": int(donor_code),
        "donor": donor_categories[int(donor_code)],
        "sample_count": len(group),
        "matched_mast_count": int(
            group["sample_code"].map(sample_counts).sum()
        )
    }

    for name in target_names:

        record[f"mast_{name}"] = np.average(
            group[f"mast_{name}"],
            weights=weights
        )

        record[f"control_{name}"] = np.average(
            group[f"control_{name}"],
            weights=weights
        )

        record[f"difference_{name}"] = (
            record[f"mast_{name}"]
            - record[f"control_{name}"]
        )

    donor_records.append(record)

donor_df = pd.DataFrame(donor_records)

print("\nDONOR-LEVEL COMPARISON")
print("=" * 65)
print("Donors:", len(donor_df))
print("Samples:", len(sample_df))

# ============================================================
# 5. PAIRED DONOR-LEVEL STATISTICAL TESTS
# ============================================================

results = []

for name in target_names:

    mast_values = donor_df[f"mast_{name}"].to_numpy()
    control_values = donor_df[f"control_{name}"].to_numpy()

    differences = mast_values - control_values

    nonzero = differences != 0

    if nonzero.sum() >= 2:
        test = wilcoxon(
            mast_values,
            control_values,
            alternative="two-sided",
            zero_method="wilcox",
            method="auto"
        )
        p_value = float(test.pvalue)
    else:
        p_value = np.nan

    # Bootstrap donors, preserving paired Mast/control values.
    boot_indices = rng_boot.integers(
        0,
        len(differences),
        size=(N_BOOTSTRAP, len(differences))
    )

    boot_differences = differences[boot_indices].mean(axis=1)

    ci_low, ci_high = np.percentile(
        boot_differences,
        [2.5, 97.5]
    )

    results.append({
        "target_population": name,
        "donor_count": len(differences),
        "mast_mean": mast_values.mean(),
        "control_mean": control_values.mean(),
        "mean_donor_difference": differences.mean(),
        "median_donor_difference": np.median(differences),
        "donors_with_positive_difference": int(
            np.sum(differences > 0)
        ),
        "donors_with_negative_difference": int(
            np.sum(differences < 0)
        ),
        "wilcoxon_p": p_value,
        "bootstrap_95ci_low": ci_low,
        "bootstrap_95ci_high": ci_high
    })

results_df = pd.DataFrame(results)

# ============================================================
# 6. BENJAMINI-HOCHBERG CORRECTION
# ============================================================

def benjamini_hochberg(p_values):

    p_values = np.asarray(p_values, dtype=float)

    adjusted = np.full(len(p_values), np.nan)

    valid = np.isfinite(p_values)

    p = p_values[valid]
    m = len(p)

    if m == 0:
        return adjusted

    order = np.argsort(p)
    sorted_p = p[order]

    sorted_adjusted = (
        sorted_p * m / np.arange(1, m + 1)
    )

    sorted_adjusted = np.minimum.accumulate(
        sorted_adjusted[::-1]
    )[::-1]

    sorted_adjusted = np.clip(
        sorted_adjusted,
        0,
        1
    )

    valid_indices = np.flatnonzero(valid)

    adjusted[valid_indices[order]] = sorted_adjusted

    return adjusted


results_df["BH_adjusted_p"] = benjamini_hochberg(
    results_df["wilcoxon_p"].to_numpy()
)

# ============================================================
# 7. MATCHED-CONTROL RESAMPLING
# ============================================================

# For every iteration, sample exactly the same number of
# controls as Mast cells within each sample-degree stratum.
#
# These are stability distributions, not biological null tests.

print("\nSTARTING MATCHED-CONTROL RESAMPLING")
print("=" * 65)

stratum_keys = sorted(mast_groups)

# Precompute candidate row indices for each stratum.
stratum_control_rows = {}
stratum_mast_counts = {}
stratum_sample_codes = {}

for key in stratum_keys:

    stratum_control_rows[key] = np.array(
        [
            control_position[int(i)]
            for i in matched_control_pools[key]
        ],
        dtype=np.int64
    )

    stratum_mast_counts[key] = len(mast_groups[key])
    stratum_sample_codes[key] = key[0]

# Store donor-level differences for each iteration.
resampling_differences = np.zeros(
    (N_RESAMPLES, len(donor_df), n_targets),
    dtype=np.float32
)

donor_code_to_position = {
    int(code): i
    for i, code in enumerate(donor_df["donor_code"])
}

# Precompute Mast means by stratum.
stratum_mast_means = {}

for key in stratum_keys:

    mast_position = {
        int(cell_index): position
        for position, cell_index in enumerate(matched_mast_indices)
    }

    rows = np.array([
        mast_position[int(i)]
        for i in mast_groups[key]
    ], dtype=np.int64)

    stratum_mast_means[key] = (
        mast_neighbor_fractions[rows].mean(axis=0)
    )

for iteration in range(N_RESAMPLES):

    # Accumulate sample-level values weighted by Mast counts.
    donor_mast_sum = np.zeros(
        (len(donor_df), n_targets),
        dtype=np.float64
    )

    donor_control_sum = np.zeros_like(donor_mast_sum)

    donor_weight_sum = np.zeros(len(donor_df), dtype=np.float64)

    for key in stratum_keys:

        sample_code = stratum_sample_codes[key]
        donor_code = sample_to_donor[sample_code]
        donor_position = donor_code_to_position[donor_code]

        n_mast = stratum_mast_counts[key]
        pool_rows = stratum_control_rows[key]

        sampled_rows = rng_resample.choice(
            pool_rows,
            size=n_mast,
            replace=False
        )

        sampled_control_mean = (
            control_neighbor_fractions[sampled_rows].mean(axis=0)
        )

        weight = n_mast

        donor_mast_sum[donor_position] += (
            stratum_mast_means[key] * weight
        )

        donor_control_sum[donor_position] += (
            sampled_control_mean * weight
        )

        donor_weight_sum[donor_position] += weight

    valid_donors = donor_weight_sum > 0

    iteration_differences = (
        donor_mast_sum[valid_donors]
        / donor_weight_sum[valid_donors, None]
        -
        donor_control_sum[valid_donors]
        / donor_weight_sum[valid_donors, None]
    )

    resampling_differences[
        iteration,
        valid_donors,
        :
    ] = iteration_differences

    if (iteration + 1) % 100 == 0:
        print(
            f"Completed {iteration + 1} / {N_RESAMPLES}"
        )

# ============================================================
# 8. RESAMPLING SUMMARY
# ============================================================

resampling_records = []

for j, name in enumerate(target_names):

    iteration_medians = np.median(
        resampling_differences[:, :, j],
        axis=1
    )

    resampling_records.append({
        "target_population": name,
        "resampling_iterations": N_RESAMPLES,
        "median_difference_across_iterations": np.median(
            iteration_medians
        ),
        "resampling_2.5_percentile": np.percentile(
            iteration_medians, 2.5
        ),
        "resampling_97.5_percentile": np.percentile(
            iteration_medians, 97.5
        )
    })

resampling_df = pd.DataFrame(resampling_records)

# ============================================================
# 9. SAVE RESULTS
# ============================================================

matched_output = os.path.join(
    output_dir,
    "Matched_Control_Donor_Comparison.csv"
)

resampling_output = os.path.join(
    output_dir,
    "Matched_Control_Resampling_Distribution.csv"
)

results_df.to_csv(matched_output, index=False)
resampling_df.to_csv(resampling_output, index=False)

donor_df.to_csv(
    os.path.join(
        output_dir,
        "Matched_Control_Donor_Level_Values.csv"
    ),
    index=False
)

# ============================================================
# 10. FINAL OUTPUT
# ============================================================

print("\n" + "=" * 75)
print("MATCHED-CONTROL ANALYSIS RESULTS")
print("=" * 75)

display(
    results_df.round(6)
)

print("\nMATCHED-CONTROL RESAMPLING SUMMARY")
print("=" * 75)

display(
    resampling_df.round(6)
)

print("\nSAVED FILES")
print(matched_output)
print(resampling_output)
print(
    os.path.join(
        output_dir,
        "Matched_Control_Donor_Level_Values.csv"
    )
)

print("\nANALYSIS COMPLETE")
```

```python
import h5py
import numpy as np
import pandas as pd
from collections import defaultdict

h5ad_path = (
    "/content/drive/MyDrive/"
    "Neurothesis Resoure Library/"
    "merfish.integrated_annotated (2).h5ad"
)

# Population codes from the restored annotation variables
population_codes = {
    "Peri": 2,
    "Perivasc Fib I": 3,
    "Perivasc Fib II": 4,
    "SM": 9
}

marker_groups = {
    "Pericyte-associated": [
        "PDGFRB", "RGS5", "MCAM", "NOTCH3"
    ],
    "Fibroblast-associated": [
        "COL1A1", "COL3A1", "PDGFRA"
    ],
    "Vascular-associated": [
        "VWF", "KDR"
    ],
    "Smooth-muscle-associated": [
        "ACTA2", "MYH11", "DES"
    ]
}

all_markers = list(dict.fromkeys(
    gene
    for genes in marker_groups.values()
    for gene in genes
))

results = defaultdict(list)
coexpression_results = []

with h5py.File(h5ad_path, "r") as f:

    gene_names = [
        x.decode("utf-8") if isinstance(x, bytes) else str(x)
        for x in f["var"]["_index"][:]
    ]

    gene_lookup = {
        gene.upper(): i
        for i, gene in enumerate(gene_names)
    }

    available_markers = [
        gene for gene in all_markers
        if gene.upper() in gene_lookup
    ]

    gene_columns = np.array([
        gene_lookup[g.upper()]
        for g in available_markers
    ])

    print("MARKERS USED:", available_markers)

    # Use normalized log1p expression
    matrix = f["layers"]["log1p"]

    indptr = matrix["indptr"]
    indices = matrix["indices"]
    data = matrix["data"]

    print("Expression matrix shape:",
          matrix.attrs["shape"])

    for population, code in population_codes.items():

        selected_rows = np.flatnonzero(
            (expanded_labels_10 == code) &
            (compartment_codes == 0)
        )

        n_cells = len(selected_rows)

        expression = np.zeros(
            (n_cells, len(available_markers)),
            dtype=np.float32
        )

        for i, row_index in enumerate(selected_rows):

            start = int(indptr[row_index])
            end = int(indptr[row_index + 1])

            row_indices = indices[start:end]
            row_data = data[start:end]

            if len(row_indices) == 0:
                continue

            positions = np.searchsorted(
                row_indices,
                gene_columns
            )

            valid = positions < len(row_indices)

            matched = np.zeros(
                len(gene_columns),
                dtype=bool
            )

            matched[valid] = (
                row_indices[positions[valid]] ==
                gene_columns[valid]
            )

            expression[i, matched] = (
                row_data[positions[matched]]
            )

        detected = expression > 0

        print(
            f"\n{population}: {n_cells:,} dermal cells"
        )

        # Individual marker summaries
        for j, gene in enumerate(available_markers):

            results["population"].append(population)
            results["marker"].append(gene)
            results["cell_count"].append(n_cells)
            results["percent_detected"].append(
                detected[:, j].mean() * 100
                if n_cells else np.nan
            )
            results["mean_expression_all_cells"].append(
                expression[:, j].mean()
                if n_cells else np.nan
            )
            results["mean_expression_positive_cells"].append(
                expression[detected[:, j], j].mean()
                if detected[:, j].any()
                else 0
            )

        # Marker-group coexpression summaries
        for group_name, group_genes in marker_groups.items():

            group_indices = [
                available_markers.index(g)
                for g in group_genes
                if g in available_markers
            ]

            if not group_indices:
                continue

            group_detected = detected[:, group_indices]
            positive_counts = group_detected.sum(axis=1)

            required = min(2, len(group_indices))

            coexpression_results.append({
                "population": population,
                "marker_group": group_name,
                "markers_available": len(group_indices),
                "cells": n_cells,
                "cells_expressing_2plus": int(
                    (positive_counts >= required).sum()
                ),
                "percent_expressing_2plus": (
                    (positive_counts >= required).mean() * 100
                    if n_cells else np.nan
                ),
                "mean_markers_detected_per_cell": (
                    positive_counts.mean()
                    if n_cells else np.nan
                )
            })

marker_df = pd.DataFrame(results)
coexpression_df = pd.DataFrame(coexpression_results)

print("\n" + "=" * 75)
print("INDIVIDUAL MARKER EXPRESSION")
print("=" * 75)

display(
    marker_df.pivot(
        index="marker",
        columns="population",
        values="percent_detected"
    ).round(2)
)

print("\n" + "=" * 75)
print("MARKER GROUP COEXPRESSION")
print("=" * 75)

display(
    coexpression_df[
        [
            "population",
            "marker_group",
            "markers_available",
            "cells",
            "percent_expressing_2plus",
            "mean_markers_detected_per_cell"
        ]
    ].round(2)
)

# Save both result tables
output_dir = (
    "/content/drive/MyDrive/"
    "Neurothesis Resoure Library"
)

marker_df.to_csv(
    f"{output_dir}/Perivascular_Population_Marker_Expression.csv",
    index=False
)

coexpression_df.to_csv(
    f"{output_dir}/Perivascular_Population_Coexpression.csv",
    index=False
)

print("\nSaved marker-validation tables.")
```

```python

import h5py
import numpy as np
import pandas as pd
from scipy.stats import wilcoxon
from statsmodels.stats.multitest import multipletests

# ============================================================
# DONOR-LEVEL PERIVASCULAR POPULATION IDENTITY VALIDATION
# ============================================================

h5ad_path = (
    "/content/drive/MyDrive/"
    "Neurothesis Resoure Library/"
    "merfish.integrated_annotated (2).h5ad"
)

output_dir = (
    "/content/drive/MyDrive/"
    "Neurothesis Resoure Library"
)

populations = {
    "Peri": 2,
    "Perivasc Fib I": 3,
    "Perivasc Fib II": 4,
    "SM": 9
}

marker_groups = {
    "Fibroblast": [
        "PDGFRA", "COL1A1", "COL3A1"
    ],
    "Pericyte": [
        "PDGFRB", "RGS5", "MCAM", "NOTCH3"
    ]
}

markers = [
    "PDGFRA", "COL1A1", "COL3A1",
    "PDGFRB", "RGS5", "MCAM", "NOTCH3"
]

# ============================================================
# 1. IDENTIFY DONOR AND GENE INDICES
# ============================================================

donor_codes = np.asarray(donor_codes_full)
donor_names = [
    x.decode("utf-8") if isinstance(x, bytes) else str(x)
    for x in donor_categories
]

gene_records = []
coexpression_records = []

with h5py.File(h5ad_path, "r") as f:

    gene_names = [
        x.decode("utf-8") if isinstance(x, bytes) else str(x)
        for x in f["var"]["_index"][:]
    ]

    gene_lookup = {
        gene.upper(): i
        for i, gene in enumerate(gene_names)
    }

    missing = [
        gene for gene in markers
        if gene.upper() not in gene_lookup
    ]

    if missing:
        raise ValueError(f"Missing required genes: {missing}")

    gene_columns = np.array([
        gene_lookup[g.upper()]
        for g in markers
    ])

    matrix = f["layers"]["log1p"]

    indptr = matrix["indptr"]
    indices = matrix["indices"]
    data = matrix["data"]

    # ========================================================
    # 2. PROCESS EACH POPULATION
    # ========================================================

    for population, population_code in populations.items():

        selected_rows = np.flatnonzero(
            (expanded_labels_10 == population_code) &
            (compartment_codes == 0) &
            (donor_codes >= 0)
        )

        print(
            f"\nProcessing {population}: "
            f"{len(selected_rows):,} dermal cells"
        )

        # Donor-level accumulators
        donor_cell_counts = {}
        donor_marker_positive = {}
        donor_marker_expression = {}
        donor_fibroblast_score = {}
        donor_pericyte_score = {}

        for row_index in selected_rows:

            donor = int(donor_codes[row_index])

            if donor not in donor_cell_counts:
                donor_cell_counts[donor] = 0
                donor_marker_positive[donor] = np.zeros(
                    len(markers), dtype=np.int64
                )
                donor_marker_expression[donor] = np.zeros(
                    len(markers), dtype=np.float64
                )
                donor_fibroblast_score[donor] = 0
                donor_pericyte_score[donor] = 0

            start = int(indptr[row_index])
            end = int(indptr[row_index + 1])

            row_indices = indices[start:end]
            row_data = data[start:end]

            expression = np.zeros(
                len(markers), dtype=np.float32
            )

            if len(row_indices):
                positions = np.searchsorted(
                    row_indices, gene_columns
                )

                valid = positions < len(row_indices)
                matched = np.zeros(
                    len(markers), dtype=bool
                )

                matched[valid] = (
                    row_indices[positions[valid]]
                    == gene_columns[valid]
                )

                expression[matched] = (
                    row_data[positions[matched]]
                )

            detected = expression > 0

            donor_cell_counts[donor] += 1
            donor_marker_positive[donor] += detected
            donor_marker_expression[donor] += expression

            # Fibroblast-associated score:
            # number of the 3 markers detected per cell
            donor_fibroblast_score[donor] += int(
                detected[[0, 1, 2]].sum()
            )

            # Pericyte-associated score:
            # number of the 4 markers detected per cell
            donor_pericyte_score[donor] += int(
                detected[[3, 4, 5, 6]].sum()
            )

        # ====================================================
        # 3. CREATE DONOR-LEVEL SUMMARIES
        # ====================================================

        for donor, n_cells in donor_cell_counts.items():

            donor_name = donor_names[donor]

            for j, marker in enumerate(markers):

                gene_records.append({
                    "donor": donor_name,
                    "population": population,
                    "cell_count": n_cells,
                    "marker": marker,
                    "positive_cells": int(
                        donor_marker_positive[donor][j]
                    ),
                    "percent_detected": (
                        donor_marker_positive[donor][j]
                        / n_cells * 100
                    ),
                    "mean_expression": (
                        donor_marker_expression[donor][j]
                        / n_cells
                    )
                })

            fibroblast_markers_detected = (
                donor_fibroblast_score[donor]
            )

            pericyte_markers_detected = (
                donor_pericyte_score[donor]
            )

            coexpression_records.append({
                "donor": donor_name,
                "population": population,
                "cell_count": n_cells,
                "fibroblast_mean_markers_per_cell": (
                    fibroblast_markers_detected
                    / n_cells
                ),
                "pericyte_mean_markers_per_cell": (
                    pericyte_markers_detected
                    / n_cells
                )
            })

# ============================================================
# 4. BUILD TABLES
# ============================================================

donor_marker_df = pd.DataFrame(gene_records)
donor_score_df = pd.DataFrame(coexpression_records)

print("\nDONOR-LEVEL MARKER DATA")
print("=" * 70)

display(
    donor_marker_df.head(15).round(3)
)

print("\nDONOR-LEVEL POPULATION COUNTS")
print("=" * 70)

display(
    donor_score_df.pivot(
        index="donor",
        columns="population",
        values="cell_count"
    ).fillna(0).astype(int)
)

# ============================================================
# 5. PAIRED DONOR COMPARISONS
# Perivasc Fib II versus Perivasc Fib I
# ============================================================

comparison_records = []

def paired_comparison(data, value_column, feature,
                      minimum_cells):

    wide = data.pivot(
        index="donor",
        columns="population",
        values=value_column
    )

    counts = donor_score_df.pivot(
        index="donor",
        columns="population",
        values="cell_count"
    )

    required = [
        "Perivasc Fib I",
        "Perivasc Fib II"
    ]

    if not all(x in wide.columns for x in required):
        return None

    valid = (
        wide[required].notna().all(axis=1)
        & (counts[required] >= minimum_cells).all(axis=1)
    )

    paired = wide.loc[valid, required]

    if len(paired) < 5:
        return None

    difference = (
        paired["Perivasc Fib II"]
        - paired["Perivasc Fib I"]
    )

    if np.allclose(difference, 0):
        p_value = 1.0
    else:
        p_value = wilcoxon(
            paired["Perivasc Fib II"],
            paired["Perivasc Fib I"],
            alternative="two-sided",
            method="auto"
        ).pvalue

    return {
        "feature": feature,
        "minimum_cells_per_population": minimum_cells,
        "donors": len(paired),
        "Fib_I_median": paired["Perivasc Fib I"].median(),
        "Fib_II_median": paired["Perivasc Fib II"].median(),
        "median_paired_difference": difference.median(),
        "mean_paired_difference": difference.mean(),
        "donors_Fib_II_higher": int((difference > 0).sum()),
        "donors_Fib_I_higher": int((difference < 0).sum()),
        "wilcoxon_p": p_value
    }

# Individual marker detection and expression
for marker in markers:

    subset = donor_marker_df[
        donor_marker_df["marker"] == marker
    ]

    for minimum in [20, 50]:

        result = paired_comparison(
            subset,
            "percent_detected",
            f"{marker}_percent_detected",
            minimum
        )

        if result:
            comparison_records.append(result)

# Coexpression scores
for score in [
    "fibroblast_mean_markers_per_cell",
    "pericyte_mean_markers_per_cell"
]:

    for minimum in [20, 50]:

        result = paired_comparison(
            donor_score_df,
            score,
            score,
            minimum
        )

        if result:
            comparison_records.append(result)

comparison_df = pd.DataFrame(comparison_records)

# BH correction separately within each cell-count threshold
comparison_df["BH_adjusted_p"] = np.nan

for threshold in comparison_df[
    "minimum_cells_per_population"
].unique():

    mask = (
        comparison_df["minimum_cells_per_population"]
        == threshold
    )

    comparison_df.loc[mask, "BH_adjusted_p"] = (
        multipletests(
            comparison_df.loc[mask, "wilcoxon_p"],
            method="fdr_bh"
        )[1]
    )

# ============================================================
# 6. RESULTS
# ============================================================

print("\n" + "=" * 75)
print("PAIRED DONOR-LEVEL IDENTITY COMPARISON")
print("Perivasc Fib II versus Perivasc Fib I")
print("=" * 75)

display(
    comparison_df.round(4)
)

print("\nDONOR-LEVEL COEXPRESSION SUMMARY")
print("=" * 75)

display(
    donor_score_df.groupby("population")[
        [
            "cell_count",
            "fibroblast_mean_markers_per_cell",
            "pericyte_mean_markers_per_cell"
        ]
    ].median().round(3)
)

# ============================================================
# 7. SAVE RESULTS
# ============================================================

donor_marker_df.to_csv(
    f"{output_dir}/Donor_Level_Perivascular_Marker_Expression.csv",
    index=False
)

donor_score_df.to_csv(
    f"{output_dir}/Donor_Level_Perivascular_Coexpression.csv",
    index=False
)

comparison_df.to_csv(
    f"{output_dir}/Donor_Level_Perivascular_Identity_Comparisons.csv",
    index=False
)

print("\nSAVED:")
print("Donor_Level_Perivascular_Marker_Expression.csv")
print("Donor_Level_Perivascular_Coexpression.csv")
print("Donor_Level_Perivascular_Identity_Comparisons.csv")
print("\nDONOR-LEVEL VALIDATION COMPLETE")
```

```python
# ============================================================
# ABUNDANCE-ADJUSTED MAST-CELL NEIGHBORHOOD ENRICHMENT
# Perivasc Fib I vs Perivasc Fib II
# ============================================================

import numpy as np
import pandas as pd
from scipy.stats import wilcoxon, spearmanr
from statsmodels.stats.multitest import multipletests

# Population codes in expanded_labels_10:
# 0 = Mast
# 3 = Perivasc Fib I
# 4 = Perivasc Fib II

MAST = 0
FIB_I = 3
FIB_II = 4
DERMIS = 0

results = []

sample_ids = np.unique(sample_codes_full[sample_codes_full >= 0])

for sample_code in sample_ids:

    # All dermal cells belonging to this sample
    sample_dermis = np.where(
        (sample_codes_full == sample_code) &
        (compartment_codes == DERMIS)
    )[0]

    if len(sample_dermis) == 0:
        continue

    labels = expanded_labels_10[sample_dermis]

    mast_local = np.where(labels == MAST)[0]
    fib_i_local = np.where(labels == FIB_I)[0]
    fib_ii_local = np.where(labels == FIB_II)[0]

    if len(mast_local) < 5:
        continue

    # Convert local indices to full-dataset indices
    mast_global = sample_dermis[mast_local]

    # Restrict graph to this sample's dermal cells
    subgraph = A_symmetric[mast_global, :][:, sample_dermis]

    # Number of graph neighbors belonging to each target population
    degree = np.asarray(subgraph.sum(axis=1)).ravel()

    fib_i_neighbors = np.asarray(
        subgraph[:, fib_i_local].sum(axis=1)
    ).ravel()

    fib_ii_neighbors = np.asarray(
        subgraph[:, fib_ii_local].sum(axis=1)
    ).ravel()

    valid = degree > 0

    if valid.sum() < 5:
        continue

    # Mean fraction of each Mast cell's dermal neighbors
    # belonging to the target population
    observed_fib_i = np.mean(
        fib_i_neighbors[valid] / degree[valid]
    )

    observed_fib_ii = np.mean(
        fib_ii_neighbors[valid] / degree[valid]
    )

    # Target abundance among all dermal cells in the sample
    abundance_fib_i = len(fib_i_local) / len(sample_dermis)
    abundance_fib_ii = len(fib_ii_local) / len(sample_dermis)

    if abundance_fib_i <= 0 or abundance_fib_ii <= 0:
        continue

    # Abundance-adjusted enrichment ratios
    enrichment_fib_i = observed_fib_i / abundance_fib_i
    enrichment_fib_ii = observed_fib_ii / abundance_fib_ii

    # Donor identity
    donor_values = np.unique(donor_codes_full[sample_dermis])
    donor_values = donor_values[donor_values >= 0]

    if len(donor_values) != 1:
        continue

    donor_code = int(donor_values[0])
    donor_name = donor_categories[donor_code]

    results.append({
        "sample_code": int(sample_code),
        "donor": donor_name,
        "mast_count": len(mast_local),
        "dermal_cell_count": len(sample_dermis),

        "fib_i_count": len(fib_i_local),
        "fib_ii_count": len(fib_ii_local),

        "fib_i_abundance": abundance_fib_i,
        "fib_ii_abundance": abundance_fib_ii,

        "fib_i_observed_fraction": observed_fib_i,
        "fib_ii_observed_fraction": observed_fib_ii,

        "fib_i_enrichment": enrichment_fib_i,
        "fib_ii_enrichment": enrichment_fib_ii,

        "log2_fib_i_enrichment": np.log2(enrichment_fib_i),
        "log2_fib_ii_enrichment": np.log2(enrichment_fib_ii)
    })

abundance_df = pd.DataFrame(results)

print("SAMPLE-LEVEL RESULTS")
print("=" * 65)
print("Samples analyzed:", len(abundance_df))
print("Unique donors:", abundance_df["donor"].nunique())

print("\nPOPULATION SUMMARY")
summary = abundance_df[
    [
        "fib_i_abundance",
        "fib_ii_abundance",
        "fib_i_observed_fraction",
        "fib_ii_observed_fraction",
        "fib_i_enrichment",
        "fib_ii_enrichment"
    ]
].agg(["count", "median", "mean", "std"])

print(summary.round(5))

# ------------------------------------------------------------
# DONOR-LEVEL AGGREGATION
# Each donor contributes one median per population.
# ------------------------------------------------------------

donor_df = (
    abundance_df
    .groupby("donor")
    .agg(
        fib_i_enrichment=("fib_i_enrichment", "median"),
        fib_ii_enrichment=("fib_ii_enrichment", "median"),
        fib_i_abundance=("fib_i_abundance", "median"),
        fib_ii_abundance=("fib_ii_abundance", "median"),
        sample_count=("sample_code", "nunique")
    )
    .reset_index()
)

donor_df["enrichment_difference"] = (
    donor_df["fib_ii_enrichment"] -
    donor_df["fib_i_enrichment"]
)

donor_df["log2_enrichment_difference"] = (
    donor_df["log2_fib_ii_enrichment"] -
    donor_df["log2_fib_i_enrichment"]
)

print("\nDONOR-LEVEL PAIRED COMPARISON")
print("=" * 65)
print("Donors:", len(donor_df))
print("Fib I median enrichment:",
      round(donor_df["fib_i_enrichment"].median(), 4))
print("Fib II median enrichment:",
      round(donor_df["fib_ii_enrichment"].median(), 4))
print("Median paired difference:",
      round(donor_df["enrichment_difference"].median(), 4))
print("Donors with Fib II higher:",
      int((donor_df["enrichment_difference"] > 0).sum()))
print("Donors with Fib I higher:",
      int((donor_df["enrichment_difference"] < 0).sum()))

if len(donor_df) >= 5:
    test = wilcoxon(
        donor_df["log2_enrichment_difference"],
        alternative="two-sided"
    )
    print("Paired Wilcoxon p-value:", test.pvalue)

# ------------------------------------------------------------
# ABUNDANCE RELATIONSHIPS
# Does target abundance track observed neighborhood fraction?
# ------------------------------------------------------------

print("\nABUNDANCE CORRELATIONS")
print("=" * 65)

correlation_results = []

for population, abundance_col, observed_col in [
    ("Perivasc Fib I", "fib_i_abundance", "fib_i_observed_fraction"),
    ("Perivasc Fib II", "fib_ii_abundance", "fib_ii_observed_fraction")
]:

    rho, p = spearmanr(
        abundance_df[abundance_col],
        abundance_df[observed_col]
    )

    correlation_results.append({
        "population": population,
        "spearman_rho": rho,
        "p_value": p
    })

    print(
        population,
        "| Spearman rho:", round(rho, 4),
        "| p:", f"{p:.6g}"
    )

# ------------------------------------------------------------
# SAVE OUTPUTS
# ------------------------------------------------------------

output_dir = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
)

abundance_df.to_csv(
    output_dir + "Mast_Abundance_Adjusted_Sample_Results.csv",
    index=False
)

donor_df.to_csv(
    output_dir + "Mast_Abundance_Adjusted_Donor_Results.csv",
    index=False
)

pd.DataFrame(correlation_results).to_csv(
    output_dir + "Mast_Abundance_Correlations.csv",
    index=False
)

print("\nSaved all three CSV files.")
```

```python
# ============================================================
# DONOR-LEVEL ENRICHMENT VALIDATION
# Test each population against expected abundance (1.0)
# + confidence intervals + leave-one-donor-out sensitivity
# ============================================================

import numpy as np
import pandas as pd
from scipy.stats import wilcoxon

# Use the existing donor-level results
assert "donor_df" in globals()

rng = np.random.default_rng(20261110)

populations = {
    "Perivasc Fib I": "fib_i_enrichment",
    "Perivasc Fib II": "fib_ii_enrichment"
}

summary_results = []

print("DONOR-LEVEL ENRICHMENT AGAINST EXPECTED ABUNDANCE")
print("=" * 75)

for population, column in populations.items():

    values = donor_df[column].dropna().to_numpy()
    log_values = np.log2(values)

    # Test whether donor-level enrichment differs from 1
    test = wilcoxon(
        log_values,
        alternative="greater",
        method="auto"
    )

    # Bootstrap donor-level median confidence interval
    bootstrap_medians = []

    for _ in range(10000):
        sampled = rng.choice(
            values,
            size=len(values),
            replace=True
        )
        bootstrap_medians.append(np.median(sampled))

    ci_low, ci_high = np.percentile(
        bootstrap_medians,
        [2.5, 97.5]
    )

    # Leave-one-donor-out sensitivity
    loo_medians = []

    for i in range(len(values)):
        remaining = np.delete(values, i)
        loo_medians.append(np.median(remaining))

    print(f"\n{population}")
    print("-" * 55)
    print("Donors:", len(values))
    print("Median enrichment:", round(np.median(values), 4))
    print("Bootstrap 95% CI:", round(ci_low, 4), "to", round(ci_high, 4))
    print("Donors above 1:", int(np.sum(values > 1)))
    print("Donors below 1:", int(np.sum(values < 1)))
    print("One-sided Wilcoxon p:", f"{test.pvalue:.6g}")
    print("Minimum leave-one-out median:", round(min(loo_medians), 4))
    print("Maximum leave-one-out median:", round(max(loo_medians), 4))
    print("All leave-one-out medians above 1:",
          bool(all(x > 1 for x in loo_medians)))

    summary_results.append({
        "population": population,
        "donors": len(values),
        "median_enrichment": np.median(values),
        "ci_lower": ci_low,
        "ci_upper": ci_high,
        "donors_above_1": int(np.sum(values > 1)),
        "donors_below_1": int(np.sum(values < 1)),
        "wilcoxon_greater_p": test.pvalue,
        "loo_min_median": min(loo_medians),
        "loo_max_median": max(loo_medians)
    })

# Correct the two population-level tests
summary_df = pd.DataFrame(summary_results)

from statsmodels.stats.multitest import multipletests

summary_df["BH_adjusted_p"] = multipletests(
    summary_df["wilcoxon_greater_p"],
    method="fdr_bh"
)[1]

print("\nMULTIPLE-TESTING-ADJUSTED SUMMARY")
print("=" * 75)
print(summary_df.round(5).to_string(index=False))

# Save
output_dir = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
)

summary_df.to_csv(
    output_dir + "Mast_Donor_Enrichment_Validation.csv",
    index=False
)

print("\nSaved donor-level enrichment validation.")
```

```python
# ============================================================
# SPATIAL ABUNDANCE-AWARE PERMUTATION TEST
# Preserve Mast-cell positions and target population counts.
# Test whether Perivasc Fib I/II are overrepresented around
# Mast cells beyond their sample-specific abundance.
# ============================================================

import numpy as np
import pandas as pd
from scipy.sparse import csr_matrix
from statsmodels.stats.multitest import multipletests

rng = np.random.default_rng(20261110)

MAST = 0
TARGETS = {
    "Perivasc Fib I": 3,
    "Perivasc Fib II": 4
}

N_PERMUTATIONS = 999

# Use the existing full-tissue graph and reconstructed labels.
# Restrict each sample to dermal cells.
sample_ids = np.unique(sample_codes_full[sample_codes_full >= 0])

permutation_results = []

for sample_code in sample_ids:

    sample_indices = np.where(
        (sample_codes_full == sample_code) &
        (compartment_codes == 0)
    )[0]

    labels = expanded_labels_10[sample_indices]

    mast_local = np.where(labels == MAST)[0]

    if len(mast_local) < 5:
        continue

    mast_global = sample_indices[mast_local]

    # Mast positions remain fixed.
    mast_graph = A_symmetric[mast_global, :][:, sample_indices]

    # Candidate labels are the two target populations.
    # All other labels remain unchanged.
    target_local = np.where(
        np.isin(labels, list(TARGETS.values()))
    )[0]

    if len(target_local) == 0:
        continue

    # Graph-neighbor denominator includes all dermal graph neighbors.
    degrees = np.asarray(mast_graph.sum(axis=1)).ravel()
    valid = degrees > 0

    if valid.sum() < 5:
        continue

    mast_graph = mast_graph[valid]
    degrees = degrees[valid]

    observed = {}

    for name, code in TARGETS.items():
        target_positions = np.where(labels == code)[0]

        observed[name] = np.asarray(
            mast_graph[:, target_positions].sum(axis=1)
        ).ravel().sum() / degrees.sum()

    # Permute the two target identities among their combined
    # positions, preserving each population's sample-level count.
    combined_positions = target_local.copy()

    n_fib_i = int(np.sum(labels == TARGETS["Perivasc Fib I"]))
    n_fib_ii = int(np.sum(labels == TARGETS["Perivasc Fib II"]))

    null_values = {
        name: np.zeros(N_PERMUTATIONS)
        for name in TARGETS
    }

    for perm in range(N_PERMUTATIONS):

        shuffled = rng.permutation(combined_positions)

        fib_i_positions = shuffled[:n_fib_i]
        fib_ii_positions = shuffled[n_fib_i:n_fib_i + n_fib_ii]

        for name, positions in [
            ("Perivasc Fib I", fib_i_positions),
            ("Perivasc Fib II", fib_ii_positions)
        ]:
            neighbor_count = np.asarray(
                mast_graph[:, positions].sum(axis=1)
            ).ravel().sum()

            null_values[name][perm] = neighbor_count / degrees.sum()

    for name in TARGETS:

        null = null_values[name]
        obs = observed[name]

        # One-sided empirical test for greater-than-null enrichment.
        p_value = (
            1 + np.sum(null >= obs)
        ) / (N_PERMUTATIONS + 1)

        permutation_results.append({
            "sample_code": int(sample_code),
            "population": name,
            "mast_count": len(mast_local),
            "observed_neighbor_fraction": obs,
            "null_mean": np.mean(null),
            "null_sd": np.std(null, ddof=1),
            "observed_to_null_ratio": (
                obs / np.mean(null) if np.mean(null) > 0 else np.nan
            ),
            "empirical_p": p_value
        })

    if len(permutation_results) % 20 == 0:
        print("Completed sample:", int(sample_code))

perm_df = pd.DataFrame(permutation_results)

# ------------------------------------------------------------
# DONOR-LEVEL SUMMARY
# ------------------------------------------------------------

donor_lookup = {
    i: donor_categories[i]
    for i in range(len(donor_categories))
}

sample_donor_lookup = {}

for sample_code in sample_ids:
    idx = np.where(sample_codes_full == sample_code)[0]
    donor_codes = np.unique(donor_codes_full[idx])
    donor_codes = donor_codes[donor_codes >= 0]

    if len(donor_codes) == 1:
        sample_donor_lookup[int(sample_code)] = donor_lookup[int(donor_codes[0])]

perm_df["donor"] = perm_df["sample_code"].map(sample_donor_lookup)

donor_perm = (
    perm_df
    .groupby(["donor", "population"])
    .agg(
        observed_fraction=("observed_neighbor_fraction", "median"),
        null_mean=("null_mean", "median"),
        enrichment_ratio=("observed_to_null_ratio", "median"),
        sample_count=("sample_code", "nunique")
    )
    .reset_index()
)

print("\nDONOR-LEVEL PERMUTATION SUMMARY")
print("=" * 75)

for population in TARGETS:

    subset = donor_perm[
        donor_perm["population"] == population
    ]

    ratios = subset["enrichment_ratio"].dropna().to_numpy()

    print("\n", population)
    print("Donors:", len(ratios))
    print("Median observed/null ratio:", round(np.median(ratios), 4))
    print("Donors above 1:", int(np.sum(ratios > 1)))
    print("Donors below 1:", int(np.sum(ratios < 1)))

    if len(ratios) >= 5:
        from scipy.stats import wilcoxon

        test = wilcoxon(
            np.log2(ratios),
            alternative="greater",
            method="auto"
        )
        print("Donor-level Wilcoxon p:", f"{test.pvalue:.6g}")

# ------------------------------------------------------------
# SAMPLE-LEVEL EMPIRICAL P-VALUES
# ------------------------------------------------------------

sample_summary = (
    perm_df
    .groupby("population")
    .agg(
        samples=("sample_code", "nunique"),
        median_observed=("observed_neighbor_fraction", "median"),
        median_null=("null_mean", "median"),
        median_ratio=("observed_to_null_ratio", "median"),
        samples_ratio_above_1=("observed_to_null_ratio",
                               lambda x: int((x > 1).sum())),
        median_empirical_p=("empirical_p", "median")
    )
    .reset_index()
)

print("\nSAMPLE-LEVEL SUMMARY")
print("=" * 75)
print(sample_summary.round(5).to_string(index=False))

# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

output_dir = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
)

perm_df.to_csv(
    output_dir + "Mast_Fibroblast_Abundance_Permutation_Samples.csv",
    index=False
)

donor_perm.to_csv(
    output_dir + "Mast_Fibroblast_Abundance_Permutation_Donors.csv",
    index=False
)

sample_summary.to_csv(
    output_dir + "Mast_Fibroblast_Abundance_Permutation_Summary.csv",
    index=False
)

print("\nSaved permutation results.")
```

```python
# ============================================================
# DERMAL FIXED-MAST, ABUNDANCE-PRESERVING SPATIAL NULL
# Randomize target identities across all non-Mast dermal cells.
# Preserve each sample's Fib I and Fib II cell counts.
# ============================================================

import numpy as np
import pandas as pd
from scipy.stats import wilcoxon
from statsmodels.stats.multitest import multipletests

rng = np.random.default_rng(20261111)

MAST = 0
TARGETS = {
    "Perivasc Fib I": 3,
    "Perivasc Fib II": 4
}

N_PERMUTATIONS = 999
permutation_results = []

sample_ids = np.unique(sample_codes_full[sample_codes_full >= 0])

for sample_code in sample_ids:

    sample_indices = np.where(
        (sample_codes_full == sample_code) &
        (compartment_codes == 0)
    )[0]

    labels = expanded_labels_10[sample_indices]

    mast_local = np.where(labels == MAST)[0]
    candidate_local = np.where(labels != MAST)[0]

    if len(mast_local) < 5 or len(candidate_local) == 0:
        continue

    mast_global = sample_indices[mast_local]

    # Keep Mast positions fixed.
    mast_graph = A_symmetric[mast_global, :][:, sample_indices]

    degrees = np.asarray(mast_graph.sum(axis=1)).ravel()
    valid = degrees > 0

    if valid.sum() < 5:
        continue

    mast_graph = mast_graph[valid]
    total_degree = degrees[valid].sum()

    n_fib_i = int(np.sum(labels == TARGETS["Perivasc Fib I"]))
    n_fib_ii = int(np.sum(labels == TARGETS["Perivasc Fib II"]))

    if n_fib_i + n_fib_ii > len(candidate_local):
        continue

    observed = {}

    for name, code in TARGETS.items():
        positions = np.where(labels == code)[0]

        observed[name] = (
            np.asarray(mast_graph[:, positions].sum(axis=1)).ravel().sum()
            / total_degree
        )

    null_values = {
        name: np.zeros(N_PERMUTATIONS)
        for name in TARGETS
    }

    # Randomize target identities across ALL non-Mast dermal cells.
    # Fib I and Fib II retain their original sample-specific counts.
    for perm in range(N_PERMUTATIONS):

        shuffled = rng.permutation(candidate_local)

        fib_i_positions = shuffled[:n_fib_i]
        fib_ii_positions = shuffled[n_fib_i:n_fib_i + n_fib_ii]

        for name, positions in [
            ("Perivasc Fib I", fib_i_positions),
            ("Perivasc Fib II", fib_ii_positions)
        ]:

            neighbor_count = np.asarray(
                mast_graph[:, positions].sum(axis=1)
            ).ravel().sum()

            null_values[name][perm] = neighbor_count / total_degree

    for name in TARGETS:

        null = null_values[name]
        obs = observed[name]
        null_mean = null.mean()

        empirical_p = (
            1 + np.sum(null >= obs)
        ) / (N_PERMUTATIONS + 1)

        permutation_results.append({
            "sample_code": int(sample_code),
            "population": name,
            "mast_count": len(mast_local),
            "target_count": (
                n_fib_i if name == "Perivasc Fib I" else n_fib_ii
            ),
            "observed_fraction": obs,
            "null_mean": null_mean,
            "null_sd": null.std(ddof=1),
            "observed_null_ratio": (
                obs / null_mean if null_mean > 0 else np.nan
            ),
            "empirical_p": empirical_p
        })

    if len(permutation_results) % 20 == 0:
        print("Completed sample:", int(sample_code))

perm_dermal_df = pd.DataFrame(permutation_results)

# ------------------------------------------------------------
# DONOR-LEVEL AGGREGATION
# ------------------------------------------------------------

sample_donor_lookup = {}

for sample_code in sample_ids:
    idx = np.where(sample_codes_full == sample_code)[0]
    codes = np.unique(donor_codes_full[idx])
    codes = codes[codes >= 0]

    if len(codes) == 1:
        sample_donor_lookup[int(sample_code)] = donor_categories[int(codes[0])]

perm_dermal_df["donor"] = (
    perm_dermal_df["sample_code"].map(sample_donor_lookup)
)

donor_perm_df = (
    perm_dermal_df
    .groupby(["donor", "population"])
    .agg(
        observed_fraction=("observed_fraction", "median"),
        null_mean=("null_mean", "median"),
        enrichment_ratio=("observed_null_ratio", "median"),
        sample_count=("sample_code", "nunique")
    )
    .reset_index()
)

# ------------------------------------------------------------
# TEST EACH POPULATION AGAINST NULL EXPECTATION OF 1
# ------------------------------------------------------------

summary = []

print("\nDERMAL DONOR-LEVEL ABUNDANCE-PRESERVING NULL")
print("=" * 75)

for population in TARGETS:

    subset = donor_perm_df[
        donor_perm_df["population"] == population
    ]

    ratios = subset["enrichment_ratio"].dropna().to_numpy()

    test = wilcoxon(
        np.log2(ratios),
        alternative="greater",
        method="auto"
    )

    print("\n", population)
    print("Donors:", len(ratios))
    print("Median observed/null ratio:", round(np.median(ratios), 4))
    print("Donors above 1:", int(np.sum(ratios > 1)))
    print("Donors below 1:", int(np.sum(ratios < 1)))
    print("One-sided donor Wilcoxon p:", f"{test.pvalue:.6g}")

    summary.append({
        "population": population,
        "donors": len(ratios),
        "median_ratio": np.median(ratios),
        "donors_above_1": int(np.sum(ratios > 1)),
        "donors_below_1": int(np.sum(ratios < 1)),
        "donor_wilcoxon_p": test.pvalue
    })

summary_df = pd.DataFrame(summary)

summary_df["BH_adjusted_p"] = multipletests(
    summary_df["donor_wilcoxon_p"],
    method="fdr_bh"
)[1]

print("\nADJUSTED SUMMARY")
print("=" * 75)
print(summary_df.round(6).to_string(index=False))

# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

output_dir = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
)

perm_dermal_df.to_csv(
    output_dir + "Dermal_Mast_Fib_Abundance_Null_Samples.csv",
    index=False
)

donor_perm_df.to_csv(
    output_dir + "Dermal_Mast_Fib_Abundance_Null_Donors.csv",
    index=False
)

summary_df.to_csv(
    output_dir + "Dermal_Mast_Fib_Abundance_Null_Summary.csv",
    index=False
)

print("\nSaved all three dermal permutation result files.")
```

```python
# ============================================================
# DEGREE-STRATIFIED, ABUNDANCE-PRESERVING PERMUTATION
# Preserve sample, dermal compartment, Mast positions,
# and graph-degree distribution of candidate target cells.
# ============================================================

import numpy as np
import pandas as pd
from scipy.stats import wilcoxon
from statsmodels.stats.multitest import multipletests

rng = np.random.default_rng(20261112)

MAST = 0
TARGETS = {
    "Perivasc Fib I": 3,
    "Perivasc Fib II": 4
}

N_PERMUTATIONS = 999
N_DEGREE_BINS = 5

results = []

sample_ids = np.unique(sample_codes_full[sample_codes_full >= 0])

for sample_code in sample_ids:

    sample_indices = np.where(
        (sample_codes_full == sample_code) &
        (compartment_codes == 0)
    )[0]

    labels = expanded_labels_10[sample_indices]

    mast_local = np.where(labels == MAST)[0]
    candidate_local = np.where(labels != MAST)[0]

    if len(mast_local) < 5 or len(candidate_local) < 20:
        continue

    mast_global = sample_indices[mast_local]

    mast_graph = A_symmetric[mast_global, :][:, sample_indices]

    degrees = np.asarray(mast_graph.sum(axis=1)).ravel()
    valid = degrees > 0

    if valid.sum() < 5:
        continue

    mast_graph = mast_graph[valid]
    total_degree = degrees[valid].sum()

    # Degree of every eligible candidate cell in the full
    # reconstructed graph.
    candidate_global = sample_indices[candidate_local]
    candidate_degree = graph_degrees[candidate_global]

    # Quantile-based degree strata.
    # Duplicate boundaries are handled by dropping duplicates.
    try:
        degree_bins = pd.qcut(
            candidate_degree,
            q=N_DEGREE_BINS,
            labels=False,
            duplicates="drop"
        )
    except ValueError:
        continue

    degree_bins = np.asarray(degree_bins)

    strata = [
        candidate_local[degree_bins == b]
        for b in np.unique(degree_bins)
    ]

    # Retain only strata with enough candidate cells.
    strata = [s for s in strata if len(s) >= 2]

    if len(strata) == 0:
        continue

    # Assign each target population its observed number of cells
    # within each degree stratum.
    target_counts = {}

    for name, code in TARGETS.items():

        target_counts[name] = []

        for stratum in strata:
            target_counts[name].append(
                int(np.sum(labels[stratum] == code))
            )

    # Observed fractions
    observed = {}

    for name, code in TARGETS.items():

        positions = np.where(labels == code)[0]

        observed[name] = (
            np.asarray(mast_graph[:, positions].sum(axis=1)).ravel().sum()
            / total_degree
        )

    null_values = {
        name: np.zeros(N_PERMUTATIONS)
        for name in TARGETS
    }

    # Permute labels within degree strata, preserving the exact
    # number of Fib I and Fib II cells in each stratum.
    for perm in range(N_PERMUTATIONS):

        assigned = {
            name: []
            for name in TARGETS
        }

        for stratum_idx, stratum in enumerate(strata):

            shuffled = rng.permutation(stratum)

            n_i = target_counts["Perivasc Fib I"][stratum_idx]
            n_ii = target_counts["Perivasc Fib II"][stratum_idx]

            assigned["Perivasc Fib I"].extend(shuffled[:n_i])
            assigned["Perivasc Fib II"].extend(
                shuffled[n_i:n_i + n_ii]
            )

        for name in TARGETS:

            positions = np.asarray(assigned[name], dtype=int)

            if len(positions) == 0:
                null_values[name][perm] = np.nan
                continue

            neighbor_count = np.asarray(
                mast_graph[:, positions].sum(axis=1)
            ).ravel().sum()

            null_values[name][perm] = neighbor_count / total_degree

    for name in TARGETS:

        null = null_values[name]
        null = null[np.isfinite(null)]

        if len(null) == 0:
            continue

        obs = observed[name]
        null_mean = null.mean()

        empirical_p = (
            1 + np.sum(null >= obs)
        ) / (len(null) + 1)

        results.append({
            "sample_code": int(sample_code),
            "population": name,
            "observed_fraction": obs,
            "null_mean": null_mean,
            "null_sd": null.std(ddof=1),
            "observed_null_ratio": (
                obs / null_mean if null_mean > 0 else np.nan
            ),
            "empirical_p": empirical_p,
            "degree_strata": len(strata)
        })

    if len(results) % 20 == 0:
        print("Completed sample:", int(sample_code))

degree_perm_df = pd.DataFrame(results)

# ------------------------------------------------------------
# DONOR-LEVEL AGGREGATION
# ------------------------------------------------------------

sample_donor_lookup = {}

for sample_code in sample_ids:

    idx = np.where(sample_codes_full == sample_code)[0]
    codes = np.unique(donor_codes_full[idx])
    codes = codes[codes >= 0]

    if len(codes) == 1:
        sample_donor_lookup[int(sample_code)] = donor_categories[int(codes[0])]

degree_perm_df["donor"] = (
    degree_perm_df["sample_code"].map(sample_donor_lookup)
)

degree_donor_df = (
    degree_perm_df
    .groupby(["donor", "population"])
    .agg(
        observed_fraction=("observed_fraction", "median"),
        null_mean=("null_mean", "median"),
        enrichment_ratio=("observed_null_ratio", "median"),
        sample_count=("sample_code", "nunique")
    )
    .reset_index()
)

summary = []

print("\nDEGREE-STRATIFIED DONOR-LEVEL RESULTS")
print("=" * 75)

for population in TARGETS:

    subset = degree_donor_df[
        degree_donor_df["population"] == population
    ]

    ratios = subset["enrichment_ratio"].dropna().to_numpy()

    test = wilcoxon(
        np.log2(ratios),
        alternative="greater",
        method="auto"
    )

    print("\n", population)
    print("Donors:", len(ratios))
    print("Median observed/null ratio:", round(np.median(ratios), 4))
    print("Donors above 1:", int(np.sum(ratios > 1)))
    print("Donors below 1:", int(np.sum(ratios < 1)))
    print("One-sided Wilcoxon p:", f"{test.pvalue:.6g}")

    summary.append({
        "population": population,
        "donors": len(ratios),
        "median_ratio": np.median(ratios),
        "donors_above_1": int(np.sum(ratios > 1)),
        "donors_below_1": int(np.sum(ratios < 1)),
        "wilcoxon_p": test.pvalue
    })

summary_df = pd.DataFrame(summary)

summary_df["BH_adjusted_p"] = multipletests(
    summary_df["wilcoxon_p"],
    method="fdr_bh"
)[1]

print("\nMULTIPLE-TESTING-ADJUSTED SUMMARY")
print("=" * 75)
print(summary_df.round(6).to_string(index=False))

# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

output_dir = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
)

degree_perm_df.to_csv(
    output_dir + "Mast_Degree_Stratified_Permutation_Samples.csv",
    index=False
)

degree_donor_df.to_csv(
    output_dir + "Mast_Degree_Stratified_Permutation_Donors.csv",
    index=False
)

summary_df.to_csv(
    output_dir + "Mast_Degree_Stratified_Permutation_Summary.csv",
    index=False
)

print("\nSaved degree-stratified permutation results.")
```

```python
# ============================================================
# DEGREE-STRATIFICATION AUDIT
# Check singleton strata and degree definitions
# ============================================================

import numpy as np
import pandas as pd

audit_rows = []

population_codes = {
    "Perivasc Fib I": population_names_10.index("Perivasc Fib I"),
    "Perivasc Fib II": population_names_10.index("Perivasc Fib II")
}

dermis_mask = (compartment_codes == 0)

for sample_code in np.unique(sample_codes_full[dermis_mask]):

    sample_mask = (
        dermis_mask &
        (sample_codes_full == sample_code)
    )

    sample_global = np.where(sample_mask)[0]

    # All eligible non-Mast dermal cells
    candidate_global = sample_global[
        expanded_labels_10[sample_global] != 0
    ]

    if len(candidate_global) < 2:
        continue

    # Full-graph degree
    full_degree = np.asarray(
        graph_degrees[candidate_global]
    ).ravel()

    # Within-sample dermal graph degree
    sample_graph = A_symmetric[
        sample_global, :
    ][:, sample_global]

    local_degree_all = np.asarray(
        sample_graph.sum(axis=1)
    ).ravel()

    local_position = {
        global_idx: local_idx
        for local_idx, global_idx in enumerate(sample_global)
    }

    candidate_local = np.array([
        local_position[idx] for idx in candidate_global
    ])

    dermal_degree = local_degree_all[candidate_local]

    # Five quantile strata, separately for each degree definition
    for degree_name, degree_values in [
        ("full_graph_degree", full_degree),
        ("within_sample_dermal_degree", dermal_degree)
    ]:

        strata = pd.qcut(
            degree_values,
            q=5,
            labels=False,
            duplicates="drop"
        )

        strata = np.asarray(strata)

        for population, code in population_codes.items():

            target_mask = (
                expanded_labels_10[candidate_global] == code
            )

            target_total = int(target_mask.sum())
            target_excluded = 0
            candidate_excluded = 0
            singleton_strata = 0

            for stratum in np.unique(strata):

                stratum_mask = strata == stratum
                stratum_size = int(stratum_mask.sum())

                if stratum_size < 2:
                    singleton_strata += 1
                    candidate_excluded += stratum_size
                    target_excluded += int(
                        (target_mask & stratum_mask).sum()
                    )

            audit_rows.append({
                "sample_code": sample_code,
                "population": population,
                "degree_definition": degree_name,
                "target_cells": target_total,
                "target_cells_in_singleton_strata": target_excluded,
                "candidate_cells_in_singleton_strata": candidate_excluded,
                "singleton_strata": singleton_strata
            })

degree_audit_df = pd.DataFrame(audit_rows)

print("DEGREE STRATIFICATION AUDIT")
print("=" * 75)

summary = (
    degree_audit_df
    .groupby(["population", "degree_definition"])
    .agg(
        samples=("sample_code", "nunique"),
        target_cells=("target_cells", "sum"),
        target_cells_in_singletons=(
            "target_cells_in_singleton_strata", "sum"
        ),
        candidate_cells_in_singletons=(
            "candidate_cells_in_singleton_strata", "sum"
        ),
        samples_with_affected_targets=(
            "target_cells_in_singleton_strata",
            lambda x: int((x > 0).sum())
        )
    )
    .reset_index()
)

summary["percent_target_cells_affected"] = (
    100 * summary["target_cells_in_singletons"] /
    summary["target_cells"]
)

print(summary.round(3).to_string(index=False))

print("\nSAMPLES WITH THE MOST AFFECTED TARGET CELLS")
print("=" * 75)

print(
    degree_audit_df[
        degree_audit_df["target_cells_in_singleton_strata"] > 0
    ]
    .sort_values(
        "target_cells_in_singleton_strata",
        ascending=False
    )
    .head(20)
    .to_string(index=False)
)

degree_audit_df.to_csv(
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Degree_Stratification_Audit.csv",
    index=False
)

summary.to_csv(
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Degree_Stratification_Audit_Summary.csv",
    index=False
)

print("\nSaved degree-stratification audit.")
```

```python

# ============================================================
# FINAL DEGREE-STRATIFIED PERMUTATION
# Within-sample dermal graph degree
# ============================================================

import numpy as np
import pandas as pd
from scipy.stats import wilcoxon

N_PERMUTATIONS = 999
N_QUANTILES = 5
RANDOM_SEED = 20261110

rng = np.random.default_rng(RANDOM_SEED)

population_codes = {
    "Perivasc Fib I": population_names_10.index("Perivasc Fib I"),
    "Perivasc Fib II": population_names_10.index("Perivasc Fib II")
}

sample_results = []

dermis_mask = (compartment_codes == 0)
sample_ids = np.unique(sample_codes_full[dermis_mask])

print("WITHIN-SAMPLE DERMAL DEGREE-STRATIFIED NULL")
print("=" * 75)

for sample_code in sample_ids:

    sample_global = np.where(
        dermis_mask & (sample_codes_full == sample_code)
    )[0]

    labels = expanded_labels_10[sample_global]

    mast_local = np.where(labels == 0)[0]
    candidate_local = np.where(labels != 0)[0]

    if len(mast_local) == 0 or len(candidate_local) < 2:
        continue

    # Construct sample-specific dermal graph
    sample_graph = A_symmetric[
        sample_global, :
    ][:, sample_global].tocsr()

    mast_rows = sample_graph[mast_local, :]
    total_mast_edges = float(mast_rows.sum())

    if total_mast_edges == 0:
        continue

    # Mast-cell edges to each non-Mast candidate
    candidate_weights = np.asarray(
        mast_rows[:, candidate_local].sum(axis=0)
    ).ravel()

    candidate_labels = labels[candidate_local]

    # Candidate degree within this sample's dermal graph
    within_dermal_degree = np.asarray(
        sample_graph.sum(axis=1)
    ).ravel()[candidate_local]

    # Divide candidates into up to five degree strata
    strata = pd.qcut(
        within_dermal_degree,
        q=N_QUANTILES,
        labels=False,
        duplicates="drop"
    )

    strata = np.asarray(strata)

    # Require every candidate to receive a degree stratum
    if pd.isna(strata).any():
        raise ValueError(
            f"Some candidate cells in sample {sample_code} "
            "were not assigned a degree stratum."
        )

    # Include ALL strata, including singleton strata
    valid_strata = np.unique(strata)

    donor_code = donor_codes_full[sample_global[0]]

    for population, target_code in population_codes.items():

        target_mask = candidate_labels == target_code
        observed_count = int(target_mask.sum())

        if observed_count == 0:
            continue

        observed_edges = float(
            candidate_weights[target_mask].sum()
        )

        observed_fraction = observed_edges / total_mast_edges

        # Preserve target counts within every degree stratum
        stratum_data = []

        for stratum in valid_strata:

            positions = np.where(strata == stratum)[0]
            target_count = int(target_mask[positions].sum())

            if target_count > 0:
                stratum_data.append(
                    (candidate_weights[positions], target_count)
                )

        null_fractions = np.zeros(N_PERMUTATIONS)

        for iteration in range(N_PERMUTATIONS):

            null_edges = 0.0

            for weights, target_count in stratum_data:

                selected = rng.choice(
                    len(weights),
                    size=target_count,
                    replace=False
                )

                null_edges += weights[selected].sum()

            null_fractions[iteration] = (
                null_edges / total_mast_edges
            )

        null_mean = float(null_fractions.mean())

        enrichment_ratio = (
            observed_fraction / null_mean
            if null_mean > 0 else np.nan
        )

        sample_results.append({
            "sample_code": sample_code,
            "donor_code": donor_code,
            "population": population,
            "target_cell_count": observed_count,
            "observed_neighbor_fraction": observed_fraction,
            "null_mean_neighbor_fraction": null_mean,
            "enrichment_ratio": enrichment_ratio,
            "null_sd": float(null_fractions.std()),
            "permutations": N_PERMUTATIONS
        })

    if len(sample_results) % 20 == 0:
        print("Processed sample:", sample_code)

degree_dermal_sample_df = pd.DataFrame(sample_results)

# Aggregate sample-level ratios to donor-level medians
degree_dermal_donor_df = (
    degree_dermal_sample_df
    .groupby(["donor_code", "population"])
    .agg(
        enrichment_ratio=("enrichment_ratio", "median"),
        sample_count=("sample_code", "nunique")
    )
    .reset_index()
)

print("\nDONOR-LEVEL RESULTS")
print("=" * 75)

for population in population_codes:

    values = degree_dermal_donor_df.loc[
        degree_dermal_donor_df["population"] == population,
        "enrichment_ratio"
    ].dropna().to_numpy()

    if len(values) < 2:
        continue

    test = wilcoxon(
        values - 1,
        alternative="greater",
        method="auto"
    )

    print("\n", population)
    print("Donors:", len(values))
    print("Median enrichment:", round(float(np.median(values)), 4))
    print("Donors above 1:", int(np.sum(values > 1)))
    print("Donors below 1:", int(np.sum(values < 1)))
    print("One-sided Wilcoxon p:", test.pvalue)

# Save both levels
output_dir = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
)

degree_dermal_sample_df.to_csv(
    output_dir + "Within_Dermal_Degree_Stratified_Sample_Results.csv",
    index=False
)

degree_dermal_donor_df.to_csv(
    output_dir + "Within_Dermal_Degree_Stratified_Donor_Results.csv",
    index=False
)

print("\nSamples analyzed:", degree_dermal_sample_df["sample_code"].nunique())
print("Donors analyzed:", degree_dermal_donor_df["donor_code"].nunique())
print("Saved sample-level and donor-level results.")
```

```python
# ============================================================
# ANATOMICAL-SITE CONSISTENCY
# Within-sample dermal degree-stratified results
# ============================================================

import numpy as np
import pandas as pd
from scipy.stats import wilcoxon
from statsmodels.stats.multitest import multipletests

# Build sample-level donor and anatomical-site metadata
sample_metadata = []

for sample_code in degree_dermal_sample_df["sample_code"].unique():

    indices = np.where(
        (sample_codes_full == sample_code) &
        (compartment_codes == 0)
    )[0]

    donors = np.unique(donor_codes_full[indices])
    sites = np.unique(anatomic_site_codes_full[indices])

    if len(donors) != 1 or len(sites) != 1:
        print("Metadata inconsistency:", sample_code, donors, sites)
        continue

    donor = donors[0]
    site_code = int(sites[0])

    sample_metadata.append({
        "sample_code": sample_code,
        "donor_code": donor,
        "site_code": site_code,
        "anatomic_site": anatomic_site_categories[site_code]
    })

sample_metadata_df = pd.DataFrame(sample_metadata)

site_sample_df = degree_dermal_sample_df.merge(
    sample_metadata_df,
    on=["sample_code", "donor_code"],
    how="inner"
)

# First aggregate multiple samples from the same donor/site
donor_site_df = (
    site_sample_df
    .groupby([
        "population",
        "anatomic_site",
        "donor_code"
    ])
    .agg(
        enrichment_ratio=("enrichment_ratio", "median"),
        sample_count=("sample_code", "nunique")
    )
    .reset_index()
)

# Summarize each anatomical site across donors
site_results = []

for population in [
    "Perivasc Fib I",
    "Perivasc Fib II"
]:

    pop_df = donor_site_df[
        donor_site_df["population"] == population
    ]

    for site, group in pop_df.groupby("anatomic_site"):

        values = group["enrichment_ratio"].dropna().to_numpy()

        if len(values) == 0:
            continue

        p_value = np.nan

        if len(values) >= 5 and np.any(values != 1):
            p_value = wilcoxon(
                values - 1,
                alternative="greater",
                method="auto"
            ).pvalue

        site_results.append({
            "population": population,
            "anatomic_site": site,
            "donor_count": len(values),
            "median_enrichment": np.median(values),
            "mean_enrichment": np.mean(values),
            "donors_above_1": int(np.sum(values > 1)),
            "donors_below_1": int(np.sum(values < 1)),
            "donor_direction_percent": (
                100 * np.mean(values > 1)
            ),
            "wilcoxon_p": p_value
        })

site_results_df = pd.DataFrame(site_results)

# Adjust p-values across all eligible population/site tests
eligible = site_results_df["wilcoxon_p"].notna()

site_results_df["BH_adjusted_p"] = np.nan

if eligible.sum() > 0:
    site_results_df.loc[eligible, "BH_adjusted_p"] = (
        multipletests(
            site_results_df.loc[eligible, "wilcoxon_p"],
            method="fdr_bh"
        )[1]
    )

print("ANATOMICAL-SITE CONSISTENCY")
print("=" * 85)

for population in [
    "Perivasc Fib I",
    "Perivasc Fib II"
]:

    print("\n", population)
    print("-" * 85)

    display_df = site_results_df[
        site_results_df["population"] == population
    ].sort_values(
        "median_enrichment",
        ascending=False
    )

    print(
        display_df.round(4).to_string(index=False)
    )

    eligible_sites = display_df[
        display_df["donor_count"] >= 5
    ]

    print("\nSites with at least 5 donors:",
          len(eligible_sites))
    print("Sites with median enrichment above 1:",
          int((eligible_sites["median_enrichment"] > 1).sum()),
          "/", len(eligible_sites))
    print("Sites with BH-adjusted p < .05:",
          int((eligible_sites["BH_adjusted_p"] < .05).sum()))

# Save results
output_dir = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
)

donor_site_df.to_csv(
    output_dir + "Within_Dermal_Degree_Donor_Site_Results.csv",
    index=False
)

site_results_df.to_csv(
    output_dir + "Within_Dermal_Degree_Anatomical_Site_Results.csv",
    index=False
)

print("\nSamples represented:", site_sample_df["sample_code"].nunique())
print("Donor-site observations:", len(donor_site_df))
print("Anatomical sites:", donor_site_df["anatomic_site"].nunique())
print("Saved anatomical-site results.")
```

```python
# ============================================================
# VASCULAR-NEIGHBORHOOD-ADJUSTED PERMUTATION
# Controls for graph degree and local vascular context
# ============================================================

import numpy as np
import pandas as pd
from scipy.stats import wilcoxon
from statsmodels.stats.multitest import multipletests

N_PERMUTATIONS = 999
RANDOM_SEED = 20261111

rng = np.random.default_rng(RANDOM_SEED)

population_codes = {
    "Perivasc Fib I": population_names_10.index("Perivasc Fib I"),
    "Perivasc Fib II": population_names_10.index("Perivasc Fib II")
}

# Vascular/perivascular context populations
vascular_names = ["Peri", "VEC", "HEC", "LEC", "SM"]
vascular_codes = [
    population_names_10.index(name)
    for name in vascular_names
]

dermis_mask = compartment_codes == 0
sample_ids = np.unique(sample_codes_full[dermis_mask])

results = []
audit = []

print("VASCULAR-NEIGHBORHOOD-ADJUSTED NULL")
print("=" * 75)

for sample_code in sample_ids:

    sample_global = np.where(
        dermis_mask & (sample_codes_full == sample_code)
    )[0]

    labels = expanded_labels_10[sample_global]

    mast_local = np.where(labels == 0)[0]
    candidate_local = np.where(labels != 0)[0]

    if len(mast_local) == 0 or len(candidate_local) < 2:
        continue

    sample_graph = A_symmetric[
        sample_global, :
    ][:, sample_global].tocsr()

    mast_rows = sample_graph[mast_local, :]
    total_mast_edges = float(mast_rows.sum())

    if total_mast_edges == 0:
        continue

    candidate_weights = np.asarray(
        mast_rows[:, candidate_local].sum(axis=0)
    ).ravel()

    candidate_labels = labels[candidate_local]

    # Candidate degree within sample-specific dermal graph
    candidate_degree = np.asarray(
        sample_graph.sum(axis=1)
    ).ravel()[candidate_local]

    # Fraction of each candidate's neighbors belonging to
    # vascular/perivascular context populations
    vascular_local = np.where(
        np.isin(labels, vascular_codes)
    )[0]

    vascular_neighbor_counts = np.asarray(
        sample_graph[candidate_local, :][:, vascular_local].sum(axis=1)
    ).ravel()

    vascular_fraction = np.divide(
        vascular_neighbor_counts,
        candidate_degree,
        out=np.zeros_like(
            vascular_neighbor_counts, dtype=float
        ),
        where=candidate_degree > 0
    )

    # Three degree bins and three vascular-context bins
    degree_bins = pd.qcut(
        candidate_degree,
        q=3,
        labels=False,
        duplicates="drop"
    )

    vascular_bins = pd.qcut(
        vascular_fraction,
        q=3,
        labels=False,
        duplicates="drop"
    )

    strata_df = pd.DataFrame({
        "degree_bin": np.asarray(degree_bins),
        "vascular_bin": np.asarray(vascular_bins)
    })

    # Construct combined strata
    stratum_positions = [
        np.asarray(group.index, dtype=int)
        for _, group in strata_df.groupby(
            ["degree_bin", "vascular_bin"],
            observed=True
        )
    ]

    valid_strata = [
        positions for positions in stratum_positions
        if len(positions) >= 2
    ]

    singleton_positions = [
        positions for positions in stratum_positions
        if len(positions) < 2
    ]

    donor_code = donor_codes_full[sample_global[0]]

    for population, target_code in population_codes.items():

        target_mask = candidate_labels == target_code
        target_total = int(target_mask.sum())

        if target_total == 0:
            continue

        excluded_targets = sum(
            int(target_mask[positions].sum())
            for positions in singleton_positions
        )

        observed_edges = float(
            candidate_weights[target_mask].sum()
        )

        observed_fraction = (
            observed_edges / total_mast_edges
        )

        # Preserve target counts within each combined stratum
        stratum_data = []
        retained_targets = 0

        for positions in valid_strata:

            target_count = int(
                target_mask[positions].sum()
            )

            retained_targets += target_count

            if target_count > 0:
                stratum_data.append((
                    candidate_weights[positions],
                    target_count
                ))

        # Record whether any target cells were excluded
        audit.append({
            "sample_code": sample_code,
            "population": population,
            "target_cells": target_total,
            "excluded_target_cells": excluded_targets,
            "retained_target_cells": retained_targets,
            "strata": len(stratum_positions),
            "valid_strata": len(valid_strata)
        })

        if retained_targets == 0:
            continue

        null_fractions = np.zeros(N_PERMUTATIONS)

        for iteration in range(N_PERMUTATIONS):

            null_edges = 0.0

            for weights, target_count in stratum_data:

                selected = rng.choice(
                    len(weights),
                    size=target_count,
                    replace=False
                )

                null_edges += weights[selected].sum()

            null_fractions[iteration] = (
                null_edges / total_mast_edges
            )

        null_mean = float(null_fractions.mean())

        results.append({
            "sample_code": sample_code,
            "donor_code": donor_code,
            "population": population,
            "target_cell_count": target_total,
            "retained_target_cells": retained_targets,
            "observed_neighbor_fraction": observed_fraction,
            "null_mean_neighbor_fraction": null_mean,
            "enrichment_ratio": (
                observed_fraction / null_mean
                if null_mean > 0 else np.nan
            ),
            "null_sd": float(null_fractions.std()),
            "permutations": N_PERMUTATIONS
        })

vascular_sample_df = pd.DataFrame(results)
vascular_audit_df = pd.DataFrame(audit)

# Donor-level aggregation
vascular_donor_df = (
    vascular_sample_df
    .groupby(["donor_code", "population"])
    .agg(
        enrichment_ratio=("enrichment_ratio", "median"),
        sample_count=("sample_code", "nunique")
    )
    .reset_index()
)

print("\nTARGET-CELL RETENTION AUDIT")
print("=" * 75)

print(
    vascular_audit_df.groupby("population").agg(
        total_targets=("target_cell_count", "sum"),
        excluded_targets=("excluded_target_cells", "sum"),
        retained_targets=("retained_target_cells", "sum"),
        samples=("sample_code", "nunique")
    ).to_string()
)

print("\nDONOR-LEVEL RESULTS")
print("=" * 75)

p_values = []

for population in population_codes:

    values = vascular_donor_df.loc[
        vascular_donor_df["population"] == population,
        "enrichment_ratio"
    ].dropna().to_numpy()

    test = wilcoxon(
        values - 1,
        alternative="greater",
        method="auto"
    )

    p_values.append(test.pvalue)

    print("\n", population)
    print("Donors:", len(values))
    print("Median enrichment:", round(float(np.median(values)), 4))
    print("Donors above 1:", int(np.sum(values > 1)))
    print("Donors below 1:", int(np.sum(values < 1)))
    print("One-sided Wilcoxon p:", test.pvalue)

adjusted = multipletests(
    p_values,
    method="fdr_bh"
)[1]

print("\nBH-ADJUSTED P-VALUES")
for population, p_adj in zip(population_codes, adjusted):
    print(population, ":", p_adj)

# Save outputs
output_dir = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
)

vascular_sample_df.to_csv(
    output_dir + "Vascular_Adjusted_Sample_Results.csv",
    index=False
)

vascular_donor_df.to_csv(
    output_dir + "Vascular_Adjusted_Donor_Results.csv",
    index=False
)

vascular_audit_df.to_csv(
    output_dir + "Vascular_Adjusted_Target_Retention_Audit.csv",
    index=False
)

print("\nSamples analyzed:",
      vascular_sample_df["sample_code"].nunique())
print("Donors analyzed:",
      vascular_donor_df["donor_code"].nunique())
print("Saved vascular-adjusted results.")
```

```python
# ============================================================
# CORRECTED VASCULAR AUDIT AND RESULTS SUMMARY
# ============================================================

print("TARGET-CELL RETENTION AUDIT")
print("=" * 75)

print(
    vascular_audit_df.groupby("population").agg(
        total_targets=("target_cells", "sum"),
        excluded_targets=("excluded_target_cells", "sum"),
        retained_targets=("retained_target_cells", "sum"),
        samples=("sample_code", "nunique")
    ).to_string()
)

print("\nPERCENTAGE OF TARGET CELLS RETAINED")
print("=" * 75)

retention = (
    vascular_audit_df.groupby("population")
    .agg(
        total=("target_cells", "sum"),
        excluded=("excluded_target_cells", "sum"),
        retained=("retained_target_cells", "sum")
    )
)

retention["percent_retained"] = (
    100 * retention["retained"] / retention["total"]
)

print(retention.round(3).to_string())

print("\nDONOR-LEVEL VASCULAR-ADJUSTED RESULTS")
print("=" * 75)

p_values = []
population_order = [
    "Perivasc Fib I",
    "Perivasc Fib II"
]

for population in population_order:

    values = vascular_donor_df.loc[
        vascular_donor_df["population"] == population,
        "enrichment_ratio"
    ].dropna().to_numpy()

    test = wilcoxon(
        values - 1,
        alternative="greater",
        method="auto"
    )

    p_values.append(test.pvalue)

    print("\n", population)
    print("Donors:", len(values))
    print("Median enrichment:", round(float(np.median(values)), 4))
    print("Donors above 1:", int(np.sum(values > 1)))
    print("Donors below 1:", int(np.sum(values < 1)))
    print("One-sided Wilcoxon p:", test.pvalue)

adjusted = multipletests(p_values, method="fdr_bh")[1]

print("\nBH-ADJUSTED P-VALUES")
for population, p_adj in zip(population_order, adjusted):
    print(population, ":", p_adj)
```

```python
# ============================================================
# PAIRED DONOR COMPARISON
# PERIVASC FIB I VS PERIVASC FIB II
# VASCULAR-ADJUSTED ENRICHMENT
# ============================================================

import numpy as np
import pandas as pd
from scipy.stats import wilcoxon, binomtest

N_BOOTSTRAP = 10000
rng = np.random.default_rng(20261112)

paired = vascular_donor_df.pivot(
    index="donor_code",
    columns="population",
    values="enrichment_ratio"
).dropna(
    subset=["Perivasc Fib I", "Perivasc Fib II"]
)

fib1 = paired["Perivasc Fib I"].to_numpy()
fib2 = paired["Perivasc Fib II"].to_numpy()

difference = fib2 - fib1

# Paired two-sided Wilcoxon test
wilcoxon_result = wilcoxon(
    difference,
    alternative="two-sided",
    method="auto"
)

# Paired bootstrap confidence interval for median difference
bootstrap_medians = np.empty(N_BOOTSTRAP)

for i in range(N_BOOTSTRAP):
    sampled = rng.choice(
        difference,
        size=len(difference),
        replace=True
    )
    bootstrap_medians[i] = np.median(sampled)

ci_low, ci_high = np.percentile(
    bootstrap_medians, [2.5, 97.5]
)

# Directional consistency
fib2_higher = int(np.sum(difference > 0))
fib1_higher = int(np.sum(difference < 0))
ties = int(np.sum(difference == 0))

non_ties = fib2_higher + fib1_higher

sign_test_p = (
    binomtest(
        fib2_higher,
        non_ties,
        p=0.5,
        alternative="two-sided"
    ).pvalue
    if non_ties > 0 else np.nan
)

print("PAIRED DONOR COMPARISON")
print("=" * 75)

print("Matched donors:", len(paired))
print("Fib II median enrichment:", round(float(np.median(fib2)), 4))
print("Fib I median enrichment:", round(float(np.median(fib1)), 4))
print("Median paired difference (Fib II - Fib I):",
      round(float(np.median(difference)), 4))
print("Bootstrap 95% CI:",
      round(float(ci_low), 4), "to", round(float(ci_high), 4))
print("Donors with Fib II higher:", fib2_higher)
print("Donors with Fib I higher:", fib1_higher)
print("Ties:", ties)
print("Paired Wilcoxon p:", wilcoxon_result.pvalue)
print("Two-sided sign-test p:", sign_test_p)

paired_output = paired.reset_index()
paired_output["Fib_II_minus_Fib_I"] = difference

paired_output.to_csv(
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Vascular_Adjusted_Paired_FibI_FibII_Donors.csv",
    index=False
)

print("\nSaved paired donor comparison.")
```

```python
# ============================================================
# LEAVE-ONE-DONOR-OUT ROBUSTNESS
# LATEST VASCULAR-ADJUSTED ENRICHMENT RESULTS
# ============================================================

import numpy as np
import pandas as pd

populations = ["Perivasc Fib I", "Perivasc Fib II"]

required = {"donor_code", "population", "enrichment_ratio"}
missing = required - set(vascular_donor_df.columns)

if missing:
    raise KeyError(f"Missing required columns: {missing}")

loo_results = []

print("VASCULAR-ADJUSTED LEAVE-ONE-DONOR-OUT AUDIT")
print("=" * 75)

for population in populations:

    subset = (
        vascular_donor_df.loc[
            vascular_donor_df["population"] == population,
            ["donor_code", "enrichment_ratio"]
        ]
        .dropna()
        .copy()
    )

    if subset["donor_code"].duplicated().any():
        raise ValueError(
            f"Duplicate donor records detected for {population}"
        )

    values = subset["enrichment_ratio"].to_numpy()
    donors = subset["donor_code"].to_numpy()

    if len(values) < 3:
        raise ValueError(
            f"Insufficient donors for leave-one-out analysis: {population}"
        )

    original_median = float(np.median(values))

    loo_medians = []
    loo_above_one = []

    for i in range(len(values)):

        remaining = np.delete(values, i)

        loo_medians.append(float(np.median(remaining)))
        loo_above_one.append(int(np.sum(remaining > 1)))

        loo_results.append({
            "population": population,
            "omitted_donor": donors[i],
            "remaining_donors": len(remaining),
            "loo_median_enrichment": float(np.median(remaining)),
            "donors_above_1": int(np.sum(remaining > 1))
        })

    loo_medians = np.asarray(loo_medians)
    loo_above_one = np.asarray(loo_above_one)

    print(f"\n{population}")
    print("-" * 55)
    print("Donors:", len(values))
    print("Original median:", round(original_median, 4))
    print(
        "LOO median range:",
        round(float(loo_medians.min()), 4),
        "to",
        round(float(loo_medians.max()), 4)
    )
    print(
        "Minimum donors above 1 after omission:",
        int(loo_above_one.min()),
        "of",
        len(values) - 1
    )
    print(
        "Maximum donors above 1 after omission:",
        int(loo_above_one.max()),
        "of",
        len(values) - 1
    )
    print(
        "All LOO medians remain above 1:",
        bool(np.all(loo_medians > 1))
    )

loo_df = pd.DataFrame(loo_results)

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Vascular_Adjusted_Leave_One_Donor_Out.csv"
)

loo_df.to_csv(output_path, index=False)

print("\nSaved:", output_path)
```

```python
# ============================================================
# SENSITIVITY TO VASCULAR-NEIGHBORHOOD ADJUSTMENT
# COMPARE WITHIN-SAMPLE DERMAL DEGREE NULL
# ============================================================

import numpy as np
import pandas as pd
from scipy.stats import wilcoxon, spearmanr

N_BOOTSTRAP = 10000
rng = np.random.default_rng(20261110)

populations = ["Perivasc Fib I", "Perivasc Fib II"]

def compare_null_models(degree_df, vascular_df, level_name):

    degree = degree_df[
        ["donor_code", "population", "enrichment_ratio"]
    ].rename(columns={
        "enrichment_ratio": "degree_enrichment"
    })

    vascular = vascular_df[
        ["donor_code", "population", "enrichment_ratio"]
    ].rename(columns={
        "enrichment_ratio": "vascular_enrichment"
    })

    merged = degree.merge(
        vascular,
        on=["donor_code", "population"],
        how="inner",
        validate="one_to_one"
    )

    print(f"\n{level_name.upper()} COMPARISON")
    print("=" * 75)
    print("Matched records:", len(merged))

    results = []

    for population in populations:

        subset = merged[
            merged["population"] == population
        ].dropna()

        degree_values = subset["degree_enrichment"].to_numpy()
        vascular_values = subset["vascular_enrichment"].to_numpy()

        difference = vascular_values - degree_values

        if len(difference) < 3:
            print(f"\nInsufficient records for {population}")
            continue

        test = wilcoxon(
            difference,
            alternative="two-sided",
            method="auto"
        )

        boot = np.empty(N_BOOTSTRAP)

        for i in range(N_BOOTSTRAP):
            sampled = rng.choice(
                difference,
                size=len(difference),
                replace=True
            )
            boot[i] = np.median(sampled)

        ci_low, ci_high = np.percentile(boot, [2.5, 97.5])

        rho, corr_p = spearmanr(
            degree_values,
            vascular_values
        )

        print(f"\n{population}")
        print("-" * 55)
        print("Matched records:", len(subset))
        print("Degree-adjusted median:",
              round(float(np.median(degree_values)), 4))
        print("Vascular-adjusted median:",
              round(float(np.median(vascular_values)), 4))
        print("Median paired change:",
              round(float(np.median(difference)), 4))
        print("Bootstrap 95% CI:",
              round(float(ci_low), 4), "to",
              round(float(ci_high), 4))
        print("Vascular estimate lower in:",
              int(np.sum(difference < 0)))
        print("Vascular estimate higher in:",
              int(np.sum(difference > 0)))
        print("Unchanged:", int(np.sum(difference == 0)))
        print("Paired Wilcoxon p:", test.pvalue)
        print("Spearman correlation:", round(float(rho), 4))
        print("Correlation p:", corr_p)

        for _, row in subset.iterrows():
            results.append({
                "level": level_name,
                "population": population,
                "donor_code": row["donor_code"],
                "degree_enrichment": row["degree_enrichment"],
                "vascular_enrichment": row["vascular_enrichment"],
                "vascular_minus_degree":
                    row["vascular_enrichment"] -
                    row["degree_enrichment"]
            })

    return results


donor_results = compare_null_models(
    degree_dermal_donor_df,
    vascular_donor_df,
    "donor"
)

sample_results = compare_null_models(
    degree_dermal_sample_df,
    vascular_sample_df,
    "sample"
)

sensitivity_comparison_df = pd.DataFrame(
    donor_results + sample_results
)

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Vascular_Adjustment_Sensitivity_Comparison.csv"
)

sensitivity_comparison_df.to_csv(
    output_path,
    index=False
)

print("\nSaved:", output_path)
```

```python
# ============================================================
# SAMPLE-LEVEL NULL MODEL SENSITIVITY
# MATCH BY SAMPLE + DONOR + POPULATION
# DESCRIPTIVE ANALYSIS ONLY
# ============================================================

import numpy as np
import pandas as pd
from scipy.stats import spearmanr

degree = degree_dermal_sample_df[
    ["sample_code", "donor_code", "population", "enrichment_ratio"]
].rename(columns={
    "enrichment_ratio": "degree_enrichment"
})

vascular = vascular_sample_df[
    ["sample_code", "donor_code", "population", "enrichment_ratio"]
].rename(columns={
    "enrichment_ratio": "vascular_enrichment"
})

keys = ["sample_code", "donor_code", "population"]

sample_comparison = degree.merge(
    vascular,
    on=keys,
    how="inner",
    validate="one_to_one"
)

print("SAMPLE-LEVEL SENSITIVITY COMPARISON")
print("=" * 75)
print("Matched sample-population records:", len(sample_comparison))
print("Unique samples:", sample_comparison["sample_code"].nunique())
print("Unique donors:", sample_comparison["donor_code"].nunique())

sample_comparison["vascular_minus_degree"] = (
    sample_comparison["vascular_enrichment"]
    - sample_comparison["degree_enrichment"]
)

for population in ["Perivasc Fib I", "Perivasc Fib II"]:

    subset = sample_comparison[
        sample_comparison["population"] == population
    ].dropna()

    degree_values = subset["degree_enrichment"].to_numpy()
    vascular_values = subset["vascular_enrichment"].to_numpy()
    differences = subset["vascular_minus_degree"].to_numpy()

    rho, corr_p = spearmanr(
        degree_values,
        vascular_values
    )

    print(f"\n{population}")
    print("-" * 55)
    print("Matched samples:", len(subset))
    print("Degree-adjusted median:",
          round(float(np.median(degree_values)), 4))
    print("Vascular-adjusted median:",
          round(float(np.median(vascular_values)), 4))
    print("Median paired change:",
          round(float(np.median(differences)), 4))
    print("25th–75th percentile change:",
          np.round(np.percentile(differences, [25, 75]), 4))
    print("Vascular estimate lower in:",
          int(np.sum(differences < 0)))
    print("Vascular estimate higher in:",
          int(np.sum(differences > 0)))
    print("Unchanged:", int(np.sum(differences == 0)))
    print("Spearman correlation:", round(float(rho), 4))
    print("Correlation p:", corr_p)

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Sample_Level_Null_Model_Sensitivity.csv"
)

sample_comparison.to_csv(output_path, index=False)

print("\nSaved:", output_path)
```

```python
# ============================================================
# FIGURE 2: ANATOMICAL-SITE ENRICHMENT
# DONOR-LEVEL MEDIANS AND INTERQUARTILE RANGES
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

out_dir = Path(
    "/content/drive/MyDrive/Neurothesis Resoure Library"
)

MIN_DONORS = 5

# Confirm one observation per donor, population, and site
if donor_site_df.duplicated(
    ["population", "anatomic_site", "donor_code"]
).any():
    raise ValueError(
        "Duplicate donor-site records detected."
    )

# Summarize donor-level enrichment distributions
site_plot_df = (
    donor_site_df
    .groupby(["population", "anatomic_site"])
    .agg(
        donor_count=("donor_code", "nunique"),
        median=("enrichment_ratio", "median"),
        q25=("enrichment_ratio", lambda x: x.quantile(0.25)),
        q75=("enrichment_ratio", lambda x: x.quantile(0.75)),
        donors_above_1=("enrichment_ratio", lambda x: int((x > 1).sum())),
        donors_below_1=("enrichment_ratio", lambda x: int((x < 1).sum()))
    )
    .reset_index()
)

site_plot_df = site_plot_df[
    site_plot_df["donor_count"] >= MIN_DONORS
].copy()

populations = ["Perivasc Fib I", "Perivasc Fib II"]

# Shared site order based on average median enrichment
site_order_df = (
    site_plot_df
    .pivot(
        index="anatomic_site",
        columns="population",
        values="median"
    )
)

site_order_df["average_median"] = site_order_df.mean(axis=1)

site_order = (
    site_order_df
    .sort_values("average_median", ascending=True)
    .index.tolist()
)

colors = {
    "Perivasc Fib I": "#2878A5",
    "Perivasc Fib II": "#C16A3D"
}

fig, axes = plt.subplots(
    1, 2,
    figsize=(14, 9),
    sharey=True
)

for ax, population in zip(axes, populations):

    subset = (
        site_plot_df[
            site_plot_df["population"] == population
        ]
        .set_index("anatomic_site")
        .reindex(site_order)
        .dropna(subset=["median"])
    )

    y_positions = np.arange(len(subset))

    medians = subset["median"].to_numpy()
    lower = medians - subset["q25"].to_numpy()
    upper = subset["q75"].to_numpy() - medians

    ax.errorbar(
        medians,
        y_positions,
        xerr=np.vstack([lower, upper]),
        fmt="o",
        color=colors[population],
        ecolor=colors[population],
        elinewidth=1.6,
        capsize=3,
        markersize=7,
        alpha=0.9
    )

    ax.axvline(
        1,
        color="gray",
        linestyle="--",
        linewidth=1.3,
        alpha=0.85
    )

    ax.set_yticks(y_positions)
    ax.set_yticklabels(subset.index)

    ax.set_title(
        population,
        fontsize=13,
        fontweight="bold",
        pad=12
    )

    ax.set_xlabel(
        "Donor-level enrichment ratio",
        fontsize=10
    )

    ax.grid(
        axis="x",
        alpha=0.2,
        linestyle=":"
    )

    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

    # Add donor counts at the right of each point
    for y, (_, row) in zip(y_positions, subset.iterrows()):
        ax.annotate(
            f"n={int(row['donor_count'])}",
            xy=(row["median"], y),
            xytext=(8, 0),
            textcoords="offset points",
            fontsize=8,
            color="dimgray",
            va="center"
        )

axes[0].set_ylabel(
    "Anatomical site",
    fontsize=11
)

fig.suptitle(
    "Anatomical-site variation in mast-cell neighborhood enrichment",
    fontsize=15,
    fontweight="bold",
    y=0.99
)

fig.text(
    0.5,
    0.015,
    "Points represent median donor-level enrichment; "
    "horizontal bars show the interquartile range. "
    "Dashed line indicates the null expectation (1). "
    f"Only sites with at least {MIN_DONORS} donors are shown.",
    ha="center",
    fontsize=9,
    color="dimgray"
)

plt.tight_layout(rect=[0, 0.045, 1, 0.96])

png_path = out_dir / "Figure_2_Anatomical_Site_Enrichment.png"
pdf_path = out_dir / "Figure_2_Anatomical_Site_Enrichment.pdf"

fig.savefig(
    png_path,
    dpi=400,
    bbox_inches="tight"
)

fig.savefig(
    pdf_path,
    bbox_inches="tight"
)

plt.show()

print("FIGURE 2 SAVED")
print(png_path)
print(pdf_path)

print("\nSITES INCLUDED")
print(
    site_plot_df.groupby("population")["anatomic_site"]
    .nunique()
)

print("\nSITE SUMMARY")
display(
    site_plot_df.sort_values(
        ["population", "median"],
        ascending=[True, False]
    ).round(3)
)
```

```python
# ============================================================
# VASCULAR-ADJUSTED ANATOMICAL-SITE ANALYSIS
# DONOR-LEVEL SUMMARY
# ============================================================

import numpy as np
import pandas as pd

# Recover anatomical-site metadata for each sample
site_metadata = (
    site_sample_df[
        ["sample_code", "donor_code", "site_code", "anatomic_site"]
    ]
    .drop_duplicates()
)

if site_metadata["sample_code"].duplicated().any():
    raise ValueError(
        "A sample maps to multiple anatomical sites or donors."
    )

vascular_site_samples = vascular_sample_df.merge(
    site_metadata,
    on=["sample_code", "donor_code"],
    how="left",
    validate="one_to_one"
)

if vascular_site_samples["anatomic_site"].isna().any():
    raise ValueError(
        "Some vascular-adjusted samples lack anatomical-site metadata."
    )

# Aggregate multiple samples from the same donor and site
vascular_donor_site_df = (
    vascular_site_samples
    .groupby(
        ["population", "anatomic_site", "donor_code"],
        as_index=False
    )
    .agg(
        enrichment_ratio=("enrichment_ratio", "median"),
        sample_count=("sample_code", "nunique")
    )
)

# Summarize sites with at least five donors
vascular_site_results_df = (
    vascular_donor_site_df
    .groupby(["population", "anatomic_site"])
    .agg(
        donor_count=("donor_code", "nunique"),
        median_enrichment=("enrichment_ratio", "median"),
        q25=("enrichment_ratio", lambda x: x.quantile(0.25)),
        q75=("enrichment_ratio", lambda x: x.quantile(0.75)),
        donors_above_1=("enrichment_ratio", lambda x: int((x > 1).sum())),
        donors_below_1=("enrichment_ratio", lambda x: int((x < 1).sum())),
        donors_equal_1=("enrichment_ratio", lambda x: int((x == 1).sum()))
    )
    .reset_index()
)

vascular_site_results_df["donor_direction_percent"] = (
    100 * vascular_site_results_df["donors_above_1"]
    / vascular_site_results_df["donor_count"]
)

eligible_vascular_sites = vascular_site_results_df[
    vascular_site_results_df["donor_count"] >= 5
].copy()

print("VASCULAR-ADJUSTED ANATOMICAL-SITE SUMMARY")
print("=" * 75)
print("Donor-site records:", len(vascular_donor_site_df))
print("Unique donors:", vascular_donor_site_df["donor_code"].nunique())
print("Unique anatomical sites:",
      vascular_donor_site_df["anatomic_site"].nunique())

for population in ["Perivasc Fib I", "Perivasc Fib II"]:

    subset = (
        eligible_vascular_sites[
            eligible_vascular_sites["population"] == population
        ]
        .sort_values("median_enrichment", ascending=False)
    )

    print(f"\n{population}")
    print("-" * 65)
    display(
        subset[
            [
                "anatomic_site",
                "donor_count",
                "median_enrichment",
                "q25",
                "q75",
                "donors_above_1",
                "donors_below_1",
                "donor_direction_percent"
            ]
        ].round(3)
    )

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Vascular_Adjusted_Anatomical_Site_Summary.csv"
)

vascular_site_results_df.to_csv(output_path, index=False)

print("\nSaved:", output_path)
```

```python
# ============================================================
# VASCULAR-ADJUSTED ANATOMICAL-SITE ANALYSIS
# CORRECTED SAMPLE METADATA MERGE
# ============================================================

import numpy as np
import pandas as pd

site_metadata = (
    site_sample_df[
        ["sample_code", "donor_code", "site_code", "anatomic_site"]
    ]
    .drop_duplicates()
    .copy()
)

# Verify one anatomical-site assignment per sample
if site_metadata["sample_code"].duplicated().any():
    raise ValueError(
        "A sample maps to multiple anatomical sites or donors."
    )

# Each sample has two population records in vascular_sample_df
vascular_site_samples = vascular_sample_df.merge(
    site_metadata,
    on=["sample_code", "donor_code"],
    how="left",
    validate="many_to_one"
)

if vascular_site_samples["anatomic_site"].isna().any():
    raise ValueError(
        "Some vascular-adjusted samples lack site metadata."
    )

if len(vascular_site_samples) != len(vascular_sample_df):
    raise ValueError("Unexpected row-count change after merging.")

# Aggregate samples within each donor, site, and population
vascular_donor_site_df = (
    vascular_site_samples
    .groupby(
        ["population", "anatomic_site", "donor_code"],
        as_index=False
    )
    .agg(
        enrichment_ratio=("enrichment_ratio", "median"),
        sample_count=("sample_code", "nunique")
    )
)

# Site-level descriptive summaries
vascular_site_results_df = (
    vascular_donor_site_df
    .groupby(["population", "anatomic_site"])
    .agg(
        donor_count=("donor_code", "nunique"),
        median_enrichment=("enrichment_ratio", "median"),
        q25=("enrichment_ratio", lambda x: x.quantile(0.25)),
        q75=("enrichment_ratio", lambda x: x.quantile(0.75)),
        donors_above_1=("enrichment_ratio", lambda x: int((x > 1).sum())),
        donors_below_1=("enrichment_ratio", lambda x: int((x < 1).sum())),
        donors_equal_1=("enrichment_ratio", lambda x: int((x == 1).sum()))
    )
    .reset_index()
)

vascular_site_results_df["donor_direction_percent"] = (
    100 * vascular_site_results_df["donors_above_1"]
    / vascular_site_results_df["donor_count"]
)

eligible_vascular_sites = vascular_site_results_df[
    vascular_site_results_df["donor_count"] >= 5
].copy()

print("VASCULAR-ADJUSTED ANATOMICAL-SITE SUMMARY")
print("=" * 75)
print("Sample-population records:", len(vascular_site_samples))
print("Donor-site records:", len(vascular_donor_site_df))
print("Unique donors:", vascular_donor_site_df["donor_code"].nunique())
print("Unique sites:", vascular_donor_site_df["anatomic_site"].nunique())

for population in ["Perivasc Fib I", "Perivasc Fib II"]:

    subset = (
        eligible_vascular_sites[
            eligible_vascular_sites["population"] == population
        ]
        .sort_values("median_enrichment", ascending=False)
    )

    print(f"\n{population}")
    display(
        subset[
            [
                "anatomic_site",
                "donor_count",
                "median_enrichment",
                "q25",
                "q75",
                "donors_above_1",
                "donors_below_1",
                "donor_direction_percent"
            ]
        ].round(3)
    )

output_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Vascular_Adjusted_Anatomical_Site_Summary.csv"
)

vascular_site_results_df.to_csv(output_path, index=False)

print("\nSaved:", output_path)
```

```python
# ============================================================
# CELL 1: ANATOMICAL-SITE ROBUSTNESS
# DEGREE-ADJUSTED VS VASCULAR-ADJUSTED
# ============================================================

import numpy as np
import pandas as pd
from scipy.stats import spearmanr
from pathlib import Path

out_dir = Path(
    "/content/drive/MyDrive/Neurothesis Resoure Library"
)

# Prepare donor-site records from both null models
degree_site = donor_site_df[
    ["population", "anatomic_site", "donor_code", "enrichment_ratio"]
].rename(columns={
    "enrichment_ratio": "degree_enrichment"
})

vascular_site = vascular_donor_site_df[
    ["population", "anatomic_site", "donor_code", "enrichment_ratio"]
].rename(columns={
    "enrichment_ratio": "vascular_enrichment"
})

keys = ["population", "anatomic_site", "donor_code"]

# Verify unique donor-site-population observations
for name, df in [
    ("Degree-adjusted", degree_site),
    ("Vascular-adjusted", vascular_site)
]:
    if df.duplicated(keys).any():
        raise ValueError(
            f"Duplicate donor-site records in {name} data."
        )

# Match the same donors at the same sites
site_comparison_df = degree_site.merge(
    vascular_site,
    on=keys,
    how="inner",
    validate="one_to_one"
)

print("ANATOMICAL-SITE ROBUSTNESS COMPARISON")
print("=" * 75)
print("Matched donor-site-population records:",
      len(site_comparison_df))
print("Matched donors:",
      site_comparison_df["donor_code"].nunique())
print("Matched anatomical sites:",
      site_comparison_df["anatomic_site"].nunique())

site_comparison_df["vascular_minus_degree"] = (
    site_comparison_df["vascular_enrichment"]
    - site_comparison_df["degree_enrichment"]
)

# Compare site summaries
site_robustness_rows = []

for population in ["Perivasc Fib I", "Perivasc Fib II"]:

    subset = site_comparison_df[
        site_comparison_df["population"] == population
    ].copy()

    rho, corr_p = spearmanr(
        subset["degree_enrichment"],
        subset["vascular_enrichment"]
    )

    print(f"\n{population}")
    print("-" * 65)
    print("Matched donor-site records:", len(subset))
    print("Spearman correlation:", round(float(rho), 4))
    print("Correlation p-value:", corr_p)

    site_summary = (
        subset.groupby("anatomic_site")
        .agg(
            donor_count=("donor_code", "nunique"),
            degree_median=("degree_enrichment", "median"),
            vascular_median=("vascular_enrichment", "median"),
            median_change=("vascular_minus_degree", "median"),
            degree_above_1=("degree_enrichment",
                            lambda x: int((x > 1).sum())),
            vascular_above_1=("vascular_enrichment",
                              lambda x: int((x > 1).sum()))
        )
        .reset_index()
    )

    site_summary["population"] = population

    site_summary["direction_changed"] = (
        (site_summary["degree_median"] > 1)
        != (site_summary["vascular_median"] > 1)
    )

    site_robustness_rows.append(site_summary)

    display(
        site_summary.sort_values(
            "vascular_median",
            ascending=False
        ).round(3)
    )

site_robustness_df = pd.concat(
    site_robustness_rows,
    ignore_index=True
)

print("\nSITE-LEVEL DIRECTION CHANGES")
print("=" * 75)

for population in ["Perivasc Fib I", "Perivasc Fib II"]:

    subset = site_robustness_df[
        site_robustness_df["population"] == population
    ]

    changed = subset[subset["direction_changed"]]

    print(f"\n{population}")
    print("Sites changing median direction relative to 1:",
          len(changed))

    if len(changed):
        display(
            changed[
                [
                    "anatomic_site",
                    "degree_median",
                    "vascular_median",
                    "median_change"
                ]
            ].round(3)
        )
    else:
        print("No sites changed direction.")

# Save matched records and site summaries
site_comparison_df.to_csv(
    out_dir / "Anatomical_Site_Null_Model_Matched_Records.csv",
    index=False
)

site_robustness_df.to_csv(
    out_dir / "Anatomical_Site_Null_Model_Robustness.csv",
    index=False
)

print("\nSaved matched records and site-level robustness summaries.")
```

```python
# ============================================================
# FINAL SENSITIVITY CHECK:
# MINIMUM TARGET-CELL COUNT THRESHOLDS
# ============================================================

import numpy as np
import pandas as pd
from scipy.stats import wilcoxon

thresholds = [0, 10, 20, 50, 100]

datasets = {
    "Degree-adjusted": degree_dermal_sample_df,
    "Vascular-adjusted": vascular_sample_df
}

sensitivity_rows = []

for method, df in datasets.items():

    print("\n" + "=" * 75)
    print(method.upper())
    print("=" * 75)

    for population in ["Perivasc Fib I", "Perivasc Fib II"]:

        pop_df = df[df["population"] == population].copy()

        print(f"\n{population}")
        print("-" * 65)

        for threshold in thresholds:

            eligible = pop_df[
                pop_df["target_cell_count"] >= threshold
            ].copy()

            donor_medians = (
                eligible.groupby("donor_code")["enrichment_ratio"]
                .median()
            )

            n_donors = len(donor_medians)

            median_enrichment = (
                float(donor_medians.median())
                if n_donors else np.nan
            )

            donors_above = int((donor_medians > 1).sum())
            donors_below = int((donor_medians < 1).sum())

            if n_donors >= 2:
                try:
                    p_value = wilcoxon(
                        donor_medians - 1,
                        alternative="greater",
                        method="auto"
                    ).pvalue
                except ValueError:
                    p_value = np.nan
            else:
                p_value = np.nan

            sensitivity_rows.append({
                "method": method,
                "population": population,
                "minimum_target_cells": threshold,
                "eligible_samples": len(eligible),
                "donors": n_donors,
                "median_donor_enrichment": median_enrichment,
                "donors_above_1": donors_above,
                "donors_below_1": donors_below,
                "one_sided_wilcoxon_p": p_value
            })

            print(
                f"Minimum cells {threshold:>3}: "
                f"samples={len(eligible):>3}, "
                f"donors={n_donors:>2}, "
                f"median={median_enrichment:.4f}, "
                f"above 1={donors_above:>2}, "
                f"below 1={donors_below:>2}, "
                f"p={p_value:.6g}"
                if n_donors else
                f"Minimum cells {threshold:>3}: no eligible donors"
            )

sensitivity_threshold_df = pd.DataFrame(sensitivity_rows)

# Save results
out_path = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/"
    "Target_Cell_Count_Threshold_Sensitivity.csv"
)

sensitivity_threshold_df.to_csv(out_path, index=False)

print("\n" + "=" * 75)
print("SAVED:", out_path)
display(sensitivity_threshold_df.round(5))
```

```python
# ============================================================
# FINAL STATISTICAL SUMMARY
# ============================================================

import numpy as np
import pandas as pd
from scipy.stats import spearmanr
from pathlib import Path

out_dir = Path(
    "/content/drive/MyDrive/Neurothesis Resoure Library"
)

# ------------------------------------------------------------
# 1. PRIMARY AND COUNT-THRESHOLD SENSITIVITY RESULTS
# ------------------------------------------------------------

selected_thresholds = [0, 50, 100]

final_threshold_summary = sensitivity_threshold_df[
    sensitivity_threshold_df["minimum_target_cells"].isin(
        selected_thresholds
    )
].copy()

final_threshold_summary = final_threshold_summary.sort_values(
    ["population", "method", "minimum_target_cells"]
)

print("DONOR-LEVEL ENRICHMENT ROBUSTNESS")
print("=" * 85)

display(
    final_threshold_summary[
        [
            "method",
            "population",
            "minimum_target_cells",
            "eligible_samples",
            "donors",
            "median_donor_enrichment",
            "donors_above_1",
            "donors_below_1",
            "one_sided_wilcoxon_p"
        ]
    ].round(5)
)

# ------------------------------------------------------------
# 2. ANATOMICAL-SITE ROBUSTNESS
# ------------------------------------------------------------

site_summary_rows = []

for population in ["Perivasc Fib I", "Perivasc Fib II"]:

    subset = site_comparison_df[
        site_comparison_df["population"] == population
    ].copy()

    rho, corr_p = spearmanr(
        subset["degree_enrichment"],
        subset["vascular_enrichment"]
    )

    site_table = site_robustness_df[
        site_robustness_df["population"] == population
    ]

    site_summary_rows.append({
        "population": population,
        "matched_donor_site_records": len(subset),
        "matched_donors": subset["donor_code"].nunique(),
        "anatomical_sites": subset["anatomic_site"].nunique(),
        "degree_vs_vascular_spearman_rho": rho,
        "spearman_p_descriptive": corr_p,
        "sites_with_median_above_1_degree": int(
            (site_table["degree_median"] > 1).sum()
        ),
        "sites_with_median_above_1_vascular": int(
            (site_table["vascular_median"] > 1).sum()
        ),
        "sites_changing_direction": int(
            site_table["direction_changed"].sum()
        )
    })

final_site_summary = pd.DataFrame(site_summary_rows)

print("\nANATOMICAL-SITE ROBUSTNESS")
print("=" * 85)
display(final_site_summary.round(5))

# ------------------------------------------------------------
# 3. SAVE FINAL SUMMARY TABLES
# ------------------------------------------------------------

threshold_path = (
    out_dir / "Final_Donor_Enrichment_Summary.csv"
)

site_path = (
    out_dir / "Final_Anatomical_Site_Robustness_Summary.csv"
)

final_threshold_summary.to_csv(
    threshold_path,
    index=False
)

final_site_summary.to_csv(
    site_path,
    index=False
)

# ------------------------------------------------------------
# 4. PRINT CONCISE INTERPRETIVE AUDIT
# ------------------------------------------------------------

print("\nFINAL ROBUSTNESS AUDIT")
print("=" * 85)

for population in ["Perivasc Fib I", "Perivasc Fib II"]:

    subset = final_threshold_summary[
        (final_threshold_summary["population"] == population)
        & (final_threshold_summary["method"] == "Vascular-adjusted")
    ].sort_values("minimum_target_cells")

    print(f"\n{population}")

    for _, row in subset.iterrows():
        print(
            f"  >= {int(row['minimum_target_cells'])} target cells: "
            f"median enrichment {row['median_donor_enrichment']:.3f}; "
            f"{int(row['donors_above_1'])}/"
            f"{int(row['donors'])} donors above 1"
        )

print("\nSaved:")
print(threshold_path)
print(site_path)
```

```python
 # ============================================================
# FINAL INTEGRATED FIGURE:
# ANATOMICAL-SITE ENRICHMENT ROBUSTNESS
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

out_dir = Path(
    "/content/drive/MyDrive/Neurothesis Resoure Library"
)

# ------------------------------------------------------------
# 1. BUILD SITE SUMMARIES
# ------------------------------------------------------------

def summarize_sites(df, method_name):
    summary = (
        df.groupby(["population", "anatomic_site"])
        ["enrichment_ratio"]
        .agg(
            median="median",
            q25=lambda x: x.quantile(0.25),
            q75=lambda x: x.quantile(0.75),
            donor_count="count"
        )
        .reset_index()
    )

    summary["method"] = method_name
    return summary


degree_summary = summarize_sites(
    donor_site_df,
    "Degree-adjusted"
)

vascular_summary = summarize_sites(
    vascular_donor_site_df,
    "Vascular-adjusted"
)

figure_summary = pd.concat(
    [degree_summary, vascular_summary],
    ignore_index=True
)

# Keep sites represented by at least five donors
figure_summary = figure_summary[
    figure_summary["donor_count"] >= 5
].copy()

# Confirm the number of sites per population
print("SITES INCLUDED IN FIGURE")
print("=" * 65)

for population in ["Perivasc Fib I", "Perivasc Fib II"]:
    subset = figure_summary[
        (figure_summary["population"] == population)
        & (figure_summary["method"] == "Vascular-adjusted")
    ]
    print(population, ":", len(subset), "sites")

# ------------------------------------------------------------
# 2. CREATE TWO-PANEL FIGURE
# ------------------------------------------------------------

fig, axes = plt.subplots(
    1, 2,
    figsize=(17, 9),
    sharex=True
)

populations = [
    "Perivasc Fib I",
    "Perivasc Fib II"
]

method_styles = {
    "Degree-adjusted": {
        "color": "#356A9A",
        "offset": -0.13
    },
    "Vascular-adjusted": {
        "color": "#D17A32",
        "offset": 0.13
    }
}

for ax, population in zip(axes, populations):

    pop_data = figure_summary[
        figure_summary["population"] == population
    ]

    # Order sites by vascular-adjusted median enrichment
    ordering = (
        pop_data[
            pop_data["method"] == "Vascular-adjusted"
        ]
        .sort_values("median", ascending=True)
        ["anatomic_site"]
        .tolist()
    )

    y_positions = np.arange(len(ordering))

    for method, style in method_styles.items():

        method_data = (
            pop_data[pop_data["method"] == method]
            .set_index("anatomic_site")
            .reindex(ordering)
        )

        medians = method_data["median"].to_numpy()
        lower = medians - method_data["q25"].to_numpy()
        upper = method_data["q75"].to_numpy() - medians

        y = y_positions + style["offset"]

        ax.errorbar(
            medians,
            y,
            xerr=np.vstack([lower, upper]),
            fmt="o",
            color=style["color"],
            ecolor=style["color"],
            markersize=7,
            capsize=3,
            elinewidth=1.8,
            label=method,
            zorder=3
        )

    ax.axvline(
        1.0,
        color="black",
        linestyle="--",
        linewidth=1.2,
        alpha=0.75,
        label="Expected abundance (1.0)"
    )

    ax.set_yticks(y_positions)
    ax.set_yticklabels(ordering)
    ax.set_title(
        population,
        fontsize=14,
        fontweight="bold",
        pad=12
    )

    ax.set_xlabel(
        "Enrichment relative to null expectation",
        fontsize=11
    )

    ax.grid(
        axis="x",
        linestyle=":",
        alpha=0.35
    )

    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

axes[0].set_ylabel(
    "Anatomical site",
    fontsize=11
)

# Shared legend
handles, labels = axes[0].get_legend_handles_labels()

fig.legend(
    handles,
    labels,
    loc="upper center",
    bbox_to_anchor=(0.5, 0.965),
    ncol=3,
    frameon=False,
    fontsize=10
)

fig.suptitle(
    "Anatomical-Site Enrichment of Perivascular Fibroblast Populations",
    fontsize=16,
    fontweight="bold",
    y=1.02
)

fig.text(
    0.5,
    0.015,
    "Points represent donor-level median enrichment; horizontal bars show "
    "the interquartile range. Sites require at least five donors. "
    "The dashed line marks the null expectation. Adjustment methods "
    "are alternative analyses of the same dataset.",
    ha="center",
    fontsize=9,
    wrap=True
)

plt.tight_layout(rect=[0, 0.06, 1, 0.91])

# ------------------------------------------------------------
# 3. SAVE HIGH-RESOLUTION OUTPUTS
# ------------------------------------------------------------

png_path = (
    out_dir / "Figure_3_Integrated_Anatomical_Site_Robustness.png"
)

pdf_path = (
    out_dir / "Figure_3_Integrated_Anatomical_Site_Robustness.pdf"
)

fig.savefig(
    png_path,
    dpi=400,
    bbox_inches="tight"
)

fig.savefig(
    pdf_path,
    bbox_inches="tight"
)

plt.show()

print("\nSAVED FINAL FIGURES")
print(png_path)
print(pdf_path)
```

```python
# ============================================================
# INVESTIGATE ZERO ENRICHMENT VALUES
# ============================================================

import numpy as np
import pandas as pd

datasets = {
    "Degree-adjusted": degree_dermal_sample_df,
    "Vascular-adjusted": vascular_sample_df
}

for method, df in datasets.items():

    print("\n" + "=" * 75)
    print(method.upper())
    print("=" * 75)

    zero_rows = df[df["enrichment_ratio"] == 0].copy()
    negative_rows = df[df["enrichment_ratio"] < 0].copy()

    print("Zero enrichment records:", len(zero_rows))
    print("Negative enrichment records:", len(negative_rows))

    if len(zero_rows):
        print("\nZERO-ENRICHMENT RECORDS")
        display(
            zero_rows[
                [
                    "sample_code",
                    "donor_code",
                    "population",
                    "target_cell_count",
                    "observed_neighbor_fraction",
                    "null_mean_neighbor_fraction",
                    "enrichment_ratio",
                    "null_sd"
                ]
            ]
        )

        valid_zero = (
            (zero_rows["observed_neighbor_fraction"] == 0)
            & (zero_rows["null_mean_neighbor_fraction"] > 0)
            & (zero_rows["target_cell_count"] > 0)
        )

        print("Mathematically valid zero records:",
              int(valid_zero.sum()), "/", len(zero_rows))

    if len(negative_rows):
        print("\nNEGATIVE RECORDS")
        display(negative_rows)

    # Check ratio arithmetic
    valid_denominator = (
        np.isfinite(df["null_mean_neighbor_fraction"])
        & (df["null_mean_neighbor_fraction"] > 0)
    )

    recalculated = (
        df.loc[valid_denominator, "observed_neighbor_fraction"]
        / df.loc[valid_denominator, "null_mean_neighbor_fraction"]
    )

    stored = df.loc[valid_denominator, "enrichment_ratio"]

    max_error = np.max(
        np.abs(recalculated.to_numpy() - stored.to_numpy())
    )

    print("\nMaximum ratio reconstruction error:",
          f"{max_error:.12g}")

# Check donor-site summaries for zero medians
print("\n" + "=" * 75)
print("DONOR-SITE ZERO ENRICHMENT AUDIT")
print("=" * 75)

for method, df in [
    ("Degree-adjusted", donor_site_df),
    ("Vascular-adjusted", vascular_donor_site_df)
]:
    zeros = df[df["enrichment_ratio"] == 0]

    print(
        method,
        "| zero donor-site records:",
        len(zeros),
        "| negative records:",
        int((df["enrichment_ratio"] < 0).sum())
    )

    if len(zeros):
        display(
            zeros[
                ["population", "anatomic_site",
                 "donor_code", "enrichment_ratio"]
            ]
        )
```

```python
# ============================================================
# CORRECTED FINAL REPRODUCIBILITY AUDIT
# ZERO ENRICHMENT IS VALID
# ============================================================

import numpy as np
import pandas as pd
from pathlib import Path

out_dir = Path(
    "/content/drive/MyDrive/Neurothesis Resoure Library"
)

audit_results = []

def record_check(name, passed, detail):
    audit_results.append({
        "check": name,
        "passed": bool(passed),
        "detail": str(detail)
    })
    print(f"[{'PASS' if passed else 'CHECK'}] {name}: {detail}")


print("=" * 75)
print("CORRECTED NEUROTHESIS REPRODUCIBILITY AUDIT")
print("=" * 75)

# ------------------------------------------------------------
# 1. SAMPLE-LEVEL RECORD MATCHING
# ------------------------------------------------------------

sample_keys = ["sample_code", "donor_code", "population"]

degree_keys = set(
    map(tuple, degree_dermal_sample_df[sample_keys].to_numpy())
)

vascular_keys = set(
    map(tuple, vascular_sample_df[sample_keys].to_numpy())
)

record_check(
    "Degree and vascular sample records match",
    degree_keys == vascular_keys,
    f"Degree={len(degree_keys)}, vascular={len(vascular_keys)}"
)

for name, df in [
    ("Degree", degree_dermal_sample_df),
    ("Vascular", vascular_sample_df)
]:
    record_check(
        f"No duplicate {name.lower()} sample records",
        not df.duplicated(sample_keys).any(),
        f"Rows={len(df)}"
    )

# ------------------------------------------------------------
# 2. DONOR-SITE RECORD MATCHING
# ------------------------------------------------------------

donor_keys = ["population", "anatomic_site", "donor_code"]

for name, df in [
    ("Degree", donor_site_df),
    ("Vascular", vascular_donor_site_df)
]:
    record_check(
        f"{name} donor-site records are unique",
        not df.duplicated(donor_keys).any(),
        f"Rows={len(df)}"
    )

record_check(
    "Matched donor-site comparison is complete",
    len(site_comparison_df) == len(donor_site_df)
        == len(vascular_donor_site_df),
    f"Matched={len(site_comparison_df)}, "
    f"degree={len(donor_site_df)}, "
    f"vascular={len(vascular_donor_site_df)}"
)

# ------------------------------------------------------------
# 3. PERMUTATION AND RETENTION CHECKS
# ------------------------------------------------------------

for name, df in [
    ("Degree", degree_dermal_sample_df),
    ("Vascular", vascular_sample_df)
]:
    record_check(
        f"{name} permutation count is 999",
        df["permutations"].eq(999).all(),
        f"Values={sorted(df['permutations'].unique())}"
    )

record_check(
    "All vascular target cells retained",
    (
        vascular_sample_df["target_cell_count"]
        == vascular_sample_df["retained_target_cells"]
    ).all(),
    f"Target={vascular_sample_df['target_cell_count'].sum():,}; "
    f"retained={vascular_sample_df['retained_target_cells'].sum():,}"
)

# ------------------------------------------------------------
# 4. CORRECTED ENRICHMENT VALIDATION
# ------------------------------------------------------------

def validate_enrichment(df):
    observed = df["observed_neighbor_fraction"].to_numpy(float)
    expected = df["null_mean_neighbor_fraction"].to_numpy(float)
    ratio = df["enrichment_ratio"].to_numpy(float)

    finite = (
        np.isfinite(observed).all()
        and np.isfinite(expected).all()
        and np.isfinite(ratio).all()
    )

    nonnegative = (
        (observed >= 0).all()
        and (expected > 0).all()
        and (ratio >= 0).all()
    )

    if not finite or not nonnegative:
        return False, "Nonfinite or invalid values"

    reconstructed = observed / expected
    max_error = np.max(np.abs(reconstructed - ratio))

    return (
        np.allclose(reconstructed, ratio, rtol=1e-7, atol=1e-10),
        f"Valid zeros={(ratio == 0).sum()}, "
        f"negative values={(ratio < 0).sum()}, "
        f"max reconstruction error={max_error:.3g}"
    )


for name, df in [
    ("Degree sample enrichment", degree_dermal_sample_df),
    ("Vascular sample enrichment", vascular_sample_df)
]:
    passed, detail = validate_enrichment(df)
    record_check(name, passed, detail)

# Donor-site records are aggregated ratios.
for name, df in [
    ("Degree donor-site enrichment", donor_site_df),
    ("Vascular donor-site enrichment", vascular_donor_site_df)
]:
    values = df["enrichment_ratio"].to_numpy(float)

    passed = (
        np.isfinite(values).all()
        and (values >= 0).all()
    )

    detail = (
        f"Rows={len(values)}, "
        f"valid zeros={(values == 0).sum()}, "
        f"negative values={(values < 0).sum()}"
    )

    record_check(name, passed, detail)

# ------------------------------------------------------------
# 5. POPULATION AND COVERAGE CHECKS
# ------------------------------------------------------------

expected_populations = {"Perivasc Fib I", "Perivasc Fib II"}

record_check(
    "Expected populations represented",
    set(vascular_sample_df["population"].unique())
        == expected_populations,
    f"Observed={sorted(vascular_sample_df['population'].unique())}"
)

record_check(
    "Expected donor coverage",
    vascular_sample_df["donor_code"].nunique() == 22,
    f"Donors={vascular_sample_df['donor_code'].nunique()}"
)

record_check(
    "Expected sample coverage",
    vascular_sample_df["sample_code"].nunique() == 112,
    f"Samples={vascular_sample_df['sample_code'].nunique()}"
)

# ------------------------------------------------------------
# 6. OUTPUT FILE CHECK
# ------------------------------------------------------------

expected_files = [
    "Figure_1_Donor_Enrichment_Robustness.png",
    "Figure_1_Donor_Enrichment_Robustness.pdf",
    "Figure_2_Anatomical_Site_Enrichment.png",
    "Figure_2_Anatomical_Site_Enrichment.pdf",
    "Figure_3_Integrated_Anatomical_Site_Robustness.png",
    "Figure_3_Integrated_Anatomical_Site_Robustness.pdf",
    "Final_Donor_Enrichment_Summary.csv",
    "Final_Anatomical_Site_Robustness_Summary.csv",
    "Target_Cell_Count_Threshold_Sensitivity.csv",
    "Anatomical_Site_Null_Model_Robustness.csv",
    "Anatomical_Site_Null_Model_Matched_Records.csv"
]

missing_files = [
    f for f in expected_files if not (out_dir / f).is_file()
]

record_check(
    "Expected output files exist",
    len(missing_files) == 0,
    f"Missing files: {missing_files if missing_files else 'None'}"
)

# ------------------------------------------------------------
# 7. SAVE CORRECTED AUDIT
# ------------------------------------------------------------

corrected_audit_df = pd.DataFrame(audit_results)

audit_path = out_dir / "Final_Reproducibility_Audit.csv"

corrected_audit_df.to_csv(audit_path, index=False)

print("\n" + "=" * 75)
print("CORRECTED AUDIT SUMMARY")
print("=" * 75)

print("Checks passed:", int(corrected_audit_df["passed"].sum()))
print("Checks requiring review:",
      int((~corrected_audit_df["passed"]).sum()))
print("Total checks:", len(corrected_audit_df))

display(corrected_audit_df)

print("\nSaved corrected audit:")
print(audit_path)
```

```python

# ============================================================
# STEP 6: RECONSTRUCT AND RECONCILE DONOR-LEVEL ENRICHMENT
# ============================================================

import os
import numpy as np
import pandas as pd

base_dir = "/content/drive/MyDrive/Neurothesis Resoure Library"

sample_path = os.path.join(
    base_dir, "Within_Dermal_Degree_Stratified_Sample_Results.csv"
)
donor_path = os.path.join(
    base_dir, "Within_Dermal_Degree_Stratified_Donor_Results.csv"
)

sample = pd.read_csv(sample_path)
donor = pd.read_csv(donor_path)

keys = ["donor_code", "population"]

# Reconstruct donor-level values from sample-level records
def weighted_mean(group):
    weights = group["target_cell_count"].to_numpy(dtype=float)
    values = group["enrichment_ratio"].to_numpy(dtype=float)
    return np.average(values, weights=weights)

reconstructed = (
    sample.groupby(keys)
    .apply(
        lambda g: pd.Series({
            "sample_rows": len(g),
            "sample_median": g["enrichment_ratio"].median(),
            "sample_mean": g["enrichment_ratio"].mean(),
            "target_weighted_mean": weighted_mean(g),
            "total_target_cells": g["target_cell_count"].sum()
        }),
        include_groups=False
    )
    .reset_index()
)

comparison = donor.merge(
    reconstructed,
    on=keys,
    how="outer",
    validate="one_to_one"
)

comparison["stored_minus_sample_median"] = (
    comparison["enrichment_ratio"] - comparison["sample_median"]
)
comparison["stored_minus_sample_mean"] = (
    comparison["enrichment_ratio"] - comparison["sample_mean"]
)
comparison["stored_minus_weighted_mean"] = (
    comparison["enrichment_ratio"] - comparison["target_weighted_mean"]
)

print("=" * 85)
print("DONOR AGGREGATION RECONCILIATION")
print("=" * 85)

print("\nSample-count agreement:")
print(
    comparison[
        ["donor_code", "population", "sample_count", "sample_rows"]
    ].to_string(index=False)
)

print("\nMaximum absolute difference by candidate aggregation:")
for label, col in [
    ("Median", "stored_minus_sample_median"),
    ("Mean", "stored_minus_sample_mean"),
    ("Target-weighted mean", "stored_minus_weighted_mean")
]:
    print(
        label + ":",
        round(comparison[col].abs().max(), 10)
    )

print("\nFull comparison:")
print(
    comparison[
        [
            "donor_code", "population",
            "enrichment_ratio",
            "sample_median",
            "sample_mean",
            "target_weighted_mean",
            "sample_count", "sample_rows"
        ]
    ].round(6).to_string(index=False)
)

print("\nOverall donor-level medians:")
print(
    comparison.groupby("population")[
        [
            "enrichment_ratio",
            "sample_median",
            "sample_mean",
            "target_weighted_mean"
        ]
    ].median().round(6).to_string()
)

print("\n" + "=" * 85)
print("RECONSTRUCTION COMPLETE")
print("=" * 85)
```

```python

# ============================================================
# STEP 7: FINAL DONOR-LEVEL INFERENCE
# Uses one median enrichment value per donor and population
# ============================================================

import os
import numpy as np
import pandas as pd
from scipy.stats import wilcoxon, binomtest

base_dir = "/content/drive/MyDrive/Neurothesis Resoure Library"
path = os.path.join(
    base_dir, "Within_Dermal_Degree_Stratified_Donor_Results.csv"
)

df = pd.read_csv(path)
rng = np.random.default_rng(20261111)
B = 20000

wide = df.pivot(
    index="donor_code",
    columns="population",
    values="enrichment_ratio"
).dropna()

fib1 = wide["Perivasc Fib I"].to_numpy()
fib2 = wide["Perivasc Fib II"].to_numpy()

def bootstrap_median_ci(values, n_boot=B):
    boot = np.empty(n_boot)
    for i in range(n_boot):
        boot[i] = np.median(
            rng.choice(values, size=len(values), replace=True)
        )
    return np.quantile(boot, [0.025, 0.975])

def loo_median_range(values):
    vals = [
        np.median(np.delete(values, i))
        for i in range(len(values))
    ]
    return min(vals), max(vals)

print("=" * 78)
print("FINAL DONOR-LEVEL INFERENCE")
print("=" * 78)

p_values = []

for name, values in [
    ("Perivasc Fib I", fib1),
    ("Perivasc Fib II", fib2)
]:
    median = np.median(values)
    ci_low, ci_high = bootstrap_median_ci(values)
    p = wilcoxon(
        values - 1,
        alternative="greater",
        method="auto"
    ).pvalue
    loo_low, loo_high = loo_median_range(values)

    p_values.append(p)

    print(f"\n{name}")
    print("-" * 50)
    print("Donors:", len(values))
    print("Median donor enrichment:", round(median, 6))
    print("Bootstrap 95% CI:", round(ci_low, 6), "to", round(ci_high, 6))
    print("Donors above 1:", int(np.sum(values > 1)))
    print("Donors below 1:", int(np.sum(values < 1)))
    print("One-sided Wilcoxon p:", p)
    print("Leave-one-donor-out median range:",
          round(loo_low, 6), "to", round(loo_high, 6))

# Benjamini-Hochberg correction across the two primary population tests
p_array = np.array(p_values)
order = np.argsort(p_array)
adjusted = np.empty(2)
ranked = p_array[order] * 2 / np.arange(1, 3)
ranked = np.minimum.accumulate(ranked[::-1])[::-1]
adjusted[order] = np.minimum(ranked, 1)

print("\nBH-adjusted p-values:")
print("Perivasc Fib I:", adjusted[0])
print("Perivasc Fib II:", adjusted[1])

# Paired donor comparison between Fib II and Fib I
difference = fib2 - fib1
paired_p = wilcoxon(
    difference,
    alternative="two-sided",
    method="auto"
).pvalue

boot_diff = np.empty(B)
for i in range(B):
    sampled = rng.choice(
        difference, size=len(difference), replace=True
    )
    boot_diff[i] = np.median(sampled)

diff_ci = np.quantile(boot_diff, [0.025, 0.975])

print("\n" + "=" * 78)
print("PAIRED FIB II VERSUS FIB I COMPARISON")
print("=" * 78)
print("Median paired difference (Fib II - Fib I):",
      round(np.median(difference), 6))
print("Bootstrap 95% CI:",
      round(diff_ci[0], 6), "to", round(diff_ci[1], 6))
print("Donors Fib II > Fib I:", int(np.sum(difference > 0)))
print("Donors Fib II < Fib I:", int(np.sum(difference < 0)))
print("Two-sided paired Wilcoxon p:", paired_p)

print("\n" + "=" * 78)
print("FINAL INFERENCE COMPLETE")
print("=" * 78)
```

# Research Synthesis and Interpretation
**Date:** November 11, 2026

## 1. Primary Research Question

How are mast cells spatially organized in relation to nerve-associated and perivascular cell populations within human skin, and do these spatial relationships vary across anatomical sites?

## 2. Dataset and Study Population

This exploratory study uses publicly available human skin MERFISH spatial transcriptomic data from Restrepo et al. (2026). The dataset contains approximately 1.2 million cells from 22 donors and multiple anatomical sites. Analyses focus on cells assigned to the dermal compartment.

## 3. Primary Spatial Enrichment Results

Degree-adjusted donor-level analyses showed that both annotated perivascular fibroblast populations had median enrichment ratios above 1 across donors.

- **Perivasc Fib I:** median enrichment 1.6414; 22 of 22 donors above 1.
- **Perivasc Fib II:** median enrichment 1.5682; 21 of 22 donors above 1.

These results indicate that both populations were more represented in mast-cell neighborhoods than expected under the degree-adjusted null model.

## 4. Robustness Analyses

Vascular-neighborhood-adjusted analyses produced broadly similar patterns:

- **Perivasc Fib I:** median enrichment approximately 1.6017; 22 of 22 donors above 1.
- **Perivasc Fib II:** median enrichment approximately 1.5765; 21 of 22 donors above 1.

These analyses provide a robustness comparison using a different null-model adjustment. They are not independent replication.

## 5. Matched-Control Results

Comparisons with sample- and graph-degree-matched non-mast-cell neighborhoods produced population-specific findings.

- Perivasc Fib I showed no statistically significant difference.
- Perivasc Fib II showed significantly greater representation around mast cells.
- Schwann cells and Retic Fib III showed lower representation around mast cells than around matched control cells.

This distinction is important: enrichment relative to a randomized null expectation does not necessarily mean enrichment relative to matched non-mast-cell neighborhoods.

## 6. Anatomical Site Variation

Enrichment patterns varied across anatomical sites. Some sites showed stronger evidence of enrichment than others. Site-level findings should be interpreted cautiously because the number of donors represented at each site differs.

## 7. Cell Identity and Annotation Limitations

The analysis uses the cell-population annotations provided in the source dataset. The dataset does not explicitly identify a distinct perineural fibroblast population. Therefore, the annotated perivascular fibroblast populations should not be described as confirmed perineural fibroblasts.

No neurons were detected in the analyzed dataset. Schwann-cell associations therefore cannot establish direct interactions with neurons.

## 8. Overall Interpretation

Across donor-level analyses, mast-cell neighborhoods showed enrichment of both annotated perivascular fibroblast populations relative to a within-dermal-degree-stratified null expectation. Similar patterns were observed after additional adjustment for vascular neighborhood composition. However, comparisons with sample- and degree-matched non-mast-cell neighborhoods showed a significant increase for Perivasc Fib II but not Perivasc Fib I. Schwann cells and Retic Fib III were less represented in mast-cell neighborhoods than in matched controls.

Together, these findings suggest that spatial relationships around mast cells are population-specific and depend partly on the comparison framework used.

## 9. Remaining Limitations

- Spatial association does not establish direct cellular interaction or causation.
- The dataset does not explicitly annotate perineural fibroblasts.
- Neurons were not detected in the analyzed data.
- Null models do not preserve every possible feature of tissue architecture.
- Matched-control and randomized-null analyses answer different questions.
- Findings from public human skin data do not independently establish a mechanism in mast cell activation syndrome (MCAS).

## 10. Next Research Steps

1. Finalize and document the spatial null-model methods.
2. Consolidate primary findings into one results table.
3. Draft the Results section.
4. Finalize figures and figure captions.
5. Prepare the analysis for mentor review.

| Perivasc Fib I | 0.024342 | 0.022240 | +0.002102 | 0.3705 | No significant difference |

**Interpretation:** The matched-control analysis identifies a population-specific pattern. Perivasc Fib II was significantly more represented in mast-cell neighborhoods, whereas Perivasc Fib I did not differ significantly. Schwann cells and Retic Fib III were less represented.

## Important Statistical Distinction

The null-model enrichment analyses and matched-control analyses answer different questions.

- Null-model analyses ask whether observed neighborhood representation exceeds a specified randomized expectation.
- Matched-control analyses ask whether mast-cell neighborhoods differ from comparable non-mast-cell neighborhoods.

Therefore, enrichment relative to a null expectation should not automatically be interpreted as enrichment relative to matched controls.

# Results

## 1. Donor-Level Spatial Enrichment of Perivascular Fibroblast Populations

Degree-adjusted spatial analyses demonstrated enrichment of both annotated perivascular fibroblast populations within mast-cell neighborhoods across human skin donors. Perivasc Fib I exhibited a median donor-level enrichment ratio of 1.6414, with all 22 donors showing enrichment ratios above 1. Perivasc Fib II exhibited a median enrichment ratio of 1.5682, with 21 of 22 donors showing enrichment ratios above 1.

The consistency of these patterns across donors suggests that the observed spatial associations were not driven solely by a small number of individuals. However, enrichment relative to the degree-adjusted null expectation does not independently establish preferential association relative to other cell neighborhoods.

## 2. Robustness to Vascular Neighborhood Adjustment

To assess whether the observed patterns were sensitive to vascular neighborhood composition, enrichment analyses were repeated using a vascular-neighborhood-adjusted null model. Both populations retained median enrichment ratios above 1. Perivasc Fib I exhibited a median ratio of 1.6017, with 22 of 22 donors above 1, while Perivasc Fib II exhibited a median ratio of 1.5765, with 21 of 22 donors above 1.

The similarity between the degree-adjusted and vascular-adjusted results indicates that the enrichment patterns were not eliminated by this additional adjustment.

## 3. Comparison With Matched Non-Mast-Cell Neighborhoods

A complementary analysis compared mast-cell neighborhoods with sample- and graph-degree-matched non-mast-cell neighborhoods. This comparison revealed population-specific differences.

Perivasc Fib II showed significantly greater neighborhood representation around mast cells than around matched control cells (mean donor-level difference = 0.005749, p = 0.000593; adjusted p = 0.000762). In contrast, Perivasc Fib I did not show a statistically significant difference (mean difference = 0.002102, p = 0.3705).

Schwann cells and Retic Fib III showed lower neighborhood representation around mast cells than around matched controls. The mean donor-level differences were −0.012499 for Schwann cells (p < 0.000001) and −0.006183 for Retic Fib III (p = 0.001091).

These findings demonstrate that enrichment relative to a randomized null expectation does not necessarily correspond to greater representation relative to matched non-mast-cell neighborhoods.

## 6. Vascular-Neighborhood-Adjusted Null Model

A stratified permutation approach was implemented to evaluate mast-cell neighborhood enrichment while accounting for candidate-cell graph degree and local vascular neighborhood composition.

Analyses were performed separately within each dermal sample using the sample-specific induced spatial graph. Mast cells were identified as neighborhood centers, and all non-mast cells were considered candidate neighboring populations.

For each candidate cell, graph degree was calculated as the number of connections within the sample-specific dermal graph. Local vascular context was quantified as the fraction of the candidate cell's graph neighbors belonging to five vascular or perivascular populations: Peri, vascular endothelial cells (VEC), high endothelial cells (HEC), lymphatic endothelial cells (LEC), and smooth muscle cells (SM).

Candidate cells were stratified independently by graph degree and vascular-context fraction using quantile-based binning with up to three bins per variable. Combined degree and vascular-context strata were then constructed.

Within each valid stratum containing at least two candidate cells, the observed number of target-population cells was preserved. For each of 999 permutations, the corresponding number of candidate cells was randomly selected without replacement within each stratum. This generated a null distribution of neighborhood edge fractions while preserving target-population counts within the combined strata.

The observed neighborhood edge fraction was calculated as the proportion of all mast-cell graph edges connecting to the target population. The expected fraction was calculated as the mean across the 999 permutations. The enrichment ratio was defined as the observed fraction divided by the mean permuted fraction.

Strata containing fewer than two candidate cells were excluded from randomization. A target-cell retention audit recorded the number of target cells excluded and retained.

Sample-level enrichment ratios were aggregated to the donor level using the median across samples for each donor and population. Donor-level enrichment was evaluated against a reference value of 1 using a one-sided Wilcoxon signed-rank test. P-values for the two target populations were adjusted using the Benjamini-Hochberg false discovery rate procedure.

The permutation procedure used a fixed random seed of 20261111. This analysis was treated as a robustness comparison rather than an independent replication.

```python
import os
import pandas as pd
import numpy as np

folder = "/content/drive/MyDrive/Neurothesis Resoure Library"

paired_path = os.path.join(
    folder, "Vascular_Adjusted_Paired_FibI_FibII_Donors.csv"
)
loo_path = os.path.join(
    folder, "Vascular_Adjusted_Leave_One_Donor_Out.csv"
)

paired = pd.read_csv(paired_path)
loo = pd.read_csv(loo_path)

# Convert the two population columns into a standard long-format table.
vascular_donor_df = paired.melt(
    id_vars="donor_code",
    value_vars=["Perivasc Fib I", "Perivasc Fib II"],
    var_name="population",
    value_name="enrichment_ratio"
)

print("RECONSTRUCTED VASCULAR DONOR RESULTS")
print("=" * 70)
print("Rows:", len(vascular_donor_df))
print("Unique donors:", vascular_donor_df["donor_code"].nunique())
print(vascular_donor_df.groupby("population")["enrichment_ratio"].agg(
    ["count", "median", "mean", "min", "max"]
))

# Verify leave-one-donor-out medians against the paired donor values.
checks = []

for _, row in loo.iterrows():
    remaining = vascular_donor_df[
        (vascular_donor_df["population"] == row["population"]) &
        (vascular_donor_df["donor_code"] != row["omitted_donor"])
    ]["enrichment_ratio"]

    reconstructed_median = remaining.median()

    checks.append({
        "population": row["population"],
        "omitted_donor": row["omitted_donor"],
        "saved_loo_median": row["loo_median_enrichment"],
        "reconstructed_loo_median": reconstructed_median,
        "absolute_difference": abs(
            row["loo_median_enrichment"] - reconstructed_median
        )
    })

check_df = pd.DataFrame(checks)

print("\nLEAVE-ONE-DONOR-OUT RECONCILIATION")
print("=" * 70)
print("Checks:", len(check_df))
print("Maximum absolute difference:",
      check_df["absolute_difference"].max())
print("All checks within 1e-8:",
      (check_df["absolute_difference"] < 1e-8).all())

print("\nFIRST FIVE RECONCILIATION ROWS")
print(check_df.head().to_string(index=False))
```

```python
import os
import numpy as np
import pandas as pd
from scipy.stats import wilcoxon, spearmanr

folder = "/content/drive/MyDrive/Neurothesis Resoure Library"

# Load the verified degree-adjusted donor results.
degree_donor_df = pd.read_csv(
    os.path.join(
        folder,
        "Within_Dermal_Degree_Stratified_Donor_Results.csv"
    )
)

# Reconstruct and save the verified vascular-adjusted donor results.
vascular_donor_df.to_csv(
    os.path.join(
        folder,
        "Vascular_Adjusted_Donor_Results_Reconstructed.csv"
    ),
    index=False
)

# Match the same donor and population across both models.
comparison = degree_donor_df[
    ["donor_code", "population", "enrichment_ratio"]
].merge(
    vascular_donor_df[
        ["donor_code", "population", "enrichment_ratio"]
    ],
    on=["donor_code", "population"],
    how="inner",
    suffixes=("_degree", "_vascular"),
    validate="one_to_one"
)

comparison["vascular_minus_degree"] = (
    comparison["enrichment_ratio_vascular"]
    - comparison["enrichment_ratio_degree"]
)

print("DONOR-LEVEL MODEL RECONCILIATION")
print("=" * 70)
print("Matched records:", len(comparison))
print("Unique donors:", comparison["donor_code"].nunique())
print("Populations:", comparison["population"].nunique())

if len(comparison) != 44:
    raise ValueError("Expected 44 matched donor-population records.")

if comparison.duplicated(["donor_code", "population"]).any():
    raise ValueError("Duplicate donor-population pairs detected.")

# Paired sensitivity comparison.
rng = np.random.default_rng(20261110)
N_BOOTSTRAP = 10000

for population in ["Perivasc Fib I", "Perivasc Fib II"]:

    subset = comparison[
        comparison["population"] == population
    ].copy()

    degree = subset["enrichment_ratio_degree"].to_numpy()
    vascular = subset["enrichment_ratio_vascular"].to_numpy()
    differences = vascular - degree

    # Paired Wilcoxon test.
    wilcoxon_result = wilcoxon(
        differences,
        alternative="two-sided"
    )

    # Bootstrap confidence interval for median paired difference.
    bootstrap_medians = np.empty(N_BOOTSTRAP)

    for i in range(N_BOOTSTRAP):
        sampled = rng.choice(
            differences,
            size=len(differences),
            replace=True
        )
        bootstrap_medians[i] = np.median(sampled)

    ci_lower, ci_upper = np.percentile(
        bootstrap_medians, [2.5, 97.5]
    )

    # Agreement between donor-level model estimates.
    correlation, correlation_p = spearmanr(
        degree, vascular
    )

    print("\n" + population)
    print("-" * 70)
    print("Matched donors:", len(subset))
    print("Degree-adjusted median:", np.median(degree))
    print("Vascular-adjusted median:", np.median(vascular))
    print("Median paired difference (vascular - degree):",
          np.median(differences))
    print("Bootstrap 95% CI:", ci_lower, "to", ci_upper)
    print("Paired Wilcoxon p-value:", wilcoxon_result.pvalue)
    print("Spearman correlation:", correlation)
    print("Spearman p-value:", correlation_p)

print("\nFIRST FIVE MATCHED RECORDS")
print(comparison.head().to_string(index=False))
```

```python

import os
import numpy as np
import pandas as pd
from scipy.stats import wilcoxon, spearmanr

folder = "/content/drive/MyDrive/Neurothesis Resoure Library"

file_path = os.path.join(
    folder,
    "Donor_Level_Null_Model_Sensitivity_Comparison.csv"
)

df = pd.read_csv(file_path)

required = [
    "donor_code",
    "population",
    "enrichment_ratio_degree",
    "enrichment_ratio_vascular"
]

missing = [c for c in required if c not in df.columns]
if missing:
    raise ValueError(f"Missing required columns: {missing}")

df["vascular_minus_degree"] = (
    df["enrichment_ratio_vascular"]
    - df["enrichment_ratio_degree"]
)

# Verify the donor-level matching structure
print("TOTAL RECORDS:", len(df))
print("UNIQUE DONORS:", df["donor_code"].nunique())
print("POPULATIONS:", df["population"].unique())

if df.duplicated(["donor_code", "population"]).any():
    raise ValueError("Duplicate donor-population pairs detected.")

# Analyze each population separately
rng = np.random.default_rng(42)
results = []

for population, group in df.groupby("population"):

    group = group.dropna(
        subset=[
            "enrichment_ratio_degree",
            "enrichment_ratio_vascular"
        ]
    )

    degree = group["enrichment_ratio_degree"].to_numpy()
    vascular = group["enrichment_ratio_vascular"].to_numpy()
    differences = vascular - degree

    n = len(differences)

    # Bootstrap confidence interval for median difference
    boot_medians = np.array([
        np.median(rng.choice(
            differences,
            size=n,
            replace=True
        ))
        for _ in range(10000)
    ])

    ci_low, ci_high = np.percentile(
        boot_medians, [2.5, 97.5]
    )

    # Paired Wilcoxon test
    if np.allclose(differences, 0):
        wilcoxon_stat = 0.0
        wilcoxon_p = 1.0
    else:
        test = wilcoxon(
            vascular,
            degree,
            alternative="two-sided"
        )
        wilcoxon_stat = test.statistic
        wilcoxon_p = test.pvalue

    # Spearman rank correlation
    if np.std(degree) == 0 or np.std(vascular) == 0:
        rho = np.nan
        rho_p = np.nan
    else:
        corr = spearmanr(degree, vascular)
        rho = corr.statistic
        rho_p = corr.pvalue

    results.append({
        "population": population,
        "n_donors": n,
        "median_degree": np.median(degree),
        "median_vascular": np.median(vascular),
        "median_difference": np.median(differences),
        "mean_difference": np.mean(differences),
        "median_absolute_difference": np.median(
            np.abs(differences)
        ),
        "vascular_higher_percent": (
            np.mean(differences > 0) * 100
        ),
        "bootstrap_CI_low": ci_low,
        "bootstrap_CI_high": ci_high,
        "wilcoxon_statistic": wilcoxon_stat,
        "wilcoxon_p": wilcoxon_p,
        "spearman_rho": rho,
        "spearman_p": rho_p
    })

summary = pd.DataFrame(results)

# Benjamini-Hochberg correction across the two population tests
p = summary["wilcoxon_p"].to_numpy()
m = len(p)
order = np.argsort(p)
adjusted = np.empty(m)

ranked = p[order] * m / np.arange(1, m + 1)
ranked = np.minimum.accumulate(ranked[::-1])[::-1]
adjusted[order] = np.minimum(ranked, 1.0)

summary["wilcoxon_BH_q"] = adjusted

output_path = os.path.join(
    folder,
    "Donor_Level_Null_Model_Sensitivity_Summary.csv"
)

summary.to_csv(output_path, index=False)

print("\nSENSITIVITY SUMMARY")
display(summary.round(5))

print("\nSAVED:", output_path)
```

```python

import os
import numpy as np
import pandas as pd
from scipy.stats import wilcoxon, spearmanr

folder = "/content/drive/MyDrive/Neurothesis Resoure Library"

input_path = os.path.join(
    folder,
    "Donor_Level_Null_Model_Sensitivity_Comparison.csv"
)

df = pd.read_csv(input_path)

# Validate paired donor structure
required = [
    "donor_code",
    "population",
    "enrichment_ratio_degree",
    "enrichment_ratio_vascular",
    "vascular_minus_degree"
]

missing = [c for c in required if c not in df.columns]
if missing:
    raise ValueError(f"Missing columns: {missing}")

if df.duplicated(["donor_code", "population"]).any():
    raise ValueError("Duplicate donor-population pairs found.")

if df[required[1:]].isna().any().any():
    raise ValueError("Missing values detected in required fields.")

# Confirm difference calculation
calculated_difference = (
    df["enrichment_ratio_vascular"]
    - df["enrichment_ratio_degree"]
)

if not np.allclose(
    calculated_difference,
    df["vascular_minus_degree"],
    atol=1e-10
):
    raise ValueError("Paired difference column does not match.")

rng = np.random.default_rng(20261112)
B = 20000

summary_rows = []
influence_rows = []
outlier_rows = []

for population, sub in df.groupby("population"):

    sub = sub.copy().reset_index(drop=True)

    donors = sub["donor_code"].to_numpy()
    degree = sub["enrichment_ratio_degree"].to_numpy()
    vascular = sub["enrichment_ratio_vascular"].to_numpy()
    differences = vascular - degree

    n = len(sub)

    if n < 5:
        raise ValueError(
            f"Too few donor pairs for {population}"
        )

    # Bootstrap donor-level paired differences
    boot_medians = np.empty(B)
    boot_mae = np.empty(B)

    for b in range(B):
        idx = rng.integers(0, n, size=n)
        sampled = differences[idx]

        boot_medians[b] = np.median(sampled)
        boot_mae[b] = np.median(np.abs(sampled))

    median_ci = np.percentile(
        boot_medians, [2.5, 97.5]
    )

    mae_ci = np.percentile(
        boot_mae, [2.5, 97.5]
    )

    # Paired Wilcoxon
    wilcoxon_result = wilcoxon(
        vascular,
        degree,
        alternative="two-sided",
        method="auto"
    )

    rho, rho_p = spearmanr(degree, vascular)

    summary_rows.append({
        "population": population,
        "n_donors": n,
        "median_paired_difference": np.median(differences),
        "median_difference_CI_low": median_ci[0],
        "median_difference_CI_high": median_ci[1],
        "mean_paired_difference": np.mean(differences),
        "median_absolute_difference": np.median(
            np.abs(differences)
        ),
        "median_absolute_difference_CI_low": mae_ci[0],
        "median_absolute_difference_CI_high": mae_ci[1],
        "mean_absolute_difference": np.mean(
            np.abs(differences)
        ),
        "spearman_rho": rho,
        "spearman_p": rho_p,
        "wilcoxon_statistic": wilcoxon_result.statistic,
        "wilcoxon_p": wilcoxon_result.pvalue
    })

    # Leave-one-donor-out influence
    full_median = np.median(differences)

    for i in range(n):
        remaining = np.delete(differences, i)

        loo_median = np.median(remaining)

        influence_rows.append({
            "population": population,
            "donor_code": donors[i],
            "difference": differences[i],
            "absolute_difference": abs(differences[i]),
            "full_median_difference": full_median,
            "leave_one_out_median": loo_median,
            "change_in_median": loo_median - full_median
        })

    # Largest individual differences
    order = np.argsort(
        np.abs(differences)
    )[::-1]

    for rank, i in enumerate(order[:5], start=1):
        outlier_rows.append({
            "population": population,
            "rank": rank,
            "donor_code": donors[i],
            "degree_enrichment": degree[i],
            "vascular_enrichment": vascular[i],
            "vascular_minus_degree": differences[i],
            "absolute_difference": abs(differences[i])
        })

# Build output tables
summary_df = pd.DataFrame(summary_rows)
influence_df = pd.DataFrame(influence_rows)
outlier_df = pd.DataFrame(outlier_rows)

# Multiple-testing correction for the two Wilcoxon tests
p = summary_df["wilcoxon_p"].to_numpy()
order = np.argsort(p)
adjusted = np.empty(len(p))

running_min = 1.0
m = len(p)

for rank, idx in enumerate(order):
    value = p[idx] * m / (rank + 1)
    running_min = min(running_min, value)
    adjusted[idx] = min(running_min, 1.0)

summary_df["wilcoxon_BH_q"] = adjusted

# Save outputs
summary_path = os.path.join(
    folder,
    "Donor_Method_Agreement_Bootstrap_Summary.csv"
)

influence_path = os.path.join(
    folder,
    "Donor_Method_Leave_One_Out_Influence.csv"
)

outlier_path = os.path.join(
    folder,
    "Donor_Method_Largest_Differences.csv"
)

summary_df.to_csv(summary_path, index=False)
influence_df.to_csv(influence_path, index=False)
outlier_df.to_csv(outlier_path, index=False)

print("BOOTSTRAP SUMMARY")
display(summary_df.round(5))

print("\nLARGEST DONOR-LEVEL DIFFERENCES")
display(outlier_df.round(5))

print("\nLEAVE-ONE-DONOR-OUT RESULTS")
display(
    influence_df.sort_values(
        "change_in_median",
        key=lambda x: x.abs(),
        ascending=False
    ).groupby("population").head(5).round(5)
)

print("\nSAVED:")
print(summary_path)
print(influence_path)
print(outlier_path)
```

```python

# STEP 7: DONOR-LEVEL METHOD AGREEMENT
# Compare degree-based and vascular-based enrichment
# across donors for Perivasc Fib I and Perivasc Fib II.

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import spearmanr, wilcoxon

# --------------------------------------------------
# 1. VALIDATE INPUT DATA
# --------------------------------------------------

required_columns = [
    "donor_code",
    "population",
    "enrichment_ratio_degree",
    "enrichment_ratio_vascular",
    "vascular_minus_degree"
]

missing = [
    col for col in required_columns
    if col not in comparison.columns
]

if missing:
    raise ValueError(f"Missing required columns: {missing}")

print("DATA VALIDATION")
print("----------------")
print("Rows:", len(comparison))
print("Unique donors:", comparison["donor_code"].nunique())
print("Populations:", comparison["population"].unique())

# --------------------------------------------------
# 2. OUTPUT DIRECTORY
# --------------------------------------------------

output_dir = (
    "/content/drive/MyDrive/"
    "Neurothesis Resoure Library"
)

os.makedirs(output_dir, exist_ok=True)

# --------------------------------------------------
# 3. ANALYZE EACH POPULATION
# --------------------------------------------------

populations = [
    "Perivasc Fib I",
    "Perivasc Fib II"
]

rng = np.random.default_rng(42)

agreement_results = []

for population in populations:

    sub = comparison[
        comparison["population"] == population
    ].copy()

    sub = sub.dropna(
        subset=[
            "enrichment_ratio_degree",
            "enrichment_ratio_vascular"
        ]
    )

    x = sub["enrichment_ratio_degree"].to_numpy()
    y = sub["enrichment_ratio_vascular"].to_numpy()

    differences = y - x
    donor_means = (x + y) / 2

    n = len(sub)

    if n < 3:
        print(f"Skipping {population}: insufficient donors.")
        continue

    # Spearman rank agreement
    rho, spearman_p = spearmanr(x, y)

    # Paired Wilcoxon test
    wilcoxon_result = wilcoxon(
        differences,
        alternative="two-sided",
        zero_method="wilcox"
    )

    # Bootstrap confidence interval for median difference
    bootstrap_medians = np.array([
        np.median(
            rng.choice(
                differences,
                size=n,
                replace=True
            )
        )
        for _ in range(10000)
    ])

    ci_lower, ci_upper = np.percentile(
        bootstrap_medians,
        [2.5, 97.5]
    )

    median_difference = np.median(differences)

    # --------------------------------------------------
    # 4. SAVE SUMMARY STATISTICS
    # --------------------------------------------------

    agreement_results.append({
        "population": population,
        "donors": n,
        "spearman_rho": rho,
        "spearman_p": spearman_p,
        "median_difference": median_difference,
        "median_difference_CI_lower": ci_lower,
        "median_difference_CI_upper": ci_upper,
        "wilcoxon_statistic": wilcoxon_result.statistic,
        "wilcoxon_p": wilcoxon_result.pvalue
    })

    # --------------------------------------------------
    # 5. FIGURE A: PAIRED METHOD AGREEMENT
    # --------------------------------------------------

    fig, ax = plt.subplots(figsize=(7, 6))

    ax.scatter(
        x,
        y,
        alpha=0.8
    )

    lower = min(x.min(), y.min())
    upper = max(x.max(), y.max())

    ax.plot(
        [lower, upper],
        [lower, upper],
        linestyle="--",
        linewidth=1.5
    )

    # Label the four donors with largest discrepancies
    largest = np.argsort(
        np.abs(differences)
    )[-4:]

    for idx in largest:
        donor_label = str(
            sub.iloc[idx]["donor_code"]
        )

        ax.annotate(
            donor_label,
            (x[idx], y[idx]),
            xytext=(5, 5),
            textcoords="offset points",
            fontsize=8
        )

    ax.set_xlabel("Degree-based enrichment")
    ax.set_ylabel("Vascular-based enrichment")
    ax.set_title(
        f"Donor-Level Method Agreement: {population}"
    )

    ax.text(
        0.04,
        0.96,
        f"n = {n}\nSpearman ρ = {rho:.3f}\np = {spearman_p:.3e}",
        transform=ax.transAxes,
        va="top"
    )

    fig.tight_layout()

    fig.savefig(
        os.path.join(
            output_dir,
            f"Method_Agreement_{population.replace(' ', '_')}.png"
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close(fig)

    # --------------------------------------------------
    # 6. FIGURE B: DIFFERENCE PLOT
    # --------------------------------------------------

    mean_difference = np.mean(differences)
    sd_difference = np.std(differences, ddof=1)

    fig, ax = plt.subplots(figsize=(7, 6))

    ax.scatter(
        donor_means,
        differences,
        alpha=0.8
    )

    ax.axhline(
        0,
        linestyle="--",
        linewidth=1.5
    )

    ax.axhline(
        mean_difference,
        linewidth=1.2
    )

    ax.axhline(
        mean_difference + 1.96 * sd_difference,
        linestyle=":",
        linewidth=1
    )

    ax.axhline(
        mean_difference - 1.96 * sd_difference,
        linestyle=":",
        linewidth=1
    )

    ax.set_xlabel("Mean enrichment across methods")
    ax.set_ylabel(
        "Vascular-based minus degree-based enrichment"
    )

    ax.set_title(
        f"Method Difference Plot: {population}"
    )

    fig.tight_layout()

    fig.savefig(
        os.path.join(
            output_dir,
            f"Method_Difference_{population.replace(' ', '_')}.png"
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close(fig)

# --------------------------------------------------
# 7. EXPORT RESULTS
# --------------------------------------------------

agreement_summary = pd.DataFrame(
    agreement_results
)

summary_path = os.path.join(
    output_dir,
    "Donor_Method_Agreement_Final_Summary.csv"
)

agreement_summary.to_csv(
    summary_path,
    index=False
)

print("\nFINAL AGREEMENT SUMMARY")
display(agreement_summary)

print("\nSaved summary:", summary_path)
print("Saved figures in:", output_dir)
```

```python

# ============================================================
# NEUROTHESIS: DONOR CODE VALIDATION + METHOD DIAGNOSTIC
# ============================================================

import os
import numpy as np
import pandas as pd
from IPython.display import display

BASE = "/content/drive/MyDrive/Neurothesis Resoure Library"

# Load source files
mapping = pd.read_csv(
    os.path.join(BASE, "Verified_Sample_Donor_Site_Mapping.csv")
)

site_results = pd.read_csv(
    os.path.join(BASE, "Within_Dermal_Degree_Donor_Site_Results.csv")
)

largest = pd.read_csv(
    os.path.join(BASE, "Donor_Method_Largest_Differences.csv")
)

vascular = pd.read_csv(
    os.path.join(BASE, "Vascular_Adjusted_Donor_Results_Reconstructed.csv")
)

# ------------------------------------------------------------
# 1. Candidate mapping: alphabetical donor IDs
# ------------------------------------------------------------

donor_ids = sorted(mapping["donor_id"].astype(str).unique())

code_map = pd.DataFrame({
    "donor_code": range(len(donor_ids)),
    "donor_id": donor_ids
})

print("CANDIDATE DONOR MAP")
display(code_map)

# ------------------------------------------------------------
# 2. Compare donor-site sample counts
# ------------------------------------------------------------

actual_counts = (
    mapping.groupby(["donor_id", "anatomic_site"])
    .agg(actual_sample_count=("sample_id", "nunique"))
    .reset_index()
)

site_counts = (
    site_results.groupby(["donor_code", "anatomic_site"])
    .agg(reported_sample_count=("sample_count", "sum"))
    .reset_index()
)

site_counts = site_counts.merge(
    code_map, on="donor_code", how="left"
)

count_check = site_counts.merge(
    actual_counts,
    on=["donor_id", "anatomic_site"],
    how="outer",
    indicator=True
)

count_check["counts_match"] = (
    count_check["reported_sample_count"].fillna(0)
    == count_check["actual_sample_count"].fillna(0)
)

print("\nDONOR-SITE SAMPLE COUNT VALIDATION")
print("Rows compared:", len(count_check))
print("Matching rows:", int(count_check["counts_match"].sum()))
print("Mismatching rows:", int((~count_check["counts_match"]).sum()))

display(
    count_check.loc[
        ~count_check["counts_match"]
    ].head(30)
)

# ------------------------------------------------------------
# 3. Check degree enrichment against saved site-level results
# ------------------------------------------------------------

degree_check = largest.merge(
    site_results,
    on=["donor_code", "population"],
    how="left",
    suffixes=("_largest", "_site")
)

degree_check["degree_value_match"] = np.isclose(
    degree_check["degree_enrichment"],
    degree_check["enrichment_ratio"],
    atol=1e-6,
    rtol=1e-6
)

print("\nDEGREE ENRICHMENT VALUE CHECK")
display(
    degree_check[
        [
            "population",
            "donor_code",
            "anatomic_site",
            "degree_enrichment",
            "enrichment_ratio",
            "degree_value_match"
        ]
    ]
)

# ------------------------------------------------------------
# 4. Check vascular-adjusted enrichment
# ------------------------------------------------------------

vascular_check = largest.merge(
    vascular,
    on=["donor_code", "population"],
    how="left"
)

vascular_check["vascular_value_match"] = np.isclose(
    vascular_check["vascular_enrichment"],
    vascular_check["enrichment_ratio"],
    atol=1e-6,
    rtol=1e-6
)

print("\nVASCULAR ENRICHMENT VALUE CHECK")
display(
    vascular_check[
        [
            "population",
            "donor_code",
            "vascular_enrichment",
            "enrichment_ratio",
            "vascular_value_match"
        ]
    ]
)

# ------------------------------------------------------------
# 5. Attach candidate donor IDs to discrepancy table
# ------------------------------------------------------------

largest_mapped = largest.merge(
    code_map,
    on="donor_code",
    how="left"
)

print("\nLARGEST METHOD DISCREPANCIES WITH CANDIDATE DONOR IDs")
display(largest_mapped)

# ------------------------------------------------------------
# 6. Summarize donor sample/site representation
# ------------------------------------------------------------

donor_site_summary = (
    mapping.groupby("donor_id")
    .agg(
        sample_count=("sample_id", "nunique"),
        site_count=("anatomic_site", "nunique"),
        site_list=(
            "anatomic_site",
            lambda x: ", ".join(sorted(set(x)))
        )
    )
    .reset_index()
)

discrepancy_donors = (
    largest_mapped[["donor_id"]]
    .drop_duplicates()
)

discrepancy_donor_summary = discrepancy_donors.merge(
    donor_site_summary,
    on="donor_id",
    how="left"
)

print("\nSAMPLE AND SITE REPRESENTATION OF FLAGGED DONORS")
display(discrepancy_donor_summary)

# ------------------------------------------------------------
# 7. Save diagnostic outputs
# ------------------------------------------------------------

outputs = {
    "Donor_Code_Mapping_Validation.csv": code_map,
    "Donor_Site_Count_Validation.csv": count_check,
    "Donor_Degree_Value_Validation.csv": degree_check,
    "Donor_Vascular_Value_Validation.csv": vascular_check,
    "Donor_Method_Discrepancies_With_IDs.csv": largest_mapped,
    "Flagged_Donor_Site_Composition.csv": discrepancy_donor_summary
}

for filename, dataframe in outputs.items():
    dataframe.to_csv(
        os.path.join(BASE, filename),
        index=False
    )

print("\nVALIDATION CELL COMPLETE")
print("All six diagnostic files saved.")
print("A mismatch is a reason to investigate, not to force a join.")
```

```python

import pandas as pd
import numpy as np

folder = "/content/drive/MyDrive/Neurothesis Resoure Library/"

# Load original verified sample metadata and degree results
samples = pd.read_csv(
    folder + "Verified_Sample_Donor_Site_Mapping.csv"
)

degree = pd.read_csv(
    folder + "Within_Dermal_Degree_Donor_Site_Results.csv"
)

# Normalize identifiers and site names
for df in [samples, degree]:
    df["donor_id"] = df["donor_id"].astype(str).str.strip()
    df["anatomic_site"] = (
        df["anatomic_site"].astype(str).str.strip().str.lower()
    )

samples["sample_id"] = samples["sample_id"].astype(str).str.strip()

# Reconstruct unique sample counts directly from sample metadata
actual_counts = (
    samples.groupby(["donor_id", "anatomic_site"])
    ["sample_id"]
    .nunique()
    .reset_index(name="actual_sample_count")
)

# Attach the confirmed original donor-code mapping
mapping = pd.read_csv(
    folder + "Verified_H5AD_Donor_Code_Mapping.csv"
)

mapping["donor_code"] = mapping["donor_code"].astype(int)

# Attach donor IDs to the degree results
degree["donor_code"] = degree["donor_code"].astype(int)

degree = degree.merge(
    mapping,
    on="donor_code",
    how="left",
    validate="many_to_one"
)

# Compare reported and reconstructed sample counts
comparison = degree.merge(
    actual_counts,
    on=["donor_id", "anatomic_site"],
    how="outer",
    indicator=True
)

comparison["counts_match"] = (
    comparison["sample_count"].fillna(-999)
    == comparison["actual_sample_count"].fillna(-999)
)

comparison["reported_to_actual_ratio"] = np.where(
    comparison["actual_sample_count"].fillna(0) > 0,
    comparison["sample_count"] /
    comparison["actual_sample_count"],
    np.nan
)

print("DONOR-SITE SAMPLE COUNT DIAGNOSTIC")
print("Total comparison rows:", len(comparison))
print("Matching counts:", comparison["counts_match"].sum())
print("Mismatching counts:", (~comparison["counts_match"]).sum())

print("\nMERGE STATUS")
print(comparison["_merge"].value_counts())

print("\nREPORTED / ACTUAL COUNT RATIOS")
print(
    comparison["reported_to_actual_ratio"]
    .value_counts(dropna=False)
    .to_string()
)

print("\nFIRST 30 MISMATCHES")
display(
    comparison.loc[
        ~comparison["counts_match"],
        [
            "donor_id",
            "anatomic_site",
            "sample_count",
            "actual_sample_count",
            "reported_to_actual_ratio",
            "_merge"
        ]
    ].head(30)
)

# Save diagnostic output
comparison.to_csv(
    folder + "Donor_Site_Sample_Count_Reconciliation.csv",
    index=False
)

print("\nSaved Donor_Site_Sample_Count_Reconciliation.csv")
```

```python

import h5py
import pandas as pd
from pathlib import Path

folder = Path("/content/drive/MyDrive/Neurothesis Resoure Library")
data_path = folder / "merfish.integrated_annotated (2).h5ad"

def decode_categories(group):
    return [
        x.decode("utf-8") if isinstance(x, bytes) else str(x)
        for x in group["categories"][:]
    ]

# Recover the original H5AD categorical codes
with h5py.File(data_path, "r") as f:
    obs = f["obs"]

    donor_group = obs["donor_id"]
    donor_labels = decode_categories(donor_group)
    donor_codes = donor_group["codes"][:]

    sample_group = obs["sample_id"]
    sample_labels = decode_categories(sample_group)
    sample_codes = sample_group["codes"][:]

    # Map each original sample code to its actual sample and donor
    code_map = pd.DataFrame({
        "sample_code": sample_codes,
        "donor_code": donor_codes
    }).drop_duplicates()

    code_map = code_map[
        (code_map["sample_code"] >= 0) &
        (code_map["donor_code"] >= 0)
    ].copy()

    code_map["sample_id"] = code_map["sample_code"].map(
        lambda x: sample_labels[int(x)]
    )
    code_map["donor_id"] = code_map["donor_code"].map(
        lambda x: donor_labels[int(x)]
    )

    code_map = code_map.drop_duplicates(
        subset=["sample_code"]
    )

# Attach anatomical site from verified metadata
metadata = pd.read_csv(
    folder / "Verified_Sample_Donor_Site_Mapping.csv"
)

code_map = code_map.merge(
    metadata[["sample_id", "anatomic_site"]],
    on="sample_id",
    how="left",
    validate="one_to_one"
)

# Load sample-level degree results
sample_results = pd.read_csv(
    folder / "Within_Dermal_Degree_Stratified_Sample_Results.csv"
)

sample_results["sample_code"] = pd.to_numeric(
    sample_results["sample_code"], errors="coerce"
)

sample_results["donor_code"] = pd.to_numeric(
    sample_results["donor_code"], errors="coerce"
)

# Attach original sample and donor identities
traced = sample_results.merge(
    code_map,
    on=["sample_code", "donor_code"],
    how="left",
    validate="many_to_one"
)

# Inspect the two cases
cases = [
    ("D082", "knee"),
    ("D151", "occipital scalp")
]

for donor, site in cases:
    print("\n" + "=" * 70)
    print(f"{donor} | {site}")

    rows = traced[
        (traced["donor_id"] == donor) &
        (traced["anatomic_site"].str.lower() == site)
    ].sort_values(["sample_id", "population"])

    display(rows)

    print("Samples represented:", rows["sample_id"].nunique())
    print("Population rows:", len(rows))

# Also inspect the degree stratification audit
audit_path = folder / "Degree_Stratification_Audit.csv"

if audit_path.exists():
    audit = pd.read_csv(audit_path)

    audit = audit.merge(
        code_map[["sample_code", "sample_id", "donor_id", "anatomic_site"]],
        on="sample_code",
        how="left"
    )

    for donor, site in cases:
        print("\nAUDIT:", donor, site)

        display(
            audit[
                (audit["donor_id"] == donor) &
                (audit["anatomic_site"].str.lower() == site)
            ]
        )

# Save traceable sample-level output
traced.to_csv(
    folder / "Verified_Sample_Level_Degree_Results_With_IDs.csv",
    index=False
)

print("\nTrace file saved.")
```

```python

import h5py
import pandas as pd
from pathlib import Path
from collections import Counter

folder = Path("/content/drive/MyDrive/Neurothesis Resoure Library")
h5ad_path = folder / "merfish.integrated_annotated (2).h5ad"

# Exact sample IDs from the original metadata
metadata = pd.read_csv(
    folder / "Verified_Sample_Donor_Site_Mapping.csv"
)

targets = metadata[
    ((metadata["donor_id"] == "D082") &
     (metadata["anatomic_site"] == "knee")) |
    ((metadata["donor_id"] == "D151") &
     (metadata["anatomic_site"] == "occipital scalp"))
].copy()

target_ids = set(targets["sample_id"].astype(str))

print("TARGET SAMPLES")
display(targets)

# --------------------------------------------------
# PART 1: Inspect original H5AD annotations
# --------------------------------------------------

with h5py.File(h5ad_path, "r") as f:
    obs = f["obs"]

    print("\nOBSERVATION FIELDS:")
    print(list(obs.keys()))

    sample_codes = obs["sample_id"]["codes"][:]
    sample_categories = [
        x.decode("utf-8") if isinstance(x, bytes) else str(x)
        for x in obs["sample_id"]["categories"][:]
    ]

    sample_lookup = {
        i: name for i, name in enumerate(sample_categories)
    }

    target_codes = {
        code for code, name in sample_lookup.items()
        if name in target_ids
    }

    # Find categorical annotation columns containing
    # the two perivascular population labels.
    candidate_fields = []

    for key in obs.keys():
        obj = obs[key]

        if not isinstance(obj, h5py.Group):
            continue

        if "categories" not in obj or "codes" not in obj:
            continue

        categories = [
            x.decode("utf-8") if isinstance(x, bytes) else str(x)
            for x in obj["categories"][:]
        ]

        matches = [
            x for x in categories
            if "Perivasc Fib I" in x or "Perivasc Fib II" in x
        ]

        if matches:
            candidate_fields.append((key, categories, obj["codes"][:]))
            print("\nPOTENTIAL CELL ANNOTATION FIELD:", key)
            print("Matching categories:", matches)

    if not candidate_fields:
        print("\nNo categorical field directly contains those labels.")
        print("We will need to identify the relevant annotation field.")

    for field, categories, annotation_codes in candidate_fields:
        print("\nCELL COUNTS FROM ORIGINAL H5AD:", field)

        category_lookup = {
            i: label for i, label in enumerate(categories)
        }

        records = []

        for sample_code in sorted(target_codes):
            sample_name = sample_lookup[sample_code]
            mask = sample_codes == sample_code

            counts = Counter(annotation_codes[mask])

            for code, label in category_lookup.items():
                if "Perivasc Fib I" in label or "Perivasc Fib II" in label:
                    records.append({
                        "sample_id": sample_name,
                        "annotation": label,
                        "cell_count": counts.get(code, 0)
                    })

        display(pd.DataFrame(records))

# --------------------------------------------------
# PART 2: Search all saved CSVs for these samples
# --------------------------------------------------

print("\n" + "=" * 70)
print("SAVED CSV TRACE")

for path in sorted(folder.glob("*.csv")):
    try:
        df = pd.read_csv(path)
    except Exception:
        continue

    matching_rows = pd.DataFrame()

    if "sample_id" in df.columns:
        matching_rows = df[
            df["sample_id"].astype(str).isin(target_ids)
        ]

    elif "sample_code" in df.columns:
        # Codes 13 and 95 correspond to the two D082 samples.
        # D151's code is retrieved from the original mapping below.
        matching_rows = df[
            pd.to_numeric(df["sample_code"], errors="coerce")
            .isin([13, 95])
        ]

    if not matching_rows.empty:
        print("\nFILE:", path.name)
        display(matching_rows.head(15))

print("\nTRACE COMPLETE")
```

```python

import h5py
import pandas as pd
from pathlib import Path
from collections import Counter

folder = Path("/content/drive/MyDrive/Neurothesis Resoure Library")
h5ad_path = folder / "merfish.integrated_annotated (2).h5ad"

metadata = pd.read_csv(
    folder / "Verified_Sample_Donor_Site_Mapping.csv"
)

targets = metadata[
    ((metadata.donor_id == "D082") &
     (metadata.anatomic_site == "knee")) |
    ((metadata.donor_id == "D151") &
     (metadata.anatomic_site == "occipital scalp"))
].copy()

target_ids = set(targets.sample_id.astype(str))

with h5py.File(h5ad_path, "r") as f:
    obs = f["obs"]

    def categories(field):
        return [
            x.decode("utf-8") if isinstance(x, bytes) else str(x)
            for x in obs[field]["categories"][:]
        ]

    sample_categories = categories("sample_id")
    sample_codes = obs["sample_id"]["codes"][:]

    compartment_categories = categories("tissue_compartment")
    compartment_codes = obs["tissue_compartment"]["codes"][:]

    detailed_categories = categories("cell_type.detailed")
    detailed_codes = obs["cell_type.detailed"]["codes"][:]

    broad_categories = categories("cell_type.broad")
    broad_codes = obs["cell_type.broad"]["codes"][:]

    sample_lookup = {
        i: name for i, name in enumerate(sample_categories)
    }

    for sample_id in sorted(target_ids):
        sample_code = next(
            code for code, name in sample_lookup.items()
            if name == sample_id
        )

        mask = sample_codes == sample_code

        print("\n" + "=" * 75)
        print("SAMPLE:", sample_id)
        print("TOTAL CELLS:", int(mask.sum()))

        # Tissue compartment counts
        comp_counts = Counter(compartment_codes[mask])
        print("\nTISSUE COMPARTMENTS:")
        display(pd.DataFrame([
            {
                "compartment": compartment_categories[code],
                "cells": count
            }
            for code, count in comp_counts.items()
            if code >= 0
        ]).sort_values("cells", ascending=False))

        # Detailed cell annotations
        detail_counts = Counter(detailed_codes[mask])
        detail_table = pd.DataFrame([
            {
                "cell_type": detailed_categories[code],
                "cells": count
            }
            for code, count in detail_counts.items()
            if code >= 0
        ])

        print("\nRELEVANT DETAILED CELL TYPES:")
        display(
            detail_table[
                detail_table.cell_type.str.contains(
                    "Mast|Perivasc Fib I|Perivasc Fib II",
                    case=False,
                    na=False
                )
            ].sort_values("cell_type")
        )

        # Broad cell annotations
        broad_counts = Counter(broad_codes[mask])
        print("\nRELEVANT BROAD CELL TYPES:")
        display(pd.DataFrame([
            {
                "cell_type": broad_categories[code],
                "cells": count
            }
            for code, count in broad_counts.items()
            if code >= 0 and
            "mast" in broad_categories[code].lower()
        ]))

# Search saved CSVs for D151 sample code 29
print("\n" + "=" * 75)
print("SEARCHING SAVED CSVs FOR SAMPLE CODE 29")

for path in sorted(folder.glob("*.csv")):
    try:
        df = pd.read_csv(path)
    except Exception:
        continue

    if "sample_code" in df.columns:
        codes = pd.to_numeric(df["sample_code"], errors="coerce")
        rows = df[codes == 29]

        if not rows.empty:
            print("\nFILE:", path.name)
            display(rows.head(10))

    elif "sample_id" in df.columns:
        rows = df[df.sample_id.astype(str).isin(target_ids)]

        if not rows.empty:
            print("\nFILE:", path.name)
            display(rows.head(10))

print("\nDIAGNOSTIC COMPLETE")
```

```python

# ============================================================
# NEUROTHESIS: REBUILD MATCHED-CONTROL STRATA
# Matching variables: sample ID + spatial graph degree
# Tissue restriction: dermis
# ============================================================

import numpy as np
import pandas as pd
from collections import defaultdict

# 1. Confirm the required arrays exist
required = [
    "expanded_labels_10",
    "compartment_codes",
    "sample_codes_full",
    "graph_degrees",
    "A_symmetric",
    "population_names_10"
]

missing = [v for v in required if v not in globals()]
if missing:
    raise NameError(f"Missing required variables: {missing}")

labels = np.asarray(expanded_labels_10)
compartments = np.asarray(compartment_codes)
samples = np.asarray(sample_codes_full)
degrees = np.asarray(graph_degrees)

assert len(labels) == len(compartments) == len(samples) == len(degrees)
assert A_symmetric.shape == (len(labels), len(labels))

# 2. Identify eligible cells
# Population code 0 = Mast
# Population codes 1-9 = non-Mast populations
# Compartment code 0 = dermis

dermis_mask = compartments == 0

mast_mask = (
    dermis_mask &
    (labels == 0) &
    (degrees > 0)
)

control_mask = (
    dermis_mask &
    (labels >= 1) &
    (labels <= 9) &
    (degrees > 0)
)

mast_indices = np.flatnonzero(mast_mask)
control_indices = np.flatnonzero(control_mask)

print("ELIGIBILITY AUDIT")
print("------------------------------")
print("Dermal Mast cells with graph edges:", len(mast_indices))
print("Eligible non-Mast control cells:", len(control_indices))

# 3. Group Mast cells by exact sample + graph degree
mast_groups = defaultdict(list)

for idx in mast_indices:
    key = (int(samples[idx]), int(degrees[idx]))
    mast_groups[key].append(int(idx))

# 4. Group non-Mast controls by the same exact strata
control_groups = defaultdict(list)

for idx in control_indices:
    key = (int(samples[idx]), int(degrees[idx]))
    control_groups[key].append(int(idx))

# 5. Keep only strata with enough controls
matched_control_pools = {}
matched_mast_groups = {}
strata_records = []

for key, mast_list in mast_groups.items():

    control_list = control_groups.get(key, [])

    if len(control_list) >= len(mast_list):

        matched_mast_groups[key] = np.asarray(
            mast_list, dtype=np.int64
        )

        matched_control_pools[key] = np.asarray(
            control_list, dtype=np.int64
        )

        strata_records.append({
            "sample_code": key[0],
            "degree": key[1],
            "mast_count": len(mast_list),
            "control_pool_size": len(control_list),
            "control_to_mast_ratio": (
                len(control_list) / len(mast_list)
            )
        })

# 6. Reconstruct combined matched Mast index array
matched_mast_indices = np.concatenate(
    list(matched_mast_groups.values())
)

matched_strata_df = pd.DataFrame(strata_records)

# 7. Calculate unique control candidate count
candidate_control_indices = np.unique(
    np.concatenate(list(matched_control_pools.values()))
)

# 8. Audit results
print("\nMATCHED-CONTROL AUDIT")
print("------------------------------")
print("Matched Mast cells:", len(matched_mast_indices))
print("Unmatched Mast cells:", len(mast_indices) - len(matched_mast_indices))
print("Matched strata:", len(matched_control_pools))
print("Samples represented:", matched_strata_df["sample_code"].nunique())
print("Unique control candidates:", len(candidate_control_indices))
print("Median control:Mast ratio:",
      matched_strata_df["control_to_mast_ratio"].median())
print("Maximum control:Mast ratio:",
      matched_strata_df["control_to_mast_ratio"].max())

# 9. Validation checks
assert len(matched_control_pools) == len(matched_mast_groups)
assert len(matched_mast_indices) > 0
assert len(candidate_control_indices) > 0
assert all(len(v) > 0 for v in matched_control_pools.values())
assert all(
    len(matched_control_pools[k]) >= len(matched_mast_groups[k])
    for k in matched_control_pools
)

print("\nSUCCESS: Matched-control objects reconstructed.")

# 10. Save the matching audit
matched_strata_df.to_csv(
    "Matched_Control_Strata_Audit.csv",
    index=False
)

print("Saved: Matched_Control_Strata_Audit.csv")
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

folder = Path("/content/drive/MyDrive/Neurothesis Resoure Library")

sample = pd.read_csv(folder / "Within_Dermal_Degree_Stratified_Sample_Results.csv")
donor = pd.read_csv(folder / "Within_Dermal_Degree_Stratified_Donor_Results.csv")
audit = pd.read_csv(folder / "Degree_Stratification_Audit.csv")

print("=" * 75)
print("1. SAMPLE-LEVEL ENRICHMENT SUMMARY")
print("=" * 75)

sample_summary = (
    sample.groupby("population")["enrichment_ratio"]
    .agg(["count", "mean", "median", "std", "min", "max"])
    .round(4)
)
display(sample_summary)

print("\n" + "=" * 75)
print("2. DONOR-LEVEL ENRICHMENT SUMMARY")
print("=" * 75)

donor_summary = (
    donor.groupby("population")["enrichment_ratio"]
    .agg(["count", "mean", "median", "std", "min", "max"])
    .round(4)
)
display(donor_summary)

print("\n" + "=" * 75)
print("3. DONOR-LEVEL CONSISTENCY")
print("=" * 75)

for population, group in donor.groupby("population"):
    values = group["enrichment_ratio"].dropna()

    print(f"\n{population}")
    print("Donors:", len(values))
    print("Above 1.0:", int((values > 1).sum()))
    print("Below 1.0:", int((values < 1).sum()))
    print("Exactly 1.0:", int((values == 1).sum()))
    print("Median:", round(values.median(), 4))
    print("IQR:", round(values.quantile(.75) - values.quantile(.25), 4))

print("\n" + "=" * 75)
print("4. MATCHING AUDIT")
print("=" * 75)

audit_summary = (
    audit.groupby(["population", "degree_definition"])
    .agg(
        samples=("sample_code", "nunique"),
        total_target_cells=("target_cells", "sum"),
        singleton_target_cells=("target_cells_in_singleton_strata", "sum"),
        singleton_candidate_cells=("candidate_cells_in_singleton_strata", "sum"),
        singleton_strata=("singleton_strata", "sum")
    )
    .reset_index()
)

audit_summary["percent_target_cells_in_singletons"] = (
    100 * audit_summary["singleton_target_cells"]
    / audit_summary["total_target_cells"].replace(0, np.nan)
)

display(audit_summary.round(3))

print("\n" + "=" * 75)
print("5. SAMPLE COVERAGE")
print("=" * 75)

print("Unique samples:", sample["sample_code"].nunique())
print("Unique donors:", sample["donor_code"].nunique())
print("Sample rows:", len(sample))
print("Donor rows:", len(donor))

print("\nSamples per population:")
display(sample.groupby("population")["sample_code"].nunique())

print("\nDonors per population:")
display(donor.groupby("population")["donor_code"].nunique())
```

```python
import numpy as np
import pandas as pd
from scipy import stats

# --------------------------------------------------
# 1. IDENTIFY THE MISSING SAMPLE
# --------------------------------------------------

print("=" * 70)
print("SAMPLE COVERAGE DISCREPANCY")
print("=" * 70)

for population in sample["population"].unique():

    result_codes = set(
        sample.loc[sample["population"] == population, "sample_code"]
    )

    audit_codes = set(
        audit.loc[
            (audit["population"] == population) &
            (audit["degree_definition"] == "within_sample_dermal_degree"),
            "sample_code"
        ]
    )

    print(f"\n{population}")
    print("Samples in result:", len(result_codes))
    print("Samples in audit:", len(audit_codes))
    print("In audit but not result:", sorted(audit_codes - result_codes))
    print("In result but not audit:", sorted(result_codes - audit_codes))

# --------------------------------------------------
# 2. DONOR-LEVEL TESTS AGAINST ENRICHMENT = 1
# --------------------------------------------------

print("\n" + "=" * 70)
print("DONOR-LEVEL INFERENCE")
print("=" * 70)

rng = np.random.default_rng(42)

for population, group in donor.groupby("population"):

    values = group["enrichment_ratio"].dropna().to_numpy()
    above = np.sum(values > 1)
    below = np.sum(values < 1)
    ties = np.sum(values == 1)

    sign_test = stats.binomtest(
        above, above + below, p=0.5, alternative="greater"
    )

    wilcoxon_test = stats.wilcoxon(
        values - 1,
        alternative="greater",
        method="auto"
    )

    boot_medians = np.array([
        np.median(rng.choice(values, size=len(values), replace=True))
        for _ in range(10000)
    ])

    ci = np.quantile(boot_medians, [0.025, 0.975])

    print(f"\n{population}")
    print("Donors:", len(values))
    print("Above 1:", above, "| Below 1:", below, "| Ties:", ties)
    print("Median enrichment:", round(np.median(values), 4))
    print("Bootstrap 95% CI for median:", np.round(ci, 4))
    print("Exact sign-test p:", sign_test.pvalue)
    print("Wilcoxon p:", wilcoxon_test.pvalue)

# --------------------------------------------------
# 3. PAIRED COMPARISON BETWEEN POPULATIONS
# --------------------------------------------------

print("\n" + "=" * 70)
print("PAIRED DONOR COMPARISON: FIB I VS FIB II")
print("=" * 70)

wide = donor.pivot(
    index="donor_code",
    columns="population",
    values="enrichment_ratio"
).dropna()

differences = (
    wide["Perivasc Fib I"] -
    wide["Perivasc Fib II"]
).to_numpy()

paired_test = stats.wilcoxon(
    differences,
    alternative="two-sided",
    method="auto"
)

boot_differences = np.array([
    np.median(rng.choice(differences, size=len(differences), replace=True))
    for _ in range(10000)
])

difference_ci = np.quantile(
    boot_differences, [0.025, 0.975]
)

print("Matched donors:", len(differences))
print("Median paired difference:", round(np.median(differences), 4))
print("Mean paired difference:", round(np.mean(differences), 4))
print("Bootstrap 95% CI:", np.round(difference_ci, 4))
print("Paired Wilcoxon p:", paired_test.pvalue)
```

```python

import os
import pandas as pd
import numpy as np

# ============================================================
# DONOR-LEVEL RESULTS: COVERAGE AND SUMMARY AUDIT
# ============================================================

BASE = "/content/drive/MyDrive/Neurothesis Resoure Library"
FILE = os.path.join(
    BASE,
    "Within_Dermal_Degree_Stratified_Donor_Results.csv"
)

df = pd.read_csv(FILE)

print("=" * 70)
print("DONOR-LEVEL RESULTS AUDIT")
print("=" * 70)

# 1. Inspect the dataset
print("\nDATASET SHAPE:", df.shape)
print("\nCOLUMNS:")
print(df.columns.tolist())

print("\nPOPULATIONS:")
print(df["population"].value_counts(dropna=False))

# 2. Check donor coverage for each population
print("\n" + "=" * 70)
print("DONOR COVERAGE")
print("=" * 70)

coverage = (
    df.groupby("population")
      .agg(
          donor_count=("donor_code", "nunique"),
          result_rows=("donor_code", "size"),
          missing_enrichment=("enrichment_ratio",
                              lambda x: x.isna().sum())
      )
      .reset_index()
)

display(coverage)

# 3. Check for duplicate donor-population combinations
duplicates = df.duplicated(
    subset=["donor_code", "population"],
    keep=False
)

print("\nDUPLICATE DONOR-POPULATION ROWS:", duplicates.sum())

if duplicates.any():
    display(df.loc[duplicates].sort_values(
        ["population", "donor_code"]
    ))
else:
    print("No duplicate donor-population combinations.")

# 4. Descriptive statistics for each population
print("\n" + "=" * 70)
print("DONOR-LEVEL ENRICHMENT SUMMARY")
print("=" * 70)

summary = (
    df.groupby("population")["enrichment_ratio"]
      .agg(
          donor_count="count",
          mean="mean",
          median="median",
          standard_deviation="std",
          minimum="min",
          maximum="max"
      )
      .reset_index()
)

summary["donors_above_1"] = (
    df.assign(above_1=df["enrichment_ratio"] > 1)
      .groupby("population")["above_1"]
      .sum()
      .values
)

summary["donors_below_1"] = (
    df.assign(below_1=df["enrichment_ratio"] < 1)
      .groupby("population")["below_1"]
      .sum()
      .values
)

display(summary.round(4))

# 5. Create a paired donor-level comparison
wide = df.pivot(
    index="donor_code",
    columns="population",
    values="enrichment_ratio"
)

print("\n" + "=" * 70)
print("PAIRED DONOR COMPARISON")
print("=" * 70)

print("Donors represented in both populations:", wide.dropna().shape[0])
print("Donors missing at least one population:", wide.isna().any(axis=1).sum())

display(wide.round(4))

# 6. Save audit outputs
coverage.to_csv(
    os.path.join(BASE, "Donor_Level_Coverage_Audit.csv"),
    index=False
)

summary.to_csv(
    os.path.join(BASE, "Donor_Level_Enrichment_Descriptive_Summary.csv"),
    index=False
)

wide.to_csv(
    os.path.join(BASE, "Donor_Level_Paired_Enrichment_Table.csv")
)

print("\n" + "=" * 70)
print("SAVED THREE OUTPUT FILES")
print("=" * 70)
print("1. Donor_Level_Coverage_Audit.csv")
print("2. Donor_Level_Enrichment_Descriptive_Summary.csv")
print("3. Donor_Level_Paired_Enrichment_Table.csv")
```

```python

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import wilcoxon, binomtest

# ============================================================
# PAIRED DONOR-LEVEL COMPARISON
# Perivasc Fib I vs Perivasc Fib II
# ============================================================

BASE = "/content/drive/MyDrive/Neurothesis Resoure Library"

FILE = os.path.join(
    BASE,
    "Within_Dermal_Degree_Stratified_Donor_Results.csv"
)

df = pd.read_csv(FILE)

# ------------------------------------------------------------
# 1. Prepare paired donor-level data
# ------------------------------------------------------------

wide = df.pivot(
    index="donor_code",
    columns="population",
    values="enrichment_ratio"
).dropna()

fib1 = wide["Perivasc Fib I"].to_numpy()
fib2 = wide["Perivasc Fib II"].to_numpy()

differences = fib1 - fib2
n = len(differences)

print("=" * 70)
print("PAIRED DONOR-LEVEL ANALYSIS")
print("=" * 70)

print("Number of paired donors:", n)
print("Difference definition: Fib I minus Fib II")

# ------------------------------------------------------------
# 2. Descriptive paired-difference statistics
# ------------------------------------------------------------

mean_diff = np.mean(differences)
median_diff = np.median(differences)
sd_diff = np.std(differences, ddof=1)

fib1_higher = np.sum(differences > 0)
fib2_higher = np.sum(differences < 0)
ties = np.sum(differences == 0)

print("\nPAIRED DIFFERENCES")
print("Mean difference:", round(mean_diff, 4))
print("Median difference:", round(median_diff, 4))
print("Standard deviation:", round(sd_diff, 4))
print("Fib I higher:", fib1_higher)
print("Fib II higher:", fib2_higher)
print("Ties:", ties)

# ------------------------------------------------------------
# 3. Bootstrap confidence intervals
# Resample donors, not individual cells or samples.
# ------------------------------------------------------------

rng = np.random.default_rng(2026)

N_BOOT = 20000

bootstrap_means = np.empty(N_BOOT)
bootstrap_medians = np.empty(N_BOOT)

for i in range(N_BOOT):
    sample = rng.choice(differences, size=n, replace=True)

    bootstrap_means[i] = np.mean(sample)
    bootstrap_medians[i] = np.median(sample)

mean_ci = np.percentile(
    bootstrap_means, [2.5, 97.5]
)

median_ci = np.percentile(
    bootstrap_medians, [2.5, 97.5]
)

print("\nBOOTSTRAP 95% CONFIDENCE INTERVALS")
print("Mean difference CI:",
      np.round(mean_ci, 4))
print("Median difference CI:",
      np.round(median_ci, 4))

# ------------------------------------------------------------
# 4. Paired statistical tests
# ------------------------------------------------------------

# Wilcoxon signed-rank test
wilcoxon_result = wilcoxon(
    fib1,
    fib2,
    alternative="two-sided",
    zero_method="wilcox",
    method="auto"
)

# Exact sign test, excluding ties
non_tied = fib1_higher + fib2_higher

if non_tied > 0:
    sign_result = binomtest(
        fib1_higher,
        n=non_tied,
        p=0.5,
        alternative="two-sided"
    )
    sign_p = sign_result.pvalue
else:
    sign_p = np.nan

print("\nSTATISTICAL TESTS")
print("Wilcoxon statistic:",
      round(wilcoxon_result.statistic, 4))
print("Wilcoxon p-value:",
      f"{wilcoxon_result.pvalue:.6f}")
print("Exact sign-test p-value:",
      f"{sign_p:.6f}")

# ------------------------------------------------------------
# 5. Save donor-level paired differences
# ------------------------------------------------------------

paired_results = wide.copy()

paired_results["Fib_I_minus_Fib_II"] = differences
paired_results["Higher_enrichment"] = np.where(
    differences > 0,
    "Perivasc Fib I",
    np.where(
        differences < 0,
        "Perivasc Fib II",
        "Tie"
    )
)

paired_results.to_csv(
    os.path.join(
        BASE,
        "Paired_Donor_Enrichment_Differences.csv"
    )
)

# ------------------------------------------------------------
# 6. Create publication-style figure
# ------------------------------------------------------------

fig, axes = plt.subplots(
    1, 2,
    figsize=(14, 6)
)

# Panel A: paired donor values
ax = axes[0]

for i in range(n):
    ax.plot(
        [0, 1],
        [fib1[i], fib2[i]],
        marker="o",
        linewidth=1.2,
        alpha=0.65
    )

ax.set_xticks([0, 1])
ax.set_xticklabels([
    "Perivasc Fib I",
    "Perivasc Fib II"
])

ax.set_ylabel("Donor-level enrichment ratio")
ax.set_title("A. Paired enrichment across donors")
ax.axhline(1, linestyle="--", linewidth=1)
ax.grid(axis="y", alpha=0.25)

# Panel B: paired differences
ax = axes[1]

ax.axhline(
    0,
    linestyle="--",
    linewidth=1
)

ax.scatter(
    np.arange(n),
    differences,
    s=45
)

ax.axhline(
    median_diff,
    linestyle="-",
    linewidth=1.5,
    label=f"Median difference = {median_diff:.3f}"
)

ax.set_xlabel("Donor")
ax.set_ylabel("Fib I minus Fib II enrichment")
ax.set_title("B. Within-donor enrichment differences")
ax.legend()
ax.grid(axis="y", alpha=0.25)

fig.suptitle(
    "Donor-Level Comparison of Perivascular Fibroblast Populations",
    fontsize=14,
    y=1.02
)

plt.tight_layout()

figure_path = os.path.join(
    BASE,
    "Figure_Paired_Donor_Fibroblast_Enrichment.png"
)

plt.savefig(
    figure_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ------------------------------------------------------------
# 7. Save statistical summary
# ------------------------------------------------------------

stats_summary = pd.DataFrame([{
    "paired_donors": n,
    "mean_difference_FibI_minus_FibII": mean_diff,
    "median_difference_FibI_minus_FibII": median_diff,
    "difference_SD": sd_diff,
    "mean_difference_CI_lower": mean_ci[0],
    "mean_difference_CI_upper": mean_ci[1],
    "median_difference_CI_lower": median_ci[0],
    "median_difference_CI_upper": median_ci[1],
    "FibI_higher": fib1_higher,
    "FibII_higher": fib2_higher,
    "ties": ties,
    "wilcoxon_p": wilcoxon_result.pvalue,
    "exact_sign_test_p": sign_p
}])

stats_summary.to_csv(
    os.path.join(
        BASE,
        "Paired_Donor_Enrichment_Statistical_Summary.csv"
    ),
    index=False
)

print("\n" + "=" * 70)
print("FILES SAVED")
print("=" * 70)

print("Paired_Donor_Enrichment_Differences.csv")
print("Paired_Donor_Enrichment_Statistical_Summary.csv")
print("Figure_Paired_Donor_Fibroblast_Enrichment.png")
```

```python

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import wilcoxon, binomtest

# ============================================================
# LEAVE-ONE-DONOR-OUT (LODO) ROBUSTNESS ANALYSIS
# Fib I vs Fib II
# ============================================================

BASE = "/content/drive/MyDrive/Neurothesis Resoure Library"

RESULTS_FILE = os.path.join(
    BASE,
    "Within_Dermal_Degree_Stratified_Donor_Results.csv"
)

MAPPING_FILE = os.path.join(
    BASE,
    "Verified_H5AD_Donor_Code_Mapping.csv"
)

df = pd.read_csv(RESULTS_FILE)

wide = df.pivot(
    index="donor_code",
    columns="population",
    values="enrichment_ratio"
).dropna()

wide = wide.sort_index()

# Optional donor-code to donor-ID mapping
if os.path.exists(MAPPING_FILE):
    mapping = pd.read_csv(MAPPING_FILE)
    mapping = mapping[["donor_code", "donor_id"]].drop_duplicates()
    wide = wide.reset_index().merge(
        mapping,
        on="donor_code",
        how="left"
    ).set_index("donor_code")
else:
    wide = wide.reset_index()
    wide["donor_id"] = wide["donor_code"].astype(str)
    wide = wide.set_index("donor_code")

fib1 = wide["Perivasc Fib I"].to_numpy()
fib2 = wide["Perivasc Fib II"].to_numpy()
donor_codes = wide.index.to_numpy()
donor_ids = wide["donor_id"].to_numpy()

differences = fib1 - fib2

# ------------------------------------------------------------
# 1. Full-dataset reference estimates
# ------------------------------------------------------------

full_mean = np.mean(differences)
full_median = np.median(differences)

full_wilcoxon = wilcoxon(
    fib1, fib2,
    alternative="two-sided",
    zero_method="wilcox",
    method="auto"
).pvalue

full_sign = binomtest(
    np.sum(differences > 0),
    n=np.sum(differences != 0),
    p=0.5,
    alternative="two-sided"
).pvalue

print("=" * 70)
print("FULL DATASET REFERENCE")
print("=" * 70)
print("Donors:", len(differences))
print("Mean paired difference:", round(full_mean, 5))
print("Median paired difference:", round(full_median, 5))
print("Wilcoxon p-value:", round(full_wilcoxon, 6))
print("Sign-test p-value:", round(full_sign, 6))

# ------------------------------------------------------------
# 2. Leave one donor out
# ------------------------------------------------------------

lodo_rows = []

for i in range(len(differences)):

    keep = np.arange(len(differences)) != i

    d1 = fib1[keep]
    d2 = fib2[keep]
    diff = d1 - d2

    w_p = wilcoxon(
        d1, d2,
        alternative="two-sided",
        zero_method="wilcox",
        method="auto"
    ).pvalue

    positive = np.sum(diff > 0)
    negative = np.sum(diff < 0)
    non_tied = positive + negative

    s_p = binomtest(
        positive,
        n=non_tied,
        p=0.5,
        alternative="two-sided"
    ).pvalue

    lodo_rows.append({
        "omitted_donor_code": donor_codes[i],
        "omitted_donor_id": donor_ids[i],
        "omitted_donor_difference": differences[i],
        "remaining_donors": len(diff),
        "mean_difference": np.mean(diff),
        "median_difference": np.median(diff),
        "wilcoxon_p": w_p,
        "sign_test_p": s_p,
        "Fib_I_higher": positive,
        "Fib_II_higher": negative
    })

lodo = pd.DataFrame(lodo_rows)

# Influence = change from full-dataset estimate
lodo["mean_estimate_change"] = (
    lodo["mean_difference"] - full_mean
)

lodo["median_estimate_change"] = (
    lodo["median_difference"] - full_median
)

lodo["absolute_mean_influence"] = (
    lodo["mean_estimate_change"].abs()
)

lodo["absolute_median_influence"] = (
    lodo["median_estimate_change"].abs()
)

# ------------------------------------------------------------
# 3. Print robustness summary
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("LEAVE-ONE-DONOR-OUT ROBUSTNESS SUMMARY")
print("=" * 70)

print("\nMEAN DIFFERENCE RANGE:")
print(
    round(lodo["mean_difference"].min(), 5),
    "to",
    round(lodo["mean_difference"].max(), 5)
)

print("\nMEDIAN DIFFERENCE RANGE:")
print(
    round(lodo["median_difference"].min(), 5),
    "to",
    round(lodo["median_difference"].max(), 5)
)

print("\nWILCOXON P-VALUE RANGE:")
print(
    round(lodo["wilcoxon_p"].min(), 6),
    "to",
    round(lodo["wilcoxon_p"].max(), 6)
)

print("\nSIGN-TEST P-VALUE RANGE:")
print(
    round(lodo["sign_test_p"].min(), 6),
    "to",
    round(lodo["sign_test_p"].max(), 6)
)

print("\nDONORS WITH LARGEST MEAN INFLUENCE:")
display(
    lodo.sort_values(
        "absolute_mean_influence",
        ascending=False
    ).head(5).round(5)
)

print("\nDONORS WITH LARGEST MEDIAN INFLUENCE:")
display(
    lodo.sort_values(
        "absolute_median_influence",
        ascending=False
    ).head(5).round(5)
)

# ------------------------------------------------------------
# 4. Check whether conclusions change after donor removal
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("ROBUSTNESS CHECK")
print("=" * 70)

print(
    "LODO analyses with Wilcoxon p < 0.05:",
    (lodo["wilcoxon_p"] < 0.05).sum(),
    "of",
    len(lodo)
)

print(
    "LODO analyses with sign-test p < 0.05:",
    (lodo["sign_test_p"] < 0.05).sum(),
    "of",
    len(lodo)
)

print(
    "LODO mean estimates above zero:",
    (lodo["mean_difference"] > 0).sum()
)

print(
    "LODO median estimates above zero:",
    (lodo["median_difference"] > 0).sum()
)

# ------------------------------------------------------------
# 5. Save results
# ------------------------------------------------------------

lodo_path = os.path.join(
    BASE,
    "Paired_Donor_Leave_One_Out_Robustness.csv"
)

lodo.to_csv(lodo_path, index=False)

# ------------------------------------------------------------
# 6. Plot leave-one-donor-out estimates
# ------------------------------------------------------------

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Mean estimates
ax = axes[0]
ax.axhline(0, linestyle="--", linewidth=1)
ax.axhline(
    full_mean,
    linestyle=":",
    linewidth=1.5,
    label=f"Full estimate = {full_mean:.3f}"
)

ax.scatter(
    np.arange(len(lodo)),
    lodo["mean_difference"],
    s=45
)

ax.set_title("A. Leave-one-donor-out mean differences")
ax.set_xlabel("Omitted donor")
ax.set_ylabel("Mean Fib I minus Fib II enrichment")
ax.legend()
ax.grid(axis="y", alpha=0.25)

# Median estimates
ax = axes[1]
ax.axhline(0, linestyle="--", linewidth=1)
ax.axhline(
    full_median,
    linestyle=":",
    linewidth=1.5,
    label=f"Full estimate = {full_median:.3f}"
)

ax.scatter(
    np.arange(len(lodo)),
    lodo["median_difference"],
    s=45
)

ax.set_title("B. Leave-one-donor-out median differences")
ax.set_xlabel("Omitted donor")
ax.set_ylabel("Median Fib I minus Fib II enrichment")
ax.legend()
ax.grid(axis="y", alpha=0.25)

fig.suptitle(
    "Sensitivity of Paired Donor Comparison to Individual Donors",
    fontsize=14
)

plt.tight_layout()

figure_path = os.path.join(
    BASE,
    "Figure_Paired_Donor_Leave_One_Out.png"
)

plt.savefig(
    figure_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nSAVED:")
print(lodo_path)
print(figure_path)
```

```python

import os
import pandas as pd
import numpy as np

# ============================================================
# ANATOMICAL-SITE COVERAGE AND DONOR REPRESENTATION AUDIT
# ============================================================

BASE = "/content/drive/MyDrive/Neurothesis Resoure Library"

FILE = os.path.join(
    BASE,
    "Verified_Sample_Level_Degree_Results_With_IDs.csv"
)

df = pd.read_csv(FILE)

# ------------------------------------------------------------
# 1. Validate required columns
# ------------------------------------------------------------

required = [
    "sample_id",
    "donor_id",
    "anatomic_site",
    "population",
    "enrichment_ratio",
    "target_cell_count"
]

missing = [col for col in required if col not in df.columns]

if missing:
    raise ValueError(
        f"Missing required columns: {missing}"
    )

print("=" * 75)
print("ANATOMICAL-SITE COVERAGE AUDIT")
print("=" * 75)

print("Rows:", len(df))
print("Unique samples:", df["sample_id"].nunique())
print("Unique donors:", df["donor_id"].nunique())
print("Anatomical sites:", df["anatomic_site"].nunique())
print("\nPopulation counts:")
print(df["population"].value_counts())

# ------------------------------------------------------------
# 2. Check sample-population uniqueness
# ------------------------------------------------------------

duplicates = df.duplicated(
    subset=["sample_id", "population"],
    keep=False
)

print("\nDuplicate sample-population rows:", duplicates.sum())

if duplicates.any():
    display(
        df.loc[duplicates].sort_values(
            ["sample_id", "population"]
        )
    )

# Confirm each sample has both populations
sample_population_counts = (
    df.groupby("sample_id")["population"]
      .nunique()
)

print(
    "\nSamples with both populations:",
    (sample_population_counts == 2).sum()
)

print(
    "Samples missing one or more populations:",
    (sample_population_counts != 2).sum()
)

# ------------------------------------------------------------
# 3. Build one-row-per-sample dataset
# ------------------------------------------------------------

sample_info = (
    df[
        ["sample_id", "donor_id", "anatomic_site"]
    ]
    .drop_duplicates()
)

print("\nSample metadata rows:", len(sample_info))
print("Unique sample IDs:", sample_info["sample_id"].nunique())

# Check whether a sample maps to multiple donors or sites
sample_mapping_check = (
    df.groupby("sample_id")
      .agg(
          donor_count=("donor_id", "nunique"),
          site_count=("anatomic_site", "nunique")
      )
      .reset_index()
)

mapping_problems = sample_mapping_check[
    (sample_mapping_check["donor_count"] != 1) |
    (sample_mapping_check["site_count"] != 1)
]

print(
    "Samples with inconsistent donor/site mapping:",
    len(mapping_problems)
)

if len(mapping_problems) > 0:
    display(mapping_problems)

# ------------------------------------------------------------
# 4. Sample and donor representation by anatomical site
# ------------------------------------------------------------

site_coverage = (
    sample_info.groupby("anatomic_site")
      .agg(
          sample_count=("sample_id", "nunique"),
          donor_count=("donor_id", "nunique")
      )
      .reset_index()
      .sort_values(
          ["donor_count", "sample_count"],
          ascending=False
      )
)

print("\n" + "=" * 75)
print("SAMPLE AND DONOR COVERAGE BY ANATOMICAL SITE")
print("=" * 75)

display(site_coverage)

# ------------------------------------------------------------
# 5. Enrichment summaries by site and population
# ------------------------------------------------------------

site_enrichment = (
    df.groupby(
        ["anatomic_site", "population"]
    )
    .agg(
        sample_count=("sample_id", "nunique"),
        donor_count=("donor_id", "nunique"),
        mean_enrichment=("enrichment_ratio", "mean"),
        median_enrichment=("enrichment_ratio", "median"),
        sd_enrichment=("enrichment_ratio", "std"),
        minimum_enrichment=("enrichment_ratio", "min"),
        maximum_enrichment=("enrichment_ratio", "max"),
        median_target_cells=("target_cell_count", "median")
    )
    .reset_index()
    .sort_values(
        ["anatomic_site", "population"]
    )
)

print("\n" + "=" * 75)
print("ENRICHMENT DESCRIPTIVES BY SITE AND POPULATION")
print("=" * 75)

display(site_enrichment.round(4))

# ------------------------------------------------------------
# 6. Identify donors contributing multiple samples
# ------------------------------------------------------------

donor_representation = (
    sample_info.groupby("donor_id")
      .agg(
          sample_count=("sample_id", "nunique"),
          site_count=("anatomic_site", "nunique"),
          anatomical_sites=(
              "anatomic_site",
              lambda x: "; ".join(
                  sorted(x.dropna().astype(str).unique())
              )
          )
      )
      .reset_index()
      .sort_values(
          ["sample_count", "site_count"],
          ascending=False
      )
)

print("\n" + "=" * 75)
print("SAMPLES AND SITES CONTRIBUTED BY EACH DONOR")
print("=" * 75)

display(donor_representation)

# ------------------------------------------------------------
# 7. Donor-by-site sample count matrix
# ------------------------------------------------------------

donor_site_counts = pd.crosstab(
    sample_info["donor_id"],
    sample_info["anatomic_site"]
)

print("\n" + "=" * 75)
print("DONOR-BY-SITE SAMPLE COUNT MATRIX")
print("=" * 75)

display(donor_site_counts)

# ------------------------------------------------------------
# 8. Save audit outputs
# ------------------------------------------------------------

site_coverage.to_csv(
    os.path.join(BASE, "Anatomical_Site_Coverage_Audit.csv"),
    index=False
)

site_enrichment.to_csv(
    os.path.join(BASE, "Anatomical_Site_Enrichment_Descriptives.csv"),
    index=False
)

donor_representation.to_csv(
    os.path.join(BASE, "Donor_Anatomical_Site_Representation.csv"),
    index=False
)

donor_site_counts.to_csv(
    os.path.join(BASE, "Donor_By_Anatomical_Site_Sample_Counts.csv")
)

print("\n" + "=" * 75)
print("FOUR AUDIT FILES SAVED")
print("=" * 75)

print("1. Anatomical_Site_Coverage_Audit.csv")
print("2. Anatomical_Site_Enrichment_Descriptives.csv")
print("3. Donor_Anatomical_Site_Representation.csv")
print("4. Donor_By_Anatomical_Site_Sample_Counts.csv")
```

```python

# ================================================================
# DONOR-BALANCED ANATOMICAL-SITE ANALYSIS
# ================================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import wilcoxon
from statsmodels.stats.multitest import multipletests

BASE = "/content/drive/MyDrive/Neurothesis Resoure Library"

INPUT = os.path.join(
    BASE,
    "Verified_Sample_Level_Degree_Results_With_IDs.csv"
)

# ------------------------------------------------
# 1. LOAD AND VALIDATE DATA
# ------------------------------------------------

df = pd.read_csv(INPUT)

required = [
    "sample_id",
    "donor_id",
    "anatomic_site",
    "population",
    "enrichment_ratio"
]

missing = [c for c in required if c not in df.columns]

if missing:
    raise ValueError(f"Missing required columns: {missing}")

df = df[required].copy()

df["enrichment_ratio"] = pd.to_numeric(
    df["enrichment_ratio"], errors="coerce"
)

df = df.dropna(
    subset=[
        "sample_id",
        "donor_id",
        "anatomic_site",
        "population",
        "enrichment_ratio"
    ]
)

print("INPUT VALIDATION")
print("Rows:", len(df))
print("Unique samples:", df["sample_id"].nunique())
print("Unique donors:", df["donor_id"].nunique())
print("Unique anatomical sites:", df["anatomic_site"].nunique())

# ------------------------------------------------
# 2. AGGREGATE REPEATED SAMPLES WITHIN DONOR-SITE
# ------------------------------------------------

# Each donor receives one value per anatomical site.
# The median prevents donors with repeated samples at a
# particular site from receiving extra statistical weight.

donor_site = (
    df.groupby(
        ["donor_id", "anatomic_site", "population"],
        as_index=False
    )
    .agg(
        enrichment_ratio=("enrichment_ratio", "median"),
        sample_count=("sample_id", "nunique"),
        mean_enrichment=("enrichment_ratio", "mean"),
        target_cell_count=(
            "target_cell_count", "median"
        ) if "target_cell_count" in df.columns
        else ("enrichment_ratio", "size")
    )
)

donor_site.to_csv(
    os.path.join(
        BASE,
        "Donor_Site_Population_Aggregated_Results.csv"
    ),
    index=False
)

print("\nDONOR-SITE AGGREGATION")
print("Rows:", len(donor_site))
print("Unique donors:", donor_site["donor_id"].nunique())
print("Unique sites:", donor_site["anatomic_site"].nunique())

# ------------------------------------------------
# 3. CREATE PAIRED DONOR-SITE COMPARISONS
# ------------------------------------------------

paired = donor_site.pivot_table(
    index=["donor_id", "anatomic_site"],
    columns="population",
    values="enrichment_ratio",
    aggfunc="first"
).reset_index()

required_populations = [
    "Perivasc Fib I",
    "Perivasc Fib II"
]

paired = paired.dropna(
    subset=required_populations
).copy()

paired["paired_difference"] = (
    paired["Perivasc Fib I"]
    - paired["Perivasc Fib II"]
)

paired["higher_population"] = np.select(
    [
        paired["paired_difference"] > 0,
        paired["paired_difference"] < 0
    ],
    [
        "Perivasc Fib I",
        "Perivasc Fib II"
    ],
    default="Tie"
)

paired.to_csv(
    os.path.join(
        BASE,
        "Donor_Site_Paired_Fibroblast_Differences.csv"
    ),
    index=False
)

print("\nPAIRED COMPARISON")
print("Paired donor-site observations:", len(paired))
print("Donors:", paired["donor_id"].nunique())
print("Sites:", paired["anatomic_site"].nunique())

# ------------------------------------------------
# 4. SUMMARIZE EACH ANATOMICAL SITE
# ------------------------------------------------

summary_rows = []

for site, group in paired.groupby("anatomic_site"):

    differences = group["paired_difference"].dropna()
    n = len(differences)

    row = {
        "anatomic_site": site,
        "donor_count": n,
        "fib_I_higher": int((differences > 0).sum()),
        "fib_II_higher": int((differences < 0).sum()),
        "ties": int((differences == 0).sum()),
        "mean_difference": differences.mean(),
        "median_difference": differences.median(),
        "sd_difference": differences.std(),
        "minimum_difference": differences.min(),
        "maximum_difference": differences.max()
    }

    # Only run inferential tests when there are enough
    # paired donors and the differences are not all zero.

    if n >= 5 and (differences != 0).any():

        try:
            test = wilcoxon(
                differences,
                alternative="two-sided",
                method="auto"
            )
            row["wilcoxon_statistic"] = test.statistic
            row["wilcoxon_p"] = test.pvalue

        except Exception:
            row["wilcoxon_statistic"] = np.nan
            row["wilcoxon_p"] = np.nan

    else:
        row["wilcoxon_statistic"] = np.nan
        row["wilcoxon_p"] = np.nan

    summary_rows.append(row)

site_summary = pd.DataFrame(summary_rows)

# Correct for multiple site-level tests.
valid = site_summary["wilcoxon_p"].notna()

site_summary["wilcoxon_q_BH"] = np.nan

if valid.sum() > 0:
    site_summary.loc[valid, "wilcoxon_q_BH"] = (
        multipletests(
            site_summary.loc[valid, "wilcoxon_p"],
            method="fdr_bh"
        )[1]
    )

site_summary = site_summary.sort_values(
    "donor_count",
    ascending=False
)

site_summary.to_csv(
    os.path.join(
        BASE,
        "Donor_Balanced_Anatomical_Site_Summary.csv"
    ),
    index=False
)

# ------------------------------------------------
# 5. SITE COVERAGE AUDIT
# ------------------------------------------------

coverage = (
    paired.groupby("anatomic_site")
    .agg(
        donor_count=("donor_id", "nunique"),
        paired_observations=("paired_difference", "count"),
        fib_I_higher=("paired_difference", lambda x: (x > 0).sum()),
        fib_II_higher=("paired_difference", lambda x: (x < 0).sum()),
        ties=("paired_difference", lambda x: (x == 0).sum())
    )
    .reset_index()
)

coverage.to_csv(
    os.path.join(
        BASE,
        "Donor_Balanced_Site_Coverage.csv"
    ),
    index=False
)

# ------------------------------------------------
# 6. DISPLAY RESULTS
# ------------------------------------------------

print("\n" + "=" * 75)
print("DONOR-BALANCED ANATOMICAL-SITE SUMMARY")
print("=" * 75)

display(site_summary.round(4))

print("\n" + "=" * 75)
print("SITE COVERAGE")
print("=" * 75)

display(coverage)

print("\n" + "=" * 75)
print("PAIRED DONOR-SITE DATA")
print("=" * 75)

display(paired.head(15).round(4))

# ------------------------------------------------
# 7. VISUALIZE SITE-LEVEL PAIRED DIFFERENCES
# ------------------------------------------------

plot_data = site_summary.sort_values(
    "median_difference"
)

plt.figure(figsize=(12, 7))

plt.axvline(
    0,
    linestyle="--",
    linewidth=1
)

plt.scatter(
    plot_data["median_difference"],
    plot_data["anatomic_site"],
    s=75
)

plt.xlabel(
    "Median paired enrichment difference "
    "(Perivasc Fib I minus Perivasc Fib II)"
)

plt.ylabel("Anatomical site")

plt.title(
    "Donor-Balanced Fibroblast Enrichment Differences "
    "Across Anatomical Sites"
)

plt.grid(axis="x", alpha=0.25)
plt.tight_layout()

plt.savefig(
    os.path.join(
        BASE,
        "Figure_Donor_Balanced_Site_Differences.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ------------------------------------------------
# 8. SAVE FINAL OUTPUTS
# ------------------------------------------------

print("\n" + "=" * 75)
print("DONOR-BALANCED ANALYSIS COMPLETE")
print("=" * 75)

print("Saved:")
print("1. Donor_Site_Population_Aggregated_Results.csv")
print("2. Donor_Site_Paired_Fibroblast_Differences.csv")
print("3. Donor_Balanced_Anatomical_Site_Summary.csv")
print("4. Donor_Balanced_Site_Coverage.csv")
print("5. Figure_Donor_Balanced_Site_Differences.png")
```

```python

# ============================================================
# DONOR-BALANCED ANATOMICAL-SITE ANALYSIS
# Paired differences with donor-level bootstrap confidence intervals
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path

BASE = Path("/content/drive/MyDrive/Neurothesis Resoure Library")

INPUT_FILE = BASE / "Verified_Sample_Level_Degree_Results_With_IDs.csv"

if not INPUT_FILE.exists():
    raise FileNotFoundError(f"Missing input file: {INPUT_FILE}")

df = pd.read_csv(INPUT_FILE)

# ------------------------------------------------------------
# 1. Validate required columns
# ------------------------------------------------------------

required = [
    "sample_id",
    "donor_id",
    "anatomic_site",
    "population",
    "enrichment_ratio"
]

missing = [c for c in required if c not in df.columns]

if missing:
    raise ValueError(f"Missing required columns: {missing}")

df = df[required].copy()

df["enrichment_ratio"] = pd.to_numeric(
    df["enrichment_ratio"], errors="coerce"
)

df = df.dropna(subset=[
    "sample_id",
    "donor_id",
    "anatomic_site",
    "population",
    "enrichment_ratio"
])

# ------------------------------------------------------------
# 2. Average repeated samples within donor-site-population
# ------------------------------------------------------------

donor_site = (
    df.groupby(
        ["donor_id", "anatomic_site", "population"],
        as_index=False
    )
    .agg(
        donor_site_enrichment=("enrichment_ratio", "mean"),
        contributing_samples=("sample_id", "nunique"),
        total_target_cells=("target_cell_count", "sum")
        if "target_cell_count" in df.columns
        else ("enrichment_ratio", "size")
    )
)

print("DONOR-SITE AGGREGATION")
print("Rows:", len(donor_site))
print("Unique donors:", donor_site["donor_id"].nunique())
print("Unique sites:", donor_site["anatomic_site"].nunique())

# ------------------------------------------------------------
# 3. Create paired Fib I and Fib II values
# ------------------------------------------------------------

paired = donor_site.pivot_table(
    index=["donor_id", "anatomic_site"],
    columns="population",
    values="donor_site_enrichment",
    aggfunc="first"
).reset_index()

required_populations = [
    "Perivasc Fib I",
    "Perivasc Fib II"
]

for pop in required_populations:
    if pop not in paired.columns:
        raise ValueError(f"Missing population: {pop}")

paired = paired.dropna(subset=required_populations).copy()

paired["paired_difference"] = (
    paired["Perivasc Fib I"] -
    paired["Perivasc Fib II"]
)

print("\nPAIRED DONOR-SITE DATA")
print("Paired donor-site observations:", len(paired))
print("Donors:", paired["donor_id"].nunique())
print("Sites:", paired["anatomic_site"].nunique())

# ------------------------------------------------------------
# 4. Bootstrap confidence intervals by site
# ------------------------------------------------------------

rng = np.random.default_rng(20261114)

N_BOOT = 10000
summary_rows = []

for site, group in paired.groupby("anatomic_site"):

    values = group["paired_difference"].to_numpy()
    n = len(values)

    observed_median = np.median(values)
    observed_mean = np.mean(values)

    if n >= 2:
        boot_medians = np.empty(N_BOOT)
        boot_means = np.empty(N_BOOT)

        for i in range(N_BOOT):
            sample = rng.choice(values, size=n, replace=True)
            boot_medians[i] = np.median(sample)
            boot_means[i] = np.mean(sample)

        median_ci = np.percentile(
            boot_medians, [2.5, 97.5]
        )

        mean_ci = np.percentile(
            boot_means, [2.5, 97.5]
        )

    else:
        median_ci = [np.nan, np.nan]
        mean_ci = [np.nan, np.nan]

    summary_rows.append({
        "anatomic_site": site,
        "donor_count": n,
        "median_paired_difference": observed_median,
        "median_ci_lower": median_ci[0],
        "median_ci_upper": median_ci[1],
        "mean_paired_difference": observed_mean,
        "mean_ci_lower": mean_ci[0],
        "mean_ci_upper": mean_ci[1],
        "fib_I_higher_donors": int((values > 0).sum()),
        "fib_II_higher_donors": int((values < 0).sum()),
        "ties": int((values == 0).sum())
    })

site_summary = pd.DataFrame(summary_rows)

site_summary = site_summary.sort_values(
    "median_paired_difference"
).reset_index(drop=True)

# ------------------------------------------------------------
# 5. Print results
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("DONOR-BALANCED SITE-LEVEL PAIRED DIFFERENCES")
print("=" * 75)

print(site_summary.round(4).to_string(index=False))

# ------------------------------------------------------------
# 6. Save outputs
# ------------------------------------------------------------

paired.to_csv(
    BASE / "Donor_Site_Paired_Differences_With_Values.csv",
    index=False
)

site_summary.to_csv(
    BASE / "Donor_Balanced_Site_Differences_With_CI.csv",
    index=False
)

# ------------------------------------------------------------
# 7. Plot median differences and bootstrap confidence intervals
# ------------------------------------------------------------

plot_data = site_summary.dropna(
    subset=["median_ci_lower", "median_ci_upper"]
).copy()

fig, ax = plt.subplots(figsize=(11, 8))

y = np.arange(len(plot_data))

x = plot_data["median_paired_difference"].to_numpy()

lower_error = (
    x - plot_data["median_ci_lower"].to_numpy()
)

upper_error = (
    plot_data["median_ci_upper"].to_numpy() - x
)

ax.errorbar(
    x,
    y,
    xerr=np.vstack([lower_error, upper_error]),
    fmt="o",
    capsize=4
)

ax.axvline(0, linestyle="--", linewidth=1)

ax.set_yticks(y)
ax.set_yticklabels(
    plot_data["anatomic_site"]
)

ax.set_xlabel(
    "Median paired enrichment difference "
    "(Perivasc Fib I minus Perivasc Fib II)"
)

ax.set_ylabel("Anatomical site")

ax.set_title(
    "Donor-Balanced Fibroblast Enrichment Differences\n"
    "Across Anatomical Sites with Bootstrap 95% CIs"
)

ax.grid(axis="x", alpha=0.25)

plt.tight_layout()

plt.savefig(
    BASE / "Figure_Donor_Balanced_Site_Differences_CI.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nSAVED:")
print("1. Donor_Site_Paired_Differences_With_Values.csv")
print("2. Donor_Balanced_Site_Differences_With_CI.csv")
print("3. Figure_Donor_Balanced_Site_Differences_CI.png")
```

```python

# ============================================================
# SITE-SPECIFIC INFERENCE: DONOR-LEVEL PAIRED DIFFERENCES
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.stats import wilcoxon, binomtest
from statsmodels.stats.multitest import multipletests

BASE = "/content/drive/MyDrive/Neurothesis Resoure Library"

INPUT = os.path.join(
    BASE,
    "Donor_Site_Paired_Differences_With_Values.csv"
)

# ------------------------------------------------------------
# 1. LOAD AND INSPECT DATA
# ------------------------------------------------------------

df = pd.read_csv(INPUT)

print("=" * 75)
print("SITE-SPECIFIC INFERENCE ANALYSIS")
print("=" * 75)

print("\nDataset shape:", df.shape)
print("Columns:", df.columns.tolist())

# Identify the paired-difference column.
difference_candidates = [
    "paired_difference",
    "enrichment_difference",
    "fibroblast_difference",
    "median_paired_difference",
    "difference",
    "Fib_I_minus_Fib_II"
]

difference_col = next(
    (c for c in difference_candidates if c in df.columns),
    None
)

if difference_col is None:
    raise ValueError(
        "Could not identify paired-difference column. "
        "Review the printed column names and update difference_col."
    )

site_col = "anatomic_site"
donor_col = "donor_id"

if site_col not in df.columns or donor_col not in df.columns:
    raise ValueError(
        "Expected anatomic_site and donor_id columns. "
        "Review the printed column names."
    )

df[difference_col] = pd.to_numeric(
    df[difference_col], errors="coerce"
)

df = df.dropna(
    subset=[site_col, donor_col, difference_col]
).copy()

# Verify one observation per donor-site combination.
duplicates = df.duplicated(
    subset=[donor_col, site_col],
    keep=False
)

print("\nDuplicate donor-site observations:", duplicates.sum())

if duplicates.any():
    raise ValueError(
        "Duplicate donor-site observations detected. "
        "Resolve these before inference."
    )

# ------------------------------------------------------------
# 2. RUN SITE-SPECIFIC TESTS
# ------------------------------------------------------------

results = []

for site, group in df.groupby(site_col):

    values = group[difference_col].to_numpy(dtype=float)
    values = values[np.isfinite(values)]

    n = len(values)

    mean_diff = np.mean(values) if n else np.nan
    median_diff = np.median(values) if n else np.nan
    sd_diff = np.std(values, ddof=1) if n > 1 else np.nan

    fib_I_higher = int(np.sum(values > 0))
    fib_II_higher = int(np.sum(values < 0))
    ties = int(np.sum(values == 0))

    # Exact two-sided sign test.
    nonzero = values[values != 0]

    if len(nonzero) >= 1:
        sign_p = binomtest(
            int(np.sum(nonzero > 0)),
            n=len(nonzero),
            p=0.5,
            alternative="two-sided"
        ).pvalue
    else:
        sign_p = np.nan

    # Wilcoxon signed-rank test.
    # Requires at least two nonzero paired differences.
    if len(nonzero) >= 2:
        try:
            wilcoxon_p = wilcoxon(
                values,
                alternative="two-sided",
                zero_method="wilcox",
                method="auto"
            ).pvalue
        except ValueError:
            wilcoxon_p = np.nan
    else:
        wilcoxon_p = np.nan

    results.append({
        "anatomic_site": site,
        "donor_count": n,
        "mean_difference": mean_diff,
        "median_difference": median_diff,
        "sd_difference": sd_diff,
        "Fib_I_higher": fib_I_higher,
        "Fib_II_higher": fib_II_higher,
        "ties": ties,
        "sign_test_p": sign_p,
        "wilcoxon_p": wilcoxon_p
    })

site_tests = pd.DataFrame(results)

# ------------------------------------------------------------
# 3. MULTIPLE-TESTING CORRECTION
# ------------------------------------------------------------

site_tests["wilcoxon_BH_q"] = np.nan
site_tests["sign_test_BH_q"] = np.nan

valid_w = site_tests["wilcoxon_p"].notna()

if valid_w.sum() > 0:
    site_tests.loc[valid_w, "wilcoxon_BH_q"] = (
        multipletests(
            site_tests.loc[valid_w, "wilcoxon_p"],
            method="fdr_bh"
        )[1]
    )

valid_s = site_tests["sign_test_p"].notna()

if valid_s.sum() > 0:
    site_tests.loc[valid_s, "sign_test_BH_q"] = (
        multipletests(
            site_tests.loc[valid_s, "sign_test_p"],
            method="fdr_bh"
        )[1]
    )

site_tests["small_donor_count"] = (
    site_tests["donor_count"] < 5
)

site_tests["wilcoxon_BH_below_0_05"] = (
    site_tests["wilcoxon_BH_q"] < 0.05
)

# ------------------------------------------------------------
# 4. DISPLAY RESULTS
# ------------------------------------------------------------

site_tests = site_tests.sort_values(
    "median_difference",
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 75)
print("SITE-SPECIFIC PAIRED-DIFFERENCE TESTS")
print("=" * 75)

display(
    site_tests.round(5)
)

print("\nSites analyzed:", len(site_tests))
print(
    "Sites with fewer than 5 donors:",
    int(site_tests["small_donor_count"].sum())
)
print(
    "Sites with BH-adjusted Wilcoxon q < 0.05:",
    int(site_tests["wilcoxon_BH_below_0_05"].sum())
)

# ------------------------------------------------------------
# 5. SAVE RESULTS
# ------------------------------------------------------------

output_path = os.path.join(
    BASE,
    "Anatomical_Site_Paired_Inference_BH.csv"
)

site_tests.to_csv(output_path, index=False)

print("\nSaved:", output_path)

# ------------------------------------------------------------
# 6. VISUALIZE SITE-SPECIFIC DIFFERENCES
# ------------------------------------------------------------

plot_df = site_tests.sort_values(
    "median_difference",
    ascending=True
).copy()

fig, ax = plt.subplots(figsize=(11, 8))

y = np.arange(len(plot_df))

ax.scatter(
    plot_df["median_difference"],
    y,
    s=75
)

ax.axvline(
    0,
    linestyle="--",
    linewidth=1
)

ax.set_yticks(y)
ax.set_yticklabels(plot_df["anatomic_site"])

ax.set_xlabel(
    "Median paired enrichment difference "
    "(Perivasc Fib I minus Perivasc Fib II)"
)

ax.set_ylabel("Anatomical site")

ax.set_title(
    "Site-Specific Donor-Level Paired Enrichment Differences"
)

ax.grid(axis="x", alpha=0.25)

plt.tight_layout()

figure_path = os.path.join(
    BASE,
    "Figure_Anatomical_Site_Paired_Inference.png"
)

plt.savefig(
    figure_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nSaved figure:", figure_path)

print("\n" + "=" * 75)
print("SITE-SPECIFIC INFERENCE COMPLETE")
print("=" * 75)
```

```python

# ============================================================
# DONOR-ADJUSTED ANATOMICAL-SITE MODEL
# Donor fixed effects + site fixed effects
# Donor-clustered standard errors
# ============================================================

import os
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf

from pathlib import Path

BASE = Path("/content/drive/MyDrive/Neurothesis Resoure Library")

INPUT = BASE / "Donor_Site_Paired_Differences_With_Values.csv"

df = pd.read_csv(INPUT)

# ------------------------------------------------------------
# 1. VALIDATE DATA
# ------------------------------------------------------------

required = [
    "donor_id",
    "anatomic_site",
    "paired_difference"
]

missing = [c for c in required if c not in df.columns]

if missing:
    raise ValueError(f"Missing columns: {missing}")

df["paired_difference"] = pd.to_numeric(
    df["paired_difference"],
    errors="coerce"
)

df = df.dropna(subset=required).copy()

df["donor_id"] = df["donor_id"].astype(str)
df["anatomic_site"] = df["anatomic_site"].astype(str)

print("=" * 75)
print("DONOR-ADJUSTED SITE MODEL")
print("=" * 75)

print("Observations:", len(df))
print("Donors:", df["donor_id"].nunique())
print("Anatomical sites:", df["anatomic_site"].nunique())

# ------------------------------------------------------------
# 2. CHECK WHICH SITES HAVE WITHIN-DONOR SUPPORT
# ------------------------------------------------------------

donor_site_counts = (
    df.groupby("anatomic_site")
    .agg(
        donor_count=("donor_id", "nunique"),
        observation_count=("paired_difference", "size")
    )
    .reset_index()
)

donors_with_multiple_sites = (
    df.groupby("donor_id")["anatomic_site"]
    .nunique()
)

print("\nDONOR REPEATED-MEASUREMENT COVERAGE")
print(
    "Donors with multiple anatomical sites:",
    int((donors_with_multiple_sites > 1).sum())
)
print(
    "Donors with only one anatomical site:",
    int((donors_with_multiple_sites == 1).sum())
)

display(donor_site_counts.sort_values(
    "donor_count", ascending=False
))

# ------------------------------------------------------------
# 3. FIT DONOR-ADJUSTED MODEL
# ------------------------------------------------------------

# Donor fixed effects control for stable differences between donors.
# Site fixed effects estimate site-associated differences.
# Clustered covariance accounts for repeated observations per donor.

formula = (
    "paired_difference ~ "
    "C(donor_id) + C(anatomic_site)"
)

model = smf.ols(
    formula=formula,
    data=df
).fit(
    cov_type="cluster",
    cov_kwds={
        "groups": df["donor_id"],
        "use_correction": True
    },
    use_t=True
)

print("\n" + "=" * 75)
print("DONOR-ADJUSTED MODEL SUMMARY")
print("=" * 75)

print(model.summary())

# ------------------------------------------------------------
# 4. TEST THE OVERALL ANATOMICAL-SITE EFFECT
# ------------------------------------------------------------

term_tests = model.wald_test_terms(
    skip_single=False,
    scalar=True
)

print("\n" + "=" * 75)
print("TERM-LEVEL ROBUST TESTS")
print("=" * 75)

print(term_tests)

# Extract site effect test.
site_test = term_tests.table.loc[
    term_tests.table.index.str.contains(
        "anatomic_site",
        case=False
    )
]

print("\n" + "=" * 75)
print("OVERALL SITE EFFECT")
print("=" * 75)

display(site_test)

# ------------------------------------------------------------
# 5. IDENTIFY SITE COEFFICIENTS
# ------------------------------------------------------------

site_coefficients = pd.DataFrame({
    "term": model.params.index,
    "coefficient": model.params.values,
    "standard_error": model.bse.values,
    "p_value": model.pvalues.values
})

site_coefficients = site_coefficients[
    site_coefficients["term"].str.contains(
        "anatomic_site",
        case=False
    )
].copy()

site_coefficients["ci_lower"] = (
    site_coefficients["coefficient"]
    - 1.96 * site_coefficients["standard_error"]
)

site_coefficients["ci_upper"] = (
    site_coefficients["coefficient"]
    + 1.96 * site_coefficients["standard_error"]
)

print("\nSITE COEFFICIENTS")
display(site_coefficients.round(5))

# ------------------------------------------------------------
# 6. MODEL DIAGNOSTICS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("MODEL DIAGNOSTICS")
print("=" * 75)

print("Model observations:", int(model.nobs))
print("Model R-squared:", round(model.rsquared, 5))
print("Design matrix rank:", np.linalg.matrix_rank(model.model.exog))
print("Design matrix columns:", model.model.exog.shape[1])
print(
    "Residual degrees of freedom:",
    round(model.df_resid, 2)
)

if np.linalg.matrix_rank(model.model.exog) < model.model.exog.shape[1]:
    print(
        "WARNING: The design matrix is rank deficient. "
        "Some donor-site effects may not be separately identifiable."
    )
else:
    print("Design matrix is full rank.")

# ------------------------------------------------------------
# 7. SAVE RESULTS
# ------------------------------------------------------------

donor_site_counts.to_csv(
    BASE / "Donor_Adjusted_Model_Site_Coverage.csv",
    index=False
)

site_coefficients.to_csv(
    BASE / "Donor_Adjusted_Model_Site_Coefficients.csv",
    index=False
)

print("\nSAVED:")
print("1. Donor_Adjusted_Model_Site_Coverage.csv")
print("2. Donor_Adjusted_Model_Site_Coefficients.csv")

print("\n" + "=" * 75)
print("DONOR-ADJUSTED SITE MODEL COMPLETE")
print("=" * 75)
```

```python

# ============================================================
# DONOR-RESTRICTED WITHIN-DONOR SITE PERMUTATION TEST
# ============================================================

import numpy as np
import pandas as pd
from pathlib import Path

BASE = Path("/content/drive/MyDrive/Neurothesis Resoure Library")

INPUT = BASE / "Donor_Site_Paired_Differences_With_Values.csv"

df = pd.read_csv(INPUT)

# ------------------------------------------------------------
# 1. VALIDATE INPUT
# ------------------------------------------------------------

required = [
    "donor_id",
    "anatomic_site",
    "paired_difference"
]

missing = [c for c in required if c not in df.columns]

if missing:
    raise ValueError(f"Missing columns: {missing}")

df = df.dropna(subset=required).copy()

df["donor_id"] = df["donor_id"].astype(str)
df["anatomic_site"] = df["anatomic_site"].astype(str)
df["paired_difference"] = pd.to_numeric(
    df["paired_difference"],
    errors="coerce"
)

df = df.dropna(subset=["paired_difference"])

if df.duplicated(["donor_id", "anatomic_site"]).any():
    raise ValueError("Duplicate donor-site observations detected.")

# ------------------------------------------------------------
# 2. IDENTIFY DONORS WITH REPEATED SITE MEASUREMENTS
# ------------------------------------------------------------

donor_site_counts = (
    df.groupby("donor_id")["anatomic_site"]
    .nunique()
)

multi_site_donors = donor_site_counts[
    donor_site_counts >= 2
].index

multi = df[
    df["donor_id"].isin(multi_site_donors)
].copy()

print("=" * 75)
print("INITIAL MULTI-SITE DONOR SUBSET")
print("=" * 75)

print("Original observations:", len(df))
print("Original donors:", df["donor_id"].nunique())
print("Donors with multiple sites:", len(multi_site_donors))
print("Observations retained:", len(multi))

# ------------------------------------------------------------
# 3. REMOVE SITES WITH ONLY ONE CONTRIBUTING DONOR
# ------------------------------------------------------------

site_support = (
    multi.groupby("anatomic_site")["donor_id"]
    .nunique()
    .rename("donor_count")
    .reset_index()
)

supported_sites = site_support.loc[
    site_support["donor_count"] >= 2,
    "anatomic_site"
]

supported = multi[
    multi["anatomic_site"].isin(supported_sites)
].copy()

# Recheck donor support after site filtering.
remaining_site_counts = (
    supported.groupby("donor_id")["anatomic_site"]
    .nunique()
)

eligible_donors = remaining_site_counts[
    remaining_site_counts >= 2
].index

analysis = supported[
    supported["donor_id"].isin(eligible_donors)
].copy()

# Remove any site that now has fewer than two donors.
final_site_support = (
    analysis.groupby("anatomic_site")["donor_id"]
    .nunique()
)

final_sites = final_site_support[
    final_site_support >= 2
].index

analysis = analysis[
    analysis["anatomic_site"].isin(final_sites)
].copy()

# Final donor eligibility check.
final_donor_sites = (
    analysis.groupby("donor_id")["anatomic_site"]
    .nunique()
)

analysis = analysis[
    analysis["donor_id"].isin(
        final_donor_sites[final_donor_sites >= 2].index
    )
].copy()

print("\n" + "=" * 75)
print("FINAL ANALYSIS SUBSET")
print("=" * 75)

print("Observations:", len(analysis))
print("Donors:", analysis["donor_id"].nunique())
print("Sites:", analysis["anatomic_site"].nunique())

display(
    analysis.groupby("anatomic_site")
    .agg(
        donor_count=("donor_id", "nunique"),
        observation_count=("paired_difference", "size")
    )
    .sort_values("donor_count", ascending=False)
)

if analysis["donor_id"].nunique() < 3:
    raise ValueError(
        "Too few multi-site donors remain for this analysis."
    )

if analysis["anatomic_site"].nunique() < 2:
    raise ValueError(
        "Too few anatomical sites remain for comparison."
    )

# ------------------------------------------------------------
# 4. DEFINE NESTED MODEL TEST STATISTIC
# ------------------------------------------------------------

def model_site_statistic(data):

    y = data["paired_difference"].to_numpy(dtype=float)

    donor_dummies = pd.get_dummies(
        data["donor_id"],
        drop_first=True,
        dtype=float
    )

    site_dummies = pd.get_dummies(
        data["anatomic_site"],
        drop_first=True,
        dtype=float
    )

    intercept = np.ones((len(data), 1))

    X_donor = np.column_stack([
        intercept,
        donor_dummies.to_numpy()
    ])

    X_full = np.column_stack([
        X_donor,
        site_dummies.to_numpy()
    ])

    beta_reduced = np.linalg.lstsq(
        X_donor, y, rcond=None
    )[0]

    beta_full = np.linalg.lstsq(
        X_full, y, rcond=None
    )[0]

    rss_reduced = np.sum(
        (y - X_donor @ beta_reduced) ** 2
    )

    rss_full = np.sum(
        (y - X_full @ beta_full) ** 2
    )

    rank_reduced = np.linalg.matrix_rank(X_donor)
    rank_full = np.linalg.matrix_rank(X_full)

    df_effect = rank_full - rank_reduced
    df_residual = len(y) - rank_full

    if df_effect <= 0 or df_residual <= 0:
        raise ValueError(
            "Site effect is not estimable in this subset."
        )

    improvement = max(0, rss_reduced - rss_full)

    if rss_full <= 1e-12:
        raise ValueError(
            "Residual variance is too small for the F statistic."
        )

    f_stat = (
        (improvement / df_effect)
        /
        (rss_full / df_residual)
    )

    partial_r2 = (
        improvement / rss_reduced
        if rss_reduced > 0 else np.nan
    )

    return {
        "F": f_stat,
        "partial_R2": partial_r2,
        "RSS_donor_only": rss_reduced,
        "RSS_donor_and_site": rss_full,
        "site_df": df_effect,
        "residual_df": df_residual,
        "design_rank": rank_full,
        "design_columns": X_full.shape[1]
    }

observed = model_site_statistic(analysis)

# ------------------------------------------------------------
# 5. PERMUTE SITE LABELS WITHIN EACH DONOR
# ------------------------------------------------------------

N_PERMUTATIONS = 9999
SEED = 20261115

rng = np.random.default_rng(SEED)

observed_f = observed["F"]
permuted_f = np.empty(N_PERMUTATIONS)

donor_groups = {
    donor: group.index.to_numpy()
    for donor, group in analysis.groupby("donor_id")
}

original_sites = analysis["anatomic_site"].copy()

for i in range(N_PERMUTATIONS):

    permuted = analysis.copy()

    for donor, indices in donor_groups.items():

        labels = original_sites.loc[indices].to_numpy()

        permuted.loc[indices, "anatomic_site"] = (
            rng.permutation(labels)
        )

    permuted_f[i] = model_site_statistic(permuted)["F"]

p_value = (
    1 + np.sum(permuted_f >= observed_f)
) / (N_PERMUTATIONS + 1)

# ------------------------------------------------------------
# 6. REPORT RESULTS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("WITHIN-DONOR SITE PERMUTATION RESULTS")
print("=" * 75)

print("Observed F statistic:", round(observed["F"], 5))
print("Permutation p-value:", round(p_value, 6))
print("Partial R-squared:", round(observed["partial_R2"], 5))
print("Site degrees of freedom:", observed["site_df"])
print("Residual degrees of freedom:", observed["residual_df"])
print("Permutations:", N_PERMUTATIONS)

print("\nDESIGN DIAGNOSTICS")

print(
    "Design matrix rank:",
    observed["design_rank"]
)

print(
    "Design matrix columns:",
    observed["design_columns"]
)

print(
    "Rank deficient:",
    observed["design_rank"] < observed["design_columns"]
)

# ------------------------------------------------------------
# 7. DONOR-CENTERED SITE DESCRIPTIVES
# ------------------------------------------------------------

analysis["donor_centered_difference"] = (
    analysis["paired_difference"]
    -
    analysis.groupby("donor_id")["paired_difference"]
    .transform("mean")
)

site_descriptives = (
    analysis.groupby("anatomic_site")
    .agg(
        donor_count=("donor_id", "nunique"),
        observation_count=("paired_difference", "size"),
        mean_centered_difference=(
            "donor_centered_difference", "mean"
        ),
        median_centered_difference=(
            "donor_centered_difference", "median"
        ),
        sd_centered_difference=(
            "donor_centered_difference", "std"
        )
    )
    .reset_index()
    .sort_values("mean_centered_difference")
)

print("\nDONOR-CENTERED SITE DESCRIPTIVES")
display(site_descriptives.round(5))

# ------------------------------------------------------------
# 8. SAVE OUTPUTS
# ------------------------------------------------------------

summary = pd.DataFrame([{
    **observed,
    "permutation_p_value": p_value,
    "permutations": N_PERMUTATIONS,
    "donor_count": analysis["donor_id"].nunique(),
    "site_count": analysis["anatomic_site"].nunique(),
    "observation_count": len(analysis),
    "random_seed": SEED
}])

summary.to_csv(
    BASE / "Within_Donor_Site_Permutation_Summary.csv",
    index=False
)

site_descriptives.to_csv(
    BASE / "Within_Donor_Site_Descriptives.csv",
    index=False
)

analysis.to_csv(
    BASE / "Within_Donor_Site_Analysis_Dataset.csv",
    index=False
)

print("\nSAVED:")
print("1. Within_Donor_Site_Permutation_Summary.csv")
print("2. Within_Donor_Site_Descriptives.csv")
print("3. Within_Donor_Site_Analysis_Dataset.csv")

print("\n" + "=" * 75)
print("ANALYSIS COMPLETE")
print("=" * 75)
```

```python

import pandas as pd
import numpy as np
from pathlib import Path

# ============================================================
# DONOR-CONTROLLED SITE EFFECT: SENSITIVITY ANALYSIS
# Compare:
# A. Primary analysis: sites represented by >=2 donors
# B. Sensitivity analysis: include every site, including forearm
# ============================================================

folder = Path("/content/drive/MyDrive/Neurothesis Resoure Library")
source_path = folder / "Donor_Site_Paired_Differences_With_Values.csv"

source = pd.read_csv(source_path)
source["donor_id"] = source["donor_id"].astype(str).str.strip()
source["anatomic_site"] = source["anatomic_site"].astype(str).str.strip()

# Retain donors represented at multiple anatomical sites
multi_site_counts = source.groupby("donor_id")["anatomic_site"].nunique()
multi_site_donors = multi_site_counts[multi_site_counts > 1].index

candidate = source[
    source["donor_id"].isin(multi_site_donors)
].copy().reset_index(drop=True)

# Primary analysis: minimum 2 independent donors per site
site_support = candidate.groupby("anatomic_site")["donor_id"].nunique()
eligible_sites = site_support[site_support >= 2].index

primary = candidate[
    candidate["anatomic_site"].isin(eligible_sites)
].copy().reset_index(drop=True)

# Sensitivity analysis: include all candidate observations
sensitivity = candidate.copy()

# ------------------------------------------------------------
# MODEL FUNCTION
# ------------------------------------------------------------

def calculate_site_model(df):
    y = df["paired_difference"].to_numpy(dtype=float)
    donors = df["donor_id"].to_numpy()
    sites = df["anatomic_site"].to_numpy()

    donor_levels = sorted(np.unique(donors))
    site_levels = sorted(np.unique(sites))

    # Intercept + donor fixed effects
    donor_dummies = pd.get_dummies(
        pd.Categorical(donors, categories=donor_levels),
        drop_first=True,
        dtype=float
    ).to_numpy()

    X_reduced = np.column_stack([
        np.ones(len(df)),
        donor_dummies
    ])

    # Add anatomical-site fixed effects
    site_dummies = pd.get_dummies(
        pd.Categorical(sites, categories=site_levels),
        drop_first=True,
        dtype=float
    ).to_numpy()

    X_full = np.column_stack([
        X_reduced,
        site_dummies
    ])

    beta_reduced = np.linalg.lstsq(X_reduced, y, rcond=None)[0]
    beta_full = np.linalg.lstsq(X_full, y, rcond=None)[0]

    rss_reduced = np.sum((y - X_reduced @ beta_reduced) ** 2)
    rss_full = np.sum((y - X_full @ beta_full) ** 2)

    rank_reduced = np.linalg.matrix_rank(X_reduced)
    rank_full = np.linalg.matrix_rank(X_full)

    df_site = rank_full - rank_reduced
    df_residual = len(y) - rank_full

    partial_r2 = (
        (rss_reduced - rss_full) / rss_reduced
        if rss_reduced > 0 else np.nan
    )

    F = (
        ((rss_reduced - rss_full) / df_site) /
        (rss_full / df_residual)
        if df_site > 0 and df_residual > 0 and rss_full > 0
        else np.nan
    )

    return {
        "F": F,
        "partial_R2": partial_r2,
        "df_site": df_site,
        "df_residual": df_residual,
        "rank_reduced": rank_reduced,
        "rank_full": rank_full,
        "rss_reduced": rss_reduced,
        "rss_full": rss_full,
        "donor_levels": donor_levels,
        "site_levels": site_levels
    }


# ------------------------------------------------------------
# WITHIN-DONOR PERMUTATION TEST
# ------------------------------------------------------------

def run_permutation_test(df, permutations=9999, seed=20261114):
    df = df.reset_index(drop=True).copy()

    observed = calculate_site_model(df)
    observed_F = observed["F"]

    donors = df["donor_id"].to_numpy()
    original_sites = df["anatomic_site"].to_numpy()
    y = df["paired_difference"].to_numpy(dtype=float)

    # Group row positions by donor
    donor_positions = [
        np.where(donors == donor)[0]
        for donor in np.unique(donors)
    ]

    rng = np.random.default_rng(seed)
    null_F = []

    for iteration in range(permutations):
        permuted_sites = original_sites.copy()

        # Shuffle site labels only within each donor
        for positions in donor_positions:
            permuted_sites[positions] = rng.permutation(
                original_sites[positions]
            )

        permuted_df = pd.DataFrame({
            "donor_id": donors,
            "anatomic_site": permuted_sites,
            "paired_difference": y
        })

        permuted_result = calculate_site_model(permuted_df)
        null_F.append(permuted_result["F"])

    null_F = np.asarray(null_F, dtype=float)
    valid_null = null_F[np.isfinite(null_F)]

    p_value = (
        (1 + np.sum(valid_null >= observed_F)) /
        (1 + len(valid_null))
    )

    return {
        "observed": observed,
        "p_value": p_value,
        "null_F": valid_null,
        "valid_permutations": len(valid_null)
    }


# ------------------------------------------------------------
# RUN BOTH ANALYSES
# ------------------------------------------------------------

N_PERMUTATIONS = 9999

print("=" * 75)
print("PRIMARY ANALYSIS")
print("=" * 75)

primary_result = run_permutation_test(
    primary,
    permutations=N_PERMUTATIONS,
    seed=20261114
)

print("Observations:", len(primary))
print("Donors:", primary["donor_id"].nunique())
print("Sites:", primary["anatomic_site"].nunique())
print("Observed F:", round(primary_result["observed"]["F"], 6))
print("Partial R-squared:", round(
    primary_result["observed"]["partial_R2"], 6
))
print("Permutation p-value:", round(primary_result["p_value"], 6))
print("Site degrees of freedom:", primary_result["observed"]["df_site"])
print("Residual degrees of freedom:", primary_result["observed"]["df_residual"])

print("\n" + "=" * 75)
print("FOREARM-INCLUDED SENSITIVITY ANALYSIS")
print("=" * 75)

sensitivity_result = run_permutation_test(
    sensitivity,
    permutations=N_PERMUTATIONS,
    seed=20261114
)

print("Observations:", len(sensitivity))
print("Donors:", sensitivity["donor_id"].nunique())
print("Sites:", sensitivity["anatomic_site"].nunique())
print("Observed F:", round(sensitivity_result["observed"]["F"], 6))
print("Partial R-squared:", round(
    sensitivity_result["observed"]["partial_R2"], 6
))
print("Permutation p-value:", round(sensitivity_result["p_value"], 6))
print("Site degrees of freedom:", sensitivity_result["observed"]["df_site"])
print("Residual degrees of freedom:", sensitivity_result["observed"]["df_residual"])

# ------------------------------------------------------------
# SIDE-BY-SIDE COMPARISON
# ------------------------------------------------------------

comparison = pd.DataFrame([
    {
        "analysis": "Primary: minimum 2 donors per site",
        "observations": len(primary),
        "donors": primary["donor_id"].nunique(),
        "sites": primary["anatomic_site"].nunique(),
        "F_statistic": primary_result["observed"]["F"],
        "partial_R2": primary_result["observed"]["partial_R2"],
        "permutation_p": primary_result["p_value"]
    },
    {
        "analysis": "Sensitivity: all sites included",
        "observations": len(sensitivity),
        "donors": sensitivity["donor_id"].nunique(),
        "sites": sensitivity["anatomic_site"].nunique(),
        "F_statistic": sensitivity_result["observed"]["F"],
        "partial_R2": sensitivity_result["observed"]["partial_R2"],
        "permutation_p": sensitivity_result["p_value"]
    }
])

print("\n" + "=" * 75)
print("SENSITIVITY COMPARISON")
print("=" * 75)
display(comparison.round(6))

# ------------------------------------------------------------
# SAVE RESULTS
# ------------------------------------------------------------

comparison.to_csv(
    folder / "Within_Donor_Site_Forearm_Sensitivity_Comparison.csv",
    index=False
)

pd.DataFrame({
    "primary_null_F": pd.Series(primary_result["null_F"]),
    "forearm_included_null_F": pd.Series(sensitivity_result["null_F"])
}).to_csv(
    folder / "Within_Donor_Site_Forearm_Permutation_Distributions.csv",
    index=False
)

print("\nSAVED:")
print("1. Within_Donor_Site_Forearm_Sensitivity_Comparison.csv")
print("2. Within_Donor_Site_Forearm_Permutation_Distributions.csv")
print("\nSENSITIVITY ANALYSIS COMPLETE")
```

```python

import pandas as pd
import numpy as np
from pathlib import Path

# ============================================================
# LEAVE-ONE-DONOR-OUT ROBUSTNESS ANALYSIS
# Primary 92-observation, 12-site dataset
# ============================================================

folder = Path("/content/drive/MyDrive/Neurothesis Resoure Library")

df = pd.read_csv(
    folder / "Within_Donor_Site_Analysis_Dataset.csv"
)

df["donor_id"] = df["donor_id"].astype(str).str.strip()
df["anatomic_site"] = df["anatomic_site"].astype(str).str.strip()

# Confirm the expected analysis dataset
print("Starting observations:", len(df))
print("Starting donors:", df["donor_id"].nunique())
print("Starting sites:", df["anatomic_site"].nunique())

# ------------------------------------------------------------
# MODEL FUNCTION
# ------------------------------------------------------------

def fit_site_model(data):
    data = data.reset_index(drop=True)

    y = data["paired_difference"].to_numpy(dtype=float)
    donors = data["donor_id"].to_numpy()
    sites = data["anatomic_site"].to_numpy()

    donor_levels = sorted(np.unique(donors))
    site_levels = sorted(np.unique(sites))

    donor_dummies = pd.get_dummies(
        pd.Categorical(donors, categories=donor_levels),
        drop_first=True,
        dtype=float
    ).to_numpy()

    site_dummies = pd.get_dummies(
        pd.Categorical(sites, categories=site_levels),
        drop_first=True,
        dtype=float
    ).to_numpy()

    X_reduced = np.column_stack([
        np.ones(len(data)),
        donor_dummies
    ])

    X_full = np.column_stack([
        X_reduced,
        site_dummies
    ])

    rank_reduced = np.linalg.matrix_rank(X_reduced)
    rank_full = np.linalg.matrix_rank(X_full)

    beta_reduced = np.linalg.lstsq(
        X_reduced, y, rcond=None
    )[0]

    beta_full = np.linalg.lstsq(
        X_full, y, rcond=None
    )[0]

    rss_reduced = np.sum(
        (y - X_reduced @ beta_reduced) ** 2
    )

    rss_full = np.sum(
        (y - X_full @ beta_full) ** 2
    )

    df_site = rank_full - rank_reduced
    df_residual = len(y) - rank_full

    partial_r2 = (
        (rss_reduced - rss_full) / rss_reduced
        if rss_reduced > 0 else np.nan
    )

    F = (
        ((rss_reduced - rss_full) / df_site) /
        (rss_full / df_residual)
        if df_site > 0 and df_residual > 0 and rss_full > 0
        else np.nan
    )

    return {
        "F_statistic": F,
        "partial_R2": partial_r2,
        "site_df": df_site,
        "residual_df": df_residual,
        "rank_full": rank_full,
        "rank_reduced": rank_reduced
    }


# ------------------------------------------------------------
# FULL DATASET MODEL
# ------------------------------------------------------------

full_result = fit_site_model(df)

# ------------------------------------------------------------
# REMOVE EACH DONOR ONE AT A TIME
# ------------------------------------------------------------

results = []

for donor in sorted(df["donor_id"].unique()):

    reduced_df = df[df["donor_id"] != donor].copy()

    result = fit_site_model(reduced_df)

    results.append({
        "donor_removed": donor,
        "observations_remaining": len(reduced_df),
        "donors_remaining": reduced_df["donor_id"].nunique(),
        "sites_remaining": reduced_df["anatomic_site"].nunique(),
        "F_statistic": result["F_statistic"],
        "partial_R2": result["partial_R2"],
        "change_in_partial_R2": (
            result["partial_R2"] - full_result["partial_R2"]
        ),
        "site_df": result["site_df"],
        "residual_df": result["residual_df"],
        "design_rank_full": result["rank_full"],
        "design_rank_reduced": result["rank_reduced"]
    })

lodo = pd.DataFrame(results)

# ------------------------------------------------------------
# REPORT RESULTS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("FULL DATASET REFERENCE")
print("=" * 75)

print("F statistic:", round(full_result["F_statistic"], 6))
print("Partial R-squared:", round(full_result["partial_R2"], 6))

print("\n" + "=" * 75)
print("LEAVE-ONE-DONOR-OUT RESULTS")
print("=" * 75)

display(
    lodo.sort_values(
        "change_in_partial_R2",
        key=lambda x: x.abs(),
        ascending=False
    ).round(6)
)

# ------------------------------------------------------------
# SUMMARY OF INFLUENCE
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("DONOR INFLUENCE SUMMARY")
print("=" * 75)

print(
    "LODO partial R-squared range:",
    round(lodo["partial_R2"].min(), 6),
    "to",
    round(lodo["partial_R2"].max(), 6)
)

print(
    "LODO F-statistic range:",
    round(lodo["F_statistic"].min(), 6),
    "to",
    round(lodo["F_statistic"].max(), 6)
)

print(
    "Maximum absolute partial R-squared change:",
    round(lodo["change_in_partial_R2"].abs().max(), 6)
)

most_influential = lodo.loc[
    lodo["change_in_partial_R2"].abs().idxmax()
]

print("\nMost influential donor by partial R-squared change:")
print("Donor removed:", most_influential["donor_removed"])
print(
    "Partial R-squared after removal:",
    round(most_influential["partial_R2"], 6)
)
print(
    "Change from full model:",
    round(most_influential["change_in_partial_R2"], 6)
)

# Check design matrix rank
print("\nRank-deficient LODO models:")
print(
    (
        lodo["design_rank_full"] <
        (
            lodo["design_rank_reduced"] +
            lodo["site_df"]
        )
    ).sum()
)

# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

lodo.to_csv(
    folder / "Within_Donor_Site_Leave_One_Donor_Out.csv",
    index=False
)

print("\nSaved:")
print(folder / "Within_Donor_Site_Leave_One_Donor_Out.csv")
print("\nLEAVE-ONE-DONOR-OUT ANALYSIS COMPLETE")
```

```python

# ============================================================
# NEUROTHESIS RESEARCH
# STEP 3: CHUNKED SAMPLE-LEVEL DATASET VALIDATION
# ============================================================

import h5py
import numpy as np
import pandas as pd
from pathlib import Path

OUTPUT_DIR = Path(
    "/content/drive/MyDrive/Neurothesis Resoure Library/research_outputs"
)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

CHUNK_SIZE = 100_000

# Fields needed for sample-level validation
FIELDS = [
    "sample_id",
    "donor_id",
    "anatomic_site",
    "cell_type.broad",
    "cell_type.detailed",
    "tissue_compartment",
    "sample_compartment",
    "center_x",
    "center_y"
]

def decode_strings(dataset, start, end):
    values = dataset[start:end]

    return np.array([
        x.decode("utf-8") if isinstance(x, bytes)
        else str(x)
        for x in values
    ], dtype=object)


def read_h5ad_column(group, field, start, end):

    obj = group[field]

    # Handle AnnData categorical columns
    if isinstance(obj, h5py.Group):

        codes = obj["codes"][start:end]
        categories = decode_strings(
            obj["categories"], 0, len(obj["categories"])
        )

        result = np.empty(len(codes), dtype=object)
        result[:] = "UNKNOWN"

        valid = codes >= 0
        result[valid] = categories[codes[valid]]

        return result

    # Handle ordinary string columns
    return decode_strings(obj, start, end)


# ------------------------------------------------------------
# READ DATA IN CHUNKS
# ------------------------------------------------------------

sample_records = {}

with h5py.File(H5AD_PATH, "r") as f:

    obs = f["obs"]
    n_cells = len(obs["sample_id"]["codes"])

    print("Total cells:", f"{n_cells:,}")

    for start in range(0, n_cells, CHUNK_SIZE):

        end = min(start + CHUNK_SIZE, n_cells)

        chunk = {
            field: read_h5ad_column(
                obs, field, start, end
            )
            for field in FIELDS
        }

        df = pd.DataFrame(chunk)

        # Group by sample
        for sample_id, group in df.groupby(
            "sample_id", sort=False
        ):

            if sample_id not in sample_records:

                sample_records[sample_id] = {
                    "total_cells": 0,
                    "mast_cells_broad": 0,
                    "schwann_cells_broad": 0,
                    "mast_cells_detailed": 0,
                    "schwann_cells_detailed": 0,
                    "retic_fib_III_detailed": 0,
                    "valid_coordinates": 0,
                    "missing_coordinates": 0,
                    "donors": set(),
                    "anatomic_sites": set(),
                    "tissue_compartments": set(),
                    "sample_compartments": set()
                }

            record = sample_records[sample_id]

            record["total_cells"] += len(group)

            broad = group["cell_type.broad"].str.lower()
            detailed = group["cell_type.detailed"].str.lower()

            record["mast_cells_broad"] += (
                broad.str.contains("mast", na=False).sum()
            )

            record["schwann_cells_broad"] += (
                broad.str.contains("schwann", na=False).sum()
            )

            record["mast_cells_detailed"] += (
                detailed.str.contains("mast", na=False).sum()
            )

            record["schwann_cells_detailed"] += (
                detailed.str.contains("schwann", na=False).sum()
            )

            record["retic_fib_III_detailed"] += (
                detailed.str.contains(
                    "retic fib iii", na=False
                ).sum()
            )

            x = pd.to_numeric(
                group["center_x"], errors="coerce"
            )

            y = pd.to_numeric(
                group["center_y"], errors="coerce"
            )

            valid = np.isfinite(x) & np.isfinite(y)

            record["valid_coordinates"] += valid.sum()
            record["missing_coordinates"] += (~valid).sum()

            for field, destination in [
                ("donor_id", "donors"),
                ("anatomic_site", "anatomic_sites"),
                ("tissue_compartment", "tissue_compartments"),
                ("sample_compartment", "sample_compartments")
            ]:

                values = group[field].dropna().unique()

                record[destination].update(
                    str(v) for v in values
                    if str(v) not in ["UNKNOWN", "nan", ""]
                )

        print(
            f"Processed {end:,} / {n_cells:,} cells"
        )


# ------------------------------------------------------------
# CREATE FINAL SAMPLE TABLE
# ------------------------------------------------------------

rows = []

for sample_id, record in sample_records.items():

    row = {
        "sample_id": sample_id,
        "total_cells": record["total_cells"],
        "mast_cells_broad": record["mast_cells_broad"],
        "schwann_cells_broad": record["schwann_cells_broad"],
        "mast_cells_detailed": record["mast_cells_detailed"],
        "schwann_cells_detailed": record["schwann_cells_detailed"],
        "retic_fib_III_detailed": record["retic_fib_III_detailed"],
        "valid_coordinates": record["valid_coordinates"],
        "missing_coordinates": record["missing_coordinates"]
    }

    for source, destination in [
        ("donors", "donor_id"),
        ("anatomic_sites", "anatomic_site"),
        ("tissue_compartments", "tissue_compartments"),
        ("sample_compartments", "sample_compartments")
    ]:

        values = sorted(record[source])

        row[destination] = (
            "; ".join(values) if values else "UNKNOWN"
        )

    rows.append(row)


validation_df = pd.DataFrame(rows)

validation_df = validation_df.sort_values(
    "sample_id"
).reset_index(drop=True)

# Add cell population proportions
for population in [
    "mast_cells_broad",
    "schwann_cells_broad",
    "mast_cells_detailed",
    "schwann_cells_detailed",
    "retic_fib_III_detailed"
]:

    validation_df[population + "_percent"] = (
        validation_df[population]
        / validation_df["total_cells"] * 100
    )

# Save
output_path = OUTPUT_DIR / "sample_level_validation.csv"

validation_df.to_csv(output_path, index=False)

print("\nVALIDATION SUMMARY")
print("=" * 60)

print("Samples:", len(validation_df))
print("Total cells:", f"{validation_df.total_cells.sum():,}")
print(
    "Unique donors:",
    validation_df.donor_id.nunique()
)

print("\nPOPULATION TOTALS")

print(
    validation_df[
        [
            "mast_cells_broad",
            "schwann_cells_broad",
            "mast_cells_detailed",
            "schwann_cells_detailed",
            "retic_fib_III_detailed"
        ]
    ].sum()
)

print("\nMISSING COORDINATES:")
print(validation_df.missing_coordinates.sum())

print("\nTOP 10 SAMPLES BY MAST CELL COUNT")

print(
    validation_df.nlargest(
        10, "mast_cells_detailed"
    )[
        [
            "sample_id",
            "anatomic_site",
            "total_cells",
            "mast_cells_detailed",
            "schwann_cells_detailed",
            "retic_fib_III_detailed"
        ]
    ].to_string(index=False)
)

print("\nSAVED:", output_path)
```

```python

from scipy.spatial import cKDTree
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# --------------------------------------------------
# CHECK REQUIRED VARIABLES
# --------------------------------------------------

required = ["coordinates", "labels", "population_codes", "OUTPUT_DIR"]

for variable in required:
    if variable not in globals():
        raise NameError(f"Missing variable: {variable}. Run Code 5 first.")

# --------------------------------------------------
# EXTRACT POPULATION COORDINATES
# --------------------------------------------------

population_coordinates = {}

for name, code in population_codes.items():
    mask = labels == code
    population_coordinates[name] = coordinates[mask]

print("POPULATION SIZES")
print("=" * 50)

for name, coords in population_coordinates.items():
    print(f"{name}: {len(coords):,}")

# --------------------------------------------------
# NEAREST-NEIGHBOR DISTANCES
# --------------------------------------------------

results = []
distance_distributions = {}

population_names = list(population_coordinates.keys())

for source_name in population_names:

    source_coords = population_coordinates[source_name]

    for target_name in population_names:

        if source_name == target_name:
            continue

        target_coords = population_coordinates[target_name]

        # Build spatial tree for target population
        tree = cKDTree(target_coords)

        # Find nearest target cell for every source cell
        distances, indices = tree.query(
            source_coords,
            k=1
        )

        distance_distributions[
            f"{source_name} to {target_name}"
        ] = distances

        results.append({
            "source_population": source_name,
            "target_population": target_name,
            "source_cell_count": len(source_coords),
            "target_cell_count": len(target_coords),
            "median_distance": float(np.median(distances)),
            "mean_distance": float(np.mean(distances)),
            "q25_distance": float(np.percentile(distances, 25)),
            "q75_distance": float(np.percentile(distances, 75)),
            "q90_distance": float(np.percentile(distances, 90)),
            "minimum_distance": float(np.min(distances))
        })

# --------------------------------------------------
# DISPLAY RESULTS
# --------------------------------------------------

distance_df = pd.DataFrame(results)

print("\nOBSERVED NEAREST-NEIGHBOR DISTANCES")
print("=" * 75)

display(
    distance_df.round(2)
)

# Save results
distance_path = OUTPUT_DIR / "face_nearest_neighbor_distances.csv"
distance_df.to_csv(distance_path, index=False)

print("\nSaved:", distance_path)

# --------------------------------------------------
# VISUALIZE DISTANCE DISTRIBUTIONS
# --------------------------------------------------

fig, ax = plt.subplots(figsize=(12, 7))

plot_names = list(distance_distributions.keys())
plot_values = [
    distance_distributions[name]
    for name in plot_names
]

ax.boxplot(
    plot_values,
    tick_labels=plot_names,
    showfliers=False
)

ax.set_title(
    "Nearest-Neighbor Distances Between Cell Populations",
    fontsize=14
)
ax.set_ylabel("Distance in dataset spatial coordinate units")
ax.set_xlabel("Directional population comparison")
ax.tick_params(axis="x", rotation=35)

plt.tight_layout()

plot_path = OUTPUT_DIR / "face_nearest_neighbor_distributions.png"
plt.savefig(plot_path, dpi=300, bbox_inches="tight")
plt.show()

print("\nSaved:", plot_path)
```

```python

import numpy as np
import pandas as pd
from scipy.spatial import cKDTree

# --------------------------------------------------
# SETTINGS
# --------------------------------------------------

N_PERMUTATIONS = 999
RANDOM_SEED = 42

rng = np.random.default_rng(RANDOM_SEED)

population_names = ["Mast", "Schwann", "Retic Fib III"]

# Use the dermal coordinates and labels from Code 9
all_coords = dermal_coordinates
all_labels = dermal_labels

n_cells = len(all_labels)

# Confirm population counts
population_sizes = {
    name: int(np.sum(all_labels == population_codes[name]))
    for name in population_names
}

print("NULL MODEL SETTINGS")
print("=" * 60)
print("Dermal cells:", n_cells)
print("Permutations:", N_PERMUTATIONS)
print("Population sizes:", population_sizes)

# --------------------------------------------------
# FUNCTION: SIX DIRECTIONAL MEDIAN DISTANCES
# --------------------------------------------------

def calculate_pairwise_medians(coords, assigned_labels):

    population_coords = {
        name: coords[assigned_labels == population_codes[name]]
        for name in population_names
    }

    medians = {}

    for source_name in population_names:

        source = population_coords[source_name]

        for target_name in population_names:

            if source_name == target_name:
                continue

            target = population_coords[target_name]

            tree = cKDTree(target)
            distances, _ = tree.query(source, k=1)

            key = f"{source_name} to {target_name}"
            medians[key] = float(np.median(distances))

    return medians

# --------------------------------------------------
# OBSERVED STATISTICS
# --------------------------------------------------

observed = calculate_pairwise_medians(
    all_coords,
    all_labels
)

print("\nOBSERVED MEDIAN DISTANCES")
for comparison, value in observed.items():
    print(f"{comparison}: {value:.2f}")

# --------------------------------------------------
# PERMUTATION TEST
# --------------------------------------------------

null_results = {key: [] for key in observed}

for permutation in range(N_PERMUTATIONS):

    # Randomly allocate the three cell identities
    # while preserving their original population sizes.
    shuffled_indices = rng.permutation(n_cells)

    randomized_labels = np.full(
        n_cells, -1, dtype=np.int32
    )

    start = 0

    for name in population_names:

        size = population_sizes[name]
        selected = shuffled_indices[start:start + size]

        randomized_labels[selected] = population_codes[name]
        start += size

    randomized_medians = calculate_pairwise_medians(
        all_coords,
        randomized_labels
    )

    for comparison, value in randomized_medians.items():
        null_results[comparison].append(value)

    if (permutation + 1) % 100 == 0:
        print(
            f"Completed {permutation + 1:,} / "
            f"{N_PERMUTATIONS:,} permutations"
        )

# --------------------------------------------------
# SUMMARIZE NULL DISTRIBUTIONS
# --------------------------------------------------

summary_rows = []

for comparison, observed_value in observed.items():

    null_values = np.asarray(null_results[comparison])

    # Lower-tail test: is observed distance unusually small?
    empirical_p = (
        1 + np.sum(null_values <= observed_value)
    ) / (N_PERMUTATIONS + 1)

    null_median = float(np.median(null_values))

    summary_rows.append({
        "comparison": comparison,
        "observed_median": observed_value,
        "null_median": null_median,
        "null_q025": float(np.percentile(null_values, 2.5)),
        "null_q975": float(np.percentile(null_values, 97.5)),
        "observed_minus_null": observed_value - null_median,
        "empirical_lower_tail_p": empirical_p
    })

permutation_df = pd.DataFrame(summary_rows)

# Correct for the six directional comparisons
# using the Holm step-down procedure.
from statsmodels.stats.multitest import multipletests

permutation_df["holm_adjusted_p"] = multipletests(
    permutation_df["empirical_lower_tail_p"],
    method="holm"
)[1]

permutation_df = permutation_df.sort_values(
    "empirical_lower_tail_p"
).reset_index(drop=True)

print("\nPERMUTATION TEST RESULTS")
print("=" * 90)
display(permutation_df.round(4))

# --------------------------------------------------
# SAVE RESULTS
# --------------------------------------------------

permutation_path = (
    OUTPUT_DIR / "face_dermis_random_label_permutation.csv"
)

permutation_df.to_csv(permutation_path, index=False)

print("\nSaved:", permutation_path)
```

```python

import h5py
import numpy as np
import pandas as pd
from scipy.spatial import cKDTree
from pathlib import Path

# --------------------------------------------------
# SETTINGS
# --------------------------------------------------

CHUNK_SIZE = 100_000
MINIMUM_THRESHOLDS = [10, 20, 50]

POPULATIONS = ["Mast", "Schwann", "Retic Fib III"]

# --------------------------------------------------
# HELPER: READ CATEGORICAL METADATA
# --------------------------------------------------

def get_categories_and_codes(obs, field):
    item = obs[field]

    if isinstance(item, h5py.Group):
        categories = item["categories"].asstr()[:]
        codes = item["codes"]
        return categories, codes

    raise ValueError(
        f"{field} is not stored as a categorical field. "
        "Inspect its HDF5 structure before proceeding."
    )

# --------------------------------------------------
# EXTRACT DERMAL TARGET POPULATIONS
# --------------------------------------------------

cells_by_sample = {}

with h5py.File(H5AD_PATH, "r") as f:

    obs = f["obs"]

    sample_categories, sample_codes = (
        get_categories_and_codes(obs, "sample_id")
    )

    donor_categories, donor_codes = (
        get_categories_and_codes(obs, "donor_id")
    )

    site_categories, site_codes = (
        get_categories_and_codes(obs, "anatomic_site")
    )

    detailed_categories, detailed_codes = (
        get_categories_and_codes(obs, "cell_type.detailed")
    )

    compartment_categories, compartment_codes = (
        get_categories_and_codes(obs, "tissue_compartment")
    )

    population_codes = {
        name: int(np.where(detailed_categories == name)[0][0])
        for name in POPULATIONS
    }

    dermis_code = int(
        np.where(compartment_categories == "dermis")[0][0]
    )

    n_cells = len(sample_codes)

    for start in range(0, n_cells, CHUNK_SIZE):

        end = min(start + CHUNK_SIZE, n_cells)

        s = sample_codes[start:end]
        d = donor_codes[start:end]
        site = site_codes[start:end]
        cell_type = detailed_codes[start:end]
        compartment = compartment_codes[start:end]

        target_mask = (
            (compartment == dermis_code) &
            np.isin(cell_type, list(population_codes.values())) &
            (s >= 0) &
            (d >= 0) &
            (site >= 0)
        )

        if not np.any(target_mask):
            continue

        x = obs["center_x"][start:end][target_mask]
        y = obs["center_y"][start:end][target_mask]

        selected_samples = s[target_mask]
        selected_donors = d[target_mask]
        selected_sites = site[target_mask]
        selected_types = cell_type[target_mask]

        selected_coords = np.column_stack((x, y))

        for sample_code in np.unique(selected_samples):

            local_mask = selected_samples == sample_code

            sample_id = str(sample_categories[sample_code])
            donor_id = str(
                donor_categories[selected_donors[local_mask][0]]
            )
            anatomic_site = str(
                site_categories[selected_sites[local_mask][0]]
            )

            if sample_id not in cells_by_sample:
                cells_by_sample[sample_id] = {
                    "donor_id": donor_id,
                    "anatomic_site": anatomic_site,
                    "coordinates": [],
                    "labels": []
                }

            cells_by_sample[sample_id]["coordinates"].append(
                selected_coords[local_mask]
            )

            cells_by_sample[sample_id]["labels"].append(
                selected_types[local_mask]
            )

        if (end % 500_000 == 0) or end == n_cells:
            print(f"Processed {end:,} / {n_cells:,} cells")

# --------------------------------------------------
# CALCULATE SAMPLE-LEVEL DISTANCES
# --------------------------------------------------

sample_results = []

for sample_id, record in cells_by_sample.items():

    coords = np.vstack(record["coordinates"])
    labels = np.concatenate(record["labels"])

    population_coords = {
        name: coords[labels == population_codes[name]]
        for name in POPULATIONS
    }

    counts = {
        name: len(population_coords[name])
        for name in POPULATIONS
    }

    for source_name in POPULATIONS:

        for target_name in POPULATIONS:

            if source_name == target_name:
                continue

            source = population_coords[source_name]
            target = population_coords[target_name]

            row = {
                "sample_id": sample_id,
                "donor_id": record["donor_id"],
                "anatomic_site": record["anatomic_site"],
                "source_population": source_name,
                "target_population": target_name,
                "source_cell_count": len(source),
                "target_cell_count": len(target)
            }

            for threshold in MINIMUM_THRESHOLDS:
                row[f"eligible_min_{threshold}"] = (
                    len(source) >= threshold and
                    len(target) >= threshold
                )

            # Calculate distances only when the primary threshold is met
            if row["eligible_min_20"]:

                tree = cKDTree(target)
                distances, _ = tree.query(source, k=1)

                row["median_distance"] = float(
                    np.median(distances)
                )
                row["mean_distance"] = float(
                    np.mean(distances)
                )
                row["q25_distance"] = float(
                    np.percentile(distances, 25)
                )
                row["q75_distance"] = float(
                    np.percentile(distances, 75)
                )

            else:
                row["median_distance"] = np.nan
                row["mean_distance"] = np.nan
                row["q25_distance"] = np.nan
                row["q75_distance"] = np.nan

            sample_results.append(row)

sample_df = pd.DataFrame(sample_results)

# --------------------------------------------------
# PRIMARY SAMPLE-LEVEL SUMMARY
# --------------------------------------------------

primary_df = sample_df[
    sample_df["eligible_min_20"]
].copy()

print("\nSAMPLE-LEVEL REPLICATION SUMMARY")
print("=" * 75)

summary_rows = []

for (source, target), group in primary_df.groupby(
    ["source_population", "target_population"]
):

    donor_summary = (
        group.groupby("donor_id")["median_distance"].median()
    )

    summary_rows.append({
        "comparison": f"{source} to {target}",
        "eligible_samples": group["sample_id"].nunique(),
        "independent_donors": group["donor_id"].nunique(),
        "median_of_sample_medians": group["median_distance"].median(),
        "mean_of_sample_medians": group["median_distance"].mean(),
        "median_of_donor_medians": donor_summary.median(),
        "min_sample_median": group["median_distance"].min(),
        "max_sample_median": group["median_distance"].max()
    })

replication_summary = pd.DataFrame(summary_rows)

display(replication_summary.round(2))

# --------------------------------------------------
# THRESHOLD SENSITIVITY
# --------------------------------------------------

print("\nSAMPLE ELIGIBILITY SENSITIVITY")
print("=" * 75)

sensitivity_rows = []

for threshold in MINIMUM_THRESHOLDS:

    eligible = sample_df[
        sample_df[f"eligible_min_{threshold}"]
    ]

    for (source, target), group in eligible.groupby(
        ["source_population", "target_population"]
    ):

        sensitivity_rows.append({
            "minimum_cells_per_population": threshold,
            "comparison": f"{source} to {target}",
            "eligible_samples": group["sample_id"].nunique(),
            "independent_donors": group["donor_id"].nunique()
        })

sensitivity_df = pd.DataFrame(sensitivity_rows)

display(sensitivity_df)

# --------------------------------------------------
# SAVE OUTPUTS
# --------------------------------------------------

OUTPUT_DIR = Path(OUTPUT_DIR)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

sample_path = OUTPUT_DIR / "all_samples_dermal_nn_distances.csv"
summary_path = OUTPUT_DIR / "all_samples_dermal_nn_summary.csv"
sensitivity_path = OUTPUT_DIR / "all_samples_threshold_sensitivity.csv"

sample_df.to_csv(sample_path, index=False)
replication_summary.to_csv(summary_path, index=False)
sensitivity_df.to_csv(sensitivity_path, index=False)

print("\nSAVED:")
print(sample_path)
print(summary_path)
print(sensitivity_path)
```

```python
import pandas as pd
import numpy as np

# Load the all-sample distance results
df = pd.read_csv(
    "/content/drive/MyDrive/Neurothesis Resoure Library/research_outputs/all_samples_dermal_nn_distances.csv"
)

print("COLUMNS:")
print(df.columns.tolist())

# Keep comparisons with at least 20 cells in both populations
eligible_col = "eligible_20"

if eligible_col not in df.columns:
    raise ValueError(
        f"Could not find {eligible_col}. Available columns: {df.columns.tolist()}"
    )

analysis = df[df[eligible_col] == True].copy()

# Confirm the fields needed for donor-level analysis
required = [
    "comparison",
    "donor_id",
    "median_distance"
]

missing = [col for col in required if col not in analysis.columns]
if missing:
    raise ValueError(f"Missing required columns: {missing}")

# One value per donor per comparison.
# If a donor has multiple samples, use the median across those samples.
donor_df = (
    analysis
    .groupby(["comparison", "donor_id"], as_index=False)
    .agg(
        donor_median_distance=("median_distance", "median"),
        samples_contributed=("median_distance", "count")
    )
)

# Bootstrap confidence interval for the median across donors
rng = np.random.default_rng(42)
results = []

for comparison, group in donor_df.groupby("comparison"):
    values = group["donor_median_distance"].dropna().to_numpy()

    bootstrap_medians = np.array([
        np.median(rng.choice(values, size=len(values), replace=True))
        for _ in range(10000)
    ])

    results.append({
        "comparison": comparison,
        "independent_donors": len(values),
        "median_donor_distance": np.median(values),
        "mean_donor_distance": np.mean(values),
        "donor_q25": np.percentile(values, 25),
        "donor_q75": np.percentile(values, 75),
        "bootstrap_95CI_lower": np.percentile(bootstrap_medians, 2.5),
        "bootstrap_95CI_upper": np.percentile(bootstrap_medians, 97.5)
    })

donor_summary = pd.DataFrame(results).sort_values(
    "median_donor_distance"
)

print("\nDONOR-LEVEL SUMMARY")
display(donor_summary.round(2))

# Save both the donor-level values and their summary
output_dir = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/research_outputs/"
)

donor_df.to_csv(output_dir + "donor_level_nn_distances.csv", index=False)
donor_summary.to_csv(output_dir + "donor_level_nn_summary.csv", index=False)

print("\nSaved donor-level analysis files.")
```

```python

import pandas as pd
import numpy as np

output_dir = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/research_outputs/"
)

df = pd.read_csv(
    output_dir + "all_samples_dermal_nn_distances.csv"
)

# Use samples meeting the 20-cell minimum
analysis = df[df["eligible_min_20"] == True].copy()

# First, inspect the available anatomical sites
print("ANATOMICAL SITES:")
print(
    analysis.groupby("anatomic_site")["sample_id"]
    .nunique()
    .sort_values(ascending=False)
)

# Aggregate repeated samples from the same donor at the same site
site_donor = (
    analysis
    .groupby(
        [
            "source_population",
            "target_population",
            "anatomic_site",
            "donor_id"
        ],
        as_index=False
    )
    .agg(
        donor_median_distance=("median_distance", "median"),
        samples_contributed=("median_distance", "count")
    )
)

# Summarize across donors within each site and comparison
site_results = []

for (source, target, site), group in site_donor.groupby(
    ["source_population", "target_population", "anatomic_site"]
):
    values = group["donor_median_distance"].dropna().to_numpy()

    site_results.append({
        "comparison": f"{source} to {target}",
        "anatomic_site": site,
        "independent_donors": len(values),
        "median_donor_distance": np.median(values),
        "mean_donor_distance": np.mean(values),
        "donor_q25": np.percentile(values, 25),
        "donor_q75": np.percentile(values, 75)
    })

site_summary = pd.DataFrame(site_results)

# Focus on the key Schwann–Retic Fib III comparison first
key_results = site_summary[
    site_summary["comparison"].isin([
        "Retic Fib III to Schwann",
        "Schwann to Retic Fib III"
    ])
].sort_values(
    ["comparison", "independent_donors"],
    ascending=[True, False]
)

print("\nSCHWANN–RETIC FIB III BY ANATOMICAL SITE")
display(key_results.round(2))

site_donor.to_csv(
    output_dir + "site_level_donor_nn_distances.csv",
    index=False
)

site_summary.to_csv(
    output_dir + "site_level_donor_nn_summary.csv",
    index=False
)

print("\nSaved anatomical-site analysis files.")
```

```python

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

output_dir = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/research_outputs/"
)

site_donor = pd.read_csv(
    output_dir + "site_level_donor_nn_distances.csv"
)

# Keep the two main directional comparisons
plot_df = site_donor[
    site_donor["source_population"].isin(
        ["Retic Fib III", "Schwann"]
    )
    & site_donor["target_population"].isin(
        ["Retic Fib III", "Schwann"]
    )
    & (site_donor["source_population"] != site_donor["target_population"])
].copy()

plot_df["comparison"] = (
    plot_df["source_population"]
    + " → "
    + plot_df["target_population"]
)

summary = (
    plot_df
    .groupby(["anatomic_site", "comparison"])
    .agg(
        median_distance=("donor_median_distance", "median"),
        donors=("donor_id", "nunique")
    )
    .reset_index()
)

# Keep sites with at least 5 donors for the main figure
summary = summary[summary["donors"] >= 5]

site_order = (
    summary[summary["comparison"] == "Retic Fib III → Schwann"]
    .sort_values("median_distance")["anatomic_site"]
    .tolist()
)

fig, ax = plt.subplots(figsize=(11, 8))

comparisons = [
    "Retic Fib III → Schwann",
    "Schwann → Retic Fib III"
]

offsets = [-0.13, 0.13]

for comparison, offset in zip(comparisons, offsets):
    subset = summary[summary["comparison"] == comparison].copy()
    subset["site_position"] = subset["anatomic_site"].map(
        {site: i for i, site in enumerate(site_order)}
    )

    ax.scatter(
        subset["median_distance"],
        subset["site_position"] + offset,
        label=comparison,
        s=65
    )

    for _, row in subset.iterrows():
        ax.annotate(
            f"n={int(row['donors'])}",
            (row["median_distance"], row["site_position"] + offset),
            xytext=(7, 0),
            textcoords="offset points",
            va="center",
            fontsize=8
        )

ax.set_yticks(range(len(site_order)))
ax.set_yticklabels(site_order)
ax.set_xlabel("Median donor-level nearest-neighbor distance (coordinate units)")
ax.set_ylabel("Anatomical site")
ax.set_title(
    "Schwann–Retic Fib III Spatial Proximity Across Human Skin Sites"
)
ax.legend()
ax.grid(axis="x", alpha=0.25)
plt.tight_layout()

plt.savefig(
    output_dir + "schwann_reticIII_anatomical_site_plot.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved anatomical-site figure.")
```

```python

import h5py
import numpy as np
import pandas as pd
import os

# --------------------------------------------------
# SETTINGS
# --------------------------------------------------

SAMPLE_ID = (
    "202304121534_MSSM-Ji-Skin3-VA227_VMSC02701_region_0"
)

TILE_WIDTH = 200  # dataset coordinate units

# Reuse the H5AD_PATH variable from your earlier notebook.
if "H5AD_PATH" not in globals():
    raise NameError(
        "H5AD_PATH is not defined. Set it to the path of your "
        "H5AD (2) file before running this code."
    )

# --------------------------------------------------
# READ CATEGORICAL OBSERVATIONS
# --------------------------------------------------

def read_h5ad_category(obs_group, field):
    group = obs_group[field]
    categories = group["categories"].asstr()[:]
    codes = group["codes"][:]

    values = np.full(len(codes), "UNKNOWN", dtype=object)
    valid = codes >= 0
    values[valid] = categories[codes[valid]]

    return values

# --------------------------------------------------
# EXTRACT FACE SAMPLE, DERMIS
# --------------------------------------------------

with h5py.File(H5AD_PATH, "r") as f:

    obs = f["obs"]

    sample_values = read_h5ad_category(obs, "sample_id")
    compartment_values = read_h5ad_category(
        obs, "tissue_compartment"
    )
    detailed_values = read_h5ad_category(
        obs, "cell_type.detailed"
    )

    coordinates = f["obsm"]["spatial"][:]

    mask = (
        (sample_values == SAMPLE_ID)
        & (compartment_values == "dermis")
    )

    face_coords = coordinates[mask]
    face_labels_original = detailed_values[mask]

print("FACE DERMIS CELLS:", len(face_coords))
print("COORDINATE ARRAY:", face_coords.shape)

# --------------------------------------------------
# COLLAPSE LABELS INTO FOUR GROUPS
# --------------------------------------------------

focal_types = [
    "Mast",
    "Schwann",
    "Retic Fib III"
]

face_labels = np.array([
    label if label in focal_types else "Other"
    for label in face_labels_original
])

print("\nPOPULATION COUNTS:")
print(pd.Series(face_labels).value_counts())

# --------------------------------------------------
# ASSIGN CELLS TO SPATIAL TILES
# --------------------------------------------------

x = face_coords[:, 0]
y = face_coords[:, 1]

x_min = x.min()
y_min = y.min()

tile_x = np.floor((x - x_min) / TILE_WIDTH).astype(int)
tile_y = np.floor((y - y_min) / TILE_WIDTH).astype(int)

tile_df = pd.DataFrame({
    "tile_x": tile_x,
    "tile_y": tile_y,
    "cell_type": face_labels
})

tile_summary = (
    tile_df
    .groupby(["tile_x", "tile_y"])
    .agg(
        total_cells=("cell_type", "size"),
        unique_labels=("cell_type", "nunique"),
        mast=("cell_type", lambda s: (s == "Mast").sum()),
        schwann=("cell_type", lambda s: (s == "Schwann").sum()),
        reticIII=("cell_type", lambda s: (s == "Retic Fib III").sum())
    )
    .reset_index()
)

# --------------------------------------------------
# VALIDATE RANDOMIZATION OPPORTUNITY
# --------------------------------------------------

mixed_tiles = tile_summary[
    tile_summary["unique_labels"] >= 2
]

focal_cells = tile_df["cell_type"].isin(focal_types)

mixed_keys = set(
    zip(mixed_tiles["tile_x"], mixed_tiles["tile_y"])
)

cell_keys = list(zip(tile_x, tile_y))

in_mixed_tiles = np.array([
    key in mixed_keys for key in cell_keys
])

print("\nTILE VALIDATION")
print("------------------------------")
print("Tile width:", TILE_WIDTH)
print("Total tiles:", len(tile_summary))
print("Tiles containing multiple label groups:", len(mixed_tiles))
print("Median cells per tile:", tile_summary["total_cells"].median())
print("Mean cells per tile:", round(tile_summary["total_cells"].mean(), 2))

print(
    "Focal cells in mixed tiles:",
    round(
        100 * in_mixed_tiles[focal_cells].mean(), 2
    ),
    "%"
)

print("\nCELL COUNTS PER TILE:")
display(
    tile_summary["total_cells"].describe().to_frame(
        "cells_per_tile"
    ).round(2)
)

print("\nFOCAL POPULATIONS PER TILE:")
display(
    tile_summary[
        ["mast", "schwann", "reticIII"]
    ].describe().round(2)
)

# Save audit
output_dir = (
    "/content/drive/MyDrive/Neurothesis Resoure Library/research_outputs/"
)

tile_summary.to_csv(
    output_dir + "face_spatial_tile_validation.csv",
    index=False
)

print("\nSaved spatial tile validation.")
```

```python

import numpy as np
import pandas as pd

# Reuse face_coords and face_labels from Code 15

tile_widths = [100, 150, 200, 300, 400]
focal_types = ["Mast", "Schwann", "Retic Fib III"]

results = []

for width in tile_widths:

    tx = np.floor(
        (face_coords[:, 0] - face_coords[:, 0].min()) / width
    ).astype(int)

    ty = np.floor(
        (face_coords[:, 1] - face_coords[:, 1].min()) / width
    ).astype(int)

    temp = pd.DataFrame({
        "tx": tx,
        "ty": ty,
        "label": face_labels
    })

    grouped = temp.groupby(["tx", "ty"])

    tile_stats = grouped.agg(
        total_cells=("label", "size"),
        mast=("label", lambda s: (s == "Mast").sum()),
        schwann=("label", lambda s: (s == "Schwann").sum()),
        retic=("label", lambda s: (s == "Retic Fib III").sum())
    ).reset_index()

    focal_counts = tile_stats[["mast", "schwann", "retic"]].to_numpy()

    # Number of focal types present in each tile
    types_present = (focal_counts > 0).sum(axis=1)

    # Number of tiles with at least two focal types
    mixed_focal_tiles = types_present >= 2

    # Number of tiles with at least two total cells
    shuffleable_tiles = tile_stats["total_cells"] >= 2

    # Identify focal cells in tiles containing at least two focal types
    mixed_keys = set(
        zip(
            tile_stats.loc[mixed_focal_tiles, "tx"],
            tile_stats.loc[mixed_focal_tiles, "ty"]
        )
    )

    cell_keys = list(zip(tx, ty))
    focal_mask = np.isin(face_labels, focal_types)

    focal_in_mixed = np.array([
        key in mixed_keys for key in cell_keys
    ])

    results.append({
        "tile_width": width,
        "total_tiles": len(tile_stats),
        "median_cells_per_tile": tile_stats["total_cells"].median(),
        "tiles_with_2plus_focal_types": int(mixed_focal_tiles.sum()),
        "tiles_with_2plus_total_cells": int(shuffleable_tiles.sum()),
        "focal_cells_in_2plus_focal_tiles_pct":
            100 * focal_in_mixed[focal_mask].mean(),
        "mast_positive_tiles": int((tile_stats["mast"] > 0).sum()),
        "schwann_positive_tiles": int((tile_stats["schwann"] > 0).sum()),
        "retic_positive_tiles": int((tile_stats["retic"] > 0).sum())
    })

tile_comparison = pd.DataFrame(results)

print("TILE SIZE SENSITIVITY")
display(tile_comparison.round(2))

tile_comparison.to_csv(
    output_dir + "face_tile_size_sensitivity.csv",
    index=False
)

print("Saved tile-size sensitivity results.")
```

```python

import numpy as np
import pandas as pd
from scipy.spatial import cKDTree

# --------------------------------------------------
# SETTINGS
# --------------------------------------------------

TILE_WIDTH = 150
N_PERMUTATIONS = 999
RANDOM_SEED = 42

# Requires face_coords, face_labels, and output_dir
rng = np.random.default_rng(RANDOM_SEED)

# --------------------------------------------------
# PREPARE POPULATIONS
# --------------------------------------------------

coords = np.asarray(face_coords)
labels_original = np.asarray(face_labels).copy()

SCHWANN = "Schwann"
RETIC = "Retic Fib III"

required_types = [SCHWANN, RETIC]

for cell_type in required_types:
    if np.sum(labels_original == cell_type) == 0:
        raise ValueError(f"No cells found for {cell_type}")

# --------------------------------------------------
# CREATE SPATIAL TILES
# --------------------------------------------------

x = coords[:, 0]
y = coords[:, 1]

tx = np.floor((x - x.min()) / TILE_WIDTH).astype(int)
ty = np.floor((y - y.min()) / TILE_WIDTH).astype(int)

tile_keys = list(zip(tx, ty))

tile_groups = {}

for i, key in enumerate(tile_keys):
    tile_groups.setdefault(key, []).append(i)

# Only tiles containing BOTH Schwann and Retic Fib III
# will have their two identities shuffled.

shuffle_groups = []

for key, indices in tile_groups.items():

    indices = np.asarray(indices, dtype=int)

    local_labels = labels_original[indices]

    has_schwann = np.any(local_labels == SCHWANN)
    has_retic = np.any(local_labels == RETIC)

    if has_schwann and has_retic:

        focal_positions = indices[
            np.isin(local_labels, required_types)
        ]

        shuffle_groups.append(focal_positions)

print("TILE WIDTH:", TILE_WIDTH)
print("TILES ELIGIBLE FOR SCHWANN/RETIC SHUFFLING:",
      len(shuffle_groups))

randomized_positions = np.concatenate(shuffle_groups)

print("CELLS ELIGIBLE FOR SHUFFLING:",
      len(randomized_positions))

print("PERCENT OF SCHWANN + RETIC CELLS ELIGIBLE:",
      round(
          100 * len(randomized_positions) /
          np.sum(np.isin(labels_original, required_types)),
          2
      ))

# --------------------------------------------------
# DISTANCE FUNCTION
# --------------------------------------------------

def directional_median(source_coords, target_coords):

    if len(source_coords) == 0 or len(target_coords) == 0:
        return np.nan

    tree = cKDTree(target_coords)
    distances, _ = tree.query(source_coords, k=1)

    return float(np.median(distances))


def calculate_pair_distances(labels):

    schwann_coords = coords[labels == SCHWANN]
    retic_coords = coords[labels == RETIC]

    return {
        "Schwann_to_Retic": directional_median(
            schwann_coords, retic_coords
        ),
        "Retic_to_Schwann": directional_median(
            retic_coords, schwann_coords
        )
    }

# --------------------------------------------------
# OBSERVED DISTANCES
# --------------------------------------------------

observed = calculate_pair_distances(labels_original)

print("\nOBSERVED DISTANCES")
print(observed)

# --------------------------------------------------
# CONSTRAINED PERMUTATIONS
# --------------------------------------------------

null_results = []

for iteration in range(N_PERMUTATIONS):

    permuted_labels = labels_original.copy()

    for indices in shuffle_groups:

        # Preserve the exact number of each cell type
        # within each eligible tile.
        shuffled = permuted_labels[indices].copy()
        rng.shuffle(shuffled)
        permuted_labels[indices] = shuffled

    distances = calculate_pair_distances(permuted_labels)

    null_results.append({
        "permutation": iteration + 1,
        **distances
    })

    if (iteration + 1) % 100 == 0:
        print(
            f"Completed {iteration + 1} / {N_PERMUTATIONS}"
        )

null_df = pd.DataFrame(null_results)

# --------------------------------------------------
# SUMMARIZE NULL DISTRIBUTIONS
# --------------------------------------------------

summary = []

for comparison, observed_value in observed.items():

    values = null_df[comparison].dropna().to_numpy()

    # Lower-tail test: are observed distances unusually small?
    p_lower = (
        1 + np.sum(values <= observed_value)
    ) / (len(values) + 1)

    # Two-sided empirical comparison around null median
    null_median = np.median(values)

    p_two_sided = (
        1 + np.sum(
            np.abs(values - null_median)
            >= abs(observed_value - null_median)
        )
    ) / (len(values) + 1)

    summary.append({
        "comparison": comparison,
        "observed_distance": observed_value,
        "null_median": null_median,
        "null_2.5_percentile": np.percentile(values, 2.5),
        "null_97.5_percentile": np.percentile(values, 97.5),
        "observed_minus_null": observed_value - null_median,
        "empirical_lower_tail_p": p_lower,
        "empirical_two_sided_p": p_two_sided
    })

summary_df = pd.DataFrame(summary)

print("\nCONSTRAINED NULL MODEL RESULTS")
display(summary_df.round(4))

# --------------------------------------------------
# SAVE OUTPUTS
# --------------------------------------------------

null_df.to_csv(
    output_dir + "face_constrained_null_permutations.csv",
    index=False
)

summary_df.to_csv(
    output_dir + "face_constrained_null_summary.csv",
    index=False
)

print("\nSaved constrained permutation results.")
```

```python

import numpy as np
import pandas as pd
from scipy.spatial import cKDTree

TILE_WIDTHS = [100, 200]
N_PERMUTATIONS = 999
rng = np.random.default_rng(123)

SCHWANN = "Schwann"
RETIC = "Retic Fib III"

coords = np.asarray(face_coords)
labels_original = np.asarray(face_labels).copy()

def median_nn(source_coords, target_coords):
    if len(source_coords) == 0 or len(target_coords) == 0:
        return np.nan
    tree = cKDTree(target_coords)
    distances, _ = tree.query(source_coords, k=1)
    return float(np.median(distances))

def get_distances(labels):
    s = coords[labels == SCHWANN]
    r = coords[labels == RETIC]

    return {
        "Schwann_to_Retic": median_nn(s, r),
        "Retic_to_Schwann": median_nn(r, s)
    }

observed = get_distances(labels_original)
results = []

for width in TILE_WIDTHS:

    tx = np.floor(
        (coords[:, 0] - coords[:, 0].min()) / width
    ).astype(int)

    ty = np.floor(
        (coords[:, 1] - coords[:, 1].min()) / width
    ).astype(int)

    tile_groups = {}

    for i, key in enumerate(zip(tx, ty)):
        tile_groups.setdefault(key, []).append(i)

    shuffle_groups = []

    for indices in tile_groups.values():

        indices = np.asarray(indices, dtype=int)
        local = labels_original[indices]

        has_s = np.any(local == SCHWANN)
        has_r = np.any(local == RETIC)

        if has_s and has_r:
            focal = indices[
                np.isin(local, [SCHWANN, RETIC])
            ]
            shuffle_groups.append(focal)

    if shuffle_groups:
        eligible_positions = np.concatenate(shuffle_groups)
    else:
        eligible_positions = np.array([], dtype=int)

    null_values = {
        "Schwann_to_Retic": [],
        "Retic_to_Schwann": []
    }

    for iteration in range(N_PERMUTATIONS):

        permuted = labels_original.copy()

        for indices in shuffle_groups:
            shuffled = permuted[indices].copy()
            rng.shuffle(shuffled)
            permuted[indices] = shuffled

        distances = get_distances(permuted)

        for comparison in null_values:
            null_values[comparison].append(
                distances[comparison]
            )

    for comparison, values in null_values.items():

        values = np.asarray(values)
        obs = observed[comparison]
        null_median = np.median(values)

        p_lower = (
            1 + np.sum(values <= obs)
        ) / (len(values) + 1)

        p_two = (
            1 + np.sum(
                np.abs(values - null_median)
                >= abs(obs - null_median)
            )
        ) / (len(values) + 1)

        results.append({
            "tile_width": width,
            "comparison": comparison,
            "eligible_tiles": len(shuffle_groups),
            "eligible_cells": len(eligible_positions),
            "eligible_focal_cells_pct":
                100 * len(eligible_positions) /
                np.sum(np.isin(
                    labels_original, [SCHWANN, RETIC]
                )),
            "observed_distance": obs,
            "null_median": null_median,
            "null_2.5_percentile": np.percentile(values, 2.5),
            "null_97.5_percentile": np.percentile(values, 97.5),
            "empirical_lower_tail_p": p_lower,
            "empirical_two_sided_p": p_two
        })

sensitivity_df = pd.DataFrame(results)

print("TILE-SIZE PERMUTATION SENSITIVITY")
display(sensitivity_df.round(4))

sensitivity_df.to_csv(
    output_dir + "face_constrained_null_tile_sensitivity.csv",
    index=False
)

print("Saved tile-size sensitivity results.")
```

```python

# ANALYSIS 19
# Donor-adjusted anatomical-site variation in
# Retic Fib III -> Schwann nearest-neighbor distance

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import chi2
import statsmodels.formula.api as smf
import warnings

# 1. Select one directional comparison
site_data = site_donor[
    (site_donor["source_population"] == "Retic Fib III") &
    (site_donor["target_population"] == "Schwann")
].copy()

site_data["donor_median_distance"] = pd.to_numeric(
    site_data["donor_median_distance"], errors="coerce"
)

site_data = site_data.dropna(
    subset=["donor_id", "anatomic_site", "donor_median_distance"]
)

# 2. Retain sites represented by at least five donors
site_counts = (
    site_data.groupby("anatomic_site")["donor_id"]
    .nunique()
    .sort_values(ascending=False)
)

eligible_sites = site_counts[site_counts >= 5].index.tolist()

analysis_df = site_data[
    site_data["anatomic_site"].isin(eligible_sites)
].copy()

analysis_df["log_distance"] = np.log(
    analysis_df["donor_median_distance"]
)

print("ANALYSIS COVERAGE")
print("-----------------")
print("Donors:", analysis_df["donor_id"].nunique())
print("Anatomical sites:", analysis_df["anatomic_site"].nunique())
print("Donor-site observations:", len(analysis_df))
print("\nDonors represented per site:")
print(
    analysis_df.groupby("anatomic_site")["donor_id"]
    .nunique()
    .sort_values(ascending=False)
)

# 3. Fit donor-random-intercept models
# Null model: no anatomical-site effect
# Full model: anatomical site as a categorical fixed effect

print("\nFITTING MIXED-EFFECTS MODELS...")

with warnings.catch_warnings(record=True) as caught:
    warnings.simplefilter("always")

    null_model = smf.mixedlm(
        "log_distance ~ 1",
        data=analysis_df,
        groups=analysis_df["donor_id"],
        re_formula="1"
    ).fit(reml=False, method="lbfgs")

    full_model = smf.mixedlm(
        "log_distance ~ C(anatomic_site)",
        data=analysis_df,
        groups=analysis_df["donor_id"],
        re_formula="1"
    ).fit(reml=False, method="lbfgs")

    for warning in caught:
        print("MODEL WARNING:", warning.message)

# 4. Likelihood-ratio comparison
lr_stat = max(
    0,
    2 * (full_model.llf - null_model.llf)
)

df_difference = (
    len(full_model.fe_params) - len(null_model.fe_params)
)

overall_p = chi2.sf(lr_stat, df_difference)

print("\nOVERALL ANATOMICAL-SITE EFFECT")
print("--------------------------------")
print("Likelihood-ratio statistic:", round(lr_stat, 4))
print("Degrees of freedom:", df_difference)
print("P-value:", overall_p)
print("Null model converged:", null_model.converged)
print("Full model converged:", full_model.converged)

# 5. Descriptive site-level results
site_summary_next = (
    analysis_df.groupby("anatomic_site")
    .agg(
        independent_donors=("donor_id", "nunique"),
        median_distance=("donor_median_distance", "median"),
        mean_distance=("donor_median_distance", "mean"),
        q25=("donor_median_distance", lambda x: x.quantile(.25)),
        q75=("donor_median_distance", lambda x: x.quantile(.75))
    )
    .sort_values("median_distance")
    .reset_index()
)

print("\nSITE-LEVEL DONOR MEDIANS")
display(site_summary_next.round(3))

# 6. Plot donor-level observations
site_order = site_summary_next["anatomic_site"].tolist()

plot_values = [
    analysis_df.loc[
        analysis_df["anatomic_site"] == site,
        "donor_median_distance"
    ].values
    for site in site_order
]

plt.figure(figsize=(13, 6))
plt.boxplot(
    plot_values,
    labels=site_order,
    showfliers=False
)

rng_plot = np.random.default_rng(42)

for position, values in enumerate(plot_values, start=1):
    jitter = rng_plot.uniform(-0.12, 0.12, len(values))
    plt.scatter(
        position + jitter,
        values,
        alpha=0.7,
        s=35
    )

plt.xticks(rotation=45, ha="right")
plt.ylabel("Donor-level median nearest-neighbor distance")
plt.xlabel("Anatomical site")
plt.title(
    "Retic Fib III to Schwann Spatial Distance Across Anatomical Sites"
)
plt.tight_layout()

figure_path = output_dir + "reticIII_schwann_site_mixed_model.png"
plt.savefig(figure_path, dpi=300, bbox_inches="tight")
plt.show()

# 7. Save results
site_summary_next.to_csv(
    output_dir + "reticIII_schwann_site_adjusted_summary.csv",
    index=False
)

print("\nSaved:")
print(figure_path)
print(output_dir + "reticIII_schwann_site_adjusted_summary.csv")
```

```python

# ANALYSIS 19B
# Donor-blocked permutation test for anatomical-site variation

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Prepare data
perm_data = site_data.copy()

perm_data = perm_data.dropna(
    subset=["donor_id", "anatomic_site", "donor_median_distance"]
).reset_index(drop=True)

perm_data["log_distance"] = np.log(
    perm_data["donor_median_distance"].astype(float)
)

# Retain sites represented by at least 5 donors
site_counts = perm_data.groupby("anatomic_site")["donor_id"].nunique()
eligible_sites = site_counts[site_counts >= 5].index

perm_data = perm_data[
    perm_data["anatomic_site"].isin(eligible_sites)
].reset_index(drop=True)

# 2. Create donor fixed-effect design matrix
donor_matrix = pd.get_dummies(
    perm_data["donor_id"],
    drop_first=True,
    dtype=float
).to_numpy()

# Intercept
intercept = np.ones((len(perm_data), 1))

X_reduced = np.column_stack([intercept, donor_matrix])

# Site design matrix
site_matrix = pd.get_dummies(
    perm_data["anatomic_site"],
    drop_first=True,
    dtype=float
).to_numpy()

X_full = np.column_stack([X_reduced, site_matrix])

y = perm_data["log_distance"].to_numpy()

# 3. Calculate observed partial F statistic
def calculate_f_statistic(y, X_reduced, X_full):
    beta_reduced = np.linalg.lstsq(X_reduced, y, rcond=None)[0]
    beta_full = np.linalg.lstsq(X_full, y, rcond=None)[0]

    residual_reduced = y - X_reduced @ beta_reduced
    residual_full = y - X_full @ beta_full

    rss_reduced = np.sum(residual_reduced ** 2)
    rss_full = np.sum(residual_full ** 2)

    rank_reduced = np.linalg.matrix_rank(X_reduced)
    rank_full = np.linalg.matrix_rank(X_full)

    df_site = rank_full - rank_reduced
    df_residual = len(y) - rank_full

    if df_site <= 0 or df_residual <= 0:
        raise ValueError("Insufficient degrees of freedom for site comparison.")

    f_stat = (
        ((rss_reduced - rss_full) / df_site) /
        (rss_full / df_residual)
    )

    return f_stat, rss_reduced, rss_full, df_site, df_residual

observed_f, rss0, rss1, df_site, df_resid = calculate_f_statistic(
    y, X_reduced, X_full
)

print("OBSERVED MODEL")
print("----------------")
print("Donors:", perm_data["donor_id"].nunique())
print("Sites:", perm_data["anatomic_site"].nunique())
print("Observations:", len(perm_data))
print("Site degrees of freedom:", df_site)
print("Residual degrees of freedom:", df_resid)
print("Observed partial F:", round(observed_f, 5))

# 4. Permute site labels WITHIN each donor
N_PERMUTATIONS = 9999
rng = np.random.default_rng(2026)

donor_groups = [
    group.index.to_numpy()
    for _, group in perm_data.groupby("donor_id")
]

site_labels = perm_data["anatomic_site"].to_numpy()

permuted_f_values = []

for iteration in range(N_PERMUTATIONS):
    shuffled_sites = site_labels.copy()

    for indices in donor_groups:
        shuffled_sites[indices] = rng.permutation(
            shuffled_sites[indices]
        )

    shuffled_site_matrix = pd.get_dummies(
        pd.Categorical(
            shuffled_sites,
            categories=perm_data["anatomic_site"].unique()
        ),
        drop_first=True,
        dtype=float
    ).to_numpy()

    X_permuted = np.column_stack([
        X_reduced,
        shuffled_site_matrix
    ])

    f_perm, _, _, _, _ = calculate_f_statistic(
        y, X_reduced, X_permuted
    )

    permuted_f_values.append(f_perm)

permuted_f_values = np.asarray(permuted_f_values)

empirical_p = (
    1 + np.sum(permuted_f_values >= observed_f)
) / (N_PERMUTATIONS + 1)

print("\nDONOR-BLOCKED PERMUTATION RESULTS")
print("---------------------------------")
print("Permutations:", N_PERMUTATIONS)
print("Observed F:", round(observed_f, 5))
print("Null median F:", round(np.median(permuted_f_values), 5))
print("Null 95th percentile:", round(np.percentile(permuted_f_values, 95), 5))
print("Empirical p-value:", round(empirical_p, 5))

# 5. Plot permutation distribution
plt.figure(figsize=(9, 5))
plt.hist(permuted_f_values, bins=45)
plt.axvline(
    observed_f,
    linestyle="--",
    linewidth=2,
    label="Observed F statistic"
)
plt.xlabel("Partial F statistic under donor-blocked permutations")
plt.ylabel("Number of permutations")
plt.title("Anatomical-Site Permutation Test")
plt.legend()
plt.tight_layout()

plt.savefig(
    output_dir + "donor_blocked_site_permutation.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

```python

# ANALYSIS 20
# Donor-blocked anatomical-site tests for mast-cell spatial relationships

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

N_PERMUTATIONS = 9999
rng = np.random.default_rng(2027)

comparisons = [
    ("Mast", "Schwann"),
    ("Mast", "Retic Fib III")
]

def run_site_permutation(source, target, seed_offset=0):

    data = site_donor[
        (site_donor["source_population"] == source) &
        (site_donor["target_population"] == target)
    ].copy()

    data = data.dropna(
        subset=["donor_id", "anatomic_site", "donor_median_distance"]
    )

    data["log_distance"] = np.log(
        data["donor_median_distance"].astype(float)
    )

    # Retain sites represented by at least five donors
    counts = data.groupby("anatomic_site")["donor_id"].nunique()
    eligible_sites = counts[counts >= 5].index.tolist()

    data = data[
        data["anatomic_site"].isin(eligible_sites)
    ].reset_index(drop=True)

    y = data["log_distance"].to_numpy()

    # Donor fixed effects
    donor_matrix = pd.get_dummies(
        data["donor_id"],
        drop_first=True,
        dtype=float
    ).to_numpy()

    X_reduced = np.column_stack([
        np.ones(len(data)),
        donor_matrix
    ])

    site_levels = data["anatomic_site"].unique()

    site_matrix = pd.get_dummies(
        pd.Categorical(
            data["anatomic_site"],
            categories=site_levels
        ),
        drop_first=True,
        dtype=float
    ).to_numpy()

    X_full = np.column_stack([
        X_reduced,
        site_matrix
    ])

    def calculate_f(y_values, full_matrix):
        beta0 = np.linalg.lstsq(
            X_reduced, y_values, rcond=None
        )[0]

        beta1 = np.linalg.lstsq(
            full_matrix, y_values, rcond=None
        )[0]

        rss0 = np.sum((y_values - X_reduced @ beta0) ** 2)
        rss1 = np.sum((y_values - full_matrix @ beta1) ** 2)

        rank0 = np.linalg.matrix_rank(X_reduced)
        rank1 = np.linalg.matrix_rank(full_matrix)

        df1 = rank1 - rank0
        df2 = len(y_values) - rank1

        if df1 <= 0 or df2 <= 0:
            raise ValueError("Insufficient degrees of freedom.")

        f_value = ((rss0 - rss1) / df1) / (rss1 / df2)

        return f_value, df1, df2

    observed_f, df1, df2 = calculate_f(y, X_full)

    # Shuffle site labels within each donor
    labels = data["anatomic_site"].to_numpy()

    donor_groups = [
        group.index.to_numpy()
        for _, group in data.groupby("donor_id")
    ]

    local_rng = np.random.default_rng(2027 + seed_offset)
    null_f = []

    for iteration in range(N_PERMUTATIONS):
        shuffled = labels.copy()

        for indices in donor_groups:
            shuffled[indices] = local_rng.permutation(
                shuffled[indices]
            )

        shuffled_matrix = pd.get_dummies(
            pd.Categorical(
                shuffled,
                categories=site_levels
            ),
            drop_first=True,
            dtype=float
        ).to_numpy()

        X_perm = np.column_stack([
            X_reduced,
            shuffled_matrix
        ])

        f_perm, _, _ = calculate_f(y, X_perm)
        null_f.append(f_perm)

    null_f = np.asarray(null_f)

    p_value = (
        1 + np.sum(null_f >= observed_f)
    ) / (N_PERMUTATIONS + 1)

    summary = {
        "comparison": f"{source} to {target}",
        "donors": data["donor_id"].nunique(),
        "sites": data["anatomic_site"].nunique(),
        "observations": len(data),
        "site_df": df1,
        "residual_df": df2,
        "observed_F": observed_f,
        "null_median_F": np.median(null_f),
        "null_95th_percentile": np.percentile(null_f, 95),
        "empirical_p": p_value
    }

    site_summary = (
        data.groupby("anatomic_site")
        .agg(
            donors=("donor_id", "nunique"),
            median_distance=("donor_median_distance", "median"),
            mean_distance=("donor_median_distance", "mean"),
            q25=("donor_median_distance", lambda x: x.quantile(.25)),
            q75=("donor_median_distance", lambda x: x.quantile(.75))
        )
        .reset_index()
        .sort_values("median_distance")
    )

    return summary, site_summary, null_f


# Run both comparisons
all_results = []
site_tables = {}
null_distributions = {}

for i, (source, target) in enumerate(comparisons):
    print(f"\nRunning: {source} -> {target}")

    summary, site_table, null_values = run_site_permutation(
        source, target, seed_offset=i
    )

    all_results.append(summary)
    site_tables[f"{source}_to_{target}"] = site_table
    null_distributions[f"{source}_to_{target}"] = null_values

results_df = pd.DataFrame(all_results)

# Holm correction across the two comparisons
order = np.argsort(results_df["empirical_p"].to_numpy())
adjusted = np.empty(len(results_df))
running_max = 0

for rank, index in enumerate(order):
    adjusted_value = min(
        1.0,
        (len(results_df) - rank) *
        results_df.loc[index, "empirical_p"]
    )
    running_max = max(running_max, adjusted_value)
    adjusted[index] = running_max

results_df["holm_adjusted_p"] = adjusted

print("\nMAST-CELL SITE ANALYSIS RESULTS")
display(results_df.round(5))

for name, table in site_tables.items():
    print("\nSITE SUMMARY:", name)
    display(table.round(3))

# Save outputs
results_df.to_csv(
    output_dir + "mast_cell_site_permutation_results.csv",
    index=False
)

for name, table in site_tables.items():
    table.to_csv(
        output_dir + name + "_site_summary.csv",
        index=False
    )

print("\nSaved mast-cell anatomical-site results.")
```

```python

# ANALYSIS 21
# Paired donor-level anatomical-site comparisons
# Exact sign-flip permutation tests, with Holm correction

import numpy as np
import pandas as pd
from itertools import product
from statsmodels.stats.multitest import multipletests

comparisons = [
    ("Mast", "Schwann"),
    ("Mast", "Retic Fib III")
]

MIN_MATCHED_DONORS = 5

all_pairwise_results = []

for source, target in comparisons:

    data = site_donor[
        (site_donor["source_population"] == source) &
        (site_donor["target_population"] == target)
    ].copy()

    data = data.dropna(
        subset=["donor_id", "anatomic_site", "donor_median_distance"]
    )

    # One donor-level observation per anatomical site
    wide = data.pivot_table(
        index="donor_id",
        columns="anatomic_site",
        values="donor_median_distance",
        aggfunc="median"
    )

    sites = list(wide.columns)

    for i in range(len(sites)):
        for j in range(i + 1, len(sites)):

            site_a = sites[i]
            site_b = sites[j]

            paired = wide[[site_a, site_b]].dropna()

            n = len(paired)

            if n < MIN_MATCHED_DONORS:
                continue

            # Paired log-distance differences
            differences = (
                np.log(paired[site_a].to_numpy()) -
                np.log(paired[site_b].to_numpy())
            )

            observed = np.mean(differences)

            # Enumerate all possible sign flips
            sign_patterns = np.asarray(
                list(product([-1, 1], repeat=n)),
                dtype=float
            )

            permuted_statistics = np.mean(
                sign_patterns * differences[None, :],
                axis=1
            )

            p_value = np.mean(
                np.abs(permuted_statistics) >= abs(observed) - 1e-12
            )

            all_pairwise_results.append({
                "comparison": f"{source} to {target}",
                "site_A": site_a,
                "site_B": site_b,
                "matched_donors": n,
                "mean_log_difference": observed,
                "geometric_mean_ratio_A_to_B": np.exp(observed),
                "median_distance_A": np.median(paired[site_a]),
                "median_distance_B": np.median(paired[site_b]),
                "exact_sign_flip_p": p_value
            })

pairwise_df = pd.DataFrame(all_pairwise_results)

# Correct across every pairwise test in both comparison families
if len(pairwise_df) > 0:
    pairwise_df["holm_adjusted_p"] = multipletests(
        pairwise_df["exact_sign_flip_p"],
        method="holm"
    )[1]

    pairwise_df = pairwise_df.sort_values(
        ["holm_adjusted_p", "exact_sign_flip_p"]
    )

print("PAIRWISE DONOR-MATCHED RESULTS")
print("--------------------------------")
print("Total tests:", len(pairwise_df))

display(pairwise_df.round(4))

pairwise_df.to_csv(
    output_dir + "paired_anatomical_site_comparisons.csv",
    index=False
)

print("\nSaved paired anatomical-site comparisons.")
```

```python

from scipy.spatial import cKDTree
import numpy as np
import pandas as pd

# ---------------------------------------------------------
# ANALYSIS 23: THREE-POPULATION NEIGHBORHOOD COMPOSITION
# ---------------------------------------------------------

K = 10
focal_types = ["Mast", "Schwann", "Retic Fib III"]

sample_records = []
cell_records = []

print("Starting 10-nearest-neighbor analysis...")
print("Samples:", len(eligible_samples))
print("Neighbors per focal cell:", K)

for sample_number, sid in enumerate(
    eligible_samples["sample_id"].tolist(), start=1
):

    # All dermal cells in this sample
    sample_mask = (
        (sample_values == sid)
        & (compartment_values == "dermis")
    )

    sample_coords = coordinates[sample_mask]
    sample_labels = detailed_values[sample_mask]

    # Keep only focal cells as neighborhood centers
    focal_mask = np.isin(sample_labels, focal_types)

    focal_coords = sample_coords[focal_mask]
    focal_labels = sample_labels[focal_mask]

    # Build tree from ALL dermal cells
    tree = cKDTree(sample_coords)

    # Query one extra neighbor so we can remove the center cell
    distances, neighbor_indices = tree.query(
        focal_coords,
        k=K + 1,
        workers=-1
    )

    # Remove each center cell's own index from its neighbor list.
    # This also handles duplicate spatial coordinates.
    focal_indices = np.flatnonzero(focal_mask)

    neighbor_labels = np.empty(
        (len(focal_indices), K),
        dtype=object
    )

    for i, center_index in enumerate(focal_indices):
        row = neighbor_indices[i]
        row = row[row != center_index][:K]

        if len(row) != K:
            raise ValueError(
                f"Insufficient neighbors in sample {sid}"
            )

        neighbor_labels[i] = sample_labels[row]

    # Summarize neighborhood composition by center population
    for center_type in focal_types:

        center_rows = np.flatnonzero(
            focal_labels == center_type
        )

        if len(center_rows) == 0:
            continue

        labels_for_center = neighbor_labels[center_rows]

        record = {
            "sample_id": sid,
            "center_type": center_type,
            "center_count": len(center_rows),
        }

        # Neighbor proportions for each focal population
        for target_type in focal_types:
            target_fraction = np.mean(
                labels_for_center == target_type
            )
            record[f"{target_type}_neighbor_fraction"] = (
                target_fraction
            )

            record[f"{target_type}_neighbor_count_mean"] = (
                np.mean(
                    np.sum(
                        labels_for_center == target_type,
                        axis=1
                    )
                )
            )

        # Neighborhood contains at least one cell of each
        # of the other two focal populations
        other_types = [
            x for x in focal_types if x != center_type
        ]

        has_both_other_types = (
            np.any(
                labels_for_center == other_types[0],
                axis=1
            )
            &
            np.any(
                labels_for_center == other_types[1],
                axis=1
            )
        )

        record["both_other_types_present_pct"] = (
            100 * np.mean(has_both_other_types)
        )

        # Neighborhood contains all three focal populations,
        # including the center cell's own population
        has_all_three = np.ones(
            len(center_rows), dtype=bool
        )

        for target_type in focal_types:
            has_all_three &= np.any(
                labels_for_center == target_type,
                axis=1
            )

        record["all_three_types_present_pct"] = (
            100 * np.mean(has_all_three)
        )

        sample_records.append(record)

    if sample_number % 10 == 0:
        print(
            f"Processed {sample_number}/"
            f"{len(eligible_samples)} samples"
        )

# ---------------------------------------------------------
# RESULTS
# ---------------------------------------------------------

neighborhood_df = pd.DataFrame(sample_records)

# Attach donor and site information
metadata = (
    distance_wide[
        ["sample_id", "donor_id", "anatomic_site"]
    ]
    .drop_duplicates("sample_id")
)

neighborhood_df = neighborhood_df.merge(
    metadata,
    on="sample_id",
    how="left"
)

print("\nANALYSIS COMPLETE")
print("Sample-population rows:", len(neighborhood_df))
print("Samples represented:", neighborhood_df["sample_id"].nunique())
print("Donors represented:", neighborhood_df["donor_id"].nunique())

print("\nSAMPLE-LEVEL RESULTS")
display(
    neighborhood_df[
        [
            "sample_id",
            "donor_id",
            "center_type",
            "center_count",
            "Mast_neighbor_fraction",
            "Schwann_neighbor_fraction",
            "Retic Fib III_neighbor_fraction",
            "both_other_types_present_pct",
            "all_three_types_present_pct",
        ]
    ].head(12).round(4)
)

print("\nSUMMARY BY CENTER POPULATION")
summary_columns = [
    "Mast_neighbor_fraction",
    "Schwann_neighbor_fraction",
    "Retic Fib III_neighbor_fraction",
    "both_other_types_present_pct",
    "all_three_types_present_pct",
]

display(
    neighborhood_df.groupby("center_type")[
        summary_columns
    ].agg(["median", "mean", "std"]).round(4)
)

print("\nDONOR-LEVEL SUMMARY")
donor_neighborhood_df = (
    neighborhood_df.groupby(
        ["donor_id", "center_type"]
    )[summary_columns]
    .median()
    .reset_index()
)

display(
    donor_neighborhood_df.groupby("center_type")[
        summary_columns
    ].agg(["median", "mean", "std"]).round(4)
)
```

```python

from scipy.spatial import cKDTree
import numpy as np
import pandas as pd

# ---------------------------------------------------------
# ANALYSIS 24A
# TILE-CONSTRAINED THREE-POPULATION NULL MODEL
# ---------------------------------------------------------

K = 10
TILE_WIDTH = 150
N_PERMUTATIONS = 199
RANDOM_SEED = 42

rng = np.random.default_rng(RANDOM_SEED)

# First sample in the eligible cohort
pilot_id = eligible_samples.iloc[0]["sample_id"]

mask = (
    (sample_values == pilot_id)
    & (compartment_values == "dermis")
)

pilot_coords = coordinates[mask]
pilot_original = detailed_values[mask]

focal_types = ["Mast", "Schwann", "Retic Fib III"]

# Collapse all non-focal identities into Other
pilot_labels = np.where(
    np.isin(pilot_original, focal_types),
    pilot_original,
    "Other"
)

print("PILOT SAMPLE:", pilot_id)
print("Total dermal cells:", len(pilot_coords))

for label in ["Mast", "Schwann", "Retic Fib III", "Other"]:
    print(label, np.sum(pilot_labels == label))

# ---------------------------------------------------------
# BUILD NEIGHBOR GRAPH
# ---------------------------------------------------------

print("\nBuilding nearest-neighbor graph...")

tree = cKDTree(pilot_coords)

_, neighbor_indices = tree.query(
    pilot_coords,
    k=K + 1,
    workers=-1
)

# Remove each cell's own index
all_indices = np.arange(len(pilot_coords))

neighbor_indices = np.array([
    row[row != i][:K]
    for i, row in enumerate(neighbor_indices)
])

assert neighbor_indices.shape == (
    len(pilot_coords), K
)

# ---------------------------------------------------------
# DEFINE SPATIAL TILES
# ---------------------------------------------------------

tile_x = np.floor(
    (pilot_coords[:, 0] - pilot_coords[:, 0].min())
    / TILE_WIDTH
).astype(int)

tile_y = np.floor(
    (pilot_coords[:, 1] - pilot_coords[:, 1].min())
    / TILE_WIDTH
).astype(int)

tile_ids = pd.Series(
    list(zip(tile_x, tile_y))
)

tile_groups = [
    np.asarray(indices, dtype=int)
    for indices in tile_ids.groupby(tile_ids).indices.values()
]

print("Spatial tiles:", len(tile_groups))

# Verify that tile-level population counts are preserved
original_tile_counts = []

for indices in tile_groups:
    original_tile_counts.append(
        pd.Series(pilot_labels[indices]).value_counts().to_dict()
    )

# ---------------------------------------------------------
# METRIC CALCULATION
# ---------------------------------------------------------

def calculate_metrics(labels, neighbors):

    neighbor_labels = labels[neighbors]

    results = {}

    for center_type in focal_types:

        center_rows = np.flatnonzero(
            labels == center_type
        )

        local = neighbor_labels[center_rows]

        record = {}

        for target_type in focal_types:
            record[f"{target_type}_neighbor_fraction"] = (
                np.mean(local == target_type)
            )

        other_types = [
            x for x in focal_types
            if x != center_type
        ]

        both_present = (
            np.any(local == other_types[0], axis=1)
            &
            np.any(local == other_types[1], axis=1)
        )

        all_three = np.ones(
            len(center_rows), dtype=bool
        )

        for target_type in focal_types:
            all_three &= np.any(
                local == target_type, axis=1
            )

        record["both_other_types_pct"] = (
            100 * np.mean(both_present)
        )

        record["all_three_types_pct"] = (
            100 * np.mean(all_three)
        )

        record["center_count"] = len(center_rows)

        results[center_type] = record

    return results


observed = calculate_metrics(
    pilot_labels,
    neighbor_indices
)

print("\nOBSERVED PILOT METRICS")
display(pd.DataFrame(observed).T.round(4))

# ---------------------------------------------------------
# CONSTRAINED PERMUTATIONS
# ---------------------------------------------------------

print("\nRunning constrained permutations...")

null_records = []

for permutation in range(N_PERMUTATIONS):

    permuted_labels = pilot_labels.copy()

    # Shuffle labels only among positions within each tile
    for indices in tile_groups:
        permuted_labels[indices] = rng.permutation(
            permuted_labels[indices]
        )

    randomized = calculate_metrics(
        permuted_labels,
        neighbor_indices
    )

    for center_type, metrics in randomized.items():
        for metric, value in metrics.items():
            null_records.append({
                "permutation": permutation,
                "center_type": center_type,
                "metric": metric,
                "value": value
            })

    if (permutation + 1) % 50 == 0:
        print(
            f"Completed {permutation + 1}/"
            f"{N_PERMUTATIONS}"
        )

null_pilot_df = pd.DataFrame(null_records)

# ---------------------------------------------------------
# COMPARE OBSERVED WITH NULL
# ---------------------------------------------------------

comparison_records = []

for center_type in focal_types:

    for metric in [
        "Schwann_neighbor_fraction",
        "Retic Fib III_neighbor_fraction",
        "both_other_types_pct",
        "all_three_types_pct"
    ]:

        observed_value = observed[center_type][metric]

        null_values = null_pilot_df.loc[
            (null_pilot_df["center_type"] == center_type)
            & (null_pilot_df["metric"] == metric),
            "value"
        ].to_numpy()

        lower_p = (
            1 + np.sum(null_values <= observed_value)
        ) / (len(null_values) + 1)

        upper_p = (
            1 + np.sum(null_values >= observed_value)
        ) / (len(null_values) + 1)

        two_sided_p = min(
            1.0,
            2 * min(lower_p, upper_p)
        )

        comparison_records.append({
            "center_type": center_type,
            "metric": metric,
            "observed": observed_value,
            "null_mean": np.mean(null_values),
            "null_2.5%": np.quantile(null_values, 0.025),
            "null_97.5%": np.quantile(null_values, 0.975),
            "observed_minus_null": (
                observed_value - np.mean(null_values)
            ),
            "two_sided_empirical_p": two_sided_p
        })

pilot_comparison_df = pd.DataFrame(
    comparison_records
)

print("\nOBSERVED VS CONSTRAINED NULL")
display(pilot_comparison_df.round(5))
```

```python

from scipy.spatial import cKDTree
import numpy as np
import pandas as pd

# ---------------------------------------------------------
# ANALYSIS 24B: DATASET-WIDE CONSTRAINED NULL
# ---------------------------------------------------------

K = 10
TILE_WIDTH = 150
N_PERMUTATIONS = 199

rng = np.random.default_rng(20261122)

focal_types = ["Mast", "Schwann", "Retic Fib III"]

metrics_to_test = [
    "Schwann_neighbor_fraction",
    "Retic Fib III_neighbor_fraction",
    "both_other_types_pct",
    "all_three_types_pct"
]

def calculate_neighborhood_metrics(labels, neighbors):

    neighbor_labels = labels[neighbors]
    results = {}

    for center_type in focal_types:

        center_rows = np.flatnonzero(
            labels == center_type
        )

        local = neighbor_labels[center_rows]

        record = {}

        for target_type in focal_types:
            record[f"{target_type}_neighbor_fraction"] = (
                np.mean(local == target_type)
            )

        other_types = [
            x for x in focal_types
            if x != center_type
        ]

        both_present = (
            np.any(local == other_types[0], axis=1)
            &
            np.any(local == other_types[1], axis=1)
        )

        all_three = np.ones(
            len(center_rows), dtype=bool
        )

        for target_type in focal_types:
            all_three &= np.any(
                local == target_type, axis=1
            )

        record["both_other_types_pct"] = (
            100 * np.mean(both_present)
        )

        record["all_three_types_pct"] = (
            100 * np.mean(all_three)
        )

        results[center_type] = record

    return results


dataset_results = []

sample_list = eligible_samples["sample_id"].tolist()

print("Running constrained permutations")
print("Samples:", len(sample_list))
print("Permutations per sample:", N_PERMUTATIONS)

for sample_number, sid in enumerate(sample_list, start=1):

    sample_mask = (
        (sample_values == sid)
        & (compartment_values == "dermis")
    )

    sample_coords = coordinates[sample_mask]
    original_labels = detailed_values[sample_mask]

    labels = np.where(
        np.isin(original_labels, focal_types),
        original_labels,
        "Other"
    )

    # Build nearest-neighbor graph once per sample
    tree = cKDTree(sample_coords)

    _, neighbors = tree.query(
        sample_coords,
        k=K + 1,
        workers=-1
    )

    # Remove each cell's own index
    neighbors = np.array([
        row[row != i][:K]
        for i, row in enumerate(neighbors)
    ])

    # Construct spatial tiles
    tx = np.floor(
        (sample_coords[:, 0] - sample_coords[:, 0].min())
        / TILE_WIDTH
    ).astype(int)

    ty = np.floor(
        (sample_coords[:, 1] - sample_coords[:, 1].min())
        / TILE_WIDTH
    ).astype(int)

    tile_series = pd.Series(list(zip(tx, ty)))

    tile_groups = [
        np.asarray(idx, dtype=int)
        for idx in tile_series.groupby(
            tile_series
        ).indices.values()
    ]

    # Observed metrics
    observed = calculate_neighborhood_metrics(
        labels, neighbors
    )

    # Store null values
    null_values = {
        (center, metric): []
        for center in focal_types
        for metric in metrics_to_test
    }

    for permutation in range(N_PERMUTATIONS):

        permuted = labels.copy()

        for indices in tile_groups:
            permuted[indices] = rng.permutation(
                permuted[indices]
            )

        randomized = calculate_neighborhood_metrics(
            permuted, neighbors
        )

        for center in focal_types:
            for metric in metrics_to_test:
                null_values[(center, metric)].append(
                    randomized[center][metric]
                )

    # Summarize sample-specific observed and null results
    for center in focal_types:

        for metric in metrics_to_test:

            obs = observed[center][metric]

            null = np.asarray(
                null_values[(center, metric)]
            )

            lower_p = (
                1 + np.sum(null <= obs)
            ) / (N_PERMUTATIONS + 1)

            upper_p = (
                1 + np.sum(null >= obs)
            ) / (N_PERMUTATIONS + 1)

            p_two = min(
                1.0,
                2 * min(lower_p, upper_p)
            )

            dataset_results.append({
                "sample_id": sid,
                "center_type": center,
                "metric": metric,
                "observed": obs,
                "null_mean": np.mean(null),
                "null_lower": np.quantile(null, 0.025),
                "null_upper": np.quantile(null, 0.975),
                "difference": obs - np.mean(null),
                "ratio": (
                    obs / np.mean(null)
                    if np.mean(null) > 0 else np.nan
                ),
                "sample_p": p_two,
            })

    if sample_number % 5 == 0:
        print(
            f"Completed {sample_number}/"
            f"{len(sample_list)} samples"
        )

# ---------------------------------------------------------
# SAMPLE-LEVEL RESULTS
# ---------------------------------------------------------

spatial_null_df = pd.DataFrame(dataset_results)

metadata = (
    distance_wide[
        ["sample_id", "donor_id", "anatomic_site"]
    ]
    .drop_duplicates("sample_id")
)

spatial_null_df = spatial_null_df.merge(
    metadata,
    on="sample_id",
    how="left"
)

print("\nCOMPLETED")
print("Rows:", len(spatial_null_df))
print("Samples:", spatial_null_df["sample_id"].nunique())
print("Donors:", spatial_null_df["donor_id"].nunique())

# ---------------------------------------------------------
# DONOR-LEVEL EFFECTS
# ---------------------------------------------------------

donor_effects = (
    spatial_null_df.groupby(
        ["donor_id", "center_type", "metric"]
    )
    .agg(
        median_difference=("difference", "median"),
        median_ratio=("ratio", "median"),
        samples=("sample_id", "nunique")
    )
    .reset_index()
)

print("\nDONOR-LEVEL EFFECT SUMMARY")

donor_summary_24 = (
    donor_effects.groupby(
        ["center_type", "metric"]
    )
    .agg(
        donors=("donor_id", "nunique"),
        median_difference=("median_difference", "median"),
        mean_difference=("median_difference", "mean"),
        donors_above_null=(
            "median_difference",
            lambda x: int(np.sum(x > 0))
        ),
        median_ratio=("median_ratio", "median")
    )
    .reset_index()
)

display(donor_summary_24.round(4))

print("\nKEY SCHWANN / RETIC FIB III COMPARISONS")

key_comparisons = donor_summary_24[
    (
        donor_summary_24["center_type"].isin(
            ["Schwann", "Retic Fib III"]
        )
    )
    &
    (
        donor_summary_24["metric"].isin(
            [
                "Schwann_neighbor_fraction",
                "Retic Fib III_neighbor_fraction"
            ]
        )
    )
]

display(key_comparisons.round(4))

# Save results
spatial_null_df.to_csv(
    f"{output_dir}/analysis24B_sample_spatial_null.csv",
    index=False
)

donor_summary_24.to_csv(
    f"{output_dir}/analysis24B_donor_effect_summary.csv",
    index=False
)

print("\nSaved Analysis 24B output files.")
```

```python

# ============================================================
# ANALYSIS 25B: DONOR-LEVEL STATISTICAL VALIDATION
# ============================================================

import numpy as np
import pandas as pd
from scipy.stats import binomtest, wilcoxon
from statsmodels.stats.multitest import multipletests
import os

# Reproducibility
RANDOM_SEED = 20261122
N_BOOTSTRAP = 20000
rng = np.random.default_rng(RANDOM_SEED)

# ------------------------------------------------------------
# 1. Validate input
# ------------------------------------------------------------

assert "donor_effects" in globals(), "donor_effects not found."

required = {
    "donor_id",
    "center_type",
    "metric",
    "median_difference",
    "median_ratio"
}

assert required.issubset(donor_effects.columns), (
    "Required columns are missing."
)

data = donor_effects.copy()

data["median_difference"] = pd.to_numeric(
    data["median_difference"], errors="coerce"
)

data = data.dropna(
    subset=["donor_id", "center_type", "metric", "median_difference"]
)

# Four primary comparisons defined before testing
primary = [
    ("Retic Fib III", "Schwann_neighbor_fraction"),
    ("Retic Fib III", "Retic Fib III_neighbor_fraction"),
    ("Schwann", "Retic Fib III_neighbor_fraction"),
    ("Schwann", "Schwann_neighbor_fraction")
]

# ------------------------------------------------------------
# 2. Donor-level tests and bootstrap confidence intervals
# ------------------------------------------------------------

results = []

for center, metric in primary:

    subset = data[
        (data["center_type"] == center) &
        (data["metric"] == metric)
    ].copy()

    # One observation per donor
    assert not subset["donor_id"].duplicated().any(), (
        f"Duplicate donor observations: {center}, {metric}"
    )

    values = subset["median_difference"].to_numpy()
    n = len(values)

    positive = int(np.sum(values > 0))
    negative = int(np.sum(values < 0))
    zero = int(np.sum(values == 0))

    # Exact two-sided sign test
    sign_result = binomtest(
        positive,
        n=positive + negative,
        p=0.5,
        alternative="two-sided"
    )

    # One-sided Wilcoxon signed-rank test
    if np.all(values == 0):
        wilcoxon_stat = 0.0
        wilcoxon_p = 1.0
    else:
        try:
            w = wilcoxon(
                values,
                alternative="greater",
                zero_method="wilcox",
                method="exact"
            )
            wilcoxon_stat = float(w.statistic)
            wilcoxon_p = float(w.pvalue)
        except Exception:
            wilcoxon_stat = np.nan
            wilcoxon_p = np.nan

    # Bootstrap median confidence interval
    boot_medians = np.empty(N_BOOTSTRAP)

    for b in range(N_BOOTSTRAP):
        sample = rng.choice(values, size=n, replace=True)
        boot_medians[b] = np.median(sample)

    ci_lower, ci_upper = np.percentile(
        boot_medians, [2.5, 97.5]
    )

    results.append({
        "center_type": center,
        "metric": metric,
        "donors": n,
        "positive_donors": positive,
        "negative_donors": negative,
        "zero_donors": zero,
        "median_difference": np.median(values),
        "mean_difference": np.mean(values),
        "median_ratio": subset["median_ratio"].median(),
        "sign_test_p": sign_result.pvalue,
        "wilcoxon_statistic": wilcoxon_stat,
        "wilcoxon_one_sided_p": wilcoxon_p,
        "bootstrap_95CI_lower": ci_lower,
        "bootstrap_95CI_upper": ci_upper
    })

results_df_25 = pd.DataFrame(results)

# ------------------------------------------------------------
# 3. Multiple-comparison correction
# ------------------------------------------------------------

results_df_25["sign_test_holm_p"] = multipletests(
    results_df_25["sign_test_p"],
    method="holm"
)[1]

results_df_25["wilcoxon_holm_p"] = multipletests(
    results_df_25["wilcoxon_one_sided_p"].fillna(1.0),
    method="holm"
)[1]

# ------------------------------------------------------------
# 4. Display results
# ------------------------------------------------------------

print("\n" + "=" * 85)
print("ANALYSIS 25B: DONOR-LEVEL VALIDATION")
print("=" * 85)

display_columns = [
    "center_type",
    "metric",
    "donors",
    "positive_donors",
    "median_difference",
    "median_ratio",
    "sign_test_holm_p",
    "wilcoxon_holm_p",
    "bootstrap_95CI_lower",
    "bootstrap_95CI_upper"
]

print(results_df_25[display_columns].to_string(index=False))

# ------------------------------------------------------------
# 5. Save outputs
# ------------------------------------------------------------

output_path = os.path.join(
    output_dir,
    "analysis25B_donor_level_validation.csv"
)

results_df_25.to_csv(output_path, index=False)

print("\nSaved:", output_path)
print("Completed donor-level statistical validation.")
```

```python

# ============================================================
# ANALYSIS 26: SPATIAL-SCALE SENSITIVITY
# Tile widths: 100, 150, 200
# K = 10; 199 constrained permutations per sample
# ============================================================

import numpy as np
import pandas as pd
import os
from scipy.spatial import cKDTree

# ------------------------------------------------------------
# SETTINGS
# ------------------------------------------------------------

K = 10
TILE_WIDTHS = [100, 200]  # 150 already completed in Analysis 24B
N_PERMUTATIONS = 199
SEED = 20261126

FOCAL_TYPES = ["Mast", "Schwann", "Retic Fib III"]

METRICS = [
    "Mast_neighbor_fraction",
    "Schwann_neighbor_fraction",
    "Retic Fib III_neighbor_fraction",
    "both_other_types_pct",
    "all_three_types_pct"
]

rng = np.random.default_rng(SEED)

# ------------------------------------------------------------
# VALIDATE INPUT ARRAYS
# ------------------------------------------------------------

required_arrays = [
    "coordinates",
    "sample_values",
    "compartment_values",
    "detailed_values"
]

for name in required_arrays:
    assert name in globals(), f"Missing required array: {name}"

n_total = len(sample_values)

assert len(coordinates) == n_total
assert len(compartment_values) == n_total
assert len(detailed_values) == n_total

assert "eligible_samples" in globals()
assert "spatial_null_df" in globals()

sample_ids = eligible_samples["sample_id"].tolist()

print("Samples:", len(sample_ids))
print("Tile widths to calculate:", TILE_WIDTHS)
print("Permutations per sample:", N_PERMUTATIONS)

# ------------------------------------------------------------
# METRIC FUNCTION
# Preserve Analysis 24B definitions
# ------------------------------------------------------------

def calculate_neighborhood_metrics(labels, neighbors):

    neighbor_labels = labels[neighbors]
    output = {}

    for center_type in FOCAL_TYPES:

        center_rows = np.where(labels == center_type)[0]

        if len(center_rows) == 0:
            continue

        local = neighbor_labels[center_rows]

        # Neighbor fractions
        for target_type in FOCAL_TYPES:
            output[
                (center_type, f"{target_type}_neighbor_fraction")
            ] = np.mean(local == target_type)

        # Both other focal populations present
        other_types = [
            x for x in FOCAL_TYPES if x != center_type
        ]

        has_first = np.any(
            local == other_types[0], axis=1
        )

        has_second = np.any(
            local == other_types[1], axis=1
        )

        output[
            (center_type, "both_other_types_pct")
        ] = 100 * np.mean(has_first & has_second)

        # All three focal populations represented
        has_all = np.ones(len(local), dtype=bool)

        for focal in FOCAL_TYPES:
            has_all &= np.any(local == focal, axis=1)

        output[
            (center_type, "all_three_types_pct")
        ] = 100 * np.mean(has_all)

    return output

# ------------------------------------------------------------
# TILE GROUP FUNCTION
# ------------------------------------------------------------

def make_tile_groups(coords, width):

    tx = np.floor(
        (coords[:, 0] - coords[:, 0].min()) / width
    ).astype(int)

    ty = np.floor(
        (coords[:, 1] - coords[:, 1].min()) / width
    ).astype(int)

    tile_codes = pd.Series(list(zip(tx, ty)))

    groups = tile_codes.groupby(tile_codes).indices

    return list(groups.values())

# ------------------------------------------------------------
# RUN EACH TILE WIDTH
# ------------------------------------------------------------

all_scale_results = []

for width in TILE_WIDTHS:

    print("\n" + "=" * 65)
    print(f"RUNNING TILE WIDTH: {width}")
    print("=" * 65)

    width_results = []

    for sample_number, sid in enumerate(sample_ids, start=1):

        # All dermal cells from this sample
        mask = (
            (sample_values == sid) &
            (compartment_values == "dermis")
        )

        coords = np.asarray(coordinates[mask], dtype=np.float64)

        original_labels = np.asarray(
            detailed_values[mask], dtype=object
        )

        # Collapse nonfocal cell labels to Other
        labels = np.where(
            np.isin(original_labels, FOCAL_TYPES),
            original_labels,
            "Other"
        ).astype(object)

        counts = {
            focal: int(np.sum(labels == focal))
            for focal in FOCAL_TYPES
        }

        # Eligibility check
        if min(counts.values()) < 20:
            print("Skipping ineligible sample:", sid)
            continue

        # Build KNN graph once per sample
        tree = cKDTree(coords)

        _, neighbor_indices = tree.query(
            coords,
            k=K + 1
        )

        # Remove each cell's self-index
        row_indices = np.arange(len(coords))

        neighbors = neighbor_indices[:, 1:]

        # Calculate observed metrics
        observed = calculate_neighborhood_metrics(
            labels, neighbors
        )

        # Construct spatial tiles
        tile_groups = make_tile_groups(coords, width)

        # Store null values
        null_values = {
            key: np.empty(N_PERMUTATIONS)
            for key in observed
        }

        # Constrained permutations
        for permutation in range(N_PERMUTATIONS):

            shuffled = labels.copy()

            for group in tile_groups:
                shuffled[group] = rng.permutation(
                    labels[group]
                )

            permuted_metrics = calculate_neighborhood_metrics(
                shuffled, neighbors
            )

            for key in observed:
                null_values[key][permutation] = (
                    permuted_metrics[key]
                )

        # Save sample-level results
        for key, observed_value in observed.items():

            center_type, metric = key
            null = null_values[key]

            null_mean = float(np.mean(null))

            null_lower = float(np.percentile(null, 2.5))
            null_upper = float(np.percentile(null, 97.5))

            difference = observed_value - null_mean

            ratio = (
                observed_value / null_mean
                if null_mean != 0 else np.nan
            )

            # Two-sided empirical permutation p-value
            p_value = (
                1 + np.sum(
                    np.abs(null - null_mean) >=
                    abs(observed_value - null_mean)
                )
            ) / (N_PERMUTATIONS + 1)

            width_results.append({
                "sample_id": sid,
                "tile_width": width,
                "center_type": center_type,
                "metric": metric,
                "observed": observed_value,
                "null_mean": null_mean,
                "null_lower": null_lower,
                "null_upper": null_upper,
                "difference": difference,
                "ratio": ratio,
                "sample_p": p_value
            })

        if sample_number % 5 == 0:
            print(
                f"Width {width}: "
                f"completed {sample_number}/{len(sample_ids)}"
            )

    width_df = pd.DataFrame(width_results)

    all_scale_results.append(width_df)

    print(
        f"Finished width {width}. "
        f"Rows: {len(width_df)}"
    )

# ------------------------------------------------------------
# COMBINE 100 AND 200 WITH EXISTING 150 RESULTS
# ------------------------------------------------------------

new_results = pd.concat(
    all_scale_results,
    ignore_index=True
)

primary_150 = spatial_null_df.copy()

primary_150["tile_width"] = 150

# Retain only metrics that were calculated in Analysis 26
primary_150 = primary_150[
    primary_150["metric"].isin(METRICS)
]

combined = pd.concat(
    [
        primary_150,
        new_results
    ],
    ignore_index=True
)

# ------------------------------------------------------------
# DONOR-LEVEL EFFECTS
# ------------------------------------------------------------

metadata = (
    spatial_null_df[
        ["sample_id", "donor_id", "anatomic_site"]
    ]
    .drop_duplicates("sample_id")
)

combined = combined.merge(
    metadata,
    on="sample_id",
    how="left",
    validate="many_to_one"
)

# One median effect per donor, width, center and metric
donor_scale = (
    combined
    .groupby(
        [
            "tile_width",
            "donor_id",
            "center_type",
            "metric"
        ],
        as_index=False
    )
    .agg(
        median_difference=("difference", "median"),
        median_ratio=("ratio", "median"),
        samples=("sample_id", "nunique")
    )
)

# Summarize across independent donors
scale_summary = (
    donor_scale
    .groupby(
        ["tile_width", "center_type", "metric"],
        as_index=False
    )
    .agg(
        donors=("donor_id", "nunique"),
        median_difference=("median_difference", "median"),
        mean_difference=("median_difference", "mean"),
        donors_above_null=(
            "median_difference",
            lambda x: int((x > 0).sum())
        ),
        median_ratio=("median_ratio", "median")
    )
)

# ------------------------------------------------------------
# FOCUS ON FOUR PRIMARY COMPARISONS
# ------------------------------------------------------------

primary_comparisons = [
    ("Retic Fib III", "Schwann_neighbor_fraction"),
    ("Retic Fib III", "Retic Fib III_neighbor_fraction"),
    ("Schwann", "Retic Fib III_neighbor_fraction"),
    ("Schwann", "Schwann_neighbor_fraction")
]

key_scale_summary = scale_summary[
    scale_summary.apply(
        lambda r: (
            r["center_type"], r["metric"]
        ) in primary_comparisons,
        axis=1
    )
].sort_values(
    ["center_type", "metric", "tile_width"]
)

print("\n" + "=" * 85)
print("ANALYSIS 26: SPATIAL-SCALE SENSITIVITY")
print("=" * 85)

print(
    key_scale_summary.to_string(index=False)
)

# ------------------------------------------------------------
# SAVE OUTPUTS
# ------------------------------------------------------------

combined.to_csv(
    os.path.join(
        output_dir,
        "analysis26_sample_scale_sensitivity.csv"
    ),
    index=False
)

donor_scale.to_csv(
    os.path.join(
        output_dir,
        "analysis26_donor_scale_effects.csv"
    ),
    index=False
)

scale_summary.to_csv(
    os.path.join(
        output_dir,
        "analysis26_scale_summary.csv"
    ),
    index=False
)

print("\nSaved all Analysis 26 files.")
print("Completed spatial-scale sensitivity analysis.")
```

```python

# ============================================================
# ANALYSIS 27: MULTI-SCALE DONOR-LEVEL INFERENCE
# Uses existing Analysis 26 results. No spatial permutations.
# ============================================================

import numpy as np
import pandas as pd
import os
import matplotlib.pyplot as plt

from scipy.stats import binomtest, wilcoxon
from statsmodels.stats.multitest import multipletests

N_BOOT = 20000
SEED = 20261127
rng = np.random.default_rng(SEED)

primary_comparisons = [
    ("Retic Fib III", "Schwann_neighbor_fraction"),
    ("Retic Fib III", "Retic Fib III_neighbor_fraction"),
    ("Schwann", "Retic Fib III_neighbor_fraction"),
    ("Schwann", "Schwann_neighbor_fraction")
]

# ------------------------------------------------------------
# 1. Validate donor-level data
# ------------------------------------------------------------

assert "donor_scale_26" in globals()

required = [
    "tile_width",
    "donor_id",
    "center_type",
    "metric",
    "median_difference"
]

assert all(c in donor_scale_26.columns for c in required)

data = donor_scale_26.copy()

results = []

# ------------------------------------------------------------
# 2. Test every comparison at each spatial scale
# ------------------------------------------------------------

for width in [100, 150, 200]:

    for center, metric in primary_comparisons:

        subset = data[
            (data["tile_width"] == width) &
            (data["center_type"] == center) &
            (data["metric"] == metric)
        ].copy()

        subset = subset.dropna(
            subset=["donor_id", "median_difference"]
        )

        assert not subset["donor_id"].duplicated().any()

        values = subset["median_difference"].to_numpy()
        n = len(values)

        positive = int(np.sum(values > 0))
        negative = int(np.sum(values < 0))
        zero = int(np.sum(values == 0))

        # Two-sided exact sign test
        sign_p = binomtest(
            positive,
            n=positive + negative,
            p=0.5,
            alternative="two-sided"
        ).pvalue

        # Two-sided Wilcoxon test
        if np.all(values == 0):
            w_stat, w_p = 0.0, 1.0
        else:
            w = wilcoxon(
                values,
                alternative="two-sided",
                zero_method="wilcox",
                method="exact"
            )
            w_stat, w_p = float(w.statistic), float(w.pvalue)

        # Bootstrap median confidence interval
        boot = np.empty(N_BOOT)

        for b in range(N_BOOT):
            sampled = rng.choice(values, size=n, replace=True)
            boot[b] = np.median(sampled)

        lower, upper = np.percentile(boot, [2.5, 97.5])

        results.append({
            "tile_width": width,
            "center_type": center,
            "metric": metric,
            "donors": n,
            "positive_donors": positive,
            "negative_donors": negative,
            "zero_donors": zero,
            "median_effect": np.median(values),
            "mean_effect": np.mean(values),
            "sign_test_p": sign_p,
            "wilcoxon_statistic": w_stat,
            "wilcoxon_p": w_p,
            "bootstrap_95CI_lower": lower,
            "bootstrap_95CI_upper": upper
        })

results_27 = pd.DataFrame(results)

# ------------------------------------------------------------
# 3. Correct for 12 comparisons
# ------------------------------------------------------------

results_27["sign_test_holm_p"] = multipletests(
    results_27["sign_test_p"],
    method="holm"
)[1]

results_27["wilcoxon_holm_p"] = multipletests(
    results_27["wilcoxon_p"],
    method="holm"
)[1]

# ------------------------------------------------------------
# 4. Display results
# ------------------------------------------------------------

print("=" * 95)
print("ANALYSIS 27: MULTI-SCALE DONOR-LEVEL INFERENCE")
print("=" * 95)

print(
    results_27.to_string(index=False)
)

# ------------------------------------------------------------
# 5. Create effect-size sensitivity figure
# ------------------------------------------------------------

fig, axes = plt.subplots(
    2, 2,
    figsize=(13, 9)
)

axes = axes.flatten()

for ax, (center, metric) in zip(axes, primary_comparisons):

    sub = results_27[
        (results_27["center_type"] == center) &
        (results_27["metric"] == metric)
    ].sort_values("tile_width")

    x = sub["tile_width"].to_numpy()
    y = sub["median_effect"].to_numpy()

    lower = sub["bootstrap_95CI_lower"].to_numpy()
    upper = sub["bootstrap_95CI_upper"].to_numpy()

    errors = np.vstack([
        y - lower,
        upper - y
    ])

    ax.errorbar(
        x,
        y,
        yerr=errors,
        marker="o",
        capsize=5,
        linewidth=1.5
    )

    ax.axhline(0, linestyle="--", linewidth=1)

    ax.set_title(f"{center}: {metric}")
    ax.set_xlabel("Spatial tile width (coordinate units)")
    ax.set_ylabel("Median donor effect (observed - null)")
    ax.set_xticks([100, 150, 200])
    ax.grid(alpha=0.25)

plt.suptitle(
    "Donor-Level Spatial Effect Across Tile Widths",
    fontsize=15
)

plt.tight_layout()

figure_path = os.path.join(
    output_dir,
    "analysis27_multiscale_effects.png"
)

plt.savefig(
    figure_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ------------------------------------------------------------
# 6. Save results
# ------------------------------------------------------------

csv_path = os.path.join(
    output_dir,
    "analysis27_multiscale_inference.csv"
)

results_27.to_csv(csv_path, index=False)

print("\nSaved:")
print(csv_path)
print(figure_path)
print("\nANALYSIS 27 COMPLETE")
```

```python

# ============================================================
# ANALYSIS 28C: DENSITY-STRATIFICATION FEASIBILITY
# Diagnostic only. No permutations yet.
# ============================================================

import numpy as np
import pandas as pd
from scipy.spatial import cKDTree

print("FOCAL TYPES:", FOCAL_TYPES)

print("\nELIGIBLE SAMPLE TABLE")
print("Columns:", list(eligible_samples.columns))
print(eligible_samples.head())

print("\nCELLS BY SAMPLE")
print("Number of samples:", len(cells_by_sample))

first_key = next(iter(cells_by_sample))
first_value = cells_by_sample[first_key]

print("Example sample key:", first_key)
print("Example stored value type:", type(first_value))
print("Example stored value shape:", np.shape(first_value))

# ------------------------------------------------------------
# Inspect all eligible samples
# ------------------------------------------------------------

WIDTH = 150
K_DENSITY = 10
N_BINS = 4

diagnostics = []

for sample_id in eligible_samples["sample_id"]:

    indices = np.asarray(cells_by_sample[sample_id])

    sample_coords = coordinates[indices]
    sample_labels = detailed_values[indices]

    # All-cell local density, based on 10th-nearest
    # other dermal cell distance.
    tree = cKDTree(sample_coords)

    distances, _ = tree.query(
        sample_coords,
        k=K_DENSITY + 1
    )

    density_distance = distances[:, -1]

    # Smaller distance = higher local density.
    density_bins = pd.qcut(
        density_distance,
        q=N_BINS,
        labels=False,
        duplicates="drop"
    )

    # Spatial tile membership
    tile_groups = make_tile_groups(
        sample_coords,
        WIDTH
    )

    tile_id = np.empty(len(sample_coords), dtype=int)

    for tile_number, rows in enumerate(tile_groups):
        tile_id[rows] = tile_number

    strata = pd.DataFrame({
        "tile": tile_id,
        "density_bin": np.asarray(density_bins)
    })

    strata["stratum"] = (
        strata["tile"].astype(str)
        + "_"
        + strata["density_bin"].astype(str)
    )

    group_sizes = strata["stratum"].value_counts()

    focal_mask = np.isin(
        sample_labels,
        FOCAL_TYPES
    )

    focal_counts = (
        pd.DataFrame({
            "stratum": strata["stratum"],
            "label": sample_labels
        })
        .loc[lambda x: x["label"].isin(FOCAL_TYPES)]
        .groupby(["stratum", "label"])
        .size()
    )

    diagnostics.append({
        "sample_id": sample_id,
        "cells": len(indices),
        "tiles": len(tile_groups),
        "strata": len(group_sizes),
        "median_stratum_size": group_sizes.median(),
        "strata_size_ge5_pct": 100 * np.mean(group_sizes >= 5),
        "strata_size_ge10_pct": 100 * np.mean(group_sizes >= 10),
        "cells_in_strata_ge5_pct": 100 * group_sizes[group_sizes >= 5].sum() / len(indices),
        "focal_cells": int(focal_mask.sum()),
        "focal_cells_in_strata_ge5_pct": (
            100 * sum(
                int(group_sizes.get(s, 0) >= 5) * int(count)
                for (s, label), count in focal_counts.items()
            ) / max(int(focal_mask.sum()), 1)
        )
    })

density_feasibility = pd.DataFrame(diagnostics)

print("\n" + "=" * 90)
print("DENSITY-STRATIFICATION FEASIBILITY")
print("=" * 90)

print(
    density_feasibility[
        [
            "cells", "tiles", "strata",
            "median_stratum_size",
            "strata_size_ge5_pct",
            "strata_size_ge10_pct",
            "cells_in_strata_ge5_pct",
            "focal_cells",
            "focal_cells_in_strata_ge5_pct"
        ]
    ].describe().round(2)
)

print("\nLowest-coverage samples:")
print(
    density_feasibility.sort_values(
        "focal_cells_in_strata_ge5_pct"
    ).head(10).to_string(index=False)
)

print("\nSaved diagnostic table in memory as density_feasibility.")
print("ANALYSIS 28C COMPLETE")
```

```python

# ============================================================
# ANALYSIS 28G: DENSITY-MATCHED SPATIAL NULL
# All dermal cells; 150-unit tiles; density quartiles.
# ============================================================

import numpy as np
import pandas as pd
from scipy.spatial import cKDTree
from statsmodels.stats.multitest import multipletests
import os
import time

WIDTH = 150
K_NEIGHBORS = 10
K_DENSITY = 10
N_DENSITY_BINS = 4
N_PERMUTATIONS = 199
SEED = 28001

rng = np.random.default_rng(SEED)

FOCAL_TYPES = ["Mast", "Schwann", "Retic Fib III"]

PRIMARY_METRICS = [
    ("Retic Fib III", "Schwann_neighbor_fraction"),
    ("Retic Fib III", "Retic Fib III_neighbor_fraction"),
    ("Schwann", "Retic Fib III_neighbor_fraction"),
    ("Schwann", "Schwann_neighbor_fraction")
]

# ------------------------------------------------------------
# Helper: calculate four primary neighborhood metrics
# ------------------------------------------------------------

def primary_neighborhood_metrics(labels, neighbors):

    output = {}

    neighbor_labels = labels[neighbors]

    for center_type, metric in PRIMARY_METRICS:

        center_rows = np.where(labels == center_type)[0]

        if len(center_rows) == 0:
            output[(center_type, metric)] = np.nan
            continue

        target_type = metric.replace(
            "_neighbor_fraction", ""
        )

        output[(center_type, metric)] = np.mean(
            neighbor_labels[center_rows] == target_type
        )

    return output


# ------------------------------------------------------------
# Helper: build density/tile strata
# ------------------------------------------------------------

def build_density_strata(sample_coords):

    tree = cKDTree(sample_coords)

    density_distances, _ = tree.query(
        sample_coords,
        k=K_DENSITY + 1
    )

    # Distance to the 10th other dermal cell.
    local_density_distance = density_distances[:, -1]

    density_bins = pd.qcut(
        local_density_distance,
        q=N_DENSITY_BINS,
        labels=False,
        duplicates="drop"
    )

    density_bins = np.asarray(density_bins)

    tile_groups = make_tile_groups(
        sample_coords,
        WIDTH
    )

    tile_ids = np.empty(len(sample_coords), dtype=int)

    for tile_number, rows in enumerate(tile_groups):
        tile_ids[rows] = tile_number

    strata = (
        pd.Series(tile_ids).astype(str)
        + "_"
        + pd.Series(density_bins).astype(str)
    ).to_numpy()

    groups = pd.Series(
        np.arange(len(strata))
    ).groupby(strata).indices

    return (
        tree,
        local_density_distance,
        density_bins,
        tile_ids,
        list(groups.values())
    )


# ------------------------------------------------------------
# Select eligible samples
# ------------------------------------------------------------

sample_ids_28 = eligible_samples["sample_id"].tolist()

results_28 = []
coverage_28 = []

start_time = time.time()

for sample_number, sample_id in enumerate(sample_ids_28, 1):

    mask = (
        (sample_values == sample_id) &
        (compartment_values == "dermis")
    )

    sample_coords = coordinates[mask]
    sample_labels = detailed_values[mask]

    # Validate complete sample
    assert len(sample_coords) == len(sample_labels)
    assert np.isfinite(sample_coords).all()

    counts = pd.Series(sample_labels).value_counts()

    if not all(counts.get(x, 0) > 0 for x in FOCAL_TYPES):
        continue

    # Build spatial and density structure
    tree, density_dist, density_bins, tile_ids, groups = (
        build_density_strata(sample_coords)
    )

    # k=10 nearest other cells
    _, neighbor_indices = tree.query(
        sample_coords,
        k=K_NEIGHBORS + 1
    )

    neighbors = neighbor_indices[:, 1:]

    observed = primary_neighborhood_metrics(
        sample_labels,
        neighbors
    )

    # Permute labels only within tile-density strata
    null_values = {
        key: [] for key in PRIMARY_METRICS
    }

    for permutation in range(N_PERMUTATIONS):

        permuted = sample_labels.copy()

        for rows in groups:
            permuted[rows] = rng.permutation(
                permuted[rows]
            )

        null_metrics = primary_neighborhood_metrics(
            permuted,
            neighbors
        )

        for key in PRIMARY_METRICS:
            null_values[key].append(null_metrics[key])

    # Record results
    focal_count = int(
        np.isin(sample_labels, FOCAL_TYPES).sum()
    )

    eligible_strata = [
        rows for rows in groups if len(rows) >= 5
    ]

    eligible_rows = sum(len(x) for x in eligible_strata)

    coverage_28.append({
        "sample_id": sample_id,
        "cells": len(sample_coords),
        "tiles": len(np.unique(tile_ids)),
        "strata": len(groups),
        "strata_ge5": len(eligible_strata),
        "cells_in_strata_ge5_pct": (
            100 * eligible_rows / len(sample_coords)
        ),
        "focal_cells": focal_count,
        "density_bins": len(np.unique(density_bins))
    })

    for center_type, metric in PRIMARY_METRICS:

        key = (center_type, metric)

        null_array = np.asarray(
            null_values[key],
            dtype=float
        )

        obs = observed[key]

        null_mean = np.nanmean(null_array)
        null_sd = np.nanstd(null_array, ddof=1)

        # Two-sided empirical permutation p-value
        p_value = (
            1 + np.sum(
                np.abs(null_array - null_mean)
                >= abs(obs - null_mean)
            )
        ) / (N_PERMUTATIONS + 1)

        results_28.append({
            "sample_id": sample_id,
            "donor_id": cells_by_sample[sample_id]["donor_id"],
            "anatomic_site": cells_by_sample[sample_id]["anatomic_site"],
            "center_type": center_type,
            "metric": metric,
            "observed": obs,
            "null_mean": null_mean,
            "null_sd": null_sd,
            "null_lower": np.nanpercentile(null_array, 2.5),
            "null_upper": np.nanpercentile(null_array, 97.5),
            "difference": obs - null_mean,
            "ratio": obs / null_mean if null_mean > 0 else np.nan,
            "sample_p": p_value
        })

    if sample_number % 10 == 0:
        print(
            f"Completed {sample_number}/{len(sample_ids_28)} "
            f"samples | elapsed {time.time()-start_time:.1f}s"
        )


# ------------------------------------------------------------
# Assemble results
# ------------------------------------------------------------

density_null_df = pd.DataFrame(results_28)
density_coverage_df = pd.DataFrame(coverage_28)

print("\n" + "=" * 90)
print("ANALYSIS 28G: DENSITY-MATCHED NULL RESULTS")
print("=" * 90)

print("Samples:", density_null_df["sample_id"].nunique())
print("Donors:", density_null_df["donor_id"].nunique())
print("Rows:", len(density_null_df))

print("\nCoverage:")
print(
    density_coverage_df[
        [
            "cells",
            "tiles",
            "strata",
            "cells_in_strata_ge5_pct",
            "focal_cells"
        ]
    ].describe().round(2)
)

print("\nSample-level results:")
print(
    density_null_df.groupby(
        ["center_type", "metric"]
    ).agg(
        samples=("sample_id", "nunique"),
        mean_difference=("difference", "mean"),
        median_difference=("difference", "median"),
        positive_samples=("difference", lambda x: (x > 0).sum()),
        median_ratio=("ratio", "median")
    ).round(5)
)

# ------------------------------------------------------------
# Save
# ------------------------------------------------------------

density_null_df.to_csv(
    os.path.join(output_dir, "analysis28_density_null.csv"),
    index=False
)

density_coverage_df.to_csv(
    os.path.join(output_dir, "analysis28_density_coverage.csv"),
    index=False
)

print("\nSaved Analysis 28 outputs.")
print("ANALYSIS 28G COMPLETE")
```

```python

# ============================================================
# ANALYSIS 28H: DONOR-LEVEL INFERENCE
# Density-matched null; one effect per donor.
# ============================================================

import numpy as np
import pandas as pd
from scipy.stats import wilcoxon, binomtest
from statsmodels.stats.multitest import multipletests
import os

N_BOOT = 20000
SEED = 28002

rng = np.random.default_rng(SEED)

# One median sample-level effect per donor and comparison
donor_density_effects = (
    density_null_df
    .groupby(
        ["donor_id", "center_type", "metric"],
        as_index=False
    )
    .agg(
        median_difference=("difference", "median"),
        median_ratio=("ratio", "median"),
        number_of_samples=("sample_id", "nunique")
    )
)

comparison_results = []

for (center_type, metric), group in donor_density_effects.groupby(
    ["center_type", "metric"]
):

    effects = group["median_difference"].dropna().to_numpy()
    ratios = group["median_ratio"].dropna().to_numpy()

    n = len(effects)
    positive = int(np.sum(effects > 0))
    negative = int(np.sum(effects < 0))
    zero = int(np.sum(effects == 0))

    # Exact two-sided sign test
    sign_p = binomtest(
        positive,
        n=n - zero,
        p=0.5,
        alternative="two-sided"
    ).pvalue

    # Two-sided Wilcoxon signed-rank test
    wilcoxon_p = wilcoxon(
        effects,
        alternative="two-sided",
        zero_method="wilcox",
        method="exact"
    ).pvalue

    # Bootstrap donors, not samples
    boot_medians = np.empty(N_BOOT)

    for b in range(N_BOOT):
        boot = rng.choice(
            effects,
            size=n,
            replace=True
        )
        boot_medians[b] = np.median(boot)

    ci_low, ci_high = np.percentile(
        boot_medians,
        [2.5, 97.5]
    )

    comparison_results.append({
        "center_type": center_type,
        "metric": metric,
        "donors": n,
        "positive_donors": positive,
        "negative_donors": negative,
        "zero_donors": zero,
        "median_donor_difference": np.median(effects),
        "mean_donor_difference": np.mean(effects),
        "median_donor_ratio": np.median(ratios),
        "bootstrap_95CI_lower": ci_low,
        "bootstrap_95CI_upper": ci_high,
        "sign_p": sign_p,
        "wilcoxon_p": wilcoxon_p
    })

donor_density_summary = pd.DataFrame(comparison_results)

# Holm correction across the four prespecified comparisons
donor_density_summary["wilcoxon_holm_p"] = multipletests(
    donor_density_summary["wilcoxon_p"],
    method="holm"
)[1]

donor_density_summary["sign_holm_p"] = multipletests(
    donor_density_summary["sign_p"],
    method="holm"
)[1]

print("=" * 95)
print("ANALYSIS 28H: DONOR-LEVEL DENSITY-MATCHED INFERENCE")
print("=" * 95)

print(
    donor_density_summary[
        [
            "center_type",
            "metric",
            "donors",
            "positive_donors",
            "median_donor_difference",
            "median_donor_ratio",
            "bootstrap_95CI_lower",
            "bootstrap_95CI_upper",
            "wilcoxon_p",
            "wilcoxon_holm_p"
        ]
    ].to_string(index=False)
)

print("\nDonor-level effect table:")
print(
    donor_density_effects.sort_values(
        ["center_type", "metric", "donor_id"]
    ).to_string(index=False)
)

donor_density_summary.to_csv(
    os.path.join(
        output_dir,
        "analysis28H_donor_inference.csv"
    ),
    index=False
)

donor_density_effects.to_csv(
    os.path.join(
        output_dir,
        "analysis28H_donor_effects.csv"
    ),
    index=False
)

print("\nSaved Analysis 28H outputs.")
print("ANALYSIS 28H COMPLETE")
```

```python

# ============================================================
# ANALYSIS 28I: MAST-CENTERED DENSITY-MATCHED NULL
# Same 150-unit tiles, density quartiles, and k=10 graph.
# ============================================================

import numpy as np
import pandas as pd
from scipy.spatial import cKDTree
from scipy.stats import wilcoxon, binomtest
from statsmodels.stats.multitest import multipletests
import os
import time

N_PERMUTATIONS = 199
rng = np.random.default_rng(28003)

MAST_METRICS = [
    ("Mast", "Schwann"),
    ("Mast", "Retic Fib III")
]

mast_results = []
start_time = time.time()

for sample_number, sample_id in enumerate(sample_ids_28, 1):

    mask = (
        (sample_values == sample_id) &
        (compartment_values == "dermis")
    )

    sample_coords = coordinates[mask]
    sample_labels = detailed_values[mask]

    # Same density-matched strata as Analysis 28G
    tree, density_dist, density_bins, tile_ids, groups = (
        build_density_strata(sample_coords)
    )

    _, neighbor_indices = tree.query(
        sample_coords,
        k=K_NEIGHBORS + 1
    )

    neighbors = neighbor_indices[:, 1:]

    mast_rows = np.where(sample_labels == "Mast")[0]

    if len(mast_rows) == 0:
        continue

    observed = {}

    for center_type, target_type in MAST_METRICS:

        target_labels = sample_labels[neighbors[mast_rows]]

        observed[(center_type, target_type)] = np.mean(
            target_labels == target_type
        )

    null_values = {
        key: [] for key in MAST_METRICS
    }

    for permutation in range(N_PERMUTATIONS):

        permuted = sample_labels.copy()

        for rows in groups:
            permuted[rows] = rng.permutation(
                permuted[rows]
            )

        for center_type, target_type in MAST_METRICS:

            permuted_mast = np.where(
                permuted == center_type
            )[0]

            if len(permuted_mast) == 0:
                null_values[(center_type, target_type)].append(
                    np.nan
                )
                continue

            target_labels = permuted[
                neighbors[permuted_mast]
            ]

            null_values[
                (center_type, target_type)
            ].append(
                np.mean(target_labels == target_type)
            )

    for center_type, target_type in MAST_METRICS:

        key = (center_type, target_type)

        null_array = np.asarray(
            null_values[key],
            dtype=float
        )

        null_array = null_array[np.isfinite(null_array)]

        obs = observed[key]
        null_mean = np.mean(null_array)

        p_value = (
            1 + np.sum(
                np.abs(null_array - null_mean)
                >= abs(obs - null_mean)
            )
        ) / (len(null_array) + 1)

        mast_results.append({
            "sample_id": sample_id,
            "donor_id": cells_by_sample[sample_id]["donor_id"],
            "anatomic_site": cells_by_sample[sample_id]["anatomic_site"],
            "center_type": center_type,
            "target_type": target_type,
            "observed": obs,
            "null_mean": null_mean,
            "null_lower": np.percentile(null_array, 2.5),
            "null_upper": np.percentile(null_array, 97.5),
            "difference": obs - null_mean,
            "ratio": obs / null_mean if null_mean > 0 else np.nan,
            "sample_p": p_value,
            "mast_cells": len(mast_rows)
        })

    if sample_number % 10 == 0:
        print(
            f"Completed {sample_number}/{len(sample_ids_28)} "
            f"| elapsed {time.time()-start_time:.1f}s"
        )

mast_density_df = pd.DataFrame(mast_results)

# Donor-level aggregation
mast_donor_effects = (
    mast_density_df
    .groupby(
        ["donor_id", "center_type", "target_type"],
        as_index=False
    )
    .agg(
        median_difference=("difference", "median"),
        median_ratio=("ratio", "median"),
        samples=("sample_id", "nunique")
    )
)

mast_donor_summary = []

for (center_type, target_type), group in mast_donor_effects.groupby(
    ["center_type", "target_type"]
):

    effects = group["median_difference"].dropna().to_numpy()

    n = len(effects)
    positive = int(np.sum(effects > 0))
    zero = int(np.sum(effects == 0))

    sign_p = binomtest(
        positive,
        n=n-zero,
        p=0.5,
        alternative="two-sided"
    ).pvalue

    wilcoxon_p = wilcoxon(
        effects,
        alternative="two-sided",
        method="exact"
    ).pvalue

    mast_donor_summary.append({
        "center_type": center_type,
        "target_type": target_type,
        "donors": n,
        "positive_donors": positive,
        "median_donor_difference": np.median(effects),
        "median_donor_ratio": np.median(
            group["median_ratio"]
        ),
        "sign_p": sign_p,
        "wilcoxon_p": wilcoxon_p
    })

mast_donor_summary = pd.DataFrame(mast_donor_summary)

mast_donor_summary["wilcoxon_holm_p"] = multipletests(
    mast_donor_summary["wilcoxon_p"],
    method="holm"
)[1]

mast_donor_summary["sign_holm_p"] = multipletests(
    mast_donor_summary["sign_p"],
    method="holm"
)[1]

print("\n" + "=" * 90)
print("ANALYSIS 28I: MAST-CENTERED DENSITY-MATCHED RESULTS")
print("=" * 90)

print("\nSample-level summary:")
print(
    mast_density_df.groupby(
        ["center_type", "target_type"]
    ).agg(
        samples=("sample_id", "nunique"),
        positive_samples=("difference", lambda x: (x > 0).sum()),
        median_difference=("difference", "median"),
        median_ratio=("ratio", "median")
    ).round(5)
)

print("\nDonor-level inference:")
print(mast_donor_summary.to_string(index=False))

mast_density_df.to_csv(
    os.path.join(output_dir, "analysis28I_mast_density_samples.csv"),
    index=False
)

mast_donor_effects.to_csv(
    os.path.join(output_dir, "analysis28I_mast_density_donors.csv"),
    index=False
)

mast_donor_summary.to_csv(
    os.path.join(output_dir, "analysis28I_mast_density_summary.csv"),
    index=False
)

print("\nSaved Analysis 28I outputs.")
print("ANALYSIS 28I COMPLETE")
```

```python

# ============================================================
# ANALYSIS 28J: FIXED MAST-CENTER DENSITY-MATCHED NULL
# Mast locations remain fixed; all other labels are shuffled
# within the existing tile x density strata.
# ============================================================

import numpy as np
import pandas as pd
from scipy.stats import wilcoxon, binomtest
from statsmodels.stats.multitest import multipletests
import os

N_PERMUTATIONS = 199
rng = np.random.default_rng(28004)

fixed_mast_results = []

for sample_number, sample_id in enumerate(sample_ids_28, 1):

    mask = (
        (sample_values == sample_id) &
        (compartment_values == "dermis")
    )

    sample_coords = coordinates[mask]
    sample_labels = detailed_values[mask].copy()

    tree, density_dist, density_bins, tile_ids, groups = (
        build_density_strata(sample_coords)
    )

    _, neighbor_indices = tree.query(
        sample_coords,
        k=K_NEIGHBORS + 1
    )

    neighbors = neighbor_indices[:, 1:]

    mast_rows = np.where(sample_labels == "Mast")[0]

    observed = {}

    for target_type in ["Schwann", "Retic Fib III"]:
        observed[target_type] = np.mean(
            sample_labels[neighbors[mast_rows]] == target_type
        )

    null_values = {
        "Schwann": [],
        "Retic Fib III": []
    }

    for permutation in range(N_PERMUTATIONS):

        permuted = sample_labels.copy()

        for rows in groups:

            # Keep observed Mast positions fixed.
            non_mast_rows = rows[
                sample_labels[rows] != "Mast"
            ]

            # Shuffle all non-Mast identities within this stratum.
            permuted[non_mast_rows] = rng.permutation(
                sample_labels[non_mast_rows]
            )

        # Evaluate the original, fixed mast-cell centers.
        for target_type in ["Schwann", "Retic Fib III"]:

            null_values[target_type].append(
                np.mean(
                    permuted[neighbors[mast_rows]] == target_type
                )
            )

    for target_type in ["Schwann", "Retic Fib III"]:

        null_array = np.asarray(
            null_values[target_type]
        )

        obs = observed[target_type]
        null_mean = np.mean(null_array)

        p_value = (
            1 + np.sum(
                np.abs(null_array - null_mean)
                >= abs(obs - null_mean)
            )
        ) / (N_PERMUTATIONS + 1)

        fixed_mast_results.append({
            "sample_id": sample_id,
            "donor_id": cells_by_sample[sample_id]["donor_id"],
            "anatomic_site": cells_by_sample[sample_id]["anatomic_site"],
            "target_type": target_type,
            "mast_cells": len(mast_rows),
            "observed": obs,
            "null_mean": null_mean,
            "null_lower": np.percentile(null_array, 2.5),
            "null_upper": np.percentile(null_array, 97.5),
            "difference": obs - null_mean,
            "ratio": obs / null_mean if null_mean > 0 else np.nan,
            "sample_p": p_value
        })

    if sample_number % 10 == 0:
        print(f"Completed {sample_number}/{len(sample_ids_28)}")


fixed_mast_df = pd.DataFrame(fixed_mast_results)

fixed_mast_donors = (
    fixed_mast_df
    .groupby(["donor_id", "target_type"], as_index=False)
    .agg(
        median_difference=("difference", "median"),
        median_ratio=("ratio", "median"),
        samples=("sample_id", "nunique")
    )
)

summary = []

for target_type, group in fixed_mast_donors.groupby("target_type"):

    effects = group["median_difference"].to_numpy()

    positive = int(np.sum(effects > 0))
    zero = int(np.sum(effects == 0))
    n = len(effects)

    sign_p = binomtest(
        positive,
        n=n-zero,
        p=0.5,
        alternative="two-sided"
    ).pvalue

    wilcoxon_p = wilcoxon(
        effects,
        alternative="two-sided",
        method="exact"
    ).pvalue

    summary.append({
        "target_type": target_type,
        "donors": n,
        "positive_donors": positive,
        "median_donor_difference": np.median(effects),
        "median_donor_ratio": np.median(
            group["median_ratio"]
        ),
        "sign_p": sign_p,
        "wilcoxon_p": wilcoxon_p
    })

fixed_mast_summary = pd.DataFrame(summary)

fixed_mast_summary["wilcoxon_holm_p"] = multipletests(
    fixed_mast_summary["wilcoxon_p"],
    method="holm"
)[1]

print("\nFIXED MAST-CENTER RESULTS")
print(
    fixed_mast_df.groupby("target_type").agg(
        samples=("sample_id", "nunique"),
        positive_samples=("difference", lambda x: (x > 0).sum()),
        median_difference=("difference", "median"),
        median_ratio=("ratio", "median")
    ).round(6)
)

print("\nDONOR-LEVEL RESULTS")
print(fixed_mast_summary.to_string(index=False))

fixed_mast_df.to_csv(
    os.path.join(output_dir, "analysis28J_fixed_mast_samples.csv"),
    index=False
)

fixed_mast_donors.to_csv(
    os.path.join(output_dir, "analysis28J_fixed_mast_donors.csv"),
    index=False
)

fixed_mast_summary.to_csv(
    os.path.join(output_dir, "analysis28J_fixed_mast_summary.csv"),
    index=False
)

print("\nANALYSIS 28J COMPLETE")
```

```python
import numpy as np, pandas as pd

base = '/content/drive/MyDrive/Neurothesis Resoure Library/research_outputs/'
don = pd.read_csv(base + 'analysis28J_fixed_mast_donors.csv')
rng = np.random.default_rng(2)
rows = []

for t, g in don.groupby('target_type'):
    for label, h in [('all', g), ('>=2 samples', g[g['samples'] >= 2])]:
        x = np.log(h['median_ratio'].values)
        obs = np.median(x)
        signs = rng.choice([-1, 1], size=(20000, len(x)))
        null = np.median(signs * x, axis=1)
        p = (np.sum(np.abs(null) >= abs(obs)) + 1) / (len(null) + 1)
        rows.append((t, label, len(x), np.exp(obs), p))

res = pd.DataFrame(rows, columns=['target', 'subset', 'n', 'median_ratio', 'perm_p'])
order = res['perm_p'].argsort().values
m = len(res)
holm = np.empty(m)
run = 0
for rank, i in enumerate(order):
    run = max(run, (m - rank) * res['perm_p'].iloc[i])
    holm[i] = min(1, run)
res['holm_p'] = holm
print(res.round(4).to_string(index=False))
```

```python
import numpy as np, pandas as pd
from scipy.stats import wilcoxon

base = '/content/drive/MyDrive/Neurothesis Resoure Library/'
d = pd.read_csv(base + 'Mast_Degree_Stratified_Permutation_Donors.csv')
loo = pd.read_csv(base + 'Mast_Degree_Stratified_Leave_One_Donor_Out.csv')
rng = np.random.default_rng(3)

for pop, g in d.groupby('population'):
    r = g['enrichment_ratio'].values
    b = [np.median(rng.choice(r, len(r))) for _ in range(10000)]
    print(pop, '| n', len(r), '| median', round(np.median(r), 4),
          '| 95% CI', np.round(np.percentile(b, [2.5, 97.5]), 4),
          '| min', round(r.min(), 3), '| max', round(r.max(), 3))

for pop, g in loo.groupby('population'):
    print(pop, '| LOO median range', round(g['median_enrichment'].min(), 4),
          round(g['median_enrichment'].max(), 4),
          '| all remaining above 1 in every exclusion:', g['all_remaining_above_1'].all())

w = d.pivot(index='donor', columns='population', values='enrichment_ratio')
diff = w['Perivasc Fib I'] - w['Perivasc Fib II']
print('Fib I minus Fib II | median', round(diff.median(), 4),
      '| Fib I higher in', int((diff > 0).sum()), 'of', len(diff),
      '| Wilcoxon two-sided p', round(wilcoxon(diff).pvalue, 4))
```

```python
import numpy as np, pandas as pd

base = '/content/drive/MyDrive/Neurothesis Resoure Library/'
a = pd.read_csv(base + 'Mast_Fibroblast_Abundance_Permutation_Samples.csv')
b = pd.read_csv(base + 'Within_Dermal_Degree_Stratified_Sample_Results.csv')

m = a.merge(b, on=['sample_code', 'population'], suffixes=('_abund', '_degree'))
print('matched rows:', len(m))
print('observed fractions identical:',
      np.allclose(m['observed_neighbor_fraction_abund'], m['observed_neighbor_fraction_degree']))

m['null_ratio'] = m['null_mean'] / m['null_mean_neighbor_fraction']
print(m.groupby('population')['null_ratio'].describe().round(3).to_string())

sd = a.pivot(index='sample_code', columns='population', values='null_sd')
print('abundance null_sd identical across Fib I and II in',
      int(np.isclose(sd.iloc[:, 0], sd.iloc[:, 1]).sum()), 'of', len(sd), 'samples')
sd2 = b.pivot(index='sample_code', columns='population', values='null_sd')
print('degree null_sd identical in',
      int(np.isclose(sd2.iloc[:, 0], sd2.iloc[:, 1]).sum()), 'of', len(sd2), 'samples')

for f in ['Dermal_Sample_Coverage_Summary.csv', 'Dermal_Three_Population_Counts_By_Sample.csv']:
    df = pd.read_csv(base + f)
    print('=====', f, df.shape)
    print(list(df.columns))
    print(df.head(4).to_string())
```

```python
import numpy as np, pandas as pd
import matplotlib.pyplot as plt

base = '/content/drive/MyDrive/Neurothesis Resoure Library/'
f = 'Within_Dermal_Degree_Stratified_Donor_Results.csv'
df = pd.read_csv(base + f)
print(df.shape); print(list(df.columns)); print(df.head(4).to_string())

pop_col = next(c for c in df.columns if 'population' in c.lower())
ratio_col = next(c for c in df.columns if 'ratio' in c.lower() or 'enrich' in c.lower())
print('using', pop_col, ratio_col)

rng = np.random.default_rng(5)
fig, ax = plt.subplots(figsize=(7, 4))
pops = ['Perivasc Fib I', 'Perivasc Fib II']
for i, p in enumerate(pops):
    r = df.loc[df[pop_col] == p, ratio_col].astype(float).values
    r = r[r > 0]
    y = i + rng.uniform(-0.18, 0.18, len(r))
    ax.scatter(r, y, s=28, alpha=0.7)
    med = np.median(r)
    b = [np.median(rng.choice(r, len(r))) for _ in range(10000)]
    lo, hi = np.percentile(b, [2.5, 97.5])
    ax.plot([lo, hi], [i, i], lw=3, color='black')
    ax.plot([med], [i], 'D', color='black')
    print(p, '| n', len(r), '| median', round(med, 3), '| 95% CI', round(lo, 3), round(hi, 3),
          '| above 1:', int((r > 1).sum()))
ax.axvline(1, ls='--', color='gray')
ax.set_xscale('log')
ax.set_yticks(range(len(pops))); ax.set_yticklabels(pops)
ax.set_xlabel('Enrichment of fibroblast neighbors around mast cells (ratio to null, log scale)')
ax.set_title('Donor-level enrichment, degree-stratified null')
plt.tight_layout()
plt.savefig(base + 'Figure1_Donor_Level_Enrichment.png', dpi=300)
plt.show()
```

```python
import numpy as np, pandas as pd
import matplotlib.pyplot as plt

base = '/content/drive/MyDrive/Neurothesis Resoure Library/'
d = pd.read_csv(base + 'Within_Dermal_Degree_Stratified_Donor_Results.csv')
fin = pd.read_csv(base + 'Final_Donor_Enrichment_Summary.csv')
fm = pd.read_csv(base + 'Fixed_Mast_Random_Labeling_Null.csv')

pops = ['Perivasc Fib I', 'Perivasc Fib II']
colors = {'Perivasc Fib I': '#1f77b4', 'Perivasc Fib II': '#d95f02'}
rng = np.random.default_rng(5)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4.8),
                               gridspec_kw={'width_ratios': [1.25, 1]})

# Panel A: donor-level ratios, dot size = samples per donor
for i, p in enumerate(pops):
    g = d[d['population'] == p]
    r = g['enrichment_ratio'].values.astype(float)
    s = g['sample_count'].values
    y = i + rng.uniform(-0.2, 0.2, len(r))
    ax1.scatter(r, y, s=25 + 14 * s, alpha=0.6, color=colors[p], edgecolor='white')
    med = np.median(r)
    b = [np.median(rng.choice(r, len(r))) for _ in range(10000)]
    lo, hi = np.percentile(b, [2.5, 97.5])
    ax1.plot([lo, hi], [i, i], color='black', lw=3)
    ax1.plot([med], [i], 'D', color='black')
    ax1.text(3.3, i, f'median {med:.2f}\n95% CI {lo:.2f} to {hi:.2f}\n'
                      f'{int((r > 1).sum())} of {len(r)} donors > 1', va='center', fontsize=8)
ax1.axvline(1, ls='--', color='gray')
ax1.set_xscale('log')
ax1.set_xlim(0.7, 6)
ax1.set_xticks([0.8, 1, 1.5, 2, 3])
ax1.set_xticklabels(['0.8', '1', '1.5', '2', '3'])
ax1.set_yticks(range(len(pops))); ax1.set_yticklabels(pops)
ax1.invert_yaxis()
ax1.set_xlabel('Fibroblast neighbors around mast cells (ratio to null, log scale)')
ax1.set_title('A. Donor-level enrichment (dot size = samples per donor)', fontsize=10)

# Panel B: medians across verified analyses
rows = []
for p in pops:
    for _, r in fin[fin['population'] == p].iterrows():
        m = int(r['minimum_target_cells'])
        lab = f"{r['method']} ({'all samples' if m == 0 else '>=' + str(m) + ' target cells'})"
        rows.append((lab, p, r['median_donor_enrichment']))
    rr = fm[fm['target'] == p]['observed_to_null_ratio'].iloc[0]
    rows.append(('Fixed-mast random labeling', p, rr))
order = list(dict.fromkeys(r[0] for r in rows))
for lab, p, v in rows:
    off = 0.12 if p == 'Perivasc Fib II' else -0.12
    ax2.scatter(v, order.index(lab) + off, color=colors[p], s=50, label=p)
h, l = ax2.get_legend_handles_labels()
ax2.legend(dict(zip(l, h)).values(), dict(zip(l, h)).keys(), fontsize=8, loc='lower right')
ax2.axvline(1, ls='--', color='gray')
ax2.set_yticks(range(len(order))); ax2.set_yticklabels(order, fontsize=8)
ax2.invert_yaxis()
ax2.set_xlim(0.8, 2.0)
ax2.set_xlabel('Median enrichment ratio')
ax2.set_title('B. Medians across verified analyses', fontsize=10)

plt.tight_layout()
plt.savefig(base + 'Figure1_Main_Result_Two_Panels.png', dpi=300)
plt.show()

print('Caption draft: Mast-cell neighborhoods in public MERFISH human skin data (22 donors) '
      'contain more Perivasc Fib I and Perivasc Fib II cells than a degree-stratified null predicts. '
      'Panel A shows donor medians with bootstrap 95% CIs; panel B shows medians across analyses. '
      'This describes spatial proximity, not interaction.')
```

```python
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import NullFormatter

base = '/content/drive/MyDrive/Neurothesis Resoure Library/'
s = pd.read_csv(base + 'Vascular_Adjusted_Anatomical_Site_Summary.csv')
print(s.sort_values(['population', 'median_enrichment']).to_string(index=False))

colors = {'Perivasc Fib I': '#1f77b4', 'Perivasc Fib II': '#d95f02'}
order = s.groupby('anatomic_site')['median_enrichment'].median().sort_values().index.tolist()

fig, ax = plt.subplots(figsize=(9, 6.5))
for p, off in [('Perivasc Fib I', -0.15), ('Perivasc Fib II', 0.15)]:
    g = s[s['population'] == p].set_index('anatomic_site')
    for site in order:
        if site not in g.index:
            continue
        r = g.loc[site]
        y = order.index(site) + off
        small = r['donor_count'] < 3
        ax.plot([r['q25'], r['q75']], [y, y], color=colors[p], lw=2, alpha=0.7)
        ax.scatter(r['median_enrichment'], y, s=20 + 10 * r['donor_count'],
                   color='white' if small else colors[p], edgecolor=colors[p], zorder=3)
        ax.text(4.3, y, f"{int(r['donors_above_1'])}/{int(r['donor_count'])}",
                fontsize=7, va='center', color=colors[p])
ax.axvline(1, ls='--', color='gray')
ax.set_xscale('log')
ax.set_xlim(0.5, 5)
ax.set_xticks([0.5, 0.8, 1, 1.5, 2, 3])
ax.set_xticklabels(['0.5', '0.8', '1', '1.5', '2', '3'])
ax.xaxis.set_minor_formatter(NullFormatter())
ax.set_yticks(range(len(order))); ax.set_yticklabels(order, fontsize=8)
ax.invert_yaxis()
ax.set_xlabel('Median donor enrichment ratio (bars show IQR; labels show donors above 1 / donors)')
ax.set_title('Enrichment by anatomical site, vascular-adjusted (open circles: fewer than 3 donors)',
             fontsize=10)
for p in colors:
    ax.scatter([], [], color=colors[p], label=p)
ax.legend(fontsize=8, loc='lower right')
plt.tight_layout()
plt.savefig(base + 'Figure2_Anatomical_Site_Consistency.png', dpi=300)
plt.show()
```

```python
import numpy as np, pandas as pd
import matplotlib.pyplot as plt

base = '/content/drive/MyDrive/Neurothesis Resoure Library/'
pairs = pd.read_csv(base + 'Three_Population_Pair_Block_Permutation.csv')
don = pd.read_csv(base + 'Schwann_ReticFibIII_Donor_Permutations_BH_Corrected.csv')
don = don.sort_values('observed_excess_pct').reset_index(drop=True)

sig = don['BH_significant_0.05']
print('BH-significant donors:', int(sig.sum()), 'of', len(don),
      '| donors with positive excess:', int((don['observed_excess_pct'] > 0).sum()))

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12.5, 5),
                               gridspec_kw={'width_ratios': [0.85, 1.15]})

cols = ['#9e9e9e' if v < 0 else '#2a9d8f' for v in pairs['observed_excess_pct']]
ax1.barh(pairs['population_pair'], pairs['observed_excess_pct'], color=cols)
ax1.axvline(0, color='black')
ax1.invert_yaxis()
for i, v in enumerate(pairs['observed_excess_pct']):
    ax1.text(v + (2 if v > 0 else -2), i, f'{v:.0f}%', va='center',
             ha='left' if v > 0 else 'right')
ax1.set_xlim(-95, 45)
ax1.set_xlabel('Observed edges vs block-permutation null (% excess)')
ax1.set_title('A. Pairwise spatial association, pooled', fontsize=10)

colors = np.where(sig, '#2a9d8f', '#bbbbbb')
ax2.barh(range(len(don)), don['observed_excess_pct'], color=colors)
ax2.set_yticks(range(len(don)))
ax2.set_yticklabels(don['donor_id'], fontsize=7)
ax2.axvline(0, color='black')
ax2.set_xlabel('Schwann to Retic Fib III edges vs null (% excess)')
ax2.set_title('B. Schwann to Retic Fib III by donor (teal: BH q < 0.05)', fontsize=10)

plt.tight_layout()
plt.savefig(base + 'Figure3_Pairwise_Association.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import numpy as np
from scipy.spatial import cKDTree

rng = np.random.default_rng(7)
N = 4000
xy = rng.random((N, 2))

# "vessels": 5 random line segments, sampled as points
segs = []
for _ in range(5):
    a, b = rng.random(2), rng.random(2)
    t = np.linspace(0, 1, 300)[:, None]
    segs.append(a + (b - a) * t)
vp = np.vstack(segs)
d = cKDTree(vp).query(xy)[0]          # distance to nearest vessel

# each type depends ONLY on vessel distance; no direct interaction
p = 0.02 + 0.30 * np.exp(-d / 0.04)
u = rng.random(N)
lab = np.zeros(N, dtype=int)          # 0 = other
lab[u < p] = 1                        # type A (mast-like)
lab[(u >= p) & (u < 2 * p)] = 2       # type B (fibroblast-like)

# neighbor graph: 6 nearest neighbors
_, idx = cKDTree(xy).query(xy, k=7)
nb = idx[:, 1:]

def frac(labels):
    A = labels == 1
    return (labels[nb[A]] == 2).mean()

obs = frac(lab)

# null 1: global label shuffle
null_g = np.array([frac(rng.permutation(lab)) for _ in range(500)])

# null 2: shuffle within 10 bands of vessel distance
edges = np.quantile(d, np.linspace(0, 1, 11))
band = np.clip(np.digitize(d, edges[1:-1]), 0, 9)
def strat_perm():
    out = lab.copy()
    for b in range(10):
        m = np.where(band == b)[0]
        out[m] = rng.permutation(lab[m])
    return out
null_s = np.array([frac(strat_perm()) for _ in range(500)])

print("counts A, B:", (lab == 1).sum(), (lab == 2).sum())
print("observed A->B neighbor fraction:", round(obs, 4))
print("global null mean, ratio, p:", round(null_g.mean(), 4),
      round(obs / null_g.mean(), 2), (null_g >= obs).mean())
print("stratified null mean, ratio, p:", round(null_s.mean(), 4),
      round(obs / null_s.mean(), 2), (null_s >= obs).mean())
```

```python
import numpy as np
from scipy.spatial import cKDTree

def make(rng, N=3000, attract=0.0):
    xy = rng.random((N, 2))
    segs = []
    for _ in range(5):
        a, b = rng.random(2), rng.random(2)
        t = np.linspace(0, 1, 300)[:, None]
        segs.append(a + (b - a) * t)
    d = cKDTree(np.vstack(segs)).query(xy)[0]
    p = 0.02 + 0.30 * np.exp(-d / 0.04)
    u = rng.random(N)
    lab = np.zeros(N, dtype=int)
    lab[u < p] = 1
    lab[(u >= p) & (u < 2 * p)] = 2
    nb = cKDTree(xy).query(xy, k=7)[1][:, 1:]
    if attract > 0:   # true interaction: nearest neighbor of some A cells becomes B
        for i in np.where(lab == 1)[0]:
            j = nb[i, 0]
            if lab[j] == 0 and rng.random() < attract:
                lab[j] = 2
    return d, lab, nb

def run(seed, attract, n_perm=99):
    rng = np.random.default_rng(seed)
    d, lab, nb = make(rng, attract=attract)
    A = lab == 1
    f = lambda L: (L[nb[A]] == 2).mean()
    obs = f(lab)
    ng = np.array([f(rng.permutation(lab)) for _ in range(n_perm)])
    edges = np.quantile(d, np.linspace(0, 1, 11))
    band = np.clip(np.digitize(d, edges[1:-1]), 0, 9)
    idxs = [np.where(band == b)[0] for b in range(10)]
    def sp():
        o = lab.copy()
        for m in idxs:
            o[m] = rng.permutation(lab[m])
        return o
    ns = np.array([f(sp()) for _ in range(n_perm)])
    pg = (1 + (ng >= obs).sum()) / (n_perm + 1)
    ps = (1 + (ns >= obs).sum()) / (n_perm + 1)
    return pg, ps

for attract in [0.0, 0.3]:
    res = np.array([run(s, attract) for s in range(100)])
    print(f"true attraction={attract}: "
          f"global null calls enrichment in {(res[:,0] < 0.05).mean():.0%} of 100 tissues, "
          f"stratified null in {(res[:,1] < 0.05).mean():.0%}")
```

```python
def run_noisy(seed, noise, n_perm=99):
    rng = np.random.default_rng(seed)
    d, lab, nb = make(rng, attract=0.0)
    d_obs = d + rng.normal(0, noise, len(d))      # imperfect proxy of the true confounder
    A = lab == 1
    f = lambda L: (L[nb[A]] == 2).mean()
    obs = f(lab)
    edges = np.quantile(d_obs, np.linspace(0, 1, 11))
    band = np.clip(np.digitize(d_obs, edges[1:-1]), 0, 9)
    idxs = [np.where(band == b)[0] for b in range(10)]
    def sp():
        o = lab.copy()
        for m in idxs:
            o[m] = rng.permutation(lab[m])
        return o
    ns = np.array([f(sp()) for _ in range(n_perm)])
    return (1 + (ns >= obs).sum()) / (n_perm + 1)

for noise in [0.0, 0.02, 0.05, 0.1, 0.3]:
    ps = np.array([run_noisy(s, noise) for s in range(100)])
    print(f"proxy noise sd={noise}: false-positive rate {(ps < 0.05).mean():.0%}")
```

```python
def make2(rng, amp, N=3000):
    xy = rng.random((N, 2))
    segs = []
    for _ in range(5):
        a, b = rng.random(2), rng.random(2)
        t = np.linspace(0, 1, 300)[:, None]
        segs.append(a + (b - a) * t)
    d = cKDTree(np.vstack(segs)).query(xy)[0]
    p = 0.02 + amp * np.exp(-d / 0.04)
    u = rng.random(N)
    lab = np.zeros(N, dtype=int)
    lab[u < p] = 1
    lab[(u >= p) & (u < 2 * p)] = 2
    nb = cKDTree(xy).query(xy, k=7)[1][:, 1:]
    return d, lab, nb

def run_amp(seed, amp, noise, n_perm=99):
    rng = np.random.default_rng(seed)
    d, lab, nb = make2(rng, amp)
    d_obs = d + rng.normal(0, noise, len(d))
    A = lab == 1
    f = lambda L: (L[nb[A]] == 2).mean()
    obs = f(lab)
    edges = np.quantile(d_obs, np.linspace(0, 1, 11))
    band = np.clip(np.digitize(d_obs, edges[1:-1]), 0, 9)
    idxs = [np.where(band == b)[0] for b in range(10)]
    def sp():
        o = lab.copy()
        for m in idxs:
            o[m] = rng.permutation(lab[m])
        return o
    ns = np.array([f(sp()) for _ in range(n_perm)])
    return (1 + (ns >= obs).sum()) / (n_perm + 1)

print("false-positive rate (no true interaction), by confounder strength and proxy noise")
for amp in [0.05, 0.10, 0.20, 0.30]:
    row = []
    for noise in [0.0, 0.02, 0.05]:
        ps = np.array([run_amp(s, amp, noise) for s in range(100)])
        row.append(f"noise {noise}: {(ps < 0.05).mean():.0%}")
    print(f"amp {amp}: " + " | ".join(row))
```

```python
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import NullFormatter

OUT = "/content/drive/MyDrive/Neurothesis Resoure Library/research_outputs/"

# Verified results copied from the outputs of cells 2 and 4
panel_a = pd.DataFrame({
    "test": ["Global shuffle", "Covariate-aware shuffle", "Covariate-aware shuffle"],
    "scenario": ["No true interaction", "No true interaction", "True attraction built in"],
    "pct_tissues_called_enriched": [100, 7, 93],
})
noise = [0.0, 0.02, 0.05]
fp = {0.05: [0, 2, 3], 0.10: [5, 7, 20], 0.20: [7, 21, 57], 0.30: [7, 27, 94]}
panel_b = pd.DataFrame([(a, n, v) for a, vals in fp.items() for n, v in zip(noise, vals)],
                       columns=["confounder_strength", "proxy_noise_sd", "false_positive_pct"])
panel_a.to_csv(OUT + "Simulation_Panel_A_Summary.csv", index=False)
panel_b.to_csv(OUT + "Simulation_Panel_B_FalsePositive_vs_ProxyNoise.csv", index=False)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5.2), gridspec_kw={"width_ratios": [1, 1.25]})

# Panel A
labels = ["Global\nshuffle", "Covariate-aware\nshuffle"]
ax1.bar([0, 1], [100, 7], color=["#9a9a9a", "#2c6fb0"], width=0.55)
ax1.bar([1.45], [93], color="#2c6fb0", width=0.55, hatch="//", edgecolor="white")
for x, v in [(0, 100), (1, 7), (1.45, 93)]:
    ax1.text(x, v + 2, f"{v}%", ha="center", fontsize=11)
ax1.set_xticks([0, 1, 1.45])
ax1.set_xticklabels(["Global\nshuffle", "Covariate-aware\nshuffle", "Covariate-aware\nshuffle"], fontsize=9)
ax1.text(0.5, -0.24, "No true interaction\n(false positives, lower is better)", ha="center", transform=ax1.get_xaxis_transform(), fontsize=9)
ax1.text(1.45, -0.24, "True attraction\n(detection, higher is better)", ha="center", transform=ax1.get_xaxis_transform(), fontsize=9)
ax1.set_ylim(0, 112)
ax1.set_ylabel("Share of 100 simulated tissues called enriched (%)")
ax1.set_title("A. Only the covariate-aware test is fair", loc="left", fontsize=12)

# Panel B
shades = {0.05: "#a9c7e6", 0.10: "#6fa3d1", 0.20: "#2c6fb0", 0.30: "#123a63"}
for a, vals in fp.items():
    ax2.plot(noise, vals, marker="o", markersize=8, linewidth=2, color=shades[a])
    ax2.text(0.0515, vals[-1], f"strength {a:.2f}", va="center", fontsize=9, color="#333333")
ax2.axhline(5, linestyle="--", color="#666666", linewidth=1)
ax2.text(0.0, 7.5, "5% expected for a fair test", fontsize=8.5, color="#555555")
ax2.set_xlim(-0.003, 0.075)
ax2.set_ylim(-2, 100)
ax2.set_xticks(noise)
ax2.xaxis.set_minor_formatter(NullFormatter())
ax2.set_xlabel("Error in the stand-in for the hidden structure (simulation units)")
ax2.set_ylabel("False positives, no true interaction (%)")
ax2.set_title("B. A noisy stand-in breaks the fair test", loc="left", fontsize=12)

for ax in (ax1, ax2):
    ax.spines[["top", "right"]].set_visible(False)
fig.suptitle("Null models can create false spatial enrichment (simulation, 100 tissues per cell)", fontsize=13, y=1.0)
plt.tight_layout()
plt.savefig(OUT + "Simulation_Figure.png", dpi=300, bbox_inches="tight")
plt.show()
print("saved figure and two CSVs")
```

```python
import numpy as np
import matplotlib.pyplot as plt

OUT = "/content/drive/MyDrive/Neurothesis Resoure Library/research_outputs/"

noise = [0.0, 0.02, 0.05]
fp = {0.05: [0, 2, 3], 0.10: [5, 7, 20], 0.20: [7, 21, 57], 0.30: [7, 27, 94]}

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12.5, 5.4), gridspec_kw={"width_ratios": [1, 1.25]})

# Panel A
xs = [0, 2.6, 3.6]
vals = [100, 7, 93]
ax1.bar(xs[0], vals[0], color="#9a9a9a", width=0.8)
ax1.bar(xs[1], vals[1], color="#2c6fb0", width=0.8)
ax1.bar(xs[2], vals[2], color="#2c6fb0", width=0.8, hatch="//", edgecolor="white")
for x, v in zip(xs, vals):
    ax1.text(x, v + 2, f"{v}%", ha="center", fontsize=11)
ax1.set_xticks(xs)
ax1.set_xticklabels(["Global\nshuffle", "Covariate-\naware", "Covariate-\naware"], fontsize=9.5)
ax1.text(1.3, -0.2, "No true interaction\nlower is better (false positives)",
         ha="center", transform=ax1.get_xaxis_transform(), fontsize=9)
ax1.text(3.6, -0.2, "True attraction built in\nhigher is better (detection)",
         ha="center", transform=ax1.get_xaxis_transform(), fontsize=9)
ax1.set_xlim(-0.7, 4.4)
ax1.set_ylim(0, 112)
ax1.set_ylabel("Share of 100 simulated tissues called enriched (%)")
ax1.set_title("A. Global shuffle fails; covariate-aware works", loc="left", fontsize=12)

# Panel B
shades = {0.05: "#a9c7e6", 0.10: "#6fa3d1", 0.20: "#2c6fb0", 0.30: "#123a63"}
for a, v in fp.items():
    ax2.plot(noise, v, marker="o", markersize=8, linewidth=2, color=shades[a])
    ax2.text(0.0515, v[-1], f"strength {a:.2f}", va="center", fontsize=9, color="#333333")
ax2.axhline(5, linestyle="--", color="#666666", linewidth=1,
            label="5% false positives expected from a fair test")
ax2.legend(loc="upper left", frameon=False, fontsize=9)
ax2.set_xlim(-0.003, 0.075)
ax2.set_ylim(-2, 100)
ax2.set_xticks(noise)
ax2.set_xlabel("Error in the stand-in for the hidden structure (simulation units)")
ax2.set_ylabel("False positives, no true interaction (%)")
ax2.set_title("B. Noise hurts most when preference is strong", loc="left", fontsize=12)

for ax in (ax1, ax2):
    ax.spines[["top", "right"]].set_visible(False)
fig.suptitle("Null models can create false spatial enrichment (simulation, 100 tissues per cell)",
             fontsize=13, y=1.0)
plt.tight_layout()
plt.savefig(OUT + "Simulation_Figure.png", dpi=300, bbox_inches="tight")
plt.show()
print("saved:", OUT + "Simulation_Figure.png")
```
